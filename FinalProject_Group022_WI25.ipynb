# Abstract

Our project explores the relationship between specific physical characteristics of dogs and the length of their stay at animal shelters in California. By analyzing data from shelters in Orange County, San Jose, and Sonoma, we aimed to identify traits that significantly influence how long dogs remain in shelters before being adopted. Our primary focus was on breed, color, and sex, and age. The goal of the research was to better understand which traits are linked to longer or shorter stays, providing insights that could help shelters improve adoption strategies.

Our analysis revealed that breed was the most influential factor in determining a dog’s shelter stay. Larger breeds, such as pit bulls and husky breeds, had significantly longer stays, while smaller and more popular breeds like chihuahuas and yorkshire terriers were adopted more quickly. Similarly, color played a crucial role, with lighter-colored dogs typically being adopted faster than dogs with darker or mixed-colored coats. This suggests that adopter preferences for certain colors may influence the time a dog spends at the shelter. Additionally, sex was found to impact shelter stay, with male dogs staying, on average, longer than female dogs, likely due to factors such as behavior and neutering/spaying status. Age had a more moderate influence, with older dogs spending slightly longer times in the shelters than younger dogs.

These findings suggest that physical characteristics, including breed, color, sex, and age significantly affect the length of stay in animal shelters. Shelters can use this information to focus their promotional efforts on dogs with longer shelter stays, particularly larger or darker-colored breeds, in order to increase adoption rates and reduce overall shelter time. By targeting specific dog characteristics that influence adoption speed, shelters may improve their operations, reduce euthanasia rates, and enhance the overall effectiveness of adoption campaigns.

Studies have been conducted studying how dog characteristics can predict the adaptability of said dog. A related study is ‘Phenotypic Characteristics Associated with Shelter Dog Adoption in the United States’ <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) that was conducted in 2017 in which records from 17 different shelters were used to determine if phenotypic characteristics were associated with higher rates of adoption. Their findings included adopters’ strong preference for puppies but a surprisingly higher chance of adoption for dogs 8 years or older in comparison those ages 5-7 years. The study also found that small to medium sized dogs had higher adoption rates but the study acknowledges this may be due to size restrictions based on owners’ living situations. Interestingly skull type was also analyzed, finding that dogs with a blockhead had lower odds of live release and that brachycephalic dogs are always less adoptable than mesocephalic dogs. This is attributed to the breed of brachycephalic dogs which include pitbulls, bulldogs, and rottweilers. The study determined that phenotypic characteristics do in fact have effects on adoption rates but that there may be modification based on the length of the dog’s stay at the shelter.

Additional studies were done on how certain dog characteristics are associated with successful adoption. In a study done on ‘Determining factors for successful adoption of dogs from an animal shelter’ <a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2), various factors were analyzed to help determine the likelihood of adoption for dogs in shelters. The study focused on a humane society in Michigan, where around 1,500 dogs were considered and present in the database at the time. They found that smaller dogs were significantly more likely to be adopted than larger dogs. They attributed this to several factors such as limited space, stigma around larger dogs, and increased popularity of smaller dogs for urban lifestyles. Breeds related to the toy, terrier, hound, or non-sporting groups were also more likely to be adopted, due to their shared characteristics of being small and physical attractiveness. Lighter colored coats (such as white, gray, gold), house-training, and reduced behavioral issues were also associated with higher adoptability. The study suggests that it cannot be generalized and is meant to help shelters determine and adjust which dogs to offer at adoption.

## Dataset #1: Orange County

```python
df = pd.DataFrame(pd.read_excel("orange_county.xlsx"))
df.head()
```

To start off our cleaning, we uploaded our file containing all the data and dropped all columns that were unnecessary to our analysis. Because we are only analyzing the last four years, we also dropped the data from before 2021.

```python
oc_clean = df.drop(columns=['impound_no', 'litter_no', 'kennel_no','intake_tot', 'kennel_stat', 'Current Status',
   'animal_id', 'blue_tag', 'activity_no', 'activity_seq','crossing', 'jurisdiction', 'intake_type', 'intake_subtype',
    'intake_cond', 'owner_source', 'os_reason', 'intake_by_id', 'intake_time', 'due_out', 'review_date',
    'hold_notify', 'outcome_request', 'outcome_subtype', 'dose', 'dose2', 'bottle_no', 'tnr_program', 
    'zip_code','outcome_by_id', 'outcome_time', 'outcome_cond', 'price','intake_receipt_no', 'outcome_receipt_no',
    'location', 'userid','stamp', 'animal_name', 'secondary_breed','secondary_color', 'animal_size', 
    'coat', 'ears', 'nose', 'tail','collar_color', 'collar_type', 'level', 'temperament', 'os_network', 'years_old', 
    'months_old']).reset_index(drop=True)

oc_clean = oc_clean[oc_clean['intake_date'].dt.year > 2020].reset_index(drop=True)

oc_clean.head()
```

We kept the variables 'Intake Date', 'Discharge Condition', 'Discharge Date', 'Sex', 'DOB', 'Animal Type', 'Breed', and 'Color'. We renamed the columns in the dataset to maintain consistency.

```python
oc_clean.rename(columns={'intake_date': 'Intake Date', 'outcome_type': 'Discharge Condition', 
                        'outcome_date': 'Discharge Date', 'sex': 'Sex', 'dob': 'DOB', 
                        'animal_type': 'Animal Type', 'primary_breed': 'Breed',
                        'primary_color': 'Color'}, inplace=True)

oc_clean.head()
```

We then calculated the stay length of each animal by subtracting their discharge date from their intake date. We added the days that each animal stayed in the shelter as a new column called 'Stay Length'.

```python
oc_clean['Intake Date'] = pd.to_datetime(oc_clean['Intake Date'], yearfirst = True)
oc_clean['Discharge Date'] = pd.to_datetime(oc_clean['Discharge Date'], yearfirst = True)

oc_clean['Stay Length'] = (oc_clean['Discharge Date'] - oc_clean['Intake Date']).dt.days
```

We then dropped all data for animals that were not dogs, as we are only analyzing the stay length of dogs in shelters. After filtering out other animals, we dropped the 'Animal Type' column.

```python
oc_clean = oc_clean[oc_clean['Animal Type'].str.contains('DOG', na=False)]
del oc_clean['Animal Type']
```

To calculate age at discharge, we subtracted the discharge date from the date of birth and added this as a new column called 'Age at Discharge' with the age in years.

```python
oc_clean['Age at Discharge'] = (oc_clean['Discharge Date'] - oc_clean['DOB']).dt.days / 365
```

The next step was to standardize the 'Sex' column. Using value_counts(), we could see there were males, females, neutered, spayed, and unknowns. We converted neutered and spayed to be male or female and standardized the string to be 'M' or 'F'. Unknowns were turned into NaNs. We applied our standardizing function to the 'Sex' column.

```python
oc_clean['Sex'].value_counts()
```

```python
def standardize_sex(sex):
    if sex == 'N' or sex == 'M':
        output = 'M'
    elif sex == 'S' or sex == 'F':
        output = 'F'
    else:
        output = np.nan
    return output

oc_clean['Sex'] = oc_clean['Sex'].apply(standardize_sex)
```

Because we are interested in analyzing stay length in relation to characteristics, adoption is the only discharge outcome that we kept because it is dependent on a dog's physical characteristics. Other discharge outcomes such as euthanization or transfers are dependent on outside factors.

```python
oc_clean = oc_clean[oc_clean['Discharge Condition'].str.contains('ADOPTION', na = False)]
oc_clean['Discharge Condition'] = oc_clean['Discharge Condition'].replace({"ADOPTION": "ADOPT" })
```

Our next column to standardize was the colors of the dogs. We first used value_counts() to look at all the different colors. We categorized the colors into eight different color categories in order to have more cohesive data. We applied our function to the 'Color' column so that the dogs were categorized into their predominant color.

```python
oc_clean['Color'].value_counts()
```

```python
def standardize_color(color):
    if 'BLACK' in color or 'BL BRINDLE' in color:
        output = 'BLACK'
    elif 'WHITE' in color or 'CREAM' in color:
        output = 'WHITE'
    elif 'BROWN' in color or 'CHOCOLATE' in color or 'BR BRINDLE' in color or 'BRN MERLE' in color or 'SABLE' in color:
        output = 'BROWN'
    elif 'RED' in color or "LIVER" in color:
        output = 'RED'
    elif 'BLUE' in color:
        output = 'BLUE'
    elif 'SILVER' in color or 'GRAY' in color:
        output = 'GRAY'
    elif 'TAN' in color or 'FAWN' in color or 'CREAM' in color or 'GOLD' in color or 'BLONDE' in color or 'GOLD' in color or 'BUFF' in color:
        output = 'TAN'
    elif '/' in color or 'TRICOLOR' in color:
        output = 'MIX'
    else:
        output = np.nan
    return output

oc_clean.loc[:, 'Color'] = oc_clean['Color'].apply(standardize_color)
```

Finally, we added a column 'County' so that after combining, we can determine which dataset each dog came from originally. We then exported our cleaned Orange County dataset as a csv to then be used for our combined dataset.

```python
oc_clean['County'] = 'ORANGE COUNTY'
oc_clean.to_csv('orangecounty_clean.csv')
oc_clean.head()
```

## Dataset #2: San Jose County

For our second dataset, we cleaned shelter data from San Jose County so that it would have the same columns and be standardized to match the Orange County dataset.

```python
#read in the San Jose County animal shelter data from each year and concat it into one DataFrame
twentyfourfive = pd.read_csv('data/24-25.csv')
twentythreefour = pd.read_csv('data/23-24.csv')
twentytwothree = pd.read_csv('data/22-23.csv')
twentyonetwo = pd.read_csv('data/21-22.csv')
sj_df = pd.concat([twentyfourfive, twentythreefour, twentytwothree, twentyonetwo])
sj_df.head()
```

```python
#drop unneeded columns
sj_clean = sj_df.drop(columns = ['_id', 'AnimalID', 'AnimalName', 'SecondaryColor', 'Age', 'IntakeType', 'IntakeCondition', 'IntakeSubtype', 'IntakeReason','OutcomeSubtype','OutcomeCondition', 'Crossing', 'Jurisdiction', 'LastUpdate'])

#drop all animals that aren't dogs
sj_clean = sj_clean[sj_clean['AnimalType'] == 'DOG']
sj_clean['AnimalType'].value_counts()

#reset the index and drop the column AnimalType now that all the animals are dogs
sj_clean = sj_clean.reset_index(drop=True)
sj_clean = sj_clean.drop(columns = 'AnimalType')

#rename the columns
sj_clean.columns = ['Color', 'Breed', 'Sex', 'DOB', 'Intake Date', 'Discharge Date', 'Discharge Condition']

#add a County column that shows that all of this data is from San Jose
sj_clean['County'] = 'SAN JOSE'

#change all columns with dates to pandas datetime
sj_clean['DOB'] = pd.to_datetime(sj_clean['DOB'], yearfirst = True)
sj_clean['Discharge Date'] = pd.to_datetime(sj_clean['Discharge Date'], yearfirst = True)
sj_clean['Intake Date'] = pd.to_datetime(sj_clean['Intake Date'], yearfirst = True)

#create a new column, Stay Length, that is discharge date - intake date
sj_clean['Stay Length'] = sj_clean['Discharge Date'] - sj_clean['Intake Date']
sj_clean['Stay Length'] = sj_clean['Stay Length'].dt.days

#drop all rows with Intake Dates from before 2021
sj_clean = sj_clean[sj_clean['Intake Date'].dt.year != 2021]

#create a new column, Age at Intake
sj_clean['Age at Discharge'] = (sj_clean['Discharge Date'] - sj_clean['DOB']).dt.days / 365

#keep only data from adopted dogs
sj_clean = sj_clean[sj_clean['Discharge Condition'].str.contains('ADOPTION', na = False)]

#sort colors into 8 main colors and apply to the Color column
color_sort = {
    'BLUE MERLE': 'BLUE',
    'TRI-LILIC': 'MIX',
    'BL BRINDLE': 'MIX',
    'TICK-YEL': 'TAN',
    'TICK-RD\LV': 'MIX',
    'BEIGE': 'TAN',
    'OTHER': 'MIX',
    'SHADE-BLUE': 'BLUE',
    'APRICOT': 'RED',
    'BR BRINDLE': 'MIX',
    'BUFF': 'TAN',
    'BRINDLE-LN': 'MIX',
    'SHADE-SLV': 'MIX',
    'ORANGE': 'RED',
    'UNKNOWN': 'MIX',
    'TICK-BLUE': 'BLUE',
    'TICK-BLACK': 'BLACK',
    'LIVER': 'RED',
    'MERLE-RED':'RED',
    'BRINDLE-TN': 'MIX',
    'CHOCOLATE':'BROWN',
    'MERLE-BLUE':'BLUE',
    'YELLOW':'TAN',
    'FAWN':'TAN',
    'SABLE':'TAN',
    'GOLD':'TAN',
    'CREAM':'WHITE',
    'TRICOLOR':'MIX',
    'BRINDLE-BN': 'MIX',
}
sj_clean['Color'] = sj_clean['Color'].replace(color_sort)

#drop the breed names with less than 10 dogs of that breed
breed_count = sj_clean['Breed'].value_counts()
keep_breed = breed_count[breed_count >= 10].index
sj_clean = sj_clean[sj_clean['Breed'].isin(keep_breed)]

#change spayed and neutered to female and male and apply in the Sex column 
def sterilized(animal):
    animal = animal.replace('SPAYED', 'F')
    animal = animal.replace('NEUTERED', 'M')
    animal = animal.replace('FEMALE', 'F')
    animal = animal.replace('MALE', 'M')
    return animal
sj_clean['Sex'] = sterilized(sj_clean['Sex'])

#drop NaNs
sj_clean = sj_clean.dropna(axis=0)
sj_clean = sj_clean.reset_index(drop = True)

sj_clean.to_csv('sanjose_clean.csv', index=True)
```

## Regression Results

```python
import statsmodels.api as sm
import pandas as pd

# Load your dataset
df = pd.read_csv("DF.csv")

# Encode categorical variables
df_encoded = df.copy()
df_encoded["Sex"] = df_encoded["Sex"].astype("category").cat.codes
df_encoded["Color"] = df_encoded["Color"].astype("category").cat.codes
df_encoded["Breed"] = df_encoded["Breed"].astype("category").cat.codes

# Define independent variables (features) and dependent variable (target)
X = df_encoded[['Age at Discharge', 'Sex', 'Color', 'Breed']]
y = df_encoded['Stay Length']

# Add constant for intercept
X = sm.add_constant(X)

# Fit multiple linear regression model
model = sm.OLS(y, X).fit()

# Display summary of the model
print(model.summary())
```

The OLS regression analysis reveals that age at discharge, sex, color, and breed all significantly influence the length of stay for dogs in animal shelters. Specifically, older dogs tend to stay longer, with an increase of 1.69 days for each additional year of age. Male dogs stay an average of 4.86 days longer than females. Lighter-colored dogs tend to be adopted more quickly, staying on average 1.37 days less than darker-colored dogs. Additionally, breed also plays a role, with certain breeds having longer stays than others. The R-squared value is low (0.018), which indicates that other unmeasured factors are contributing to the variation in shelter stay times aside from just the physical variables that were accounted for in this study. The low model fit shows that, while the physical attribute variables we measured in this study are significant, other important variables are missing that would better account for variation in dog shelter stay length.

# Ethics & Privacy

Our data will be relevant for Californian animal shelters to create targeted campaigns that increase awareness and positive public perception for certain types of dogs with increased shelter stay times. To address this, we will take animal shelter data from three major California counties (Sonoma County, San Jose County, Orange County). ## Discussison and Conclusion

Our project aimed to investigate the impact of physical characteristics of dogs, such as breed, age, sex, and coat color, on their length of stay in animal shelters in California. By leveraging a comprehensive dataset from shelters across multiple counties, our aim was to identify patterns that could guide shelters in reducing stay lengths and improving adoption rates. From our analysis, we were able to reject a null-hypothesis. Our findings support our hypothesis that younger dogs, while not the most significant trait, do have shorter shelter stays and that smaller breeds do spend less time in shleters. The findings from our analyses indicate that in addition to age and breed, color and sex are also influential, demonstrating that physical traits do indeed play a significant role in determining how long dogs remain in shelters.

While age appeared the least significant, our analysis still shows a relationship between age and the length of stay. The regression analysis revealed that older dogs tend to stay longer in shelters, with each additional year of age associated with an increase of approximately 1.69 days in stay length. This finding supports existing literature suggesting that older dogs face more significant challenges in finding adopters, who may be concerned about potential health issues or shorter life expectancy. This bias toward younger dogs emphasizes the need for shelters to increase efforts to promote the adoption of older dogs, possibly by highlighting their calmer demeanor and the advantages of adopting a dog with less energy. Shelters could tailor their messaging to encourage adopters to consider the benefits of giving an older dog a second chance.

Sex also emerged as a key predictor of stay length, with male dogs spending an average of 4.86 days longer in shelters than females. This is consistent with previous research suggesting a preference among adopters for female dogs. This difference could be due to perceived behavioral differences, with male dogs often being stereotyped as more difficult to manage or more aggressive. Shelters should consider addressing these biases, highlighting the positive attributes of male dogs through targeted campaigns and educational outreach to challenge the assumptions that may delay male dog adoptions.

Color and breed were the most significant factors. Lighter-colored dogs tend to stay shorter periods in shelters, likely due to aesthetic preferences of potential adopters. The negative coefficient for color (-1.37) suggests that darker-colored dogs, as well as mixed-breed dogs, tend to stay longer, possibly due to biases or misconceptions surrounding these characteristics. To combat this, shelters could initiate campaigns that celebrate the uniqueness of darker and mixed-color dogs, emphasizing their beauty and individual personality traits to attract more adopters. Additionally, the finding that certain breeds stay longer or shorter aligns with popular perceptions of breed desirability. Larger breeds, for example, tend to stay longer, likely due to concerns about space, maintenance, and the general perception that larger dogs are more difficult to manage. By promoting less-adopted breeds and addressing misconceptions, shelters may be able to shorten the stay times of these dogs.

Although the factors identified—age, sex, color, and breed—were statistically significant in predicting stay length, the low R-squared value (0.018) indicates that there are other important factors influencing shelter stay times that were not captured in this study. Even though the proportion of variance explained by these factors is relatively small, we acknowledge that this may be due to other external factors such as dog personality, health status and behavior, which is a limitation of our project. Because our project revolved around studying only physical characteristics, these results still provide valuable insights for shelter managers and adoption marketing campaigns hoping to reduce the amount of dogs at the shelter. 

In conclusion, this study provides a deeper understanding of how physical traits influence the length of time dogs spend in California animal shelters. While the low explanatory power of the model suggests that other factors are at play, the results lay the foundation for targeted marketing and publicity strategies that could improve adoption rates and reduce shelter overcrowding. Shelters can use these findings to design more effective adoption campaigns, focusing on promoting older dogs, addressing biases related to male dogs and darker-colored breeds, and fostering more awareness of the advantages of adopting larger or less popular breeds. With these targeted efforts, shelters can help increase adoption rates and shorten shelter stays, ultimately benefiting both dogs and potential adopters.
