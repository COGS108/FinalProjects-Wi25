## Background and Prior Work

## Introduction

As the internet rapidly expanded during the 21st century, online gambling has become increasingly accessible and popular among young people. In Portugal, where internet penetration is high, online gambling offers both entertainment opportunities and significant risks, such as addiction, financial distress, and emotional strain. Understanding the factors that influence online gambling behavior is crucial for policymakers, researchers, and mental health professionals. This study aims to explore the demographic, psychological, and behavioral characteristics most associated with online gambling among young people in Portugal, providing insights that can inform targeted interventions and policy measures.

## Previous Research on Online Gambling

Research on online gambling has grown substantially since the early 2000s, revealing various factors that contribute to gambling behaviors. Studies consistently highlight the roles of demographic attributes, psychological traits, and behavioral patterns, but a lack of region-specific data limits the effectiveness of prevention strategies.

## Summary of Previous Research

- **Montiel et al. (2021):** Conducted a systematic review of 658 studies on problematic online gambling among adolescents. The study identified a significant gender gap in gambling participation and emphasized the lack of consensus in diagnosing online gambling disorders, highlighting the need for further research into demographic and psychological influences. <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)
- **Calado and Griffiths (2016):** Investigated risk factors for online gambling in a European context. They found that young age, male gender, and impulsivity were consistent predictors of problem gambling, reinforcing the notion that these elements interact to drive gambling behaviors. <a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2)
- **Hing et al. (2018):** Explored the impact of socioeconomic status and personality traits on gambling habits. Their research demonstrated that lower socioeconomic status correlates with higher rates of problem gambling, while risk-taking personality traits amplify these issues. <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3)

By building on these studies, this research will offer a localized understanding of online gambling behaviors among young people in Portugal. The findings aim to support the development of evidence-based policies and targeted interventions to mitigate gambling-related harm.

### Demographic Factors and their relationship with Gambling

Naturally, Demographic Information likely contains important clues to why some people prefer to gamble more often then others. By exploring demographic information, we can get a deeper look into why people gamble.

```python
def convert_range_to_tuple(x):
    if isinstance(x, str):
        # Remove parentheses and split by comma
        cleaned = x.strip('()')
        # Handle empty strings (in case of a single element tuple)
        if cleaned[:2] == '41':
            return (41,)
        if cleaned:
            return tuple(map(int, cleaned.split(',')))
        else:
            return (int(cleaned),)  # Single value tuple (e.g., '(41,)')
    return x  # If it's NaN or any non-string, return it as-is
```

The stacked bar chart illustrates the distribution of lifetime online gambling participation across different levels of paternal education. Each bar represents a specific education category (e.g., "0-6 years," "7-12 years"), while the segments within the bars show the frequency of gambling occasions. This visualization helps identify whether higher parental education is associated with lower or higher gambling participation trends.

```python
data['GM_2_on'] = data['GM_2_on'].apply(convert_range_to_tuple)

education_labels = {
    1: '0-6 years',
    2: '7-12 years',
    3: '13-16 years',
    4: 'More than 17 years',
    5.0: np.nan  # Handle NaN values if necessary
}

data['SD_3a_FATHER'] = data['SD_3a_FATHER'].map(education_labels)
data['SD_3b_MOTHER'] = data['SD_3b_MOTHER'].map(education_labels)

mother_counts = data.groupby(['SD_3b_MOTHER', 'GM_2_on']).size().unstack(fill_value=0)
father_counts = data.groupby(['SD_3a_FATHER', 'GM_2_on']).size().unstack(fill_value=0)

mother_counts_prop = mother_counts.div(mother_counts.sum(axis=1), axis=0)
father_counts_prop = father_counts.div(father_counts.sum(axis=1), axis=0)

# Create figure
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Grouped Bar Plot - Mother's Education
mother_counts_prop.plot(kind='bar', stacked=True, ax=axes[0], colormap='coolwarm')
axes[0].set_title("Online Gambling Occasions by Mother's Education")
axes[0].set_xlabel("Mother's Education (Binned)")
axes[0].set_ylabel("Count of Individuals")
axes[0].legend(title="Gambling Occasions", bbox_to_anchor=(1.05, 1))

# Grouped Bar Plot - Father's Education
father_counts_prop.plot(kind='bar', stacked=True, ax=axes[1], colormap='coolwarm')
axes[1].set_title("Online Gambling Occasions by Father's Education")
axes[1].set_xlabel("Father's Education (Binned)")
axes[1].set_ylabel("Count of Individuals")
axes[1].legend(title="Gambling Occasions", bbox_to_anchor=(1.05, 1))

plt.tight_layout()
plt.show()
```

Here we can see that while a mother's education does not appear to have much correlation to how often someone gambles, father's education very much does. As father's education increases, the participants seem disproportionately more likely to gamble more often. Specifically, the 41+ times group increased as father's education increased.

## Machine Learning Analysis

Here, we wanted to see the qualities that were the greatest predictors of participating in online gambling. Based off of our exploratory data analysis, we were most interested in using gender, offline gambling history, father's education, and self-perceived luck as features. The last factor was added to incorporate a psychological feature in addition to demographic varaibles. 

We decided to choose a random forest classifier as we wanted a model that wouldn't overfit. Random forest classifiers are generally good at avoiding overfitting because of the use of ensemble learning and multiple decision trees leading to a model that typically generalizes better.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn.dummy import DummyClassifier
from sklearn.svm import SVC

#Initialize Random Forest Classifier
forest = RandomForestClassifier(n_estimators=100,criterion='log_loss')
```

Below, we prepared the data to train our machine learning model. We first converted all of the values in the self-perceived luck column to float values (GB_A1.3) and proceeded to create a variable (selected_columns) representing the features we were going to use: father's education (SD_3a_FATHER), gender (SD_1), offline gambling history (GM_1a), and self-perceived luck(GB_A1.3). 

Finally, we filtered out rows that were missing our target variable (online gambling history (GM_1b)) and filled null values in our feature columns with the value 0. We used an 80-20 split to split our data into training and test datasets respectively.

```python
# Process the data to the correct type for ML and remove invalid rows
cleaned_df['GB_A1.3'] = cleaned_df['GB_A1.3'].astype(float)
selected_columns = ['SD_3a_FATHER','SD_1','GM_1a','GB_A1.3']
valid_subset = cleaned_df[~cleaned_df['GM_1b'].isna()]
X = valid_subset[selected_columns].fillna(0)
y = valid_subset['GM_1b']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=33)
```

We establish a baseline to compare against by using the DummyCassifier, which strictly predicts based off the most frequent class in the training data for all samples. From the classification report, we can see that this approach received an accuracy of ~56%.

 When we train our random forest classifier, we can reliably see how learning from actual patterns in data improves performance over this naive approach. 

```python
# Naive Model
dummy = DummyClassifier()

dummy.fit(X_train,y_train)

print(classification_report(y_test,dummy.predict(X_test),zero_division=1))
```

Our model's training performance is relatively decent. With high precision, recall, and F1-scores for both non-gamblers (class 0) and gamblers (class 1), the model demonstrates effective learning from the training data. 

These strong metrics suggest the model has successfully recognized patterns that distinguish between the two classes with relatively few misclassifications. The model's ability to correctly identify instances from each class within the training dataset indicates it has captured meaningful relationships in the features that characterize gambling behavior.

```python
forest.fit(X_train,y_train)

pred = forest.predict(X_test)

print(classification_report(y_train,forest.predict(X_train)))
```

When we train our random forest classifier on our training data, and use it to predict on our test data, we see that our accuracy improves from ~56% to ~71%, showing that our model performs substantially better than the baseline. 

For predicting when people have not gambled online (class 0), the precision increased from .56 to .75, and still remained relatively high at .66 when predicting if people have gambled online. The macro and weighted averages also show balanced performance across both classes, as both show .71 across precision, recall, and f1-score. 

```python
print(classification_report(y_test,pred))
```

 Below, we plotted a barplot using seaborn to show feature importance (how each of the features we used to train our model contributed to its subsequent accuracy). From the graph, we can see that gender had a significant influence on our model's decision-making, followed by offline gambling history, self-perceived luck, and finally father's education. 
 
 Therefore, when determining if someone is likely to gamble online we should foremost consider their gender and whether they have a history of offline gambling. The other 2 factors of self-perceived luck and father's education may not be as important.

```python
sns.set_theme()
sns.barplot(x=selected_columns,y=forest.feature_importances_,palette='coolwarm')
plt.title("Random Forest Classifier's Feature Importances")
plt.xlabel("Feature")
plt.ylabel("Feature Importance")
plt.xticks(ticks=[0, 1, 2, 3], labels=["Father's Education", "Gender", "Offline Gambling History", "Self-perceived Luck"], rotation = 45)
plt.plot()
```

When using data about gambling, making sure all of the data is properly anonymized is very important to respect privacy. Since gambling is a sensitive topic and it is important that people give honest answers, their answers must be private to ensure accuracy. Another major bias present in gambling may be Nonresponse bias. People who do not want to give out their information about gambling or to a particular question about gambling may instead decide to leave the form or question blank.

As well, there are potential biases in the dataset which we use. For instance, if we use an online survey, we may get a very different demographic who responds compared to a phone survey. This may lead to certain groups such as older people being excluded. Furthermore, if the survey is taken in a particular region or country, we could only generalize to that area. As well, the way that the questions are written may influence respondents to answer in a certain way to avoid being judged.

To identify these biases before our analysis, we will make sure to choose a dataset that is collected in a manner that avoids most non-response bias and mitigates selection bias. During our analysis, we will make sure to impute data that is not there or remove missing data to ensure that all responses are represented and accurate.
