# Abstract

This study examines the relationship between a goalkeeper’s height, age, and their performance metrics in Major League Soccer (MLS) over the past ten years. Specifically, we investigate whether these physical attributes significantly impact save percentage (Save%) and post-shot expected goals (PSxG)—two key indicators of goalkeeping effectiveness. While taller goalkeepers may have an advantage in shot-blocking due to greater reach, agility and reflexes also play a crucial role. Similarly, while experience could benefit goalkeeping decision-making, aging may reduce reaction speed. This study seeks to determine whether height and age have a statistically significant effect (p < 0.05) on goalkeeping performance. <br><br>

To explore this relationship, we aggregated and cleaned goalkeeper data from multiple seasons, compiling statistics on height, age, Save%, and PSxG. We applied Pearson correlation analysis to measure the strength and significance of these associations, both with and without outliers identified through interquartile range (IQR) filtering. Initial analyses yielded weak or insignificant correlations, but after expanding the dataset to include additional seasons, such results improved. <br><br>

Based on our analysis and exploration, there does not appear to be a significant correlation between goalkeeper height and their performance, as measured by Save Percentage (Save%). While it is often assumed that taller goalkeepers have an advantage due to their extended reach, the data does not support this hypothesis. Visualizing the data through scatter plots, histograms, and correlation heatmaps revealed no positive relationship. Instead, the results suggest a negative correlation between goalkeeper height and their average Save%, indicating that taller goalkeepers may not necessarily outperform their shorter counterparts in this specific metric. <br><br>

These findings challenge the common perception that height is a dominant factor in goalkeeper performance. The negative correlation observed could be influenced by other factors, such as agility, reaction time, positioning, and decision-making, which are also critical for success in goalkeeping. It is also possible that their team's resources like coaching, equipment, and healthcare alongside their team's defensive strategies play a substantial role in determining a goalkeeper's Save%. Further research incorporating additional variables like the quality of opposition, number of shots faced, and specific save situations would provide a more comprehensive understanding of goalkeeper performance.

## Background and Prior Work


Soccer is a popular, worldwide sport with an estimate of over 240 million registered players. It is a big part of culture and entertainment in many countries. As of today, Major League Soccer consists of 30 teams. Each team generates, on average, about 10-50 million USD yearly from season matches.

In soccer, the goalkeeper’s role is to prevent the ball from entering the goal. Goalkeepers must perform jumps, kicks, and dives to catch or deflect the ball from the goal. A goalkeeper’s performance can be influenced by height; taller goalkeepers have greater reach and may be able to more effectively block shots. However, height might also impact agility and speed, which are also important traits for goalies. Additionally, research shows that older pro-soccer players generally decline in their physical performance year by year however they generally experience increased pass accuracy as they gain more experience. These traits can offer valuable insight into success during matches such as saves per shot. However, due to the explosive nature of the actions goalies must perform, size or age alone cannot predict the accuracy of goalkeeper saves.

Post-shot expectancy goal (PSxG) is a trained model that evaluates the average value of each target shot (determined by angle, speed, trajectory, etc.) and produces the probability of that shot becoming a goal. Any shot that ends up outside of the goal range has a PSxG of 0% because there is no chance the shot would lead to a goal. Additionally, PSxG only considers actions from the range after the ball is kicked by the shooter to when it passes or reaches the goalkeeper.

Previous research has analyzed individual factors such as the impact of height on goalie block rates. However, our research aims to synthesize a variety of important physical characteristics that reflect on goalkeeper performance metrics. By integrating historical goalkeeper statistics in the MLS with physical attribute datasets, our study aims to better understand the relationship between multiple physical features and their impact on on-field performance. This may offer valuable insight into game strategy as well as player development and recruitment.

```python
#necessary imports
import pandas as pd
from scipy.stats import pearsonr
import numpy as np
```

```python
master = pd.read_csv('stats_keeper_adv_18.csv')
master = master.rename(columns=master.iloc[0]).iloc[1:].reset_index(drop=True)  # Fix headers'
master = master[['Player', "Save%", "Saves", "SoTA", "GA"]]
master['Year'] = 2018

# Iterate through other CSV files
for i in range(19, 25):
    df = pd.read_csv(f'stats_keeper_adv_{i}.csv')  # Use f-string for cleaner formatting
    df = df.rename(columns=df.iloc[0]).iloc[1:].reset_index(drop=True)  # Fix headers
    df['Year'] = 2000 + i
    df = df[["Player", "Save%", "Saves", "SoTA", "GA", "Year"]] # Ensure column names match master
    master = pd.concat([master, df], ignore_index=True)  # Append rows

master.columns = ['Player', 'Save%', "PK Save%", "Saves", "SoTA","GA","Year"]
master = master.drop(columns= ["PK Save%"])
print(master)
```

## Dataset #2 : Keeper Heights (cm)

```python
master2 = pd.read_csv('height_18.csv')
master2 = master2[['Player', "Height"]]
master2['Year'] = 2018

#Iterate through other CSV files   
for i in range(19, 25):
    df = pd.read_csv(f'height_{i}.csv') 
    df = df[["Player", "Height"]] # Ensure column names match master
    df['Year'] = 2000 + i
    master2 = pd.concat([master2, df], ignore_index=True)  # Append rows
    
# clean data
master2['Height'] = master2['Height'].replace({'cm': '', ' ': ''}, regex=True)
master2['Height'] = pd.to_numeric(master2['Height'], errors='coerce')
# fill nan values
master2['Height'] = master2['Height'].fillna(0)
#convert height column into int
master2['Height'] = master2['Height'].astype(int)
master2.rename(columns={'Height': 'Height (cm)'}, inplace=True)
master2
```

Given that a goalkeeper, like Jeff Attinella, can play for multiple years, we added a year column to separate his height data throughout the years to prevent skewness in the experiment.

## Correlation Analysis Pt.1

Next, we calculate Pearson's correlation (r) between goalkeeper height/age and save percentage to determine if there is a statistically significant relationship between the variables. A correlation closer to 1 or -1 might suggest that height does/doesn't plays a significant role in goalkeeper performance, and a correlation close to 0 would mean that there is no correlation. We want to look for a correlation closer to -1 or 1 for our experiment.

We also calculate the p-value to validate whether the probability of our results is more extreme or not than the observed results, leading us to either reject or accept the null hypothesis. We want to look for a p < 0.05 to reject the null hypothesis.

```python
#calculate correlation between height and save percentage
merged_data['Save%'] = pd.to_numeric(merged_data['Save%'], errors='coerce')
merged_data['Save%'] = merged_data['Save%'].fillna(0)
merged_data['Height (cm)'] = pd.to_numeric(merged_data['Height (cm)'], errors='coerce')


correlation1,p1 = pearsonr(merged_data['Height (cm)'],(merged_data['Save%']))
print(f'Correlation between Height (cm) and Save Percentage: {correlation1}')
print(f'p-value:', p1)
```

```python
#calculate correlation between height and xg
correlation2,p2 = pearsonr(merged_data['Height (cm)'],(merged_data['PSxG']))
print(f'Correlation between Height (cm) and PSxG: {correlation2}')
print(f'p-value:', p2)
```

```python
#calculate correlation between age and save percentage
correlation3,p3 = pearsonr(merged_data['Age'],(merged_data['Save%']))
print(f'Correlation between Age and Save%: {correlation3}')
print(f'p-value:', p3)
```

```python
#calculate correlation between age and xg
correlation4,p4 = pearsonr(merged_data['Age'],(merged_data['PSxG']))
print(f'Correlation between age and expected goals: {correlation4}')
print(f'p-value:', p4)
```

## Correlation Analysis Pt.2
**It seems that there is a significant relationship between age and a keeper's save%. However, the correlation is weak.**


**It seems that there is no significant relationship between height and a keeper's performance.** This is also supported by online sources, but we wondered if the issue could also be rooted in two reasons: <br>
    - height does not vary much in the league <br>
    - many keepers stay for multiple seasons, meaning that we get many duplicate height values, which may confound results as keepers' performances often differ widely per season <br>

To mitigate this, we decided to do the following analysis, where we average the save percentage values at each height, and analyze their correlation. This, we hope, reduces the confounding effect that duplicate heights may have by creating a singular plot point for each height.

```python
#create average save percentage for each height and age
height_save_avg = merged_data.groupby('Height (cm)')['Save%'].mean().reset_index()
age_save_avg = merged_data.groupby('Age')['Save%'].mean().reset_index()
```

# Discussion and Conclusion

## Results Discussion

Our initial research question aimed to answer, "does age and/or height have an effect on a goalkeeper's save percentage in the MLS?" Overall, our research yielded statistically significant results. Based on our analysis, a goalkeeper's age is positively correlated with their save percentage, meaning that the older a goalkeeper is, the higher the probability that they save any given shot. Moreover, a goalkeeper's height is negatively correlated with their age, such that the shorter a goalkeeper is, the higher the probability that they save any given shot. <br> <br>

The positive correlation between age and save percentage aligns with prevailing perspectives in football, where experience is often considered a valuable asset for goalkeepers. Tactical awareness, positioning, and decision-making abilities generally improve with age, compensating for any potential decline in agility. However, this relationship is not absolute, as the physical demands of the position may offset the advantages of experience over time. Further research is necessary to examine whether this trend persists in older goalkeepers nearing the end of their careers. <br><br>

The negative correlation between height and save percentage is inconsistent with the widely held belief that taller goalkeepers possess a competitive advantage. Height is traditionally viewed as a key metric in goalkeeper recruitment, as it provides greater reach and aerial dominance. However, our results suggest that shorter goalkeepers may demonstrate superior reflex-based shot-stopping ability, possibly due to quicker reaction times, enhanced mobility, or a lower center of gravity. These findings indicate that scouting and recruitment strategies should place greater emphasis on technical proficiency rather than height alone.

## Limitations and Space for Future Analysis

There are multiple limitations to our analysis. The first of those is in the breadth of our dataset. We only cover goalkeepers in the MLS-which may not capture the relationships between height/age and save% on a global scale, meaning that we cannot extrapolate our results past the MLS. This is  important as different leagues (e.g. the Champions League) are often considered to have different skill levels, with harder/easier shots as well as better/worse goalkeepers. For future analysis, more goalkeepers from various leagues should be considered.<br><br>

Additionally, this study evaluates only height and age as potential predictors of save percentage. Goalkeeper performance, however, is influenced by multiple factors, including reaction time, positioning, defensive organization, and team tactics. A more comprehensive approach incorporating advanced performance metrics, such as expected saves (xSaves) and reaction speed, would provide deeper insights into the determinants of goalkeeping success. <br><br>

Another consideration is the potential influence of outliers and sample size constraints on the observed correlations. Although we employed outlier detection and removal techniques, variability in playing time and data collection inconsistencies may still impact the results. A comprehensive study tracking goalkeepers over multiple seasons would allow for a more robust examination of performance trends and their association with age and height. <br><br>

One limitation of this study is the relatively low variability in the height and age of goalkeepers within the dataset. The majority of professional goalkeepers fall within a narrow height and age range. This restricted variability may reduce the statistical power of correlation analyses, as meaningful trends are more difficult to detect when the dataset lacks sufficient diversity in the variables of interest.

## Conclusion

Our project explored how height and age influence goalkeeper performance in Major League Soccer by analyzing correlations between these factors and key performance metrics, including save percentage, save percentage average, and post-shot expected goals (PSxG). Our findings indicated a negative correlation between height and average save percentage (r = -0.50 without outliers in the dataset), supporting existing research suggesting that taller goalkeepers may take longer to refine their skills. In contrast, a positive correlation was observed between age and average save percentage (r = 0.56 without outliers in the dataset), implying that experience plays a significant role in enhancing performance. <br><br>

However, several limitations should be acknowledged in our data exploration. The study faced restricted age variation among Major League Soccer goalkeepers, most of whom are in their mid-to-late 20s or early 30s. We also omitted physical attributes like wingspan, vertical jump, and reaction time that were not available on the Internet for us, which could provide a more comprehensive evaluation of goalkeeper abilities. It is also important to recognize that success in professional soccer is influenced by external factors such as access to superior training, coaching, and facilities, often linked to wealthier teams. Ethically, height and other physical characteristics should not be used as sole determinants in excluding goalkeepers from opportunities, as skill development and experience are equally valuable contributors to a goalkeeper’s performance.
