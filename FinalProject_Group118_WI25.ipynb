# Abstract
In this project, our group explored the relationship between financial stability and mental health, as financial well being plays a critical role in mental health, which is often reflected in suicide rates. Suicide is a complex issue influenced by various factors, with a strong connection to the economic health of a country. One key indicator of a country's economic health is GDP. A declining GDP signals economic recessions or financial crises, which can lead to financial insecurity for individuals, potentially increasing the risk of suicide. Additionally, unemployment causes significant financial distress, which negatively impacts mental health and, as a result, tends to increase suicide rates.

To investigate this, we explored the correlation between GDP and unemployment rates with suicide rates using datasets that contain information on these variables from 2000 to 2019 across various countries. Through our Exploratory Data Analysis (EDA), we found a negative correlation between GDP and suicide rates, suggesting that when economic conditions worsen, suicide rates tend to rise. On the other hand, we discovered a positive correlation between unemployment rates and suicide rates, showing that higher unemployment is linked to higher suicide rates.

In addition, we created predictive machine learning models to help quantify and forecast the impact of economic conditions (GDP and unemployment rates) on suicide rates. These models allow us to better understand how changes in these economic factors might influence mental health outcomes.

## Background and Prior Work


The decline of mental health has been a critical topic of discussion for decades, with psychology and social sciences striving to understand the complexities of mental well-being. Research has consistently shown that poor mental health is influenced by an intricate interplay of social, economic, and cultural factors<a href="#fn1" name="ref1">1</a>. One of the most alarming consequences of poor mental health is its strong link to suicide, as studies estimate that approximately 60 percent of individuals who die by suicide have had a mental disorder, such as depression or bipolar disorder<a href="#fn2" name="ref2">2</a>. Furthermore, the connection between financial hardship and psychological stress is well-documented. For example, In 2008, the burst of the housing bubble and lax financial regulations along with predatory loans from banks culminated in the 2008 financial crisis, causing many to lose their homes, jobs and property. From this, a significant increase of antidepressant usage was reported<a href="#fn3" name="ref3">3</a>, highlighting the impact of economic distress on mental health. Similarly, the Great Depression, one of the most severe economic crises in history, earned its name not only for its financial devastation but also for the widespread psychological toll it took on individuals. These historical examples demonstrate the clear correlation between financial hardship and rising psychological stress as many people are becoming more and more uncertain about their future.

However, across countries the perception and cultural context of suicide vary widely across different countries, as in some countries suicide is stigmatized or even criminalized against, which can influence reporting rates and social attitudes towards mental health. For example, countries like Antigua and Barbuda have some of the lowest suicide rates globally<a href="#fn4" name="ref4">4</a>, which is likely attributed to the strong social stigma against suicide present in their cultural and religious beliefs<a href="#fn5" name="ref5">5</a>. Meanwhile in other areas such as the Carribean, attempted suicide is legally a criminal offense with a penalty of up to two years in prison or a fine<a href="#fn6" name="ref6">6</a>. Meanwhile, South Korea is a country that boasts one of the strongest economies worldwide, and yet has one of the highest suicide rates at 26 per 100,000<a href="#fn7" name="ref7">7</a>. So, while the relationship between economic wellness and mental health seems clear on the surface, there is yet a high degree of variability that comes with the human condition and its interplay with not only society and culture, but also personal circumstance.

## Exploratory Data Analysis



### Data Transformation

When we firsted merge the three dataset, gdp.csv, unemployment.csv, and suicide.csv, each row represented a country, and the timestep from 2000 to 2019 of gdp, suicide rate, and unemployment rate. Resulting in excess amounts of columns, since each row had suicide rate, unemployment rate, and gdp from 2000 to 2019. 
In the new dataset, each row in the dataset represents a specific country in a given year, with range 2000-2019. Each row contains GDP, suicide rate, and unemployment rate. The rows are set up such that it is chronically ordered by year within each country. Going from 61 columns to only 4. 


```python
import pandas as pd
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
gdp = pd.read_csv("merge/data/gdp.csv", skiprows=3)
suicide = pd.read_csv("merge/data/suicide.csv", skiprows=3)
unemployment = pd.read_csv("merge/data/unemployment.csv")
#pd.set_option('display.max_rows', None)
#pd.set_option('display.max_columns', None)

years_to_keep = [str(year) for year in range(2000, 2020)]
non_year_cols = ['Country Name', 'Country Code']
keep_cols = non_year_cols + years_to_keep
gdp = gdp[keep_cols] # Keep years 2000-2019
gdp = gdp.dropna() # Drop any countries without data for these years

suicide = suicide[keep_cols]
suicide = suicide.dropna()

merged_df = pd.merge(
    gdp,
    suicide,
    on=['Country Name', 'Country Code'],
    suffixes=('_gdp', '_suicide')
)

unemployment = unemployment[keep_cols]
unemployment = unemployment.dropna()

unemployment.rename(
    columns={year: f"{year}_unemp" for year in years_to_keep},
    inplace=True
)

df = pd.merge(
    merged_df,
    unemployment,
    on=['Country Name', 'Country Code'],
    how='inner'
)

df = df.drop(columns= "Country Code")
df.head()
```

```python
df_long = df.melt(id_vars=['Country Name'], var_name='Metric_Year', value_name='Value')

df_long[['Year', 'Metric']] = df_long['Metric_Year'].str.split('_', expand=True)
df_long['Year'] = df_long['Year'].astype(int) 

df_pivot = df_long.pivot_table(index=['Country Name', 'Year'], 
                               columns='Metric', values='Value').reset_index()

df_pivot.columns.name = None  
df_pivot.rename(columns={'gdp': 'GDP', 'unemp': 'Unemployment', 'suicide': 'Suicide Rate'}, inplace=True)
df_pivot
```

We decided to perform min-max normalization on the dataset as well, given that GDP and unemployment are both ratio while GDP is a very big numerical number


```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df_pivot[['GDP', 'Unemployment', 'Suicide Rate']] = scaler.fit_transform(df_pivot[['GDP', 'Unemployment', 'Suicide Rate']])
df_pivot.describe()
```

### GDP vs. suicide rate by country
The regression plots show the relationship between GDP (x-axis) and suicide rates per country (y-axis). We observe that most countries exhibit a generally negative slope between suicide rate and GDP, which supports our hypothesis.

### OLS Regression 1

```python
outcome_2, predictors_2 = patsy.dmatrices('Suicide_Rate ~ Unemployment', data=df_pivot)
model_2 = statsmodels.api.OLS(outcome_2, predictors_2)
results_2 = model_2.fit()

print(results_2.summary())
```

### Mixed-Effects Model
We have observed from the visualizations of GDP vs. suicide rate by country that there seemed to be generally downward trends in each chart. However, when we looked at the generalizations of the overall data around the globe, there were positive correlations between GDP and suicide rate, and not very strong correlations for either suicide rate or unemployment. This phenomenon is called Simpson’s Paradox, where a trend appears in different groups of data but disappears or reverses when the groups are combined. So, we have decided to use the Mixed-Effects Model, which should account for the nested structure of the data where each country has multiple plot points.
From this test, we can see a negative coefficient between suicide rates vs. GDP, and the correlation is also stronger. However, for unemployment rates we see a slightly weaker positive correlation. The z-value is very low, meaning that the null hypothesis can be rejected.

```python
import statsmodels.api as sm
import statsmodels.formula.api as smf

model = smf.mixedlm("Q('Suicide_Rate') ~ GDP", df_pivot, groups=df_pivot['Country Name'])
result = model.fit()
print(result.summary())

model2 = smf.mixedlm("Q('Suicide_Rate') ~ Unemployment", df_pivot, groups=df_pivot['Country Name'])
result2 = model2.fit()
print(result2.summary())
```

## Predictive Models

Finally, we will be creating some predictive models for our dataset. We already normalization earlier so we do not have to worry about scaling our features again before training the models. Since our features and target variable are both numerical, we will use regressor models. Our feature variable will be GDP and unemployment rate, and our target variable will be suicide rate. 

### Performing Train Test Split

```python
from sklearn.model_selection import train_test_split

X = df_pivot[['GDP', 'Unemployment']].values
y = df_pivot['Suicide Rate'].values

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

### Model 1: Linear Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

reg_model = LinearRegression().fit(X_train, y_train)

y_pred = reg_model.predict(X_test)

mse = mean_squared_error(y_test, y_pred)
print(f"Mean Squared Error: {mse:.4f}")
```

Already, we can see that Linear Regression does a good job of fitting our dataset, as indicated by the low Mean Squared Error (MSE) of 0.0084. This means that the Linear Regression model's predictions are very close to the actual values in the dataset. On the other hand, a relatively low loss in a linear model can potentially be an indicator of overfitting, we will attempt to provide some non-linearity to our dataset to hopefully make a more predictive model.

### Model 2: Random Forest

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators = 100, 
    max_depth = None,   
    min_samples_split = 2,
    min_samples_leaf = 1
)

rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)

mse_rf = mean_squared_error(y_test, y_pred_rf)
print(f"Mean Squared Error: {mse_rf:.4f}")
```

Our next model is a random forest regressor. After training and evaluating, we arrive at a MSE of 0.0059. Performing slightly better at capturing the patterns of our dataset then the linear regression model. While it captures complex patterns better, it may still have limitations. To further enhance predictive performance, we now turn to our final model, a neural network. 

### Model 3: Feedforward Neural Network

```python
import torch.nn as nn

class FFN(nn.Module):
    def __init__(self):
        super(FFN, self).__init__()
        self.layer1 = nn.Linear(2, 32)
        self.layer2 = nn.Linear(32, 32)
        self.layer3 = nn.Linear(32, 32)
        self.layer4 = nn.Linear(32, 1)
        self.relu = nn.ReLU()

    def forward(self, x):
        x1 = self.relu(self.layer1(x))
        x2 = self.relu(self.layer2(x1))
        x3 = self.relu(self.layer3(x2))
        x4 = self.layer4(x3)

        return x4 
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader, TensorDataset
from tqdm import tqdm

X_train = X_train.clone().detach().float()
y_train = y_train.clone().detach().float().view(-1, 1)
X_test = X_test.clone().detach().float()
y_test = y_test.clone().detach().float().view(-1, 1)


batch_size = 32
train_dataset = TensorDataset(X_train, y_train)
train_loader = DataLoader(train_dataset, batch_size = batch_size, shuffle = True)

model = FFN()
criterion = nn.MSELoss()
optimizer = optim.AdamW(model.parameters(), lr = 0.01)

epochs = 100
train_losses = []

for epoch in tqdm(range(epochs)):
    model.train()
    epoch_loss = 0

    for x, y in train_loader:
        optimizer.zero_grad()
        output = model(x)
        loss = criterion(output, y)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()

    avg_epoch_loss = epoch_loss / len(train_loader)
    train_losses.append(avg_epoch_loss)

    if (epoch + 1) % 50 == 0:
        print(f"Epoch {epoch + 1}, Avg Loss: {avg_epoch_loss:.4f}, LR: {optimizer.param_groups[0]['lr']:.6f}")

model.eval()
with torch.no_grad():
    y_pred = model(X_test)
    test_loss = criterion(y_pred, y_test)
    print(f"Test Loss: {test_loss.item():.4f}")
```

With the worst loss after training, resulting in an MSE of 0.0083 on the test set, the FFNN performs slightly worse than both of our previous models. This suggests that the dataset may not be "complex" enough, making the introduction of non-linearity unnecessary. In conclusion, the FFNN model's added complexity doesn't offer significant improvements over the simpler models for this particular task.

```python
plt.figure(figsize = (8, 5))
plt.plot(range(1, epochs + 1), train_losses, label = "Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss per Epoch")
plt.legend()
plt.grid()
plt.show()
```
