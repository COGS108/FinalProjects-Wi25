# Abstract

Our study investigated how specific elements of employees' work-from-home environments impact productivity in knowledge-based professions, addressing a gap in existing research that has primarily compared remote and in-office work without examining specific workspace factors. Analyzing data from three comprehensive sources (n = 12,512), we employed correlation analysis and comparative visualization techniques to identify relationships between environmental factors and productivity. Our findings revealed that employees with structured, well-equipped home workspaces reported productivity levels up to 50% higher than those without adequate setups, with strong correlations between workspace quality, motivation levels, and employee well-being. Additional analysis showed that caregiving responsibilities and technological resources significantly influenced remote work effectiveness. These results confirm our hypothesis that structured home environments positively impact productivity and suggest organizations should prioritize supporting employees in creating dedicated workspaces with appropriate ergonomic features, technological tools, and strategies for managing household distractions, providing actionable insights for organizations developing remote and hybrid work policies.



# Research Question


Does remote work lead to more productive work, and if so what specific factors (job security, childcare responsibility, etc) most likely lead to that productivity?



## Background and Prior Work

Remote work has become increasingly common since the onset of the COVID-19 pandemic, with many studies suggesting that flexibility and reduced commute times can enhance productivity. However, there are also concerns that working outside of a traditional office setting may lead to distractions, lack of supervision, and communication challenges. A 2022 Microsoft Hybrid Report found that 85% of business leaders believe hybrid work diminishes productivity due to reduced control and engagement. Additionally, increased reliance on virtual meetings was cited as a factor that fragments employee focus.

In contrast, a study by Stanford University analyzed more than 1,600 employees at Trip.com and observed that employees who worked from home two days per week were equally productive and equally likely to be promoted as their in-person peers. Furthermore, this partial work-from-home model was associated with a 33% decrease in resignations, especially among employees with long commutes. These results suggest that a well-structured hybrid arrangement can benefit both employees and employers by combining flexibility with sufficient in-person engagement. [¹]

A Harvard Business School study indicates that while fully remote work may be 10% less productive than onsite work, a thoughtfully designed hybrid model could mitigate much of this loss by preserving aspects of in-person collaboration and informal learning. Taken together, these findings underscore that whether remote or hybrid work boosts productivity often depends on company structure, management strategies, and specific job roles. [²]

Despite the breadth of research on remote and hybrid work, few studies examine how particular work-from-home setups—such as dedicated workspace configurations, level of digital communication, and work-life balance—impact individual productivity. Our research aims to address this gap. By focusing on specific, measurable conditions of the home work environment, we hope to clarify which factors contribute most significantly to productivity and provide actionable insights for organizations navigating a new era of remote and hybrid work. ### Correlation Analysis

```python
selected_cols = {
    'Caring Responsibilities': 'Have the following barriers to remote working improved or worsened for you over the past 6 months?   - My caring responsibilities',
    'Workspace': 'Have the following barriers to remote working improved or worsened for you over the past 6 months?   - My workspace (e.g. suitable chair, lighting, noise levels, facilities)',
    'Motivation': 'Have the following barriers to remote working improved or worsened for you over the past 6 months?   - Motivation',
    'Living Situation': 'Have the following barriers to remote working improved or worsened for you over the past 6 months?   - My living situation (e.g. location, home size, who I live with)',
    'Health & Safety': 'Have the following barriers to remote working improved or worsened for you over the past 6 months?   - My health and safety when working remotely',
}

ordinal_map = {
    # Barriers: 1 = worsened, 5 = improved
    "Significantly worsened": 1,
    "Somewhat worsened": 2,
    "Stayed about the same": 3,
    "Somewhat improved": 4,
    "Significantly improved": 5,
    "Not a barrier for me": 3,  # Or you could drop these rows

    # Productivity: 1 = much less productive, 5 = much more productive
    "Much less productive when working remotely": 1,
    "I’m 50% less productive when working remotely": 2,
    "I’m 50% less productive when working remotely (or worse)": 1.5,
    "I’m 40% less productive when working remotely": 2,
    "I’m 30% less productive when working remotely": 2.5,
    "I’m 20% less productive when working remotely": 2.7,
    "I’m 10% less productive when working remotely": 2.9,
    "My productivity is about same when I work remotely": 3,
    "My productivity is about same when I work remotely as when I work in my employer’s workplace": 3,
    "I’m 10% more productive when working remotely": 3.1,
    "I’m 20% more productive when working remotely": 3.3,
    "I’m 30% more productive when working remotely": 3.5,
    "I’m 40% more productive when working remotely": 3.7,
    "I’m 50% more productive when working remotely": 4,
    "I’m 50% more productive when working remotely (or more)": 4.5,
    "Much more productive when working remotely": 5
}


import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Replace this with your actual DataFrame
# df = pd.read_csv("your_file.csv")

# Rename for easier access
df_subset = df[[v for v in selected_cols.values()]].copy()
df_subset.columns = list(selected_cols.keys())  # Rename columns for readability

# Map all ordinal responses to numeric
df_subset = df_subset.applymap(lambda x: ordinal_map.get(x, None))

# Drop rows with any missing values in selected columns
df_subset.dropna(inplace=True)

# Correlation heatmap
plt.figure(figsize=(10, 7))
sns.heatmap(df_subset.corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Correlation Between Remote Work Barriers and Productivity", fontsize=14)
plt.tight_layout()
plt.show()
```

This heatmap illustrates the correlations between different barriers to remote working, such as caring responsibilities, workspace quality, motivation, living situation, and health & safety. The values indicate positive correlations across all variables, suggesting that improvements or deteriorations in one area (e.g. workspace) are often associated with similar changes in others (e.g. motivation or health & safety). Notably, motivation shows strong correlations with health & safety (0.49) and living situation (0.46), highlighting the interconnected nature of remote work conditions.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Mapping responses to ordinal values
ordinal_map = {
    "Significantly worsened": 1,
    "Somewhat worsened": 2,
    "Stayed about the same": 3,
    "Somewhat improved": 4,
    "Significantly improved": 5,
    
    "Strongly disagree": 1,
    "Somewhat disagree": 2,
    "Neither agree nor disagree": 3,
    "Somewhat agree": 4,
    "Strongly agree": 5
}

# Define column names exactly as they appear
workspace_col = 'Have the following barriers to remote working improved or worsened for you over the past 6 months?   - My workspace (e.g. suitable chair, lighting, noise levels, facilities)'
feeling_better_col = "How strongly do you agree or disagree with the following statements?   - On days when I work remotely I feel better"

# Make a copy of relevant columns
df_plot = df[[workspace_col, feeling_better_col]].copy()

# Clean and map
df_plot = df_plot.applymap(lambda x: ordinal_map.get(str(x).strip()) if pd.notnull(x) else None)
df_plot.dropna(inplace=True)

# Plot
plt.figure(figsize=(8, 6))
sns.barplot(x=workspace_col, y=feeling_better_col, data=df_plot, errorbar=None)
plt.title("Impact of Workspace Quality on Feeling Better During Remote Work")
plt.xlabel("Workspace Improvement (1=Worsened, 5=Improved)")
plt.ylabel("Agreement: 'I feel better on remote days' (1–5)")
plt.tight_layout()
plt.show()
```

This bar plot shows the relationship between changes in employees' remote workspace quality and how strongly they agree with the statement "I feel better on days when I work remotely." The results suggest that as workspace conditions improve, employees are more likely to report feeling better on remote workdays, indicating a positive link between the physical work-from-home setup and well-being.

## How a better work environment can lead to increased satisfaction

This section analyzes how enhanced workspace quality can affect the feelings of employees on remote workdays.

```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Load the dataset
df = pd.read_csv("arya.csv")

# Define question categories
yes_no_columns = [col for col in df.columns if col.startswith(('1.', '2.', '3.', '4.', '5.', '8.'))]
multiple_choice_columns = [col for col in df.columns if col.startswith(('6.', '7.', '9.', '10.'))]

# Set plotting style
plt.style.use('seaborn-v0_8-whitegrid')

# Create a better layout
fig, axes = plt.subplots(len(yes_no_columns)//2, 2, figsize=(14, 12))
axes = axes.flatten()

# 1. Yes/No Questions - Improved visualization
for i, col in enumerate(yes_no_columns):
    ax = axes[i]
    
    # Extract counts
    counts = df[:100][col].value_counts()
    total = counts.sum()
    percentages = [(count/total)*100 for count in counts]
    
    # Bar plot
    sns.barplot(x=counts.index, y=counts.values, palette="Set2", ax=ax)
    
    # Add percentage labels
    for j, p in enumerate(percentages):
        ax.annotate(f'{p:.1f}%', 
                   (j, counts.values[j]), 
                   ha='center', va='bottom', fontweight='bold')

    ax.set_title(f"Q{i+1}", fontsize=12, fontweight='bold')
    ax.set_xlabel('')
    ax.set_ylabel('Count')

# Remove any unused subplots
for i in range(len(yes_no_columns), len(axes)):
    fig.delaxes(axes[i])

plt.tight_layout(pad=2.0)

# 2. Work Environment Preferences - Improved
fig, ax = plt.subplots(figsize=(10, 6))

work_options = ['Working from home', 'Working in an office', 'A mixed mode of working']
work_env_cols = ['6', '7', '9', '10']
summary_plot_data = {wtype: [] for wtype in work_options}

for col_num in work_env_cols:
    col_name = next(c for c in df.columns if c.startswith(f'{col_num}.'))
    counts = df[:100][col_name].value_counts(normalize=True) * 100
    for wtype in work_options:
        summary_plot_data[wtype].append(counts.get(wtype, 0))

# Bar plot
x = np.arange(len(work_env_cols))
width = 0.2

for i, wtype in enumerate(work_options):
    ax.bar(x + (i-1)*width, summary_plot_data[wtype], width, label=wtype)

ax.set_ylabel('Percentage')
ax.set_title('Work Environment Preferences', fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels([f'Q{num}' for num in work_env_cols])
ax.legend()

plt.show()
```
