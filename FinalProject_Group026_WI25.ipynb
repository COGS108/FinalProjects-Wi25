# Abstract

The project began with a shared interest in the events of the 2025 Los Angeles/Palisades wildfires. Los Angeles county is home to diverse socioeconomic groups <a name="cite_ref-6"></a>[<sup>1</sup>](#cite_note-1.1) (Los Angeles County Department of Public Health, 2018) and we were particularly curious about how the wildfire impacted different communities and what recovery looked like for each. To explore the economic burden across counties of different income levels in California, we collected data from the Federal Emergency Management Agency (FEMA), Department of Forestry and Fire Protection (CAL FIRE), Bureau of Economic Analysis (BEA) GDP & Personal Income, and California Open Data Portal (California County Data). The FEMA data allowed us to explore aid distributions for wildfire disasters between 2000 and 2023. CAL FIRE’s dataset gave us insight on fire size, frequency, and geographical distribution. As for studying financial burden, BEA’s data on GDP and personal income allowed us to classify CA counties into low, medium, and high-income groups. The California county data describes county size and population density which allows us to compare wildfire impacts relative to size of counties of various income. Together, these datasets provided a comprehensive framework for understanding wildfire impacts across income groups and regions in California.

Our initial data analysis reflected that low-income groups experience the highest amount of economic burden while higher-income groups experience lower burdens despite larger fires. In order to isolate whether income is a statistically significant factor in determining economic burden, we also looked at factors such as population density, acres burned, fire size, etc. In analyzing wildfire frequency in different counties, we recognized that lower-income counties faced higher wildfire frequencies which can contribute to economic burden. Land as a percent burned is also a strong predictor of economic burden while income and population density show weaker correlations. Additionally, the distributions of aid and resources are not strongly tied to income or population size as both low and high income counties received the most declared aid. Through regression analysis, we concluded that income level is not a significant predictor of economic burden. So while low-income groups do experience higher economic burden, this outcome is not solely attributed to income alone.

# Research Question

How does the economic burden of California wildfires (2000-2023) differ across low, medium, and high-income counties, measured by the ratio of property damage costs to median household income and average displacement time?


## Background and Prior Work


Wildfires in California have grown in frequency and severity over the past few years as a result of climate change. Wildfires not only create physical devastation but significant economic and social challenges for those affected For example, Diaz (2012) highlights the true impact of wildfires beyond typical indicators such as acres burned, exploring impacts on community, businesses, state, economy of natural areas, and property damage. <a name="cite_ref-5"></a>[<sup>1</sup>](#cite_note-5) This research aims to investigate the economic burden of California wildfires spanning from the years 2000 to 2023, across communities in California counties of varying income levels. As stated by the NOAA (2023), ten of the largest and most destructive California wildfires have occurred in the past 20 years alone, in part due to climate change.<a name="cite_ref-4"></a>[<sup>1</sup>](#cite_note-4) We are particularly curious about how property damage costs to median household income affects average displacement times across low, medium, and high-income counties. The impacts of wildfires, specifically in terms of property damage, are related to how resources are spread among communities- as stated by Jones (2023). <a name="cite_ref-6"></a>[<sup>1</sup>](#cite_note-6) Understanding the economic impact of wildfires in relation to income disparities in California can inform decisions for equitable disaster response and recovery strategies.

Several studies have been conducted on the specific economic impact of wildfires. For example, Li and Yu (2025) take on the economic impact of the devastating wildfires that hit Los Angeles as of January 2025; estimating total property and capital losses between $95 billion and $164 billion.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Their findings suggest an increase in insurance premiums and rental prices across LA County. However, there is more to be said about the strain of these increases on the diverse communities of Los Angeles with varying income levels. 

Other research, for example, as done by Roy (2024), discusses the inequities across communities in wildfire response and recovery.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) In California, counties with larger Hispanic, Black, elderly, and disabled populations get less aid and resource allocation when wildfires strike. (2024) emphasizes the inequities in wildfire recovery, specifically in Ventura County.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Those of the Ventura County community are low income and marginalized, suggesting a disproportionate effort by the government to help with recovery as a result of wildfires.

## Calculate Relative Burden of Each Fire using Cal Fire Dataset
> Relative burden formula: (Square Miles X Population Density) / Per Capita Income (dollars)

> Interpretation: The relative burden score represents the economic strain from wildfires per dollar of income, combining burned area exposure (square miles × population density) with a county's financial capacity—higher values indicate greater burden relative to local income. We only calculate the relative burden with the Cal Fire dataset because the fema dataset does not have the features to do so.

```python
# make a copy to add new data
cal_fire_with_county_data = filtered_cal_fire.copy()
```

```python
# Only have income data from income_pop_levels (LineCode = 1.0)
income_data = income_pop_levels[income_pop_levels['LineCode'] == 1.0].melt(id_vars = ['GeoName'], var_name= 'Year', value_name='personal_income')

income_data['GeoName'] = income_data['GeoName'].str.replace(', CA', '', regex=False)

# Renaming columns for merging
income_data = income_data.rename(columns = {'GeoName': 'County', 'Year': 'YEAR_'})
income_data['YEAR_'] = pd.to_numeric(income_data['YEAR_'], errors='coerce')
income_data['County'] = income_data['County'].str.strip().str.title()

income_data = income_data[(income_data['YEAR_']>=2000) & (income_data['YEAR_']<=2023)]
income_data['YEAR_'] = income_data['YEAR_'].astype(int)
```

```python
# Only have population data from income_pop_levels (LineCode = 2.0)
population_data = income_pop_levels[income_pop_levels['LineCode'] == 2.0].melt(id_vars = ['GeoName'], var_name= 'Year', value_name='population')

# Keeps the county data without the state abbreviations 
population_data['GeoName'] = population_data['GeoName'].str.replace(', CA', '', regex=False)

# Rename columns for merging 
population_data  = population_data.rename(columns={'GeoName': 'County', 'Year': 'YEAR_'})
population_data['YEAR_'] = pd.to_numeric(population_data['YEAR_'], errors='coerce')
population_data['County'] = population_data['County'].str.strip().str.title()

# Ensures that the data stores the years we care about
population_data = population_data[(population_data['YEAR_']>=2000) & (population_data['YEAR_']<=2023)]
population_data['YEAR_'] = population_data['YEAR_'].astype(int)
```

>### Interpretation of Regression Results
The linear regression is to investigate whether income levels significantly impact the economic burden caused by wildfires. Using ordinary least squares (OLS) regression, we looked at how relative economic burden measured by person acres per dollar is influenced by income quartile, population density, and fire size by GIS_ACRES. Though we are particularly interested in income levels, we included population density and fire size in our regression to account for confounds that may affect burden variation. First, we look at model fit and significance. The r-squared value is 0.009, which means that income quartiles explain less than 1% of the variance in burden. This suggests that income quartiles alone does not significantly explain the economic burden caused by wildfires. The F-statistic, 17.93, derived from p-value = 1.14e-14 tells us that the model as a whole is statistically significant, but the low R^2 value suggests that the predictors are not strongly explaining variation in economic burden. The model also tells us that as population density increases, relative economic burden also increases, though by a small margin. This suggests that more densely populated areas experience higher economic burden per capita, but the effect size is small. The p-values for all income quartile variables are above 0.05, which means that none of them are statistically significant predictors- the data does not provide strong evidence that a county’s income level has a direct effect on the per-person economic burden from wildfires. Overall, we can conclude that population density has a small but significant positive effect on economic burden and that income quartile does not appear to significantly impact economic burden.

The project was successful however, in highlighting the fact that there are more factors than the obvious that influence relative economic burden. At the start of the project, we initially assumed that low income would equate to greater challenges and high income to easier recovery. But through our analysis, we discovered that there are many confounding factors that shape what defines relative economic burden. In the future, we hope this data contributes to a deeper understanding that factors beyond finances, such as policy, preparedness, and local advocacy, play a significant role in wildfire recovery.
