###  Loading the Data:

First step to cleaning the data is to load the data of course! We lead the data, and we drop the columns which we won't be needing since they're not part of our hypothesis.

```python
# load the data by importing pandas and reading the file

import pandas as pd

ls_set = pd.read_csv('include/lifestyle_dataset.csv')
ls_set = ls_set.drop(columns=['Extracurricular_Hours_Per_Day', 'Stress_Level'])
ls_set.head()
```

### Checking for nulls / NaNs:

We will check for nulls now. If 0, we have a good dataset, unless the NaNs/nulls are stored as 0 or as a string, but we will check more on that later. 

```python
# checking the total amount of null values in the entire dataset
# if 0, good dataset, else, we clean the data

num_nulls_NaNs = ls_set.isnull().sum().sum()
print(f"Total number of NaNs in the dataset: {num_nulls_NaNs}")
```

### Outliers:

In order to combat against the outliers, we will set hard limits to all of the columns. By doing so, we limit our range of numbers, but by doing that, we are also getting rid of all of any weird extreme outliers.

#### Study Hours:
- Bottom Limit set to 0, since you can't have negative study hours.
- Top Limit set to 10 since 10 hours is already a really large number.

#### Sleep Hours:
- Bottom Limit set to 0, since you can't have negative sleep hours.
- Top Limit set to 10 since 10 hours of sleep is enough in order to operate throughout the day.

#### Social Hours:
- Bottom Limit set to 0, since you can't have negative social hours.
- Top Limit set to 8 since 8 hours seems like a lot of social hours. This takes into consideration for those full-time talkers, especially those that think it's okay to have a full conversation mid-class (sorry...)!

#### Physical Activity Hours:
- Bottom Limit set to 0, since you can't have negative physical activity hours.
- Top Limit set to 10 since 10 hours is already a really large number.

#### GPA:
- Bottom Limit set to 0, since you can't have negative GPA.
- Top Limit set to 4, since GPA range is only from [0-4].

```python
# checking for outliers

# checking outliers for study hours per day
# if value is less than 0, set to 0
# if value is greater than 10, set to 10
ls_set['Study_Hours_Per_Day'] = ls_set['Study_Hours_Per_Day'].apply(lambda x: max(x, 0))
ls_set['Study_Hours_Per_Day'] = ls_set['Study_Hours_Per_Day'].apply(lambda x: min(x, 10))

# checking outliers for sleep hours per day
# if value is less than 0, set to 0
# if value is greater than 10, set to 10
ls_set['Sleep_Hours_Per_Day'] = ls_set['Sleep_Hours_Per_Day'].apply(lambda x: max(x, 0))
ls_set['Sleep_Hours_Per_Day'] = ls_set['Sleep_Hours_Per_Day'].apply(lambda x: min(x, 10))

# checking outliers for social hours per day
# if value is less than 0, set to 0
# if value is greater than 8, set to 8
ls_set['Social_Hours_Per_Day'] = ls_set['Social_Hours_Per_Day'].apply(lambda x: max(x, 0))
ls_set['Social_Hours_Per_Day'] = ls_set['Social_Hours_Per_Day'].apply(lambda x: min(x, 7))

# checking outliers for physical activity hours per day
# if value is less than 0, set to 0
# if value is greater than 10, set to 10
ls_set['Physical_Activity_Hours_Per_Day'] = ls_set['Physical_Activity_Hours_Per_Day'].apply(lambda x: max(x, 0))
ls_set['Physical_Activity_Hours_Per_Day'] = ls_set['Physical_Activity_Hours_Per_Day'].apply(lambda x: min(x, 10))

# checking outliers for GPA
# if value is less than 0, set to 0
# if value is greater than 4, set to 4
ls_set['GPA'] = ls_set['GPA'].apply(lambda x: max(x, 0))
ls_set['GPA'] = ls_set['GPA'].apply(lambda x: min(x, 4))
```

### Uniqueness:

Here, we check that there is only one unique entry for every student. By doing so, we make sure that there isn't any overlapping data that we have to take care of.

```python
 # checking if there are 2000 unique students

if ls_set['Student_ID'].nunique() == ls_set.shape[0]:
    print("There are 2000 unique students in the dataset.")
else:
    print("There isn't 2000 unique entries in the dataset.")
```

### Checking for nulls/NaNs:

In order to know how many nulls/NaNs are in our dataset, we will find how many nulls/NaNs are in each column. Then, we will add all of them. This means that if we have any number > 0, then we have nulls in our dataset, therefore, we will need to deal with those.

A problem that might occur in datasets is that there could be many features given for one subject. However, this will mess with our data, therefore, we have to make sure that each subject is unique to avoid any overlapping of any sort.

```python
# checking for unique subjects

if sleep['subject_id'].nunique() == sleep.shape[0]:
    print('All subjects are unique!')
else:
    print("Oh no! There seems to be multiple data for certain subjects.")
```

### Data Wrangling
We have to modify some columns in order to get what we want. First, we want the sleep_duration to be averaged per night, in order to be able to compare with the first dataset:

```python
# dividing by 60 as a result of 60 days in order to get average sleep per night

sleep['sleep_duration_avg'] = sleep['sleep_duration']/60
sleep['sleep_duration_avg'].describe()
```

Something else we will want to fix is the `overall_score` column since we can't really understand what a score of, for example, 273 points means. Is it good? Is it a passing grade? Therefore, we will divide the column by the total points available, in order to get a final score:

```python
# dividing by 370 total points to get total %

sleep['final_score'] = sleep['overall_score']/370
sleep['final_score'].describe()
```

Something else we can do now is to drop the columns that won't be needed. Since we modified `overall_score` and `sleep_duration`, we can drop those. We can also drop other coumns that we won't be using:

```python
# dropping columns that won't be needed

sleep = sleep.drop(columns=['overall_score', 'sleep_duration', 'sleep_std_minutes', 'sleep_quality_std', 'gender'])
```

### Outliers

Something really important that we have to deal with is outliers. Therefore, we will be dealing with them by:

#### Mean Sleep Quality:
- Bottom Limit set to 0, since that's the lowest possible choice.
- Top Limit set to 10, since that's the highest possible choice.

#### Sleep Duration AVG:
- Bottom Limit set to 0, since you can't have negative sleep hours.
- Top Limit set to 12 since 12 hours is a significantly large number, taking in consideration that a "healthy" amount of sleep is 8 hours.

#### Final Score:
- Bottom Limit set to .5, since .5 and below is considered failing.
- Top Limit set to 1 since you can't have higher than a 100% in this class.

```python
# checking outliers for Mean Sleep Quality
# if value is less than 0, set to 0
# if value is greater than 10, set to 10
sleep['mean_sleep_quality'] = sleep['mean_sleep_quality'].apply(lambda x: max(x, 0))
sleep['mean_sleep_quality'] = sleep['mean_sleep_quality'].apply(lambda x: min(x, 10))

# checking outliers for Sleep Duration avg
# if value is less than 0, set to 0
# if value is greater than 12, set to 12
sleep['sleep_duration_avg'] = sleep['sleep_duration_avg'].apply(lambda x: max(x, 0))
sleep['sleep_duration_avg'] = sleep['sleep_duration_avg'].apply(lambda x: min(x, 12))

# checking outliers for Final Score
# if value is less than .5, set to .5
sleep['final_score'] = sleep['final_score'].apply(lambda x: max(x, .5))
sleep['final_score'] = sleep['final_score'].apply(lambda x: min(x, 1))
```

## Interesting models!

For this section, we're going to train a model by splitting the data into train data and test data. We are going to train it usnig the train data, and then we will see how well it does with data that it hasn't seen. Then, if the model has a good accuracy, we will use the model to see if our hypothesis (6 hours of sleep, 1 hour of physical activity, 1 hour of social activity, 5 hours of study time) to see if this is a good hypothesis thusfar.

To start this, we are going to use the `statsmodels.api` and `sklearn` in order to train this model.

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score


# convert GPA to binary classification (1 if GPA ≥ 3.0, else 0)
ls_set["GPA_Binary"] = (ls_set["GPA"] >= 3.0).astype(int)

X = ls_set[["Sleep_Hours_Per_Day", "Study_Hours_Per_Day", 
            "Social_Hours_Per_Day", "Physical_Activity_Hours_Per_Day"]]
y = ls_set["GPA_Binary"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# standardize features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# add constant (for intercept)
X_train_ols = sm.add_constant(X_train_scaled)
X_test_ols = sm.add_constant(X_test_scaled)

ols_model = sm.OLS(y_train, X_train_ols).fit()
print(ols_model.summary())

y_pred_prob = ols_model.predict(X_test_ols)
y_pred_binary = (y_pred_prob >= 0.5).astype(int)

accuracy = accuracy_score(y_test, y_pred_binary)
print(f"Model Classification Accuracy: {accuracy:.2%}")

def predict_gpa_binary(sleep_hours, study_hours, social_hours, physical_activity_hours):
    input_data = np.array([[sleep_hours, study_hours, social_hours, physical_activity_hours]])

    input_scaled = scaler.transform(input_data)
    input_scaled = sm.add_constant(input_scaled, has_constant='add')

    predicted_prob = ols_model.predict(input_scaled)[0]
    predicted_class = int(predicted_prob >= 0.5)
    return predicted_class

# 6 hours of sleep, 5 hours of study, 1 hour of social, 1 hour of physical activity
gpa_class_hypothesis = predict_gpa_binary(6, 5, 1, 1)
print(f"Predicted GPA Class (given our hypothesis): {gpa_class_hypothesis} (1 = Good GPA, 0 = Low GPA)")
# 8 hours of sleep, 8 hours of study, 1 hour of social, 1 hour of physical activity
gpa_class_prediction = predict_gpa_binary(8, 8, 1, 1)
print(f"
Predicted GPA Class: {gpa_class_prediction} (1 = Good GPA, 0 = Low GPA)")
```

### Observations:

We see that the most important variable to a student's GPA is the total amount of study hours that they do throughout a day. We can see that our hypothesis wasn't the best prediction. We really underestimated how much people actually study throughout the day. Therefore, we thought about the "healthiest" sleep hours is 8 hours, since that's what doctors actually recommend, and then we have 8 hours of studying, since being a full-time student means that it's a full-time commitment.

### About the model:

The first thing that I did in order to start this model is to import various packages. Some of these include sklearn's `StandardScaler()` which is really important since we are working with different units when it comes to GPA and hours. GPA works on a range of 0-4 while hours per day has a range of 0-24, which is completely different. Therefore, we set the mean of all columns to be 0 and then find the std of each, standardizing them, in order to be able to add a model to it.

A problem that we ran into was finding how accurate our model is. So, to combat this, we made the decision boundary to be when `GPA == 3.0`. So, if `GPA >= 3.0`, we'll have a `1`. And vice versa, but when it's less than, we'll have a `0`. This helps us just have a binary classifier by having discrete variables isntead of continuous. So, when we have our test data, if `GPA >= 3.0`,or `GPA_binary == 1`, if the model predicts `1`, then that is an accurate prediction, and same thing for when `GPA_binary == 0`. And in order to make this calculation a lot easier, we just use the package given by `sklearn`, which is `accuracy`. And with this model, we get an accuracy which ranges between 78-83%, depending on which data points it uses during the splits.

The rest is pretty simple, including adding a constant in order to take in count for the intercept. Then, we have a function which we can input our own values and it returns a `1` or a `0`.
