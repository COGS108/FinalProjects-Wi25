## Background and Prior Work

The implementation of Seattle’s local minimum wage policy, phased in from 2015 to 2023, has sparked significant debate over its impact on employment rates, particularly when compared to cities like Houston, Texas, which adhered to the federal minimum wage during the same period. Before the policy’s introduction, Seattle’s minimum wage was ＄9.47 per hour in 2014, which was already higher than the federal minimum wage of ＄7.25 but still considered insufficient to meet the city’s high cost of living<a name="cite_ref-1"></a><sup>1</sup> . As one of the fastest-growing metropolitan areas in the U.S., Seattle faced escalating housing costs and inflation, prompting policymakers to adopt the $15 minimum wage as a measure to address income inequality and improve worker well-being. The policy applied distinct requirements for large employers (500+ employees) and smaller businesses, with the former facing steeper wage increases earlier in the timeline.

Proponents of the policy argued that raising wages would reduce income inequality and improve living standards for low-wage workers, while opponents warned of potential job losses and operational strain, particularly for small businesses operating on thin margins<a name="cite_ref-2"></a><sup>2</sup> . Prior research on Seattle’s policy has yielded conflicting conclusions, reflecting methodological differences and scope limitations. A landmark 2017 study by the University of California, Berkeley analyzed employment trends across various industries and found that wage hikes led to a 1–2.3% increase in wages with no significant job losses, particularly among large employers in Seattle<a name="cite_ref-3"></a><sup>3</sup> . However, a contemporaneous University of Washington (UW) study contradicted these findings, reporting a 10% reduction in hours worked and a ＄125 monthly earnings loss for low-wage workers after the second wage increase to ＄13/hour. Critics noted that the UW study excluded multi-location businesses, which account for approximately 40% of Seattle’s workforce, potentially skewing results toward smaller employers<a name="cite_ref-4"></a><sup>4</sup> . These divergent outcomes highlight the importance of considering labor market structure: in concentrated markets, employers with wage-setting power may absorb higher wages without cutting jobs, while smaller businesses may face greater challenges.

Further insights into the effects of minimum wage increases come from comparisons with cities like Houston, Texas, which adhered to the federal minimum wage during the same period. Houston provides a useful counterfactual for evaluating Seattle’s experience, as it shares some economic similarities but differs significantly in its wage policy. Research has shown that regions without significant minimum wage increases often experience slower wage growth but also fewer immediate disruptions to employment levels<a name="cite_ref-5"></a><sup>5</sup> . By comparing Seattle’s employment trends to those in Houston, this project aims to contribute new insights into the nuanced relationship between minimum wage laws and overall employment outcomes, while addressing methodological challenges such as selection bias and geographic variability.

Attention must be given to threats to validity when interpreting these studies. For example, the exclusion of multi-location businesses in the UW study introduces selection bias, while survivorship bias in datasets like Yelp may undercount business closures. Additionally, the geographic focus of Seattle’s policy limits generalizability to other regions with differing economic conditions. By building on these studies and comparing Seattle’s experience to Houston’s, this project aims to provide a comprehensive analysis of the broader employment impacts of minimum wage policies.

## Merging

We merge the BLS employment dataset with the DOL minimum wage dataset to analyze how changes in minimum wage correlate with employment trends. This integration would allow us to examine the relationship between wage policies and employment levels over time within a single framework. The merged dataset preserves key variables for our analysis, enabling us to draw more meaningful insights from the data.

Here, we introduce the employment index, which we use for our later analysis. The employment index is a normalized measure that allows us to compare employment trends over time relative to a baseline year. Instead of looking at absolute employment levels, which may vary greatly due to population size and economic differences between cities, the index helps us analyze relative employment growth or decline in Seattle and Houston. We set 2015 as the baseline year (index = 100) and calculated the employment index for each subsequent year using the formula:

$$\text{Employment Index}_{t} = \frac{\text{Employment Level}_t}{\text{Employment Level}_{2015}} \times 100 ,$$

where $t$ representing the year between 2015 and 2023. This was computed separately for both Seattle and Houston, allowing for a direct comparison of employment trends over time. 

Since Seattle and Houston have different population sizes and economies, comparing absolute employment levels would not be meaningful. By indexing employment to 2015 = 100, we can directly compare relative changes in employment over time. 

```python
# aggregate employment data by year
seattle_emp_by_year = seattle.groupby('year')['annual_avg_emplvl'].sum().reset_index()
seattle_emp_by_year.columns = ['Year', 'Employment']

houston_emp_by_year = houston.groupby('year')['annual_avg_emplvl'].sum().reset_index()
houston_emp_by_year.columns = ['Year', 'Employment']

# calculate indexed employment growth (2015 = 100)
seattle_base_emp = seattle_emp_by_year[seattle_emp_by_year['Year'] == 2015]['Employment'].values[0]
houston_base_emp = houston_emp_by_year[houston_emp_by_year['Year'] == 2015]['Employment'].values[0]

seattle_emp_by_year['Index'] = (seattle_emp_by_year['Employment'] / seattle_base_emp) * 100
houston_emp_by_year['Index'] = (houston_emp_by_year['Employment'] / houston_base_emp) * 100

# calculate year-over-year percentage changes
seattle_emp_by_year['YoY_Change'] = seattle_emp_by_year['Employment'].pct_change() * 100
houston_emp_by_year['YoY_Change'] = houston_emp_by_year['Employment'].pct_change() * 100

# merge minimum wage and employment data
seattle_merged = pd.merge(seattle_emp_by_year, seattle_minwage[['Year', 'Wage']], on='Year')
houston_merged = pd.merge(houston_emp_by_year, tx_minwage[['Year', 'Wage']], on='Year')

# display the merged datasets
display('seattle employment summary', seattle_merged)
display('houston employment summary', houston_merged)
```

### OLS Seattle MinWage and DOL Washington and Texas Data: Outliers


```python
# function to detect outliers using Z-score
def find_outliers_zscore(data, column, threshold=2.5):
    # check if the column has any variability
    if np.std(data[column]) == 0:
        print(f"Warning: Column '{column}' has no variability. Skipping Z-score calculation.")
        return data.iloc[0:0]  # return an empty DataFrame
    
    z_scores = np.abs(stats.zscore(data[column]))
    return data[z_scores > threshold]

# function to detect outliers using IQR
def find_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return data[(data[column] < lower_bound) | (data[column] > upper_bound)]

# find outliers in Seattle, Washington, and Texas datasets using both methods
print("Outliers in Seattle Minimum Wage Data (Z-score method):")
print(find_outliers_zscore(seattle_minwage, 'Wage'))

print("\nOutliers in Seattle Minimum Wage Data (IQR method):")
print(find_outliers_iqr(seattle_minwage, 'Wage'))

print("\nOutliers in Texas Minimum Wage Data (Z-score method):")
print(find_outliers_zscore(tx_minwage, 'Wage'))

print("\nOutliers in Texas Minimum Wage Data (IQR method):")
print(find_outliers_iqr(tx_minwage, 'Wage'))
```

### Realtionship Between Variables

In Seattle’s minimum wage data:
- Dependent Variable: Wage (minimum wage over time)
-  Independent Variable: Year (time)
In Texas’s minimum wage data:
- Dependent Variable: Wage (minimum wage) (renamed from STTMINWGTX)
- Independent Variable: Year (time)


```python
# extract the Wage data for Seattle and Texas
seattle_wages = seattle_minwage['Wage']
texas_wages = tx_minwage['Wage']

# Step 1: check normality (Shapiro-Wilk Test)
# for Seattle
shapiro_seattle = stats.shapiro(seattle_wages)
print(f"Shapiro-Wilk Test for Seattle: Statistic={shapiro_seattle.statistic}, P-value={shapiro_seattle.pvalue}")

# for Texas, skip the test because the data is constant
if np.std(texas_wages) == 0:
    print("Texas data is constant. Skipping Shapiro-Wilk test for normality.")
else:
    shapiro_texas = stats.shapiro(texas_wages)
    print(f"Shapiro-Wilk Test for Texas: Statistic={shapiro_texas.statistic}, P-value={shapiro_texas.pvalue}")

# Step 2: check variance (Levene's Test)
with warnings.catch_warnings():
    # suppress RuntimeWarnings caused by constant data
    warnings.filterwarnings("ignore", category=RuntimeWarning)
    levene_test = stats.levene(seattle_wages, texas_wages)

print(f"Levene's Test for Equal Variances: Statistic={levene_test.statistic}, P-value={levene_test.pvalue}")

# Step 3: perform the t-Test
if levene_test.pvalue < 0.05:
    print("Variances are unequal. Using Welch's t-test.")
    with warnings.catch_warnings():
        # suppress RuntimeWarnings caused by constant data
        warnings.filterwarnings("ignore", category=RuntimeWarning)
        t_stat, p_value = stats.ttest_ind(seattle_wages, texas_wages, equal_var=False)
else:
    print("Variances are equal. Using Student's t-test.")
    with warnings.catch_warnings():
        # suppress RuntimeWarnings caused by constant data
        warnings.filterwarnings("ignore", category=RuntimeWarning)
        t_stat, p_value = stats.ttest_ind(seattle_wages, texas_wages, equal_var=True)

print(f"T-Test Results: T-statistic={t_stat}, P-value={p_value}")
```

1. Shapiro-Wilk Test (Seattle):

    - Result: p-value = 0.966 (fails to reject normality at α=0.05).
    - Interpretation: Seattle’s wage data approximately follows a normal distribution, allowing parametric tests.

2. Levene’s Test:
    - Result: p-value = 0.0003 (rejects equal variances).
    - Interpretation: Variances between Seattle and Texas wage data are unequal, necessitating Welch’s t-test.

3. Welch’s t-test:

    - Result: p-value = 0.0000198 (rejects equal means).
    - Interpretation: Statistically significant difference in mean wages between Seattle and Texas, confirming divergent wage policies.

### Difference-in-Differences Analysis

```python
# create a combined dataframe for DiD analysis
did_data = pd.DataFrame({
    'Year': seattle_merged['Year'],
    'Seattle_Employment': seattle_merged['Index'],
    'Houston_Employment': houston_merged['Index'],
    'Seattle_Wage': seattle_merged['Wage'],
    'Houston_Wage': houston_merged['Wage']
})

# calculate differences
did_data['Employment_Diff'] = did_data['Seattle_Employment'] - did_data['Houston_Employment']
did_data['MinWage_Diff'] = did_data['Seattle_Wage'] - did_data['Houston_Wage']

# split it up
above0 = np.maximum(did_data['Employment_Diff'], 0)
below0 = np.minimum(did_data['Employment_Diff'], 0)

# plot the differences over time
plt.figure(figsize=(10, 6))
plt.bar(did_data['Year'], above0, alpha=0.7, width=0.6)
plt.bar(did_data['Year'], below0, alpha=0.7, width=0.6)
plt.axhline(y=0, color='r', linestyle='-', alpha=0.3)
plt.title('Difference in Employment Index: Seattle - Houston', fontsize=16)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Difference in Employment Index', fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# summary statistics for the difference in differences
print("\nDifference-in-Differences Summary:")
display(did_data[['Year', 'Employment_Diff', 'MinWage_Diff']])

# statistical summary of the differences
print("\nStatistical Summary of Employment Differences:")
display(did_data[['Employment_Diff']].describe())

# check for outliers in the differences
q1 = did_data['Employment_Diff'].quantile(0.25)
q3 = did_data['Employment_Diff'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = did_data[(did_data['Employment_Diff'] < lower_bound) | 
                     (did_data['Employment_Diff'] > upper_bound)]

if not outliers.empty:
    print("\nOutliers in Employment Differences:")
    print(outliers[['Year', 'Employment_Diff']])
else:
    print("\nNo outliers detected in the employment differences.")

# add final correlation between minimum wage difference and employment difference
mw_empl_diff_corr = did_data['MinWage_Diff'].corr(did_data['Employment_Diff'])
print(f"\nCorrelation between Min Wage Difference and Employment Difference: {mw_empl_diff_corr:.4f}")
```

## Approaches Used for Analysis
1. Exploratory Data Analysis (EDA):
- Why: EDA is essential to understand the structure, patterns, and anomalies in the data before formal hypothesis testing. It helps identify trends, outliers, and relationships between variables.
- Approach:
    - Summarized key distributions (minimum wage, employment index, year-over-year changes).
    - Visualized trends using line plots and scatterplots.
    - Detected outliers using Z-score and IQR methods.
    - Calculated summary statistics (mean, standard deviation, etc.) for employment and wage data.
- Results:
    - Seattle’s minimum wage increased steadily from $11 in 2015 to $18.69 in 2023, with a mean of $14.82 and a standard deviation of $2.56.
    - Houston’s minimum wage remained constant at $7.25 throughout the period.
    - Both cities experienced a sharp drop in employment during the pandemic (2020), followed by recovery.
    - The difference in employment indices (Seattle - Houston) had a mean of -1.39%, indicating Houston slightly outperformed Seattle on average.

2. Statistical Testing (t-Test):
    - Why: To determine whether the observed differences in minimum wages between Seattle and Texas were statistically significant.
    - Approach:
        - Conducted a Shapiro-Wilk test for normality and Levene’s test for variance equality.
        - Performed Welch’s t-test due to unequal variances.
    - Results:
    - The t-test confirmed a significant difference in mean wages between Seattle and Texas (T-statistic = 8.9185, P-value = 1.98e-05). This reflects systemic differences in wage policies.

3. Correlation Analysis:
    - Why: To quantify the relationship between minimum wage increases and employment outcomes.
    - Approach:
        - Calculated Pearson correlation coefficients between minimum wage and employment index for Seattle and Houston.
    - Results:
        - Seattle showed a strong positive correlation (0.8065) between minimum wage and employment index, suggesting that rising wages may have contributed to employment growth or stability.
        - Houston’s correlation was NaN due to zero variance in its minimum wage.

4. Difference-in-Differences (DiD) Analysis:
    - Why: DiD is a quasi-experimental method to estimate the causal effect of an intervention (Seattle’s minimum wage policy) by comparing treated (Seattle) and control (Houston) groups over time.
    - Approach:
        - Created a combined dataset with employment and wage differences between Seattle and Houston.
        - Plotted and analyzed trends in employment differences and wage gaps over time.
        - Calculated the correlation between wage differences and employment differences.
    - Results:
        - Employment differences fluctuated over time:
            - Positive in 2018–2019 (Seattle outperformed Houston).
            - Negative in 2020–2023 (Houston outperformed Seattle post-pandemic).
        - The mean employment difference was -1.39%, with a standard deviation of 3.40.
        - Wage differences widened consistently from $3.75 in 2015 to $11.44 in 2023.
        - A moderate negative correlation (-0.6666) was observed between wage differences and employment differences, indicating that as Seattle’s wage advantage grew, its relative employment performance worsened compared to Houston.

# Ethics & Privacy

Ethical Considerations for Analyzing Seattle’s ＄15 Minimum Wage Policy Using New Datasets:

1. **Data Biases and Representativeness**
    - **Aggregation Bias** : Government datasets often aggregate data at broad geographic levels (e.g., county or metro area). This may obscure neighborhood-level disparities, particularly in marginalized communities.
    - **Industry Misclassification** : Employment data categorized by industry codes (e.g., NAICS) might not accurately distinguish between - small, independent businesses and large chains within the same sector.
    - **Temporal Lag** : Annual or quarterly data may not capture short-term adjustments caused by minimum wage increases, leading to incomplete conclusions about immediate impacts.
    - **Mitigation** :
        - Use high-frequency data (e.g., monthly employment reports) where available to analyze short-term effects.
        - Stratify results by business size, location, and demographic characteristics to identify inequitable impacts.
        - Disclose limitations related to data aggregation and temporal resolution in the methodology section.
2. **Privacy and Terms of Use**
    - **Business Identifiability** : While aggregated government data does not typically identify individual businesses, there is still a risk of indirectly revealing sensitive information about specific establishments through detailed analyses.
    - **Data Licensing** : Government datasets are generally public domain but may have restrictions on redistribution or commercial use.
    - **Mitigation** :
        - Ensure all findings are reported at sufficiently aggregated levels to avoid identifying specific businesses.
        - Comply with data-sharing agreements and provide proper attribution when using government datasets.
        - Avoid redistributing raw data unless explicitly permitted by the source.
3. **Equitable Impact and Interpretation**
    - **Worker Demographics** : Low-wage workers disproportionately include minorities, women, and immigrants. Any findings suggesting job losses due to the minimum wage increase could be misused to argue against such policies, potentially harming these vulnerable groups.
    - **Omitted Variables** : Government datasets may lack granular details about worker demographics, business ownership types (e.g., minority-owned vs. corporate), or informal labor markets, which could mask differential impacts across groups.
    - **Mitigation** :
        - Contextualize findings with broader literature showing positive outcomes of higher wages, such as reduced poverty rates and improved living standards.
        - Highlight that correlation does not imply causation and emphasize the importance of considering policy trade-offs.
        - Supplement quantitative findings with qualitative studies or surveys capturing lived experiences of affected workers.
4. **Political Sensitivity and Communication**
    - **Policy Implications** : Results from this analysis could influence highly politicized debates about minimum wage laws. Misinterpretation or selective reporting of findings might sway public opinion or policy decisions.
    - **Media Representation** : Simplistic narratives (e.g., “minimum wage kills jobs”) can overshadow nuanced discussions about long-term benefits versus short-term costs.
    - **Mitigation** :
        - Provide transparent documentation of the methodology, including code, assumptions, and limitations, to foster reproducibility and trust.
        - Collaborate with advocacy groups representing workers and businesses to ensure balanced messaging.
        - Clearly communicate both the potential benefits (e.g., higher wages, reduced inequality) and challenges (e.g., possible employment reductions) associated with the policy.
5. **Compliance with Ethics Frameworks**
    - **Documented Biases** : Disclosed issues such as aggregation bias, temporal lag, and potential underrepresentation of marginalized communities.
    - **Protected Privacy** : Ensured no personally identifiable information (PII) or business-specific identifiers were included in the analysis.
    - **Assessed Social Impact** : Evaluated how our findings might affect low-wage workers, small businesses, and disadvantaged communities.
