# Research Question

What is the correlation between transit stop density and passenger volume with the local incidence of violent and property crimes near transit nodes in Chicago and New York City from 2014 to 2021? Specifically, does higher transit stop density and increased ridership correspond to elevated crime counts, and how does normalizing crime by passenger volume affect this relationship?

## Background and Prior Work

Urban planners and criminologists have long sought to answer how public transportation systems influence local crime rates, especially in major metropolitan areas. The common hypothesis is that transit accessibility can function either as a deterrent or attractor for opportunistic crimes where large crowds gather every day. By exploring the interplay between different factors researchers hope to inform evidence based transportation and policing in order to create safer and better cities. 

Several studies illustrate the complexity of this relationship. Ridgeway and MacDonald<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) investigated Los Angeles Metro Rail expansions between 1988 and 2014. They leveraged major strikes as natural experiments to examine crime trends in neighborhoods that are affected by disrupted transit systems. Furthermore, they examined the effects of newly established transit access. Contrary to earlier assumptions that transit stops serve as a crime magnet, they failed to find significant impact of rail services on local crime rates.

This stands in contrast to the result Neiss<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) evaluated with short term effects of the new bus line introduced in Cleveland. From this incident Neiss was able to detect a measurable uptick in property crime following the introduction of that new line. Meanwhile, Newton’s broader survey<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) of crime on public transport underscored the importance of context from his observations that crimes often cluster at specific stations or nodes rather than uniformly throughout the transit system. 

Considering these findings together, it would be necessary to implement large data driven approaches that include geographic and socioeconomic conditions alongside the transit and crime metrics, to determine how public transportation contributes to changes in crime patterns. As prior works have mixed results, and focus on single cities or short term effects, our research should aim to analyze multiple major cities and long term data such that offers broader perspective on how transit accessibility affects crime rate. Through this we expect to fill the gap in the academic and policy discourse, clarifying whether certain patterns hold across diverse urban environments and its transportation.

We hypothesize that higher transit stop density and increased passenger volume are positively correlated with the incidence of both violent and property crimes near transit nodes in Chicago and New York City from 2014 to 2021. Specifically, we expect that areas with greater transit accessibility will exhibit higher raw crime counts due to increased foot traffic; however, normalizing crime by passenger volume will likely attenuate this relationship, indicating that the apparent increase in crime is largely driven by the volume of people present rather than an inherent propensity for crime. While socioeconomic factors undoubtedly influence crime rates, our analysis focuses on transit metrics and crime data to isolate this relationship.

## Dataset 1 : NYPD Arrests Data (Historic)

```python
# data
nypd = pd.read_csv('nypd.csv')

# Cleaning
nypd['ARREST_DATE'] = pd.to_datetime(nypd['ARREST_DATE'], errors='coerce')
nypd['YEAR'] = nypd['ARREST_DATE'].dt.year
nypd = nypd[(nypd['YEAR'] >= 2014) & (nypd['YEAR'] <= 2021)]
print("NYPD dataset shape:", nypd.shape)

# sort
nypd = nypd.sort_values(by='ARREST_DATE', ascending=True)
display(nypd)
```

## Dataset 3 : MTA Subway Stations

```python
# data
mta_stations = pd.read_csv('MTA_Subway_Stations.csv')

# Cleaning
mta_stations = mta_stations.dropna(subset=['GTFS Latitude', 'GTFS Longitude'])
mta_stations['GTFS Latitude'] = pd.to_numeric(mta_stations['GTFS Latitude'], errors='coerce')
mta_stations['GTFS Longitude'] = pd.to_numeric(mta_stations['GTFS Longitude'], errors='coerce')

print("MTA stations dataset shape:", mta_stations.shape)
display(mta_stations)
```

## Dataset 4 : MTA Subway Turnstile Usage Data

```python
def processTurnstileData(csv_file):
    # data
    df = pd.read_csv(csv_file, low_memory=False)

    # Cleaning
    df.columns = df.columns.str.strip()
    df['Date'] = pd.to_datetime(df['Date'], format='%m/%d/%Y', errors='coerce')
    df = df.dropna(subset=['Date'])
    df = df.sort_values(by=['C/A', 'Unit', 'SCP', 'Date', 'Time'])

    # Calculate differences per group for each turnstile (C/A + Unit + SCP)
    df['Entry_diff'] = df.groupby(['C/A', 'Unit', 'SCP'])['Entries'].diff().fillna(0)
    df['Exit_diff'] = df.groupby(['C/A', 'Unit', 'SCP'])['Exits'].diff().fillna(0)

    # Counter resets or errors
    df['Entry_diff'] = df['Entry_diff'].clip(lower=0)
    df['Exit_diff'] = df['Exit_diff'].clip(lower=0)

    return df
```

```python
# Loop over all the CSV files from 2014-2021
dataList = []
for year in range(2014, 2022): 
    csv_file = f"MTA_Subway_Turnstile_Usage_Data__{year}.csv"
    print(f"Processing file: {csv_file}")
    
    daily_summary = processTurnstileData(csv_file)
    dataList.append(daily_summary)

# Merge all yearly dataframes into one
mergedDailyTurnstile = pd.concat(dataList, ignore_index=True)

print(f"CTA Turnstile dataset shape (2014-2021): {mergedDailyTurnstile.shape}")
print("Merged Turnstile Daily Data (2014-2021):")
display(mergedDailyTurnstile)
```

## A. NYC Quantitative Modeling Workflow

### 1. Spatial Aggregation for NYC

```python
# Group by stations and sum the "Entry_diff"
passenger_volume = mergedDailyTurnstile.groupby('Station')['Entry_diff'].sum().reset_index()
passenger_volume.columns = ['Station', 'total_volume']

# Cleaning
mta_stations['clean_stop'] = mta_stations['Stop Name'].str.strip().str.lower()
passenger_volume['clean_station'] = passenger_volume['Station'].str.strip().str.lower()

# GeoDataFrame for MTA
gdf_mta = gpd.GeoDataFrame(
    mta_stations,
    geometry=gpd.points_from_xy(mta_stations['GTFS Longitude'], mta_stations['GTFS Latitude']),
    crs="EPSG:4326"
)
gdf_mta = gdf_mta.to_crs(epsg=3857)
gdf_mta['buffer'] = gdf_mta.geometry.buffer(500)

# Merge
gdf_mta = gdf_mta.merge(passenger_volume, left_on='clean_stop', right_on='clean_station', how='left')
gdf_mta['total_volume'] = gdf_mta['total_volume'].fillna(0)

# GeoDataFrame for NYC Crimes
nypd_valid = nypd.dropna(subset=['Longitude', 'Latitude'])
gdf_nypd = gpd.GeoDataFrame(
    nypd_valid,
    geometry=gpd.points_from_xy(nypd_valid['Longitude'], nypd_valid['Latitude']),
    crs="EPSG:4326"
)
gdf_nypd = gdf_nypd.to_crs(epsg=3857)

# Count Crimes around the station's buffer
def count_crimes(buffer_geom, crime_gdf):
    total = crime_gdf[crime_gdf.within(buffer_geom)].shape[0]
    property_count = crime_gdf[(crime_gdf.within(buffer_geom)) & (crime_gdf['CRIME_TYPE'] == 'Property Crime')].shape[0]
    violent_count = crime_gdf[(crime_gdf.within(buffer_geom)) & (crime_gdf['CRIME_TYPE'] == 'Violent Crime')].shape[0]
    return pd.Series({'total_crimes': total, 'property_crimes': property_count, 'violent_crimes': violent_count})

# Apply the function to each MTA station buffer
crime_counts = gdf_mta['buffer'].apply(lambda buf: count_crimes(buf, gdf_nypd))
gdf_mta = pd.concat([gdf_mta, crime_counts], axis=1)

# Crime Rates 
gdf_mta['property_crime_rate'] = gdf_mta['property_crimes'] / (gdf_mta['total_volume'] + 1) * 1e6
gdf_mta['violent_crime_rate'] = gdf_mta['violent_crimes'] / (gdf_mta['total_volume'] + 1) * 1e6


plt.figure(figsize=(8,5))
plt.hist(gdf_mta['property_crime_rate'], bins=20)
plt.xlabel('Property Crime Rate (per 1e6 passengers)')
plt.ylabel('Number of Stations')
plt.title('Distribution of Property Crime Rates at NYC Stations')
plt.show()
```

Figure 11: The distribution of property crime rates at NYC subway stations is shown in Figure 11. The majority of stations display slight property crime rates; however, a few demonstrate substantially elevated crime rates. This biased distribution indicates that while transportation hubs may promote property crime, only some sites encounter considerable criminal activity. Additional research is required to identify the variables leading to these elevated-crime stations.

```python
# Data for modeling NYC
df_nyc_model = gdf_mta[['total_volume', 'property_crimes', 'violent_crimes']].copy()
df_nyc_model['log_total_volume'] = np.log(df_nyc_model['total_volume'] + 1)
df_nyc_model['log_property_crimes'] = np.log(df_nyc_model['property_crimes'] + 1)
df_nyc_model['log_violent_crimes'] = np.log(df_nyc_model['violent_crimes'] + 1)
```

```python
# OLS Regression for NYC Property Crimes 
X_prop = df_nyc_model[['log_total_volume']]
X_prop = sm.add_constant(X_prop)
y_prop = df_nyc_model['log_property_crimes']
model_prop = sm.OLS(y_prop, X_prop).fit()
print("NYC - Property Crimes Regression:")
print(model_prop.summary())
```

```python
# OLS Regression for NYC Violent Crimes
X_viol = df_nyc_model[['log_total_volume']]
X_viol = sm.add_constant(X_viol)
y_viol = df_nyc_model['log_violent_crimes']
model_viol = sm.OLS(y_viol, X_viol).fit()
print("NYC - Violent Crimes Regression:")
print(model_viol.summary())
```

# Spatial Autoregressive Models

First, we collect the locations of transit stations from our data. Then, we create a spatial weights matrix using the K nearest neighbors method, linking each station to its four closest neighbors. We also standardize these weights to make comparisons fair across all stations. This step helps us measure how crime at one station might be related to crime at nearby stations. The weights matrix defines which stations are neighbors, and standardizing ensures the influence of each neighbor is balanced. If this setup is incorrect, our results might be misleading.

Next, we use a Spatial Autoregressive (SAR) model to study how crime levels at one station are influenced by its neighbors. The model estimates crime based on factors like transit volume and nearby crime rates, using a technique called Maximum Likelihood Estimation.
Crime often clusters in space, meaning crime at one station can impact nearby stations. The SAR model accounts for this spillover effect, giving a more realistic picture of the relationship between transit use and crime.

```python
import libpysal
from spreg import ML_Lag
```

### SAR NYC Property Crime

Here we are setting up the spatial context by extracting station coordinates from the gdf_mta and generating a KNN spatial weights matrix. Using k=4 means each station’s neighborhood is defined by its four closest stations. Row standardizing ensures that each station's influence is scaled to sum to one. A well defined weights matrix is critical because it directly affects the estimation of spatial spillover effects in the subsequent SAR models.

```python
coords = np.array(list(zip(gdf_mta.geometry.x, gdf_mta.geometry.y)))


kWeightsNYC = libpysal.weights.KNN.from_array(coords, k=4)
kWeightsNYC.transform = 'r'
y = df_nyc_model['log_property_crimes'].values.reshape(-1, 1)
X = df_nyc_model[['log_total_volume']].values  
sarNYCProperty = ML_Lag(y, X, w=kWeightsNYC, name_y='log_property_crimes', name_x=['log_total_volume'])

print(sarNYCProperty.summary)
```

The dependent variable y is the log transformed property crime counts, and the independent variable X is the log transformed total transit volume. The model equation is y=ρWy+Xβ+ϵ, where ρ is the spatial lag coefficient which holds the influence of neighboring stations. Using Maximum Likelihood estimation via ML_Lag function, this accounts for both direct effects such as local influence of transit volume and indirect effects like spillover effects from nearby stations. 

The results show that a 1% increase in transit volume leads to a small increase about 0.011% in property crimes at a station. This suggests that busier stations tend to experience slightly more crime. The spatial lag coefficient 0.682 is strong and highly significant, meaning that crime at one station is closely linked to crime at nearby stations. In other words, crime patterns spread across locations rather than being isolated to a single station.

### SAR NYC Violent Crime

```python
yNYCViolent = df_nyc_model['log_violent_crimes'].values.reshape(-1, 1)
sarNYCViolent = ML_Lag(yNYCViolent, X, w=kWeightsNYC, name_y='log_violent_crimes', name_x=['log_total_volume'])
print(sarNYCViolent.summary)
```

 Here we examines violent crime at transit stations, similar to the property crime analysis. The dependent variable is the log of violent crime counts at each station. This model estimates violent crimes at NYC transit stations, using the log of violent crime counts as the dependent variable. Similar to the property crime analysis, the results show that transit volume has a small but significant effect on crime. 1% increase in transit volume is associated with a 0.016% rise in violent crimes, suggesting that busier stations experience slightly more incidents. The spatial lag coefficient of 0.606 confirms that violent crimes are not isolated but tend to cluster across nearby stations. This means that crime at one location is influenced by crime levels at surrounding stations.
