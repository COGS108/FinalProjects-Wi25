# Hypothesis


Our hypothesis is that AI-generated text can be identified based on word frequency, sentence structure, and overall complexity. AI text tends to use less diverse vocabulary, more uniform sentence structures, and predictable word choices compared to human writing. It also overuses unnecessarily formal words and phrases that a human would naturally simplify. Additionally, AI often avoids contractions and follows rigid sentence structures, making its writing feel unnatural.  

We predict that detection tools leveraging these measures might achieve 80–90% accuracy, with false positive rates around 15–20% and false negatives between 10–15%.

```python
##Import all require Python Packages
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## AI Content Detection Responses Dataset Cleaning

```python
# Import csv file from 
# https://edintegrity.biomedcentral.com/articles/10.1007/s40979-023-00140-5/tables/3
# The website does not derectly provide a csv file so had to paste the data on google sheet and convert to csv
df  = pd.read_csv('~/private/Group082_WI25/ai_detection_responses.csv')

# Renaming the columns and index of the dataframe
df.columns = [
    "Generated_Content", "OpenAI_Classifier", "WRITER", 
    "Crossplag_Percentage", "Crossplag_Classification", 
    "Copyleaks_Percentage", "Copyleaks_Classification", 
    "GPTZero_Percentage", "GPTZero_Classification"
]

# Data cleaning
#Check if there is null data and remove the rows
rows_with_null = df[df.isnull().any(axis=1)]
rows_with_null.shape
df = df.dropna()

#Covert percent to number
for col in ["Crossplag_Percentage", "Copyleaks_Percentage", "GPTZero_Percentage"]:
    df[col] = df[col].str.replace('%', '').astype(float)
df_detector = df
df_detector.head()
```

## GPT WIKI Intro Dataset Cleaning

```python
# 1. Initial Setup
from datasets import load_dataset
from textstat import flesch_reading_ease

sns.set(font_scale=1.2)
plt.rcParams['figure.figsize'] = (12, 6)


# 2. Load & Validate Data
dataset = load_dataset("aadityaubhat/GPT-wiki-intro")
df = dataset['train'].to_pandas()

print("First 5 rows:")
display(df[['wiki_intro', 'generated_intro']].head(5))
print("\nMissing values:")
print(df[['wiki_intro', 'generated_intro']].isnull().sum())


# 3. Combine Samples
human = df[['wiki_intro']].assign(label='Human')
ai = df[['generated_intro']].rename(columns={'generated_intro':'text'}).assign(label='AI')
combined = pd.concat([human.rename(columns={'wiki_intro':'text'}), ai])
print(f"Total samples: {len(combined):,}")
display(combined.sample(5))
```

## Detailed Text Analysis: AI-Generated Text vs. Human-Written Text

In this section, we compare **readability scores, vocabulary diversity, sentence structures, and contraction usage** to gain a more detailed understanding of which specific text patterns best distinguish AI-generated text from human-written text.

#### **2.1 Readability Scores**

Flesch Reading Ease score measures text complexity:
- Scores are range from -200 to 100
- Higher scores = easier to read
- Lower scores = harder to read
- A score of 60 corresponds to "High School Level" readability

```python
combined['flesch'] = combined['text'].apply(
    lambda x: flesch_reading_ease(x) if isinstance(x, str) else None
)

# Handle missing values in readability scores
combined = combined.dropna(subset=['flesch'])

# Create density plot for readability scores
plt.figure(figsize=(12,6))
sns.kdeplot(data=combined, x='flesch', hue='label', fill=True, alpha=0.5, palette=['#1f77b4', '#ff7f0e'])
plt.axvline(60, color='red', linestyle='--', linewidth=1.5, label='High School Level (Score = 60)')

# Add plot annotations
plt.title('Readability Scores Distribution\n(Higher = Easier to Read)', fontsize=14)
plt.xlabel('Flesch Reading Ease Score', fontsize=12)
plt.ylabel('Density', fontsize=12)
plt.legend(title='Legend', labels=['AI', 'Human', 'High School Level'])

plt.show()
```

**Discussion on KDE plot:**

The KDE plot compares the readability scores (Flesch Reading Ease) of human-written and AI-generated Wikipedia introductions. Overall, human texts clustered around lower scores as compared to AI-generated texts. This shows that AI texts might be more consistent and formulaic in their sentences which makes it easier to read than human-generated wikipedia entries on various complex topics.

#### **2.2 Vocabulary Diversity Analysis**

Calculate Type-Token Ratio (unique words / total words) measures vocabulary diversity: 
- higher values indicate richer vocabulary.

```python
combined['unique_ratio'] = combined['text'].apply(
    lambda x: len(set(x.split())) / len(x.split()) if len(x.split()) > 0 else 0
)

# Create boxplot to visualize distribution of vocabulary diversity
plt.figure(figsize=(12,6))
sns.boxplot(x='label', y='unique_ratio', data=combined, palette=['#1f77b4', '#ff7f0e'], showfliers=False)
plt.title('Vocabulary Diversity Distribution\n(Type-Token Ratio)', fontsize=14)
plt.xlabel('Text Type', fontsize=12)
plt.ylabel('Unique Words / Total Words', fontsize=12)
plt.show()

# Display statistical summary table for mean and standard deviation
summary = combined.groupby('label')['unique_ratio'].agg(['mean', 'std']).reset_index()
display(summary)
```

**Discussion on Vocabulary Diversity (Type-Token Ratio) Plot:**

The boxplot compares the vocabulary diversity (Type-Token Ratio) between human-written and AI-generated texts. Human texts show slightly lower variability (narrower interquartile range) but comparable median values to AI texts. This indicates similar overall vocabulary richness. However, AI texts show greater variability, suggesting inconsistent word usage patterns across samples.

#### **2.3 Sentence Structure Analysis**

```python
combined['word_count'] = combined['text'].str.split().str.len()
combined['sentence_count'] = combined['text'].str.split('.').str.len()
combined['avg_sent_len'] = combined['word_count'] / combined['sentence_count']

plt.figure(figsize=(10,4))
sns.violinplot(x='label', y='avg_sent_len', data=combined, inner='quartile')
plt.title('Average Sentence Length Distribution')
plt.show()
```

**Discussion on the distribution of average sentence lengths**

The violin plot shows the distribution of average sentence lengths for human and AI texts. Both distributions are highly skewed, with most sentences clustering around shorter lengths. However, occasional outliers suggest rare instances of very long sentences in both categories. As a whole, the sentence lengths are more consistent in AI generated intros with fewer extreme outliers and more consistent sentence lengths as seen through the greater width of the AI plot.

#### **2.4 Contraction Usage**

```python
combined['contractions'] = combined['text'].str.count(r"\w+'\w+")

plt.figure()
sns.countplot(x='contractions', hue='label', data=combined[combined['contractions'] > 0])
plt.title('Contraction Usage Comparison')
plt.show()
```

**Discussion on contraction usage plot:**

The bar plot compares contraction usage frequency between human-written and AI-generated texts. Human texts consistently use more contractions across all frequency levels. AI texts show significantly fewer contractions overall, supporting the common hypothesis that AI-generated content tends to avoid informal language patterns.

#### **Conclusion on Text Pattern Analysis**

The analysis reveals that AI-generated Wikipedia introductions exhibit higher readability scores, more consistent sentence lengths, and significantly fewer contractions compared to human-written texts, while showing slightly lower vocabulary diversity. These patterns suggest AI-generated content tends to be more formulaic and structurally uniform.
