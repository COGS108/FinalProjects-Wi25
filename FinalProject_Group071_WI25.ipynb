# Abstract
This study examines the relationship between Federal Reserve interest rate changes and the monthly returns and trading volumes of major technology stocks—Meta, Apple, Amazon, Netflix, and Google (MAANG)—from 2013 to 2023. Given the sensitivity of growth stocks to macroeconomic conditions, we hypothesized that interest rate hikes would negatively impact MAANG stock returns and reduce trading volumes due to investor uncertainty. Using a dataset combining stock market performance, Federal Funds Rate, inflation, unemployment, and other economic indicators, we performed regression analysis to quantify these effects. Contrary to our hypothesis, our findings indicate a slightly positive relationship between interest rate increases and MAANG stock returns, while trading volumes showed no significant change after controlling for macroeconomic variables. The results suggest that monetary policy influences market behavior but does not operate in isolation; additional economic factors contribute to investor sentiment and stock performance. These findings provide insights into the complexities of financial market responses to interest rate shifts and highlight the need for further research incorporating nonlinear modeling and lagged effects.

# Research Question

What is the relationship between Federal Reserve interest rate changes and MAANG stocks' monthly returns and trading volumes from 2013-2023?

<p>
    The performance of technology stocks, particularly those of major firms such as Meta, Apple, Amazon, Netflix, and Google (MAANG), is influenced by a multitude of economic factors.

    The Federal Reserve’s interest rate decisions play a crucial role in shaping investor sentiment and stock market trends. 
    <a href="#cite_note-2"><sup>[2]</sup></a> 
    The Federal Reserve adjusts interest rates to manage inflation, economic growth, and employment levels, thereby affecting corporate borrowing costs, consumer spending, and overall market liquidity.

    Given the significant role of MAANG companies in the global economy, understanding the relationship between interest rate changes and their stock performance is important for investors, policymakers, and financial analysts.
</p>

<p>
    Interest rate adjustments impact stock markets in multiple ways. Higher interest rates increase borrowing costs for companies, potentially reducing their investment in innovation and expansion. 
    Under the discounted cash flow (DCF) model, rising interest rates decrease the present value of future earnings, making stocks less attractive to investors. 
    <a href="#cite_note-4"><sup>[4]</sup></a> <a href="#cite_note-5"><sup>[5]</sup></a> 
    Additionally, investor preferences often shift toward safer fixed-income securities under higher interest rates. 
    <a href="#cite_note-6"><sup>[6]</sup></a> 
    This pattern has been observed historically, indicating that past monetary tightening cycles have resulted in capital reallocation from emerging markets to developed economies. 
    <a href="#cite_note-7"><sup>[7]</sup></a> 
</p>

<p>
    The effect of interest rates is particularly significant for growth stocks, such as those in the MAANG group, since a large portion of their value is derived from projected future earnings. 
    These technology companies are often more sensitive to changes in interest rates than value stocks because their business models rely on long-term growth potential rather than immediate profitability. 
    <a href="#cite_note-8"><sup>[8]</sup></a> 
    Previous research has shown that technology sector stocks tend to exhibit higher volatility in response to interest rate fluctuations. 
    <a href="#cite_note-9"><sup>[9]</sup></a> <a href="#cite_note-10"><sup>[10]</sup></a> 
</p>

<p>
    Beyond price movements, trading volumes may also fluctuate as investors react to changing monetary conditions, influencing liquidity and volatility in the stock market. 
    Prior research has examined the effects of interest rate changes on stock market volatility and asset allocation, finding that rising rates often lead to reduced trading volumes as borrowing costs increase and risk aversion grows. 
    <a href="#cite_note-11"><sup>[11]</sup></a> 
    Higher interest rates can reduce market liquidity by increasing the cost of borrowing, leading to lower trading activity, particularly in growth stocks that rely on future earnings projections. 
    <a href="#cite_note-12"><sup>[12]</sup></a> 
    Conversely, during periods of low interest rates, investors may be more inclined to engage in speculative trading, increasing both trading volumes and market volatility. 
    <a href="#cite_note-13"><sup>[13]</sup></a> 
    Additionally, research suggests that liquidity providers play a key role in stabilizing markets during periods of heightened volatility, as they adjust their trading strategies based on perceived market risk. 
    <a href="#cite_note-14"><sup>[14]</sup></a> 
    These dynamics are especially relevant for high-growth stocks like MAANG, which tend to experience pronounced fluctuations in trading volume due to their sensitivity to macroeconomic conditions. 
    <a href="#cite_note-15"><sup>[15]</sup></a> 
</p>

<p>
    The period from 2013 to 2023 consists of multiple phases of monetary policy and economic cycles. 
    Following the 2008 financial crisis, the Fed maintained near-zero interest rates to stimulate economic recovery. 
    Beginning in 2012, the Fed gradually raised rates as the economy strengthened. 
    Due to the COVID-19 pandemic, the Fed cut rates to near-zero levels to support economic stability in 2020. 
    Since 2022, in response to post-pandemic inflation, the Fed has aggressively raised interest rates, hoping to curb rising consumer prices.

### Basic OLS Regression on MAANG's Average Monthly Return

In order to numerically examine the effect of interest rate on MAANG's Average Monthly Return, we initially tested a simple regression model without additional control variables to assess the direct impact of interest rates on MAANG's returns.

```python
import statsmodels.api as sm
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(combined_df['MAANG_Monthly_Return'], bins=30, kde=True, color='blue')
plt.title('Distribution of MAANG Monthly Return')
plt.xlabel('MAANG Monthly Return')
plt.ylabel('Frequency')
plt.show()
```

As we can see from the histogram, the distribution of MAANG's monthly return is normal, so we can directly use this value in our regression.

```python
X = combined_df[['DFF']]
y = combined_df['MAANG_Monthly_Return']
X = sm.add_constant(X)
model1 = sm.OLS(y, X).fit()
print(model1.summary())
```

The regression results indicate that the coefficient for DFF (0.0014) is small and statistically insignificant (p = 0.732). This means that changes in the Federal Funds Rate had little to no direct impact on MAANG’s average monthly stock return. The R-squared value (0.001) implies that only 0.1% of the variation in MAANG’s monthly returns is explained by the Federal Funds Rate. Additionally, the F-statistic (0.1182, p = 0.732) confirms that the overall model lacks statistical significance. This finding suggests that Stock returns for MAANG companies are likely influenced more by firm-specific factors and broader macroeconomic conditions rather than just interest rates.

### Mutiple Regression on MAANG's Average Monthly Return

Therefore, we also tested the multiple regressions to examine the effect of fed fund rate on MAANG's average monthly return by adding the control variables to account for macroeconomic factors.

```python
X = combined_df[['DFF', 'Unemployment Rate', 'Inflation Rate', 'USD_index', 'Gold_price']]
y = combined_df['MAANG_Monthly_Return']
X = sm.add_constant(X)
model1 = sm.OLS(y, X).fit()
print(model1.summary())
```

Among the predictors, DFF (the interest rate) and the unemployment rate are statistically significant at the 5% significance level. As interest rate increases by 1%, there is a 1.51% increase in MAANG's average monthly return. Similarly, when unemployment rate incerases by 1%, there is a 1.34% increase in MAANG's average monthly return. However, variables like the inflation rate, USD index, and gold price are not statistically signicant so there is no effect on MAANG's average monthly return.

# Discusison and Conclusion

This study investigated the relationship between Federal Reserve interest rate changes and MAANG stocks' monthly returns and trading volumes from 2013 to 2023. Our hypothesis predicted that interest rate hikes would have a strong negative impact on MAANG stock returns, particularly in the months immediately following a rate increase, and that trading volumes would decrease due to heightened investor uncertainty and repositioning. We also anticipated that the magnitude of this effect would diminish over time as markets adjusted to new rate environments.

Our findings do not support this hypothesis. While we observed a slightly positive relationship between interest rate increases and monthly stock returns, there was no effect on trading volume when controlling for key macroeconomic factors. Regression analysis showed that interest rate changes alone did not significantly impact MAANG stock returns, though incorporating additional macroeconomic variables improved the model's explanatory power. This suggests that while monetary policy influences market behavior, it does not operate in isolation; factors such as inflation, unemployment, and currency fluctuations also play a role in shaping investor sentiment and stock performance.

However, the effect of interest rate changes on trading volume was weaker than anticipated. Higher interest rates were expected to reduce MAANG stock trading activity, but our results suggest that once macroeconomic conditions are accounted for, the direct effect of interest rates on trading behavior is minimal. This finding challenges the assumption that investor uncertainty around monetary policy leads to temporary reductions in market liquidity.

Although our study accounts for key macroeconomic variables, there are limitations to consider. Another limitation is that the stock monthly return points are very dispersed. Therefore, the relationship between interest rates and stock returns may be nonlinear or influenced by interaction terms that were not included. Additionally, the exclusion of company-specific factors, such as earnings reports and industry developments, may have constrained our ability to fully capture stock price movements. Furthermore, our analysis focused on contemporaneous effects, meaning potential delayed responses to interest rate changes were not fully explored. Future research could extend this work by incorporating lagged effects, real-time investor sentiment data, and sector-specific comparisons to provide a more nuanced understanding of stock market responses to monetary policy.

Overall, this study provides valuable insights into how interest rate fluctuations influence MAANG stocks. These findings suggest that monetary policy effects on the stock market may be more complex than initially hypothesized, potentially requiring nonlinear modeling approaches or additional interaction terms for deeper analysis. This study contributes to the broader understanding of economic influences on financial markets and offers practical implications for investors navigating periods of monetary tightening or easing.
