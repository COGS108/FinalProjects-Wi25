```python
# Load the dataset
file_path = '/Users/paulineramos/Group106_WI25/Group106_WI25/sustainable_fashion_trends_2024.csv'
df = pd.read_csv(file_path)

# Display the first few rows of the dataset
df.head()
```

The **Certifications** column contains 1004 null values, and since it is not essential for our primary analyses or calculations, we will remove this column from the dataset to maintain data quality and clarity.

```python
# dropping the Certifications Column
df = df.drop(columns=['Certifications'])
```

We will also remove the **Brand_Name** column, as it does not contain the actual names of brands and essentially serves as a duplicate of the existing identifier (**Brand_ID**). Removing this column will help reduce clutter and simplify our analysis.

```python
# dropping the Brand_Name Column 
df = df.drop(columns=['Brand_Name'])

# looking at the dataset after dropping the columns 
df
```

```python
# Check unique values for categorical columns to confirm consistent labeling
categorical_cols = ["Eco_Friendly_Manufacturing", "Recycling_Programs", "Sustainability_Rating"]

for col in categorical_cols:
    print(f"{col} unique values:", df[col].unique())
```

```python
# Checking for negative values
negative_values = (df[['Carbon_Footprint_MT', 
                       'Water_Usage_Liters', 
                       'Waste_Production_KG']] < 0).sum()

# Checking for unrealistic extremes using descriptive statistics
extreme_values_summary = df[['Carbon_Footprint_MT', 
                             'Water_Usage_Liters', 
                             'Waste_Production_KG']].describe(percentiles=[0.01, 0.99])

print("Negative values:\n", negative_values)
print("\nSummary statistics for extremes:\n", extreme_values_summary)
```

The Categorical variables (**Eco_Friendly_Manufacturing**, **Recycling_Programs**, and **Sustainability_Rating**) were checked for consistent labeling. No spelling variations or inconsistencies were found, and thus these variables are appropriate for analysis without further adjustments.

Additionally, the dataset was examined for logical consistency in numerical variables related to environmental impact (**Carbon_Footprint_MT**, **Water_Usage_Liters**, and **Waste_Production_KG**). No negative values or clearly unrealistic extremes were found, indicating the data is reliable, logically consistent, and appropriate for further analysis.

To help our analysis, we are converting the categorical sustainability ratings (A, B, C, D) to numeric values (4, 3, 2, 1) as doing so allows us to perform quantitative analyses, such as correlations, comparisons, and statistical tests more easily. Numeric values help to simplify interpreting trends, relationships, and impacts associated with different sustainability levels.

```python
# renaming columns
df.rename(columns={
    'Carbon_Footprint_MT': 'Carbon_Footprint',
    'Water_Usage_Liters': 'Water_Usage',
    'Waste_Production_KG': 'Waste_Production',
    'Average_Price_USD': 'Average_Price',
}, inplace=True)

# checking dataset after renaming 
df
```

Finally, we renamed some of the columns for clarity and brevity. This has been done by removing the units in the following variables: **Carbon_Footprint_MT**, **Water_Usage_Liters**, **Waste_Production_KG**, and **Average_Price_USD**. The variables have been renamed to: **Carbon_Footprint**, **Water_Usage**, **Waste_Production**, and **Average_Price**, respectively.

### Regression Analysis 
For our analysis, we chose to do a multivariate regression analysis. We used **Sustainability_Rating** as our dependent variable and **Eco_Friendly_Manufacturing**,  **Carbon_Footprint**, **Water_Usage**, **Waste_Production**, and **Recycling_Programs** as our independent variables. 

```python
mod = smf.ols(formula='Sustainability_Rating ~ Eco_Friendly_Manufacturing + Carbon_Footprint + Water_Usage + Waste_Production + Recycling_Programs', data=df)

results = mod.fit()

print(results.summary())
```

After running and fitting the model, we can see from the model's summary that this is a very weak model. The coefficients can be interprented as the following:
- Holding all other variables constant, brands that have eco-friendly manufacturing are associated with a 0.0513 point decrease in their sustainability rating.
- Holding all other variables constant, brands that have recycling programs are associated with a -0.0385 point decrease in their sustainability rating. 
- Holding all other variables constant, brands that see an an increase in carbon footprint are associated with a 0.00008256 point increase in their sustainability rating. 
- Holding all other variables constant, brands that see an an increase in water usage are associated with a 0.000000007282 point increase in their sustainability rating. 
- Holding all other variables constant, brands that see an an increase in waste production are associated with a 0.0000002577 point decrease in their sustainability rating. 

However, the p-values of all the coefficients are large and indicate that these results are not statiscally significant. Further, the model shows that it is weak because its r-square value is 0.001 and its adjusted r-squared value is 0.000. As these values are very small and closer to 0, it highlights that the model does not really explain any variance happening. 

### Regression Analysis With Standardization 
In an effort to improve the linear regression model, we standardized some of the variables (**Carbon_Footprint**, **Water_Usage**, and **Waste_Production**).Then, we ran the same model again, with the same dependent and independent variables. 

```python
# define what columns to scale
columns_to_scale = ['Carbon_Footprint', 'Water_Usage', 'Waste_Production']

# initialize the scaler
scaler = StandardScaler()

# fit and transform the data for scaling
df[columns_to_scale] = scaler.fit_transform(df[columns_to_scale])

# refit model with scaled data
model = smf.ols(formula='Sustainability_Rating ~ Eco_Friendly_Manufacturing + Carbon_Footprint + Water_Usage + Waste_Production + Recycling_Programs', data=df).fit()

print(model.summary())
```

The scaled linear regression can be interpreted as the following:

- Holding all other variables constant, brands that have eco-friendly manufacturing are associated with a 0.0513 point decrease in their sustainability rating.
- Holding all other variables constant, brands that have recycling programs are associated with a 0.0385 point decrease in their sustainability rating. 
- Holding all other variables constant, brands that see an an increase in carbon footprint are associated with a 0.0118 point increase in their sustainability rating. 
- Holding all other variables constant, brands that see an an increase in water usage are associated with a 0.0104 point increase in their sustainability rating. 
- Holding all other variables constant, brands that see an an increase in waste production are associated with a 0.0074 point decrease in their sustainability rating. 

Standardizing the variables resulted in making it easier to compare the results with each other. The numbers are not as small as they were in the first model. However, like the previous model, the p-values of all the coefficients are still very large, indicating that these results are not statiscally significant. Again, the model's r-square value is 0.001 and its adjusted r-squared value is 0.000. These values are still very small and practically zero, indicating that the model is very weak.

There are several ethics and privacy concerns that we kept in mind as we did our analysis. The dataset has information from various parts of the world and we may encounter problems of overrepresentation/underrepresentation of certain geographical areas. These biases in the geographical representation may skew our results, especially as our analysis may favor brands from countries that have stricter fashion sustainability regulations. While the brands are not explicitly named, bigger companies with more resources may be more likely to be represented in the dataset compared to smaller companies who have less resources to get their name out. Finally, there can be further biases within the dataset, especially when it comes to Certifications and Sustainability Ratings as the ratings may not always reflect the brand’s commitment to sustainability, especially as some brands practice greenwashing, which happens when brands mislead consumers about the true environmental impact of their products. The ratings and certifications may also be susceptible to different rating systems, so it can be hard to compare the brands.  As the dataset does not contain personal data and does not explicitly mention brands by their names, the privacy risks are minimal. However, it is still important to keep in mind privacy concerns that may rise as we continue with our analysis. 

While we did not use geographical distribution of the brands, one thing we could have done was using a similar number of regions in our analysis to ensure proper representation. Further, we compared the sustainability ratings with metrics that reflect the brands’ true impact on the environment, such as carbon footprint, water usage, and waste production, as seen in our exploratory data analysis and overall analysis. To address these many of the concerns, we were transparent about our whole process by ackowledging the limitations we encountered with the data and how it has affected our analysis. Explaining our thought process helps to keep us accountable. 

# Discussion and Conclusion

In our project, we wanted to discover whether sustainability initiatives in the fashion industry lead to reductions in environmental impact or simply served as a marketing strategy. We hypothesized that brands that adopt sustainability initiatives such as eco-friendly manufacturing and recycling programs would significantly reduce carbon emissions, water usage, and waste. Through our data analysis that we conducted on 5,000 fashion brands and their sustainability attributes, the results showed that there were many challenges to our hypothesis. While previous research has shown that eco-friendly practices result in the reduction of environmental damage, our findings do not suggest significant environmental improvements. 

First, our analysis of eco-friendly manufacturing suggests that brands who report manufacturing eco-friendly practices do not have a significant impact on reductions in carbon emissions or water usage in comparison to brands that do not. Although there was a slight reduction in waste production, these measured results show that the impact was very marginal. This finding raises concerns about greenwashing which is when brands use marketing strategies to label themselves as environmentally friendly, without eco-friendly operations. The lack of significant differences in the analysis of carbon footprint, water usage, and waste reduction suggest that sustainability initiatives taken by brands have no significant impact on the environment compared to brands that do not adopt eco-friendly practices. 

Second, our analysis of the relationship between sustainability ratings and environmental impact metrics show that there were insignificant correlations between a brand’s sustainability rating and its eco-friendly manufacturing results. The highest observed correlation was a weak negative relationship that was between carbon footprint and water usage (r = -0.027), which is statistically insignificant. Our findings raise concerns on whether these sustainability ratings are reliable in metrics of environmental performance. This is an indication that there are outlying factors that are not related to the measurement of environmental outputs. 

Our findings suggest that sustainability initiatives may be adopted into brand’s labels, without the execution of eco-friendly practices that would result in significant measures of reducing environmental harm. Our project’s data analysis emphasizes the intentions behind brands and their eco-friendly labels. However, a limitation in our study is that the data relies on observations, which introduces the possibility of bias and misreporting. In addition, the statistical analysis using OLS regression resulted in a low R-squared value of 0.001 which means our model yields no variation in environmental impact. These limitations could be improved by collecting standardized data and expanding our data to include other datasets based on time-series data could help improve visualizations to observe changes in sustainability initiatives and the impact over time.

In conclusion, our project suggests that sustainability initiatives in the fashion industry often focus more on the eco-friendly label than achieving measurable outcomes for environmental efforts. Ultimately, our hypothesis – that brands implementing eco-friendly practices would reduce carbon emissions, water usage, and waste – was not supported by the data analysis. While some practices like waste reduction show minor improvements, there is still a disconnect between the sustainability claims in comparison to the measurable outcomes. This gap can be an indication that many sustainability initiatives serve as a marketing strategy rather than pushing for real environmental change. As consumers become more aware of misleading environmental claims, the fashion industry may be driven to improve by prioritizing transparency in eco-friendly manufacturing and making meaningful environmental progress.
