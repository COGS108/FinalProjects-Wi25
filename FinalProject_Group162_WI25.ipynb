## Background and Prior Work

TikTok serves as entertainment and a news source for many of the younger generation. With the global shift towards short-form content and the attention economy, we are interested in analyzing the “why” behind the virality. Why do certain TikToks go viral and others don't? What are the characteristics of short-form content that can be optimized to boost virality? In today’s content creation and influencer world, virality means money, and those who can unlock the secret to virality can amass large sums of it. In this project, we aim to investigate the key factors that contribute to the success behind videos that find popularity on TikTok. We believe that by looking at the content and creator characteristics surrounding the videos that rapidly gain traction, we will be able to predict the likelihood of reaching 1 million views.

One past study that aimed to analyze the virality of TikTok's suggested features in its extremely personalized recommendation algorithm was conducted by Ren Zhou of Tsinghua University. The study conducted a quantitative analysis on metadata and user interactions (like likes, comments, and shares) for specific TikToks over 6 months, revealing that metrics such as like ratios, trending hashtags, and video length were the most powerful predictors of a video’s viral success. The study also analyzed what type of content a user likes, focusing on the user side as well to get a complete view of the sophisticated recommendation system that allows even newer users to go viral [1]. While our studies are quite similar, we diverge in the variables we control. For example, we focus on more immediate virality, limiting the videos to their first 48 hours rather than 6 months, and we focus more on predicting the outcome of the algorithm itself, affording less attention to user-centric metrics.

Similar research was conducted by ACM, where authors Chen Ling, Jeremy Blackburn, Emiliano De Cristofaro, and Gianluca Stringhini found that follower count was the strongest predictor, trivially, for virality in short-form TikToks. However, they also found that close-up and medium shots were reliable predictors for how viral a video was, alongside the lifespan of the video, presence of text, and point of view [2]. Our studies diverge here, as we (as mentioned earlier) account for the lifespan of a video by limiting our data to their first 48 hours of existence, and as we focus on different predictors and time frames to collect our data, which we believe will uncover different relationships and allow for stronger predictors.

# Setup

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import re
# import emoji
from datetime import datetime

#EXPLAIN HERE
from transformers import pipeline #RoBERTa Transformer (handles sarcasm and emojis, very important for tiktok captions)
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.svm import SVR
from sklearn.metrics import classification_report, mean_squared_error, r2_score #necessary metrics for testing



sentiment_model = pipeline("sentiment-analysis", model="cardiffnlp/twitter-roberta-base-sentiment")
```

The reason we use Roberta here is because it is a Twitter-sentiment analysis lexicon which is able to factor in emoji's and do much better text-processing. Instead of Vader, which we used in class that tells you positive/negative, Roberta is much better for analyzing social-media text, wich contains emoji's. A lot of captions in social media have emoji's which strongly convey sentiment. In additiion, it's able to identify sarcasm a lot better.

```python
# the following code is from chatgpt, we asked for a function which standardizes unix timestamp to an int (num of minutes passed since 12AM)

def unix_to_minutes(timestamp):
    dt = datetime.utcfromtimestamp(timestamp)
    return dt.hour * 60 + dt.minute
```

This function converts the given Unix timestamp to a Time-of-day integer (in minutes) for better graphing.

```python
# the following code was generated by chatgpt, we asked for a

def clean_tiktok_caption(text):
    if not isinstance(text, str):  # Ensure text is a string
        return ""  # Replace NaN/missing captions with an empty string
    # text = text.lower()  # Convert to lowercase
    text = re.sub(r'@\w+', '', text)  # Remove @mentions
    text = re.sub(r'#\w+', '', text)  # Remove hashtags
    text = re.sub(r'http\S+', '', text)  # Remove URLs
    text = text.strip()  # Remove leading/trailing whitespace
    return text
```

This function cleans the string sent inside for sentiment analysis, removing hashtags and unnecessary whitespace. As we are using `RoBERTa`, we will not need to remove emojis or casing.

```python
def clean_tiktok_caption_tfidf(text):
    return re.sub(r'[^a-zA-Z\s]', '', text).strip()
```

Cleans code for tfidf, removing all non-alphabetical values from string.

# Fashion

```python
#preproc

#load and wrangle "fashion"
fashion_pre = pd.read_csv('http://raw.githubusercontent.com/pranavRedBus/cogs108_162data/refs/heads/main/fashion-452.csv') #datahub load csv file

fashion = fashion_pre[fashion_pre['textLanguage']=='en'][['id','commentCount','authorMeta/verified', 'createTime', 'text', 'shareCount', 'playCount','videoMeta/duration', 'diggCount']]
fashion.columns = ['id','commentCount','isVerified', 'postDate', 'caption', 'shareCount', 'playCount','duration', 'likeCount']
fashion['timePosted'] = fashion['postDate'].apply(unix_to_minutes)

```

Loading of the data. We load data csv from our public data repository, choose specific columns from English-only TikToks, rename them, and make them able to be analyzed.

```python
#cleaning
#sentiment analysis cleaning
fashion['clean_caption'] = fashion['caption'].apply(clean_tiktok_caption)

#deepseek addition for cleaning
# Check for empty captions after cleaning
fashion['clean_caption'] = fashion['clean_caption'].apply(
    lambda x: "[EMPTY]" if len(x.strip()) == 0 else x
)


#tfidf cleaning
top_30_percent = fashion['commentCount'].quantile(0.70)
bottom_30_percent = fashion['commentCount'].quantile(0.30)

fashion['clean_caption_tfidf'] = fashion['clean_caption'].apply(clean_tiktok_caption_tfidf)
```

Cleaning data for sentiment analysis and tfidf. We apply the aforementioned cleaning functions, remove empty captions, and select the top and bottom 30% of data in terms of `commentCount` (for tfidf) to see which keywords boost engagement the most and least in the captions.



```python
#avg duration
avg_duration_viral = fashion[fashion['playCount'] > 1000000]['duration'].mean()
print(avg_duration_viral) #avg duration for viral video in fashion dataset
```

We calculate the `avg_duration_viral`, which is the average length of a TikTok with over 1 million views, for this dataset. THis will help us see how long a viral TikTok typically is. For this dataset, the most viral TikToks lasted around **22 seconds**.

```python
#tfidf
most_commented = fashion[fashion['commentCount'] >= top_30_percent]['clean_caption_tfidf']
least_commented = fashion[fashion['commentCount'] <= bottom_30_percent]['clean_caption_tfidf']

tfidf = TfidfVectorizer(stop_words='english', max_features=30)  # Limit to 30 most important words

# Fit TF-IDF separately for both groups
tfidf_most = tfidf.fit_transform(most_commented)
tfidf_least = tfidf.fit_transform(least_commented)

# Convert to DataFrame for better readability
most_commented_tfidf_df = pd.DataFrame(tfidf_most.toarray(), columns=tfidf.get_feature_names_out()).mean().sort_values(ascending=False)
least_commented_tfidf_df = pd.DataFrame(tfidf_least.toarray(), columns=tfidf.get_feature_names_out()).mean().sort_values(ascending=False)

print('Caption TFIDF for Most Commented')
print(most_commented_tfidf_df.head())
print()
print('Caption TFIDF for Least Commented')
print(least_commented_tfidf_df.head())
```

```python
# trending niche analysis - correlation between engagement and audience size, do accounts with more followers get
# more engagement?
# weighted engagement distribution: commentCount and shareCount weigh more than likeCount
trending['engagement_score'] = (trending['likeCount'] + (2 * trending['commentCount']) + (2 * trending['shareCount'])) / trending['playCount']
small_audience_threshold = trending['audienceSize'].quantile(0.3)  # 33rd percentile
medium_audience_threshold = trending['audienceSize'].quantile(0.7) # 66th percentile

trending['audience_category'] = 'Large'  # large as default
trending.loc[trending['audienceSize'] < small_audience_threshold, 'audience_category'] = 'Small'
trending.loc[(trending['audienceSize'] >= small_audience_threshold) & (trending['audienceSize'] < medium_audience_threshold), 'audience_category'] = 'Medium'

audience_engagement = trending.groupby('audience_category')['engagement_score'].mean()
sns.barplot(x=audience_engagement.index, y=audience_engagement.values)
plt.title("Average Engagement Score by Audience Size")
plt.ylabel("Engagement Score")
plt.xlabel("Audience Category")
plt.show()
```

The way we measured engagement was based on an engagement score, the weighted formula for which gave more emphasis on the amount of comments and shares per video, rather than the number of likes, since it requires much more effort for a user to either comment something or share the video with a friend. We also separated the audience size into 3 different categories: "Small", "Medium", and "Large".

The results seem to illustrate that the accounts with a much smaller audience size tend to get a lot more engagement than thoes with a larger audience. These results seem to make a lot sense, especially when considering the fact there is an abudance of accounts on TikTok with a smaller number of followers. # Creating Predictive Model

```python
# join the datasets

all_tiktoks = pd.concat([fashion, couple, funny, music, trending, tiktokmademe], axis=0)
```

The above cell is aggregating the data after we analyzed it separately to ensure that performance will not wary between different categories of hashtags.

```python
# prepare timePosted for all datasets
all_tiktoks['timePosted'] = all_tiktoks['postDate'].apply(unix_to_minutes)

# create an engagement_score column to properly track virality of a tiktok
all_tiktoks['total_engagement'] = (all_tiktoks['likeCount'] + (2 * all_tiktoks['commentCount']) + (2 * all_tiktoks['shareCount']) + 0.25*all_tiktoks['playCount'])

# make virality column, allows for easy predicting for SVM
#exact method suggested by chatgpt
threshold = all_tiktoks['total_engagement'].quantile(0.70) # 30% chosen to have a good spread, based on EDA should have mostly viral tiktoks
all_tiktoks['virality'] = (all_tiktoks['total_engagement'] > threshold).astype(int) #cast true or false to 0,1; prepares data to be predicted
```

This cell creates/initializes columns that will later be used for the model across the newly aggregated dataset, such as `timePosted` (mentioned above), `total_engagement` (for a uniform way to conclude virality across engagement metrics), and `virality` (binary value indicating if a TikTok went viral or not based on the threshold of its corresponding `total_engagement` score).

```python
# we decided to move forward with sentiment analysis, duration, and timePosted as our predictors as they are the most informative and interesting for every tiktok
all_tiktoks_model = all_tiktoks[['virality', 'sentiment_score', 'duration', 'timePosted']]

all_tiktoks_model.head()
```

The above cell creates a new dataframe with solely the columns we need for analysis
