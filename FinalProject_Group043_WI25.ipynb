# Abstract

As social media has become more integrated into society, we wanted to explore how its use affects sleep duration, sleep quality, and cognitive functioning. To investigate this, we analyzed two datasets; one from the United States and another from Singapore. By comparing these two populations, we sought to identify potential regional differences in how social media habits influence sleep quality and cognitive performance. The U.S. dataset examines different types of social media platforms (such as Facebook and YouTube), their usage frequency, and various sleep metrics. The Singapore dataset includes variables related to social media use, sleep patterns, and the dominant social media platform. After gathering and analyzing the data, we found a correlation between social media use and sleep.

To achieve this, we employed various data science techniques, including data wrangling, exploratory data analysis (EDA), and visualization methods, to uncover patterns and correlations between social media usage and sleep-related metrics. Using Python-based tools such as Pandas for data cleaning, Seaborn and Matplotlib for data visualization, and statistical modeling techniques, we thoroughly examined the extent to which social media consumption influences sleep behaviors across different populations. By leveraging bar, line and box plots as well as correlation matrices, we evaluated the predictive strength of social media engagement on sleep disruptions and cognitive functioning impairments.

Our findings reveal that increased social media usage, particularly before bedtime, is associated with poorer sleep quality and decreased sleep duration. Individuals who engaged in prolonged social media use late at night reported greater difficulty falling asleep and maintaining a consistent sleep schedule. Additionally, excessive engagement correlated with sleep disturbances such as waking up frequently and experiencing non-restorative sleep. However, the strength of this relationship varied between regions. In the U.S. dataset, social media usage alone did not show a strong linear correlation with sleep quality, whereas in the Singapore dataset, prolonged engagement—especially before sleep—was significantly associated with sleep disruptions. Overall, this study underscores the complex correlation between social media habits, sleep quality, and cognitive health. While excessive engagement—especially before sleep—appears to have negative consequences, the nature of engagement, rather than just duration, plays a crucial role in determining its effects.

From dataset: "Most participants identified as female (63.1% female, 35.9% male, 1.0% non-binary)". Thus, let us see if our sample is representative by checking how the gender distribution of our subsample of participants compares to the entire dataset. Let's do that through a chi-square test.  

```python
gender_counts_us = df_us['GENDER'].value_counts()
print("Gender counts in df_us:")
print(gender_counts_us)
```

```python
df_us_whole = pd.read_csv("https://raw.githubusercontent.com/camrynwilliams/Datasets/refs/heads/main/24Jun13_SocialMediaSleepDataset%20(1).csv")
df_us_whole['GENDER'] = df_us_whole['GENDER'].replace({1:'Male', 2: 'Female', 3: 'Nonbinary'})
gender_counts_us_whole = df_us_whole['GENDER'].value_counts()
print("Gender counts in df_us_whole:")
print(gender_counts_us_whole)
```

```python
from scipy.stats import chisquare

# Gender proportions for df_us
gender_proportions_us = gender_counts_us / gender_counts_us.sum()

# Gender proportions for df_us_whole
gender_proportions_us_whole = gender_counts_us_whole / gender_counts_us_whole.sum()

# Observed counts in df_us
observed = gender_counts_us.values

# Expected counts in df_us_whole (based on proportions)
expected = gender_proportions_us_whole.values * gender_counts_us.sum()

# Perform Chi-square test
chi2_stat, p_val = chisquare(observed, expected)

print(f"Chi-square Statistic: {chi2_stat}")
print(f"P-value: {p_val}")
```

Conclusion: Since p-value = 0.3 > 0.05, then the gender proportions in df_us are not significantly different from the proportions in df_us_whole. Thus, the proportions in our subsample are likely representative of the entire sample.

```python
df_singapore_whole =  pd.read_csv("https://raw.githubusercontent.com/camrynwilliams/Datasets/refs/heads/main/SocialMediaUsage_SleepLatencyAnalysis_Singapore.csv")
gender_counts_singapore_whole = df_singapore_whole['Gender'].value_counts()
print("Gender counts in df_singapore_whole:")
print(gender_counts_singapore_whole)
```

Conclusion: Since p-value = 0.85 > 0.05, then the gender proportions in df_singapore are not significantly different from the proportions in df_singapore_whole. Thus, the proportions in our subsample are likely representative of the entire sample.

A next step would be to compare the difference in representation across both sub-samples, the US and Singapore. This can be achieved by plotting their proportions side by side.

```python
gender_counts_us = gender_counts_us.rename(index={'Nonbinary': 'Other'})
gender_proportions_us = gender_counts_us / gender_counts_us.sum()
gender_proportions_sg = gender_counts_SG / gender_counts_SG.sum()

gender_df = pd.DataFrame({'US': gender_proportions_us, 'Singapore': gender_proportions_sg})

gender_df.plot(kind='bar')
plt.title("Comparison of Gender Proportions: US vs. Singapore")
plt.ylabel("Proportion")
plt.xlabel("Dataset")
plt.xticks(rotation=0)
plt.legend(title="Gender")
plt.show()
```

Evidently, there is a significant difference in the proportions between the gender distribution across samples. Let us do a quick chi-square test to confirm this.

```python
observed_us = gender_counts_us.values
expected_us = gender_proportions_sg.values * gender_counts_us.sum()
chi2_stat, p_val = chisquare(observed_us, expected_us)

print(f"Chi-square Statistic: {chi2_stat}")
print(f"P-value: {p_val}")

if p_val < 0.05:
    print("The gender distributions in the US and Singapore datasets are significantly different.")
else:
    print("There is no significant difference in gender distributions between the US and Singapore datasets.")
```

Therefore, there is a significant variation in the proportion of males, females and nonbinary/other individuals across the two samples. The demographic difference may be important when looking at the impact of social media use on sleep qualiy, if there is evidence to support that social media consumption patterns vary by gender -- for example, if one gender spends more time on certain platforms or engages in different types of content -- then this could influence how social media affects sleep.
```python
#Relationship between Average Daily Social Media Time (minutes) and Sleep Quality Rating
sns.scatterplot(data=df_singapore, y='Average Daily Social Media Use Time (minutes)', x = 'Sleep Quality Rating', color='skyblue')
plt.title('Relationship Between Daily Social Media Use and Sleep Quality (Singapore)')
```

The above visualization indicates a strong linear relationship: as average social media use time decreases, sleep quality increases! This supports our hypothesis. Given the indication of a linear relationship, we can proceed with plotting a regression line using Seaborn's regplot.

```python
sns.regplot(data=df_singapore, y='Average Daily Social Media Use Time (minutes)', x = 'Sleep Quality Rating', color='skyblue')
plt.title("Sleep Quality Rating vs Average Daily Social Media Use Time (Singapore)")
```

This seems to be a relatively strong negative correlation. To verify, we can perform OLS regression.

```python
X = df_singapore['Average Daily Social Media Use Time (minutes)']
y = df_singapore['Sleep Quality Rating']
X = sm.add_constant(X)
model = sm.OLS(y,X).fit()
print(model.summary())
```

The R-squared value of 0.85 suggests a strong correlation between the two variables. Moreover, the constant 4.7511 suggests the following: If a person does not use social media daily, the expected sleep quality rating would be around 4.75, but for every additional minute of daily social media use, the sleep quality rating decreases by 0.0177. This suggests that more social media usage is associated with lower sleep quality. Moreover, a p-value of 0.000 suggests that there is strong statistical significance and that this relationship is not a result of chance alone.

Conclusion: as the amount of time spent on social media increases, sleep quality tends to decrease.
