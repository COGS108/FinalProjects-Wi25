# Abstract

This study examines whether monthly gasoline prices and seasonal weather conditions predict changes in grocery prices across different U.S. regions from 2012 to 2018. Drawing on Bureau of Labor Statistics data for grocery categories (e.g., meat, produce, dairy), U.S. Energy Information Administration records of fuel prices, and NOAA’s monthly temperature averages, we conduct correlation analyses, linear regressions, and Granger causality tests. Overall, fuel costs show a notable capacity to forecast grocery inflation—most consistently for beverages and produce, and at shorter lags of one to four months. A key finding involves the meat category, where certain scatter plots reveal an unexpected negative correlation with gas prices. However, Granger tests suggest that, in some regions (notably the West), higher gas prices still predict meat prices at a one-period lag, indicating more complex industry dynamics than a simple one-to-one cost passthrough. Meanwhile, average seasonal temperature exerts weaker but still measurable delayed effects on produce and occasionally meat, appearing at longer lags of two to four months and reflecting the time needed for climate impacts to affect agricultural supply. Other categories, such as dairy and bakery, remain largely insensitive to both gas and weather fluctuations, underscoring a more stable pricing structure. These results highlight the dominant yet nuanced role of transportation costs in short-term grocery inflation, while confirming that weather extremities can trigger delayed but meaningful price shifts for perishable items. Understanding these time-lagged mechanisms can help policymakers, retailers, and consumers better anticipate food price volatility and mitigate its effects.

```python
import os
import pandas as pd


def fetch_bls_data(series_id):
    # mock a browser request (may have to change the cookie values if they expire)
    url = "https://data.bls.gov/pdq/SurveyOutputServlet"
    headers = {
        "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8,application/signed-exchange;v=b3;q=0.7",
        "Accept-Language": "en-US,en;q=0.9,und;q=0.8,sk;q=0.7,nl;q=0.6",
        "Cache-Control": "no-cache",
        "Connection": "keep-alive",
        "Content-Type": "application/x-www-form-urlencoded",
        "DNT": "1",
        "Origin": "https://data.bls.gov",
        "Pragma": "no-cache",
        "Referer": "https://data.bls.gov/pdq/SurveyOutputServlet",
        "Sec-Fetch-Dest": "document",
        "Sec-Fetch-Mode": "navigate",
        "Sec-Fetch-Site": "same-origin",
        "Sec-Fetch-User": "?1",
        "Upgrade-Insecure-Requests": "1",
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/133.0.0.0 Safari/537.36",
        "sec-ch-ua": '"Not(A:Brand";v="99", "Google Chrome";v="133", "Chromium";v="133"',
        "sec-ch-ua-mobile": "?0",
        "sec-ch-ua-platform": "Windows",
    }
    cookies = {
        "JSESSIONID": "286F5B7844BD0BACEEE53209B230EA8B._t4_08v",
        "_ga": "GA1.1.127825815.1740173305",
        "nmstat": "3c7bff1b-c28b-1c9d-4ea9-0034ce2d9a0d",
        "_ga_CSLL4ZEK4L": "GS1.1.1740173305.1.1.1740173367.0.0.0",
    }
    data = {
        "output_type": "default",
        "output_view": "data",
        "years_option": "specific_years",
        "from_year": "2012",
        "to_year": "2018",
        "periods_option": "all_periods",
        "output_format": "text",
        "delimiter": "comma",
        "reformat": "true",
        "request_action": "get_data",
        "initial_request": "false",
        "data_tool": "latest_numbers",
        "series_id": series_id,
        "original_include_graphs": "true",
        "original_output_view": "data||||||||",
    }

    response = requests.post(url, headers=headers, cookies=cookies, data=data)

    if response.status_code == 200:
        filename = f"{series_id}.html"
        with open(filename, "w", encoding="utf-8") as file:
            file.write(response.text)
        print(f"Saved data to {filename}")
    else:
        print(f"Failed to fetch data for {series_id}, status code: {response.status_code}")


# grab actual data from the BLS html page responses
def extract_data_from_html(html_file):
    with open(html_file, "r", encoding="utf-8") as file:
        content = file.read()

    match = re.search(r"</PRE><H4>(.*?)</PRE>", content, re.DOTALL)
    if match:
        extracted_data = match.group(1).strip()
        txt_filename = html_file.replace(".html", ".txt")
        with open(txt_filename, "w", encoding="utf-8") as txt_file:
            txt_file.write(extracted_data)
        print(f"Extracted data saved to {txt_filename}")
    else:
        print("No matching data found in the HTML file.")


# and from those text files extracted from the html file, parse it into a dataframe and save it as a CSV
def extract_data_from_txt(folder_path):
    data_list = []
    column_names = ["Series_ID", "Series_Title"]
    for year in range(2012, 2019):
        for month in ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]:
            column_names.append(month + "-" + str(year))

    for filename in os.listdir(folder_path):
        if filename.endswith(".txt"):
            file_path = os.path.join(folder_path, filename)

            with open(file_path, "r", encoding="utf-8") as file:
                content = file.read()

                series_id_match = re.search(r"Series Id:\s+([A-Z0-9]+)", content)
                series_id = series_id_match.group(1) if series_id_match else None

                series_title_match = re.search(r"Series Title:\s+(.+)", content)
                series_title = series_title_match.group(1) if series_title_match else None

                table_match = re.search(r"Year,.*\n([\s\S]*)", content)
                if table_match:
                    table_data = table_match.group(1).strip().split("\n")
                    row_data = [series_id, series_title]
                    year_month_dict = {}

                    for row in table_data:
                        values = row.split(",")
                        year = values[0]
                        month_values = values[1:]
                        if len(month_values) != 12:
                            print(f"Warning, likely bad data: {series_title}")
                        for month, price in zip(["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"], month_values):
                            col_name = f"{month}-{year}"
                            year_month_dict[col_name] = price

                    row_data.extend([year_month_dict.get(col, None) for col in column_names[2:]])
                    data_list.append(row_data)

    df = pd.DataFrame(data_list, columns=column_names)
    return df
```

```python
# continuing data cleaning & wrangling here with the bls-*.csv files

import numpy as np
import pandas as pd

# Helper functions to clean data and standardize the food categories for later analysis


def get_food(row):
    food = row.split(",")[:2]
    # Join the list elements into a single string
    return " ".join(food)


def categorize_food(item):
    categories = {
        "meat": ["chicken", "beef", "pork", "lamb", "chuck", "ham", "steak", "chops", "round roast", "bacon", "turkey"],
        "produce": [
            "apple",
            "banana",
            "carrot",
            "lettuce",
            "lemon",
            "strawberries",
            "tomato",
            "orange",
            "pears",
            "peaches",
            "pepper",
            "potato",
            "cabbage",
            "celery",
            "grapes",
            "cherries",
            "grapefruit",
            "eggs",
            "broccoli",
        ],
        "dairy": ["milk", "cheese", "yogurt", "butter", "margarine"],
        "bakery": ["bread", "cake", "muffin"],
        "beverages": ["coffee", "tea", "juice", "soda", "wine", "malt", "cola", "water", "beer"],
    }

    for category, items in categories.items():
        if any(food in item.lower() for food in items):
            return category
        if "peanut butter" in item.lower():
            return "other"
    return "other"
```

```python
# West BLS DataFrame

bls_w = pd.read_csv("bls-west.csv")
bls_w.drop(columns=["Unnamed: 0"], inplace=True)
bls_w["Food"] = bls_w["Series_Title"].apply(get_food)
bls_w["Category"] = bls_w["Food"].apply(categorize_food)
bls_w.replace(["-", "", " "], np.nan, inplace=True)

bls_w_clean = bls_w[bls_w.isna().sum(axis=1) <= 50]
bls_w_clean.interpolate(method="linear", limit_direction="both", inplace=True)
bls_w_clean.iloc[:, 2:-3] = bls_w_clean.iloc[:, 2:-3].apply(pd.to_numeric, errors="coerce")
bls_w_clean.iloc[:, 2:-3] = bls_w_clean.iloc[:, 2:-3].apply(lambda row: row.fillna(row.median()), axis=1)
print("Number of missing values: ", bls_w_clean.isna().sum().sum())

print("Number of rows: ", bls_w_clean.shape[0])
print("Number of columns: ", bls_w_clean.shape[1])
print("West BLS DataFrame")
bls_w_clean.head()
```

# Dataset Name: NOAA Climate at a Glance – Regional Mapping


```python
import os
import requests


def download_noaa_data(start_year=2012, end_year=2018):
    base_url = "https://www.ncei.noaa.gov/access/monitoring/climate-at-a-glance/regional/mapping/tavg-{year}{month:02d}-1-map-data.json"
    output_dir = "noaa_data"
    os.makedirs(output_dir, exist_ok=True)

    for year in range(start_year, end_year + 1):
        for month in range(1, 13):  # Months from 01 to 12
            url = base_url.format(year=year, month=month)
            response = requests.get(url)

            if response.status_code == 200:
                file_name = f"{output_dir}/tavg-{year}{month:02d}.json"
                with open(file_name, "w", encoding="utf-8") as file:
                    file.write(response.text)
                print(f"Downloaded: {file_name}")
            else:
                print(f"Failed to download data for {year}-{month:02d}, Status Code: {response.status_code}")
```

```python
import os
import json
import pandas as pd


def load_noaa_data(folder_path="noaa_data"):
    data_rows = []

    for year in range(2012, 2019):
        for month in range(1, 13):
            file_name = f"tavg-{year}{month:02d}.json"
            file_path = os.path.join(folder_path, file_name)

            if not os.path.exists(file_path):
                print(f"File not found: {file_path}, skipping...")
                continue

            with open(file_path, "r") as f:
                json_data = json.load(f)

            for region_id, region_info in json_data.items():
                name = region_info.get("name")
                mean_value = region_info.get("value")

                data_rows.append({"Year": year, "Month": month, "Region_ID": region_id, "Name": name, "Mean": mean_value})

    # Create a DataFrame from the collected rows
    df = pd.DataFrame(data_rows)
    return df


df_noaa = load_noaa_data()
print(f'Number of regions: {len(list(df_noaa["Name"].unique()))}')
print(df_noaa.shape)
```

```python
# filter to only regions with name containing "Region"
df_noaa = df_noaa[df_noaa["Name"].str.contains("Region")]
print(list(df_noaa["Name"].unique()))

# only keep West Climate Region, Upper Midwest Climate Region, and Ohio Valley Climate Region
df_noaa = df_noaa[df_noaa["Name"].isin(["West Climate Region", "Upper Midwest Climate Region", "Ohio Valley Climate Region"])]
# Because there is no Midwest Climate Region, we will average out the Upper Midwest and Ohio Valley Climate Regions
df_noaa["Name"] = df_noaa["Name"].replace({"Upper Midwest Climate Region": "Midwest Climate Region", "Ohio Valley Climate Region": "Midwest Climate Region"})
# Drop Region_ID column
df_noaa = df_noaa.drop(columns="Region_ID")
df_noaa = df_noaa.groupby(["Year", "Month", "Name"]).mean().reset_index()

df_noaa = df_noaa.sort_values(by=["Year", "Month"])

# check for missing values
missing_values = df_noaa.isnull().sum().sum()
assert missing_values == 0, f"Missing values found: {missing_values}"

# ensure that all columns are of the correct data type
df_noaa["Year"] = df_noaa["Year"].astype(int)
df_noaa["Month"] = df_noaa["Month"].astype(int)
df_noaa["Name"] = df_noaa["Name"].astype(str)
df_noaa["Mean"] = df_noaa["Mean"].astype(float)

df_noaa.to_csv("noaa_climate_data.csv", index=False)

# ensure all years are between 2012 and 2018, and that all months are between 1 and 12
assert df_noaa["Year"].between(2012, 2018).all()
assert df_noaa["Month"].between(1, 12).all()

# ensure the mean is not nan and between -50F and 100F
assert not df_noaa["Mean"].isnull().any()
assert df_noaa["Mean"].between(-100, 100).all()

print(df_noaa.sample(20).sort_values(by=["Year", "Month"]))
print("Data looks good!")
```

```python
import pandas as pd


# Reshape the BLS datasets
def reshape_bls(df):
    id_vars = ["Series_ID", "Series_Title", "Food", "Category"]
    value_vars = df.columns.difference(id_vars)
    df_long = df.melt(id_vars=id_vars, var_name="Month_Year", value_name="Value")
    df_long["Date"] = pd.to_datetime(df_long["Month_Year"], format="%b-%Y")
    # drop Series data, will interfere with later df merging
    df_long.drop(columns=["Month_Year", "Series_ID", "Food"], inplace=True)
    # remove any occurences of region-specific wording in Series_Title for proper merging
    region_wordlist = [" in West urban", " in Midwest urban", " in U.S. city average"]
    for word in region_wordlist:
        df_long["Series_Title"] = df_long["Series_Title"].str.replace(word, "")
    return df_long


bls_w_long = reshape_bls(bls_w_clean)
bls_mw_long = reshape_bls(bls_mw_clean)
bls_avg_long = reshape_bls(bls_avg_clean)

ers_fmap_df = df1.copy()
ers_fmap_regions_to_keep = ["National", "Census Region 2: Midwest", "Census Region 4: West"]
ers_fmap_df = ers_fmap_df[ers_fmap_df["region_name"].isin(ers_fmap_regions_to_keep)]
if "Date" not in ers_fmap_df.columns:
    ers_fmap_df["Date"] = pd.to_datetime(ers_fmap_df["year"].astype(str) + "-" + ers_fmap_df["month"].astype(str), format="%Y-%m")
    ers_fmap_df.drop(columns=["year", "month"], inplace=True)

gas_df["Date"] = pd.to_datetime(gas_df["Date"])
diesel_df["Date"] = pd.to_datetime(diesel_df["Date"])
fuel_regions_to_keep = ["U.S.", "Midwest", "West Coast", "Date"]
gas_df = gas_df[fuel_regions_to_keep]
diesel_df = diesel_df[fuel_regions_to_keep]

if "Date" not in df_noaa.columns:
    df_noaa["Date"] = pd.to_datetime(df_noaa["Year"].astype(str) + "-" + df_noaa["Month"].astype(str), format="%Y-%m")
    df_noaa.drop(columns=["Year", "Month"], inplace=True)

df_noaa = df_noaa[["Date", "Name", "Mean"]]

bls_w_long["Date"] = bls_w_long["Date"].dt.strftime("%m-%Y")
bls_mw_long["Date"] = bls_mw_long["Date"].dt.strftime("%m-%Y")
bls_avg_long["Date"] = bls_avg_long["Date"].dt.strftime("%m-%Y")
ers_fmap_df["Date"] = ers_fmap_df["Date"].dt.strftime("%m-%Y")
gas_df["Date"] = gas_df["Date"].dt.strftime("%m-%Y")
diesel_df["Date"] = diesel_df["Date"].dt.strftime("%m-%Y")
df_noaa["Date"] = df_noaa["Date"].dt.strftime("%m-%Y")

df_noaa_pivot = df_noaa.pivot(index="Date", columns="Name", values="Mean")
df_noaa_pivot.columns = ["Midwest_Mean_Temp", "West_Mean_Temp"]
df_noaa_pivot = df_noaa_pivot.reset_index()
```

#### RESULTS
Across the scatter plots, the strength of the relationship between grocery prices and gas prices or weather differs noticeably by category and region. Nevertheless, when there is a relationship, it tends to be linear, with quadratic R^2 scores largely remaining less than or similar to their corresponding linear R^2 score. Certain categories (e.g., produce or meat) in some regions show a moderate positive slope when regressed against gas prices, suggesting a possible link to transportation or distribution costs. Meanwhile, other categories (like bakery or dairy) exhibit weaker or inconsistent patterns. When comparing grocery prices to mean temperature, only a few categories reveal a mild correlation, indicating that weather may influence supply conditions (especially for perishable goods), but the relationship is less consistent. Finally, the scatter plots of gas prices vs. weather show only a modest association, implying that temperature alone does not strongly drive fuel costs.

#### INTERPRETATION
These findings suggest that neither gas prices nor weather uniformly dictates grocery prices across all categories and regions. While certain plots indicate a noticeable linear (or slightly non-linear) trend—implying that higher gas prices or temperatures can coincide with increased grocery prices—the correlations are not universally strong. Seasonal and supply chain factors may interact with weather, particularly for fresh produce, and rising transportation costs could be more pronounced in some categories (e.g., meat in one region) than others.

#### RELEVANCE
By examining these scatter plots and regression fits, we gain insight into which factors—fuel costs or weather—might have a more direct influence on grocery prices. Although there are pockets of moderate correlation, the data suggest that other variables (such as broader economic conditions, specific supply chain disruptions, or local production capabilities) could also play critical roles. This analysis lays the groundwork for deeper time-series methods (like Granger causality) to determine whether changes in gas prices or weather patterns truly precede shifts in grocery prices.

#### RESULTS

For context, Granger causality tests evaluate whether past values of one time series provide statistically significant information about predicting another series beyond its own past values. If one variable "Granger-causes" another, including its historical data improves the forecast of the target variable. A lag value indicates how many past time periods are considered in the test; for instance, a lag of 1 uses the immediately preceding period, while a lag of 4 incorporates data from four periods ago. This method helps identify predictive relationships, although it does not confirm true causation.

```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import grangercausalitytests
import warnings
import random

random.seed(42)

warnings.filterwarnings("ignore")

df_merged["Date"] = pd.to_datetime(df_merged["Date"], format="%m-%Y")
df_merged = df_merged.sort_values("Date")

region_info = {
    "Census Region 2: Midwest": {"gas": "Midwest_Gas", "weather": "Midwest_Mean_Temp"},
    "Census Region 4: West": {"gas": "West Coast_Gas", "weather": "West_Mean_Temp"},
}

food_categories = df_merged["BLS_Category"].unique()

# maximum lag for Granger causality tests
maxlag = 4

for region, vars_map in region_info.items():
    gas_col = vars_map["gas"]
    weather_col = vars_map["weather"]

    region_df_merged = df_merged[df_merged["ERS_region_name"] == region]

    for food_cat in food_categories:
        price_column = "BLS_Value_" + ("West" if region == "Census Region 4: West" else "Midwest")

        # only keep columns that we will use for the analysis and remove duplicates
        df = region_df_merged[(region_df_merged["BLS_Category"] == food_cat)].copy()
        df = df[["Date", "BLS_Series_Title", price_column, gas_col, weather_col]]
        df = df.drop_duplicates()

        average_grocery_price = df.groupby("Date")[price_column].mean()

        df = df.sort_values("Date")

        # -------------------------
        # Plot 1: Grocery Prices vs. Gas Prices
        # -------------------------
        plt.figure()
        fig, ax1 = plt.subplots(figsize=(10, 6))

        ax1.plot(average_grocery_price.index, average_grocery_price, marker="o", label="Average Grocery Price", color="blue", linewidth=1)
        ax1.set_ylabel("Average Grocery Price", color="blue")
        ax1.tick_params(axis="y", labelcolor="blue")

        ax2 = ax1.twinx()
        ax2.plot(df["Date"], df[gas_col], marker="x", label="Gas Price", color="red", linewidth=1)
        ax2.set_ylabel("Gas Price", color="red")
        ax2.tick_params(axis="y", labelcolor="red")

        plt.title(f"{food_cat.title()} in {region}\n Grocery Price vs. Gas Price")
        ax1.legend(loc="upper left")
        ax2.legend(loc="upper right")
        fig.tight_layout()
        plt.show()

        # -------------------------
        # Plot 2: Grocery Prices vs. Weather
        # -------------------------
        fig, ax1 = plt.subplots(figsize=(10, 6))
        ax1.plot(average_grocery_price.index, average_grocery_price, marker="o", label="Average Grocery Price", color="blue", linewidth=1)
        ax1.set_ylabel("Average Grocery Price", color="blue")
        ax1.tick_params(axis="y", labelcolor="blue")

        ax2 = ax1.twinx()
        ax2.plot(df["Date"], df[weather_col], marker="s", label="Weather", color="green", linewidth=1)
        ax2.set_ylabel("Weather (Mean Temp)", color="green")
        ax2.tick_params(axis="y", labelcolor="green")

        plt.title(f"{food_cat.title()} in {region}\nGrocery Price vs. Weather")
        ax1.legend(loc="upper left")
        ax2.legend(loc="upper right")
        fig.tight_layout()
        plt.show()

        avg_gas = df.groupby("Date")[gas_col].mean()
        avg_weather = df.groupby("Date")[weather_col].mean()

        # Merge the aggregated series into a single DataFrame
        granger_df = pd.DataFrame({"Avg_Grocery": average_grocery_price, "Gas": avg_gas, "Weather": avg_weather}).dropna()

        print(f"\nGranger Causality Test for {food_cat} in {region}: Gas Prices predicting Average Grocery Prices")
        granger_result_gas = grangercausalitytests(granger_df[["Avg_Grocery", "Gas"]], maxlag=maxlag, verbose=True)

        print(f"\nGranger Causality Test for {food_cat} in {region}: Weather predicting Average Grocery Prices")
        granger_result_weather = grangercausalitytests(granger_df[["Avg_Grocery", "Weather"]], maxlag=maxlag, verbose=True)

        print("\n" + "=" * 80 + "\n")
```

# Discussion

Our analysis from 2012 to 2018 indicates that fluctuations in gasoline prices were indeed linked with changes in grocery prices, albeit to varying degrees across food categories and regions. In both the Midwest and West, we generally observed that as gas became more expensive, the retail prices of many grocery items trended upward—a cost-push effect via higher transportation and production costs. However, one notable exception emerged in certain scatter plots for the meat category, where gas price and meat price unexpectedly showed a negative (or inverse) correlation. This contradictory result suggests that other forces may overshadow the immediate transportation-cost logic in meat supply chains, such as seasonal production cycles, feed costs, or inventory management strategies. In other words, while higher fuel costs often align with more expensive groceries overall, meat pricing in some cases followed a different trajectory that the scatter plots alone could not fully explain.

To clarify these patterns, we conducted Granger causality tests, which shed light on whether past changes in gas prices or weather could predict future grocery prices. In the Midwest, gas prices do not predict meat prices at any lag—a finding consistent with the scatter plots that either showed weak or inverse associations between these variables. Meanwhile, gas price changes do show a borderline or significant influence on produce, “other,” beverages, and (marginally) dairy at certain lags, indicating that cost shocks may transmit to prices in a delayed manner for some items (e.g., two to four periods) or more immediately for others (e.g., beverages at one period). By contrast, in the West, gas prices significantly predict meat and beverage prices—often at a one-period lag—despite the unexpected negative correlation for some meat observations. This may suggest that when time-lag effects are accounted for, rising fuel costs do, in fact, lead to higher meat prices later. Dairy and bakery categories remain largely insensitive to both gas and weather in a forward-looking sense, aligning with earlier observations of weaker correlations.

Looking at seasonal temperatures, the correlation measures showed a generally weaker relationship with grocery prices. Nevertheless, the Granger causality results refine our perspective: while immediate or same-month temperature effects are small, certain categories—especially produce and occasionally meat—do exhibit significant lagged predictive power from weather. In the Midwest, for instance, weather forecasts produce and meat prices with lags of two to four periods. Similarly, in the West, produce prices show strong predictive ties to weather at lags two through four. This pattern indicates that although average monthly temperature alone does not drive immediate grocery price swings, climatic conditions can become important over longer leads, particularly for agriculture-based items that depend on prior months’ growing conditions.

In terms of the form of these relationships, our findings point predominantly to linear trends. Generally, when fuel costs rose or fell, grocery prices tended to shift in proportion; similarly, any weather-related impacts, if present, manifested in a roughly linear fashion when lagged effects were considered. We tested for possible non-linearities (e.g., quadratic fits) in our exploratory analysis, but the data rarely supported a substantially better fit than the linear model. We did detect occasional mild curvature—suggesting that extreme weather conditions might sometimes trigger sharper price changes—but these were exceptions rather than the consensus.

Crucially, the analysis suggests that gasoline prices exert a stronger and more consistent influence on grocery inflation than seasonal temperatures do, an insight bolstered by both the correlation measures (positive or negative in some isolated cases) and the Granger causality tests. Fuel costs showed moderate to strong associations with many food categories—explaining, in some instances, a sizable share of monthly price variation—whereas ordinary seasonal temperature effects accounted for only a modest fraction of that variance. Even for categories that are climate-sensitive, such as produce, the impact of gas price swings often outweighed the effect of routine temperature fluctuations. The Granger results underscore this further: where gas prices do predict grocery costs, they tend to do so at shorter lags (e.g., one period for beverages or meat in the West), whereas weather influences—though real—appear primarily at longer lags and are less uniform across categories.

## Interpretation in Context of Prior Research

These findings align with broader research on cost drivers of food inflation. Studies from agencies like the USDA’s Economic Research Service have shown that rising fuel and energy costs can exert notable upward pressure on food prices through multiple points in the supply chain. Our results echo these conclusions, as evidenced by both positive correlations in most categories and by the Granger evidence that many items respond to gas price changes within one to four months. Still, the unexpected negative correlation between gas and meat in certain plots underscores that simple pairwise comparisons can be misleading. Sector-specific factors—such as feed costs, livestock supply cycles, or import-export conditions—may cause meat prices to move independently of, or even opposite to, short-term gas price trends.

Meanwhile, the finding that weather can predict produce (and sometimes meat) prices over two-to-four-month lags tracks with well-known agricultural dynamics: climate-related changes in yield or quality typically manifest slowly and may not be reflected in shelf prices until after harvest, processing, and distribution phases. Furthermore, items like dairy and bakery showed minimal responsiveness to either gas prices or weather, suggesting that some categories have more stable input streams or pricing mechanisms that buffer them from month-to-month fluctuations.

## Limitations and Directions for Future Research

Though our analyses reveal useful patterns, several limitations apply. First, we focused on just two U.S. regions over a six-year period. Meat markets, for instance, may behave differently in other parts of the country or under distinct macroeconomic conditions. Second, while correlation and Granger causality are powerful tools, they cannot fully rule out underlying confounders like shifting consumer demand, exchange rates, or labor costs. Future research might explore more advanced time-series frameworks (e.g., vector autoregression with additional controls) or expand the dataset to capture broader economic variables. Third, average monthly temperature may overlook critical climate elements—such as rainfall, storms, or drought severity—that can sharply affect agricultural production. Incorporating event-driven indices could clarify whether the delayed weather impacts we observe are primarily tied to specific extremities rather than generalized seasonality.

In addition, further disaggregating grocery items could help explain paradoxical findings, like the negative gas–meat correlation. Certain meat products or cuts might have shorter or more regionally concentrated supply chains, leading to divergent price movements. Finally, variables such as labor costs, commodity market fluctuations, and policy changes (e.g., agricultural subsidies) were outside our model scope but could be integrated in future analyses to provide a more holistic view of how food prices evolve.

Despite these constraints, our combined correlation and Granger methods offer a clearer perspective into how gas and weather variables shape grocery price movements in different categories and regions. The consistent, if sometimes lagged, effect of fuel costs on multiple items underscores the importance of transportation-related expenses in retail food markets, while the intermittent yet significant weather impacts highlight longer-term agricultural and climate influences. Taken together, these findings suggest that short-term grocery inflation is often dominated by changes in gas prices, yet weather extremities can induce delayed but consequential shifts—particularly for perishable goods like produce.

# Conclusion

Contrary to our initial hypothesis, we did not find uniformly strong positive correlations between gas prices and grocery costs in every category or region—nor did we detect clear nonlinear effects of seasonal temperatures on produce and meat. These findings imply that—while fuel costs remain an important driver of short-term grocery inflation—neither gas prices nor weather conditions dictate retail prices in a strictly predictable manner across all foods or regions. Our results highlight a more nuanced picture, in which individual supply chains, market structures, and timing considerations complicate the straightforward notion that grocery prices would always rise linearly with gas costs or that temperature-induced supply shocks would follow a nonlinear pattern. As a result, we reject our original hypothesis in its entirety and conclude that grocery price dynamics are both more fragmented and less neatly shaped by fuel or weather alone than we predicted.

Despite this outcome, our Granger causality analysis does reveal that fuel price changes can significantly forecast future cost increases for certain categories over one to four-month lags, while weather conditions sometimes produce lagged effects for perishable items. Taken together, these findings suggest that in the short to medium term, stabilizing transportation costs and monitoring climate extremes may still help limit inflation in specific grocery sectors—albeit under more complex, context-dependent circumstances than our hypothesis presumed.

Finally, recognizing the limitations of our study underscores that food pricing remains complex.
