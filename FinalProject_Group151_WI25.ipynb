# Abstract

Caffeine consumption is largely a common practice among undergraduate students and is often used to combat fatigue and enhance focus, or even done unconsciously as a habit. As students belonging to this demographic ourselves, we see this play out in our lives as well as our peers. However, we particularly wanted to analyze the potential impact of caffeine on academic performance, when considering sleep duration as a mediating factor. Hence our study investigates the relationship between daily caffeine consumption patterns and GPA among undergraduate students, primarily from India, with an aim to understand whether increased caffeine intake negatively affects sleep and, in turn, academic success.

To examine this, we integrated two publicly available datasets: the Student Sleep Patterns dataset, which provides information on university students’ sleep habits, caffeine consumption, and study behaviors, and the Student Lifestyle dataset, which contains key variables such as GPA, sleep duration, and study hours. After preprocessing the datasets separately first, for any correlations amongst variables, we went ahead and merged them based on sleep duration. We conducted exploratory data analysis, including correlation analysis and visualizations, to assess potential relationships and trends between caffeine intake, sleep, study hours, and academic performance. 

Our findings indicate no strong linear correlation between caffeine consumption and GPA. We found a moderate positive correlation between caffeine intake and study hours, however, it was not substantial enough to suggest a significant mediating effect of caffeine on academic performance for this dataset. Additionally, we saw that there was a stronger relationship between study hours per day and a student’s GPA.

# Research Question

What is the relationship between daily caffeine consumption patterns and academic performance (measured by GPA) among undergraduate students, primarily from India, accounting for sleep duration and quality as mediating factors?



## Background and Prior Work

Trying to keep up with academic schedules can be challenging for students in university.  Balancing workload and extracurricular activities while also trying to stay healthy is not a simple task. Many students are relying more and more on different forms of caffeine. Caffeine consumption, which helps to overcome exhaustion and preserve focus, is rising along with academic stress. However, less is known about how it affects students’ academic performance and health especially when it comes to quality of sleep.

Caffeine consumption is a widespread habit among college students often linked to academic performance and study habits. Many students use this to counteract sleep deprivation, enhance cognitive function, and stay alert during long study sessions. However, the impact of daily caffeine consumption on academic performance remains a topic of interest, especially when considering sleep quality and duration as mediating factors.

Prior research done by Smith (2002)</a>[<sup>1</sup>](#cite_note-1) found that moderate caffeine consumption improved cognitive performance and alertness, particularly in sleep-deprived individuals. Similarly, a study by McLellan, Caldwell, and Lieberman (2016)</a>[<sup>2</sup>](#cite_note-2) suggested that caffeine enhances reaction time and attention, which could benefit students in academic settings. However, excessive caffeine intake, particularly in the evening, has been shown to negatively impact sleep quality, potentially counteracting its cognitive benefits (Drake et al., 2013)</a>[<sup>3</sup>](#cite_note-3). As demonstrated by a study conducted by Hershner and Chervin (2014)</a>[<sup>4</sup>](#cite_note-4), poor sleep quality is strongly associated with decreased academic performance, showing that sleep deprivation negatively affects memory, attention, and learning capacity. Other studies have explored the direct link between caffeine consumption and GPA. For instance, a study by Truong et al. (2022) examined caffeine consumption patterns among college students and found that while moderate coffee consumption correlated with improved focus and study efficiency, excessive intake led to higher stress levels and reduced academic performance</a>[<sup>5</sup>](#cite_note-5).

## 1. Student Sleep Patterns 

### Checking the overall data:

```python
print("\nDescriptive Statistics:")
print(df1.describe())
```

### Filtering unrelated data:

We dropped the columns not related to our research topic, and focus on the variables we need.

```python
df1_filtered = df1.drop(columns=['Gender', 'University_Year', 'Student_ID', 'Age', 'Screen_Time', 'Physical_Activity', 
                                    'Weekday_Sleep_Start', 'Weekend_Sleep_Start', 'Weekday_Sleep_End', 'Weekend_Sleep_End'])

print("Filtered DataFrame:")
print(df1_filtered.head())
```

We want to first display the distribution of each variable. The shape, center, and spread of the bars helps us identify typical values and any potential outliers or skewness.

```python
df1_filtered.hist(bins=30, figsize=(6, 5), color='green')
plt.tight_layout()
plt.show()
```

### Checking overall correlation between variables present:

```python
numeric_df1 = df1_filtered.select_dtypes(include=['number'])
corr_matrix = numeric_df1.corr()
plt.figure(figsize=(5, 4))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix for Numeric Columns in Student Sleep Patterns")
plt.show()
```

Each cell in this heatmap represents the correlation coefficient between a pair of numeric variables. Darker colors near ±1 indicate stronger linear relationships, while values near zero suggest weak or no linear correlation. Here, we do not see any strong relationships visually represented.

### Checking correlation of Caffeine Intake with Study Hours and Sleep Duration:

```python
sns.regplot(x = df1['Caffeine_Intake'], y = df1['Study_Hours'], scatter_kws={'color':'green'}, line_kws={'color':'black'})
plt.xlabel('Caffeine_Intake')
plt.ylabel('Study_Hours')
plt.title("Caffeine Intake vs Study Hours with Regression Line")
plt.show()
```

```python
sns.regplot(x = df1['Caffeine_Intake'], y = df1['Sleep_Duration'], scatter_kws={'color':'green'}, line_kws={'color':'black'})
plt.xlabel('Caffeine_Intake')
plt.ylabel('Sleep_Duration')
plt.title("Caffeine Intake vs Sleep Duration with Regression Line")
plt.show()
```

The line regression line (black line in the middle) is almost horizontal for both. This suggests that caffeine intake has very little or no significant effect on study hours or sleep duration. But we can only be sure of these relationships once we check the numeric data.

### Numerical correlation:

```python
def calculate_correlations(data):
   cols = data.columns
   correlations = pd.DataFrame(index=cols, columns=cols)
   p_values = pd.DataFrame(index=cols, columns=cols)
   for i in cols:
       for j in cols:
           if i == j:
               correlations.loc[i, j] = 1.0
               p_values.loc[i, j] = 0.0
           else:
               corr, p_val = pearsonr(data[i], data[j])
               correlations.loc[i, j] = corr
               p_values.loc[i, j] = p_val
   return correlations, p_values

correlations, p_values = calculate_correlations(df1_filtered)

print("Correlation Coefficients:\n", correlations)
print("\nP-Values:\n", p_values)
```

We performed a correlation analysis on the numerical variables within our filtered dataset using Pearson’s correlation coefficients and corresponding p-values.
The results indicate weak linear relationships between the variables. The correlation between sleep duration and caffeine Intake was nearly zero (r = -0.0145, p = 0.7458), the correlation between study hours and caffeine intake (r = 0.0325, p=0.4688), as well as the correlation between sleep duration and study hours (r = 0.8045 p = 0.0111). All three suggest that there is no significant linear relationship between caffeine consumption, time spent sleeping or time spent studying.

## 3. Student Sleep Patterns and Lifestyle Dataset

### Merging Student Sleep Patters dataset and Student Lifestyle dataset:

Finding any possible relations between variables. We merge it along the column that calculates number of hours of sleep per day since it is common and exists as a float in both datasets.

```python
df2_renamed = df2_filtered.rename(columns={'Sleep_Hours_Per_Day': 'Sleep_Duration'})
df_merged = pd.merge(
    df1_filtered,
    df2_renamed,
    on='Sleep_Duration',
    how='inner'
)
```

```python
print("Merged Dataset (First 5 Rows):")
print(df_merged.head())
```

### Checking overall data:

```python
print("\nMerged Dataset Info:")
df_merged.info()
```

### Checking overall correlation between variables present:

```python
#Correlation Matrix
numeric_df = df_merged.select_dtypes(include=[np.number])
corr_matrix = numeric_df.corr()

plt.figure(figsize=(8, 5))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix of Merged Dataset")
plt.show()
```

```python
correlations, p_values = calculate_correlations(df_merged)

print("Correlation Coefficients:\n", correlations)
print("\nP-Values:\n", p_values)
```

This merged correlation matrix plot indicates that there is no strong linear relationship between caffeine intake, sleep hours, and GPA. We see that the p-value of _Sleep_Duration_ and _Caffeine_Intake_ is p=0.000003 (p < 0.05) and their correlational coefficient is r=-0.037592. This indicates a non-random association, meaning the relationship is unlikely to be due to chance. However, because the correlation coefficient is so close to zero, the effect is practically negligible.

In our hypothesis, we expected that higher caffeine consumption would be associated with shorter or lower-quality sleep, which in turn would negatively impact academic performance. However, the matrix shows correlation coefficients near zero among _Caffeine_Intake_ and _GPA_, suggesting that, at least in a linear framework, caffeine intake does not exhibit the anticipated negative association with GPA through its effect on sleep. This result implies that other factors may be influencing these relationships or that a more complex, non-linear model might be needed to uncover any underlying patterns.
