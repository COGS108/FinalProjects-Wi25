# Research Question

In the NBA since 2015, are players who use isolation plays (one-on-one playstyle where the rest of the players do not interfere) for more than 10% of their scoring possessions less efficient (as measured by points per possession, effective field goal percentage, and free throw conversions) compared to players who score primarily through off-ball actions (playtypes which rely on teamwork) such as catch-and-shoot, cuts, and screens?

## Background and Prior Work

Isolation plays, where a player goes one-on-one against a defender, are often used by high-usage stars, but they tend to be less efficient on average compared to team-oriented offensive sets. Studies in basketball analytics have shown that isolation plays generally result in lower points per possession (PPP) than plays involving ball movement and off-ball actions like catch-and-shoot and cuts. For instance, Synergy Sports and NBA tracking data indicate that the league-average PPP for isolation plays tends to be around **0.91**, while other play types such as off-screen actions and spot-up shooting yield higher efficiency numbers [[1]](https://heavy.com/sports/nbas-best-iso-players/).

Several researchers and analysts have explored the impact of isolation basketball on team success. Kirk Goldsberry, a prominent basketball analyst, has highlighted in his work that teams with a high frequency of isolation possessions often have lower offensive efficiency unless they feature elite one-on-one scorers like Kevin Durant or James Harden [[2]](https://bestballstats.com/2021/06/13/isolation-basketball-an-inefficient-yet-necessary-offense/). Meanwhile, studies from basketball data platforms like Cleaning the Glass and Basketball Reference suggest that off-ball movement, particularly in systems like those of the Golden State Warriors and San Antonio Spurs, leads to higher team efficiency [[3]](https://www.bruinsportsanalytics.com/post/isolation-vs-teamwork-nba/).

Additionally, effective field goal percentage (eFG%)—a more comprehensive efficiency metric that accounts for three-point shooting-provides further insights into the effectiveness of different scoring styles. Players who rely on isolation often take more contested mid-range shots, which tend to have a lower expected value compared to three-pointers or shots at the rim generated through ball movement. One analysis found that the effective field goal percentage (eFG%) of players in isolation hovers around the mid-40s, while off-screen plays can generate an **eFG% closer to 55%** [[4]](https://medium.com/@luisdepradine/unlocking-nba-efficiency-the-hidden-value-of-isolation-vs-off-screen-play-types-bb59897140d0/).



Given these existing findings, our research aims to quantitatively assess whether players who use isolation plays for more than 10% of their scoring possessions are indeed less efficient than those who rely on off-ball actions. By analyzing NBA player data since 2015, we will compare key efficiency metrics such as PPP and TS% to determine if a strong statistical relationship exists.

Each of these data sets contain various metrics which provide information regarding basketball play types. Each data set shares the same variables, and the important ones are the percentage columns which help us determine efficiency for each play type. PLAYER_ID, PLAYER_NAME, and TEAM_ABBREVIATION are used to identify the player and the team they are on. PLAY_TYPE lists the playtype that the data is for. POSS_PCT is possession percentage which measures the percentage of a player's total possesions that are the playtype being analyzed. PPP stands for points per possesion which measures on average how many points a player scores per possesion that is categorized as the playtype. FT_POSS_PCT measures the percentage of possessions labled as the playtype which end in free throw opportunities. FG_PCT measures a player's field goal percentage (percentage of shots made) during the playtype possesion. TOV_POSS_PCT measures the percentage that the playtype possession results in a turnover (the player turning the ball over to the other team). SCORE_POSS_PCT measures the percentage of playtype possessions that result in any type of score. EFG_PCT is similar to FG_PCT but takes into account the value of three point shots (shots from the free throw line are worth 1 point, inside the arch are worth 2, and beyond the arch are worth 3). Apart from the columns that identify players, teams, and playtype, the variables we will be analyzing are all floats between 0 and 1 representing a percentage. As for the data itself, we just need to remove all rows which have less than 10% isolation play type possessions and remove any unnecessary columns that do not help in calculating efficiency.

```python
import requests
import pandas as pd
import os

import matplotlib.pyplot as plt
%matplotlib inline
```

```python
################################################################
############ USED TO RETRIEVE DATA FROM NBA WEBSITE ############
################################################################
YEAR_SUFFIXES = [i for i in range(15,25)]
PLAYTYPES = ["Isolation", "OffScreen", "Cut", "PRBallHandler", "Spotup"]
PARENT_FOLDER_NAME = "COGS108_Project_Data"
SAVE_DIRECTORY = os.getcwd() +"\\Downloads\\"

# Method used to create folders for data storage
def createFolder(location, name):
    try:
        os.mkdir(location + name)
    except FileExistsError:
        print("Folder Already Exists")

# Method used for formatting API endpoint url
def getUrl(playType, startYearSuffix):
    endYearSuffix = startYearSuffix+1
    return "https://stats.nba.com/stats/synergyplaytypes?LeagueID=00&PerMode=PerGame&PlayType=" + playType + "&PlayerOrTeam=P&SeasonType=Regular%20Season&SeasonYear=20" + str(startYearSuffix) + "-" + str(endYearSuffix) + "&TypeGrouping=offensive"

def getData():
    # Create Folder to hold data and save it to the users downloads directory
    createFolder(SAVE_DIRECTORY, PARENT_FOLDER_NAME)
    SUB_DIRECTORY = SAVE_DIRECTORY + PARENT_FOLDER_NAME +"\\"

    for year_suffix in YEAR_SUFFIXES:
        # Create a subfolder for the given period
        subFolderName = "20"+str(year_suffix)+"-"+str(year_suffix+1)
        createFolder(SUB_DIRECTORY, subFolderName)
        current_directory = SUB_DIRECTORY + "\\" + subFolderName + "\\"

        # Retrieve data for each play type by modifiying API endpoint
        for playType in PLAYTYPES:

            # Define the API endpoint
            url = getUrl(playType,year_suffix)

            # Set the parameters for the API request
            params = {
                "LeagueID": "00",
                "SeasonYear": "2024-25",
                "SeasonType": "Regular Season",
                "PlayerOrTeam": "P",
                "TypeGrouping": "Isolation",
                "PerMode": "PerGame"
            }

            # Define the headers, including the User-Agent
            headers = {
                "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3",
                "Referer": "https://www.nba.com/",
                "Origin": "https://www.nba.com"
            }

            # Make the API request
            response = requests.get(url, headers=headers, params=params)

            # Check if the request was successful
            if response.status_code == 200:
                data = response.json()
                # Extract the relevant data
                headers = data['resultSets'][0]['headers']
                rows = data['resultSets'][0]['rowSet']
                # Create a DataFrame
                df = pd.DataFrame(rows, columns=headers)
                # Save DataFrame as CSV File
                df.to_csv(current_directory + "data_"+playType+"(20" +str(year_suffix)+ "-" +str(year_suffix+1) + ")" + ".csv")
            else:
                print(f"Request failed with status code {response.status_code}")
################################################################
```

```python
def standardize_ints(str_in):
    if type(str_in) != int:
        return int(str_in)
    return str_in

def standardize_types(str_in):
    if type(str_in) != str:
        str_in = str(str_in)
    str_in = str_in.lower()
    str_in = str_in.strip()
    if str_in not in ['isolation', 'cut', 'offscreen', 'prballhandler', 'spotup']:
        return None
    return str_in

def standardize_teams(str_in):
    if type(str_in) != str:
        str_in = str(str_in)
    str_in = str_in.upper()
    str_in = str_in.strip()
    if len(str_in) != 3:
        return None
    return str_in

def standardize_floats(str_in):
    if type(str_in) != float:
        return float(str_in)
    return str_in
```

```python
df_isolation = pd.DataFrame()
df_cut = pd.DataFrame()
df_OffScreen = pd.DataFrame()
df_PRBallHandler = pd.DataFrame()
df_Spotup = pd.DataFrame()

folder_path = "Data"

for foldername in os.listdir(folder_path):
    for filename in os.listdir(folder_path + '/' + foldername):
        sub_df = pd.read_csv(folder_path + '/' + foldername + '/' + filename)
        sub_df = sub_df[sub_df['POSS_PCT'] >= 0.1]
        sub_df = sub_df[['PLAYER_ID', 'PLAYER_NAME', 'TEAM_ABBREVIATION', 'PLAY_TYPE', 'POSS_PCT', 'PPP', 'FT_POSS_PCT', 'FG_PCT', 'TOV_POSS_PCT', 'SCORE_POSS_PCT', 'EFG_PCT']].dropna(how='all') #Only include those we want to focus on
        
        #Cleaning data
        sub_df.dropna() #Remove missing data
        sub_df['PLAYER_ID'] = sub_df['PLAYER_ID'].apply(standardize_ints)
        sub_df['PLAY_TYPE'] = sub_df['PLAY_TYPE'].apply(standardize_types)
        sub_df['TEAM_ABBREVIATION'] = sub_df['TEAM_ABBREVIATION'].apply(standardize_teams)
        sub_df['POSS_PCT'] = sub_df['POSS_PCT'].apply(standardize_floats)
        sub_df['PPP'] = sub_df['PPP'].apply(standardize_floats)
        sub_df['FT_POSS_PCT'] = sub_df['FT_POSS_PCT'].apply(standardize_floats)
        sub_df['FG_PCT'] = sub_df['FG_PCT'].apply(standardize_floats)
        sub_df['TOV_POSS_PCT'] = sub_df['TOV_POSS_PCT'].apply(standardize_floats)
        sub_df['SCORE_POSS_PCT'] = sub_df['SCORE_POSS_PCT'].apply(standardize_floats)
        sub_df['EFG_PCT'] = sub_df['EFG_PCT'].apply(standardize_floats)
        sub_df.dropna() #Remove missing data


        if 'Isolation' in filename:
            df_isolation = pd.concat([df_isolation, sub_df], ignore_index=True)
        elif 'Cut' in filename:
            df_cut = pd.concat([df_cut, sub_df], ignore_index=True)
        elif 'OffScreen' in filename:
            df_OffScreen = pd.concat([df_OffScreen, sub_df], ignore_index=True)
        elif 'PRBallHandler' in filename:
            df_PRBallHandler = pd.concat([df_PRBallHandler, sub_df], ignore_index=True)
        elif 'Spotup' in filename:
            df_Spotup = pd.concat([df_Spotup, sub_df], ignore_index=True)

print(df_isolation.shape)
df_isolation.head()
```

## GRAPHS

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Base directory containing all years
base_dir = "Data"

# Play types (same structure for all years)
play_types = ["data_PRBallHandler.csv", "data_Spotup.csv", "data_Cut.csv", "data_Isolation.csv", "data_OffScreen.csv"]

# List to store data
df_list = []

# Loop through all available years
for year_folder in os.listdir(base_dir):
    year_path = os.path.join(base_dir, year_folder)
    
    # Ensure it is a directory
    if os.path.isdir(year_path):
        for play_type in play_types:
            file_path = os.path.join(year_path, play_type)
            
            if os.path.exists(file_path):
                df = pd.read_csv(file_path)
                df["PLAY_TYPE"] = play_type.replace("data_", "").replace(".csv", "")  # Extract play type name
                df["SEASON"] = year_folder  # Add the season info
                df_list.append(df)

# Combine all seasons into one DataFrame
df_all = pd.concat(df_list, ignore_index=True)
df_all.head()




# Plot 1: Average eFG% by Play Type
efg_by_playtype = df_all.groupby("PLAY_TYPE")["EFG_PCT"].mean().sort_values()
plt.figure(figsize=(10, 6))
plt.barh(efg_by_playtype.index, efg_by_playtype.values, color="orange")
plt.xlabel("Effective Field Goal Percentage (eFG%)")
plt.ylabel("Play Type")
plt.title("Average eFG% by Play Type")
plt.grid(axis="x", linestyle="--", alpha=0.7)
plt.show()

# Plot 2: Top 10 Players by eFG% Across All Play Types
top_players_efg = df_all.nlargest(10, "EFG_PCT")[["PLAYER_NAME", "PLAY_TYPE", "EFG_PCT"]]
plt.figure(figsize=(12, 6))
plt.barh(top_players_efg["PLAYER_NAME"] + " (" + top_players_efg["PLAY_TYPE"] + ")", top_players_efg["EFG_PCT"], color="orange")
plt.xlabel("Effective Field Goal Percentage (eFG%)")
plt.ylabel("Player (Play Type)")
plt.title("Top 10 Players by eFG% Across All Play Types")
plt.grid(axis="x", linestyle="--", alpha=0.7)
plt.show()

# Plot 3: Correlation Between eFG% and Scoring Possession Percentage
plt.figure(figsize=(10, 6))
plt.scatter(df_all["EFG_PCT"], df_all["SCORE_POSS_PCT"], alpha=0.5, color="orange")
plt.xlabel("Effective Field Goal Percentage (eFG%)")
plt.ylabel("Scoring Possession Percentage")
plt.title("Correlation Between eFG% and Scoring Possession Percentage")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()

# Plot 4: eFG% vs. Number of Possessions
plt.figure(figsize=(10, 6))
plt.scatter(df_all["POSS"], df_all["EFG_PCT"], alpha=0.5, color="orange")
plt.xlabel("Possessions (POSS)")
plt.ylabel("Effective Field Goal Percentage (eFG%)")
plt.title("eFG% vs. Number of Possessions")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

The graphs above display a general overview of efficient field goal percentage across all play types. We see that the isolation play type has a lower average eFG% than the off-ball playtypes. We also notice how the cut play type seems to have much greater eFG% than the rest of the play types, this can be easily explained because the scoring in a cut playtype consists entirely of layups or dunks which always have a greater efficiency percentage than mid-range shots or 3-point shots. This will be explored more later on. We also notice a linear relationship between eFG% and Scoring Possession Percentage which makes sense since the the eFG% measures the percentage that shots that go in while scoring possession percentage is the percentage that the play ends in some type of score. The final graph shows how eFG% converges to around 45% eFG% the more possessions one has which also makes sense as this is the around the average eFG% of the enitre NBA league.

### Playtype Total Scoring and Free Throw Conversions
In this section of our analysis we want to look more into how many points are scored with these playtypes by looking at points per possession (PPP) and how often players get to shoot free throws (FT_POSS_PCT) rather than looking at field goal percentages.

```python
# Base directory containing all years
base_dir = "Data"

# Play types (same structure for all years)
play_types = ["data_PRBallHandler.csv", "data_Spotup.csv", "data_Cut.csv", "data_Isolation.csv", "data_OffScreen.csv"]

# List to store data
df_list = []

# Loop through all years
for year in os.listdir(base_dir):
    year_path = os.path.join(base_dir, year)
    
    if os.path.isdir(year_path):
        for play_type in play_types:
            file_path = os.path.join(year_path, play_type)
            
            if os.path.exists(file_path):
                df = pd.read_csv(file_path)
                df["SEASON"] = year  # Add season info
                df_list.append(df)

# Combine all years
df_all = pd.concat(df_list, ignore_index=True)

# Display first few rows
df_all.head()
```

```python
# Group by play type and compute mean
scoring_ft_stats = df_all.groupby("PLAY_TYPE")[["PPP", "FT_POSS_PCT"]].mean().sort_values("PPP", ascending=False)


fig, ax1 = plt.subplots(figsize=(10,6))

ax1.bar(scoring_ft_stats.index, scoring_ft_stats["PPP"], color="orange", label="Points Per Possession (PPP)")

# Line chart for FT Possession Percentage
ax2 = ax1.twinx()
ax2.plot(scoring_ft_stats.index, scoring_ft_stats["FT_POSS_PCT"], color="blue", marker="o", linestyle="-", label="FT Possession %")
ax1.set_ylabel("Points Per Possession (PPP)", color="orange")
ax2.set_ylabel("FT Possession Percentage", color="blue")
ax1.set_xlabel("Play Type")
plt.title("Playtype Total Scoring and Free Throw Conversions Across All Seasons")
ax1.grid(axis="y", linestyle="--", alpha=0.7)
ax1.legend(loc="upper left")
ax2.legend(loc="upper right")

plt.show()
```

The graph presents the total scoring and free throw conversions across all seasons by play type, comparing Points Per Possession (PPP) (orange bars) and Free Throw Possession Percentage (FT_POSS_PCT) (blue line). The Cut play type has the highest PPP, indicating it is the most efficient scoring play and its FT Possession Percentage is relatively high, suggesting that free throws are also a significant factor in its scoring efficiency. Isolation plays, in contrast, generate a moderate PPP but have the second highest FT Possession Percentage, meaning they lead to more free throw opportunities. Spot-up and Off-screen plays show a dip in PPP compared to cut play type and have the lowest FT Possession Percentage, making them less effective in generating scoring through free throws. We see that Pick & Roll Ball handling has a pretty moderate PPP and FT possession percentage. This analysis highlights how different play types contribute to scoring efficiency and drawing fouls in the game.

This aligns with our hypothesis that players relying heavily on isolation plays tend to have lower scoring efficiency than those benefiting from team-generated scoring opportunities.

Our research and findings have extended any previous research stated in our background section by considering the effectiveness and importance of free throw possessions per each play type. The previous research mainly looked into shooting percentages and points per possession while our project dove into which plays have higher free throw possession percentage and what that could mean for scoring and efficiency. We concluded that a high free throw possession percentage can give a positive effect on scoring as it gives the player an opportunity at free points. It also gives a strategic advantage by pausing the clock and giving players time to rest.

Overall, our findings highlight the importance of off-ball movement and team-oriented play in maximizing scoring efficiency, reinforcing the strategic value of creating open shot opportunities rather than relying heavily on one-on-one isolation plays.
