With the invention of the TV, its popularity further increased as people around the world could tune in to the national teams and show off their skills and abilities. Our group was interested in seeing if the placement of a national team in the World Cup had an impact on audience attendance in their respective domestic Leagues. 

We decided to investigate this correlation for the most watched leagues in the world (commonly referred to as the big five) which include England, Germany, France, Italy, and Spain. In order to learn if there was a correlation, we gathered data that contains the attendance of the domestic leagues, their placement in the World Cups, and what awards they received at the World Cups. ### Tidy and clean data for the placement parts of the dataset. Use the wcl datasets, wc datasets, and any outside sources to create a clean dataset for the first half of df_1 with the following:

#### Year: What year this country's placement in the world cup occured
#### Country: The country we are looking at for that year
#### Placement: How far the country got in the World Cup. Possible Values are 1st, 2nd, 3rd, 4th, 8th, 16, DNQ


```python
#Making life easier by creating a "winner" column in wc_stats_2. 
#This winner column will have either the name of the winner or "N/A" (in case of ties or other circumstances). 
#Winners are determined by who has the higher goals (seperated by hyphen) 
#Unless a tie game is reached in a game where a winner is needed. Then penalties are used (denoted by paranthesis)
home_team_array = np.array(wc_stats_2["home_team"])
away_team_array = np.array(wc_stats_2["away_team"])
score_array = np.array(wc_stats_2["score"])

def delete_substrings(string, substrings):
    for substring in substrings:
        string = string.replace(substring, "")
    return string


#winner = np.array([""] * len(score_array))
winner = np.array([])
for i, score in enumerate(score_array):
    if("(" in score):
        score = score.strip()
        score_list = score.split(" ")
        try:
            home_penalty = int(delete_substrings(score_list[0], [")","("]).strip())
            away_penalty = int(delete_substrings(score_list[2], [")","("]).strip())
        except:
            raise Exception("Error with penalty handling")
        if(home_penalty > away_penalty):
            winner = np.append(winner, away_team_array[i])
            #winner[i] = away_team_array[i]
        elif(home_penalty < away_penalty):
            winner = np.append(winner, home_team_array[i])
            #winner[i] = home_team_array[i]
        else:
             winner = np.append(winner, "N/A")
            #winner[i] = "Tie"
        
    elif("–" in score):
        score = score.strip()
        score_list = score.split("–")
        try:
            home_score = int(score_list[0].strip())
            away_score = int(score_list[1].strip())
        except:
            raise Exception("Error with score handling")
        
        if(home_score > away_score):
            winner = np.append(winner, home_team_array[i])
            #winner[i] = home_team_array[i]
            
        elif(home_score < away_score):
            winner = np.append(winner, away_team_array[i])
            #winner[i] = away_team_array[i]
        else:
            winner = np.append(winner, "N/A")
            #winner[i] = "Tie"
    else:
        winner[i] = "N/A"
    winner[i] = winner[i].strip()

wc_stats_2 = wc_stats_2.assign(winner= winner)
```

```python
#The check placement function should utilize the datasets provided to find the placement of a country during a specific year's world cup. 
#The placements (both regular and encoded) are defined as the following:
    #1st or 1: Won the World Cup as champion. Played 7 matches in total
    #2nd or 2: Lost World Cup in final match. Played 7 matches in  total
    #3rd and 4th or 3 and 4: Lost World Cup in the semi-finals. Played 7 matches in total due to the 3rd place match that occurs
    #8th or 8: Lost World Cup in the Quarter Finals. There is no placement matches after 4th, so these countries only played 5 matches
    #16th or 16: Lost World Cup in the Group of 16 matches. Only played 4 matches, as during group matches you must play a minimum of 3 matches to earn points to move on.
    #Qualified or 32: Lost Word Cup in the general Group macthes. Must play a minimum of 3 matches no matter what.
    #DNQ or 64: This country didn't qualify for the world cup in the year it was played. 0 matches

def check_placement(country_name, year, encoded):
    if(year < 2022 and year > 1997):
        temp_matches_df = wc_stats_1[(wc_stats_1["year"] == year) & ((wc_stats_1["home_team"] == country_name) | (wc_stats_1["away_team"] == country_name))]
        matches_appeared = temp_matches_df.shape[0]
        if(matches_appeared == 7):
            if(wcl_data_1[wcl_data_1["year"] == year]["winner"].values[0] == country_name):
                
                if(encoded): 
                    return 8
                else: 
                    return "1st"
            elif(wcl_data_1[wcl_data_1["year"] == year]["second"].values[0] == country_name):
                if(encoded): 
                    return 7 
                else: 
                    return "2nd"
            elif(wcl_data_1[wcl_data_1["year"] == year]["third"].values[0] == country_name):
                if(encoded): 
                    return 6 
                else: 
                    return "3rd"
            elif(wcl_data_1[wcl_data_1["year"] == year]["fourth"].values[0] == country_name):
                if(encoded): 
                    return 5 
                else: 
                    return "4th"
            else:
                return np.nan

        elif(matches_appeared == 5):
            if(encoded): 
                return 4 
            else: 
                return "8th"
        elif(matches_appeared == 4):
            if(encoded): 
                return 3 
            else: 
                return "16th"
        elif(matches_appeared == 3):
            if(encoded): 
                return 2 
            else: 
                return "Qualified"
        else: 
            if(encoded): 
                return 1 
            else: 
                return "DNQ"
            
    elif(year == 2022):
        temp_matches_df = wc_stats_2[((wc_stats_2["home_team"] == country_name) | (wc_stats_2["away_team"] == country_name))]
        matches_appeared = temp_matches_df.shape[0]
        if(matches_appeared == 7):
            if(wcl_data_2[wcl_data_2["Year"] == year]["Champion"].values[0] == country_name):
                if(encoded): 
                    return 8 
                else: 
                    return "1st"
            elif(wcl_data_2[wcl_data_2["Year"] == year]["Runner-Up"].values[0] == country_name):
                if(encoded): 
                    return 7 
                else: 
                    return "2nd"
            else:
                if(temp_matches_df[-1]["winner"].values[0] == country_name):
                    if(encoded): 
                        return 6
                    else: 
                        return "3rd"
                else:
                    if(encoded): 
                        return 5 
                    else: 
                        return "4th"
        elif(matches_appeared == 5):
            if(encoded): 
                return 4 
            else: 
                return "8th"
        elif(matches_appeared == 4):
            if(encoded): 
                return 3 
            else: 
                return "16th"
        elif(matches_appeared == 3):
            if(encoded): 
                return 2 
            else: 
                return "Qualified"
        else: 
            if(encoded): 
                return 1 
            else: 
                return "DNQ"
    else:
        return np.nan
```

#### Tournament Year Column Code

Will be used to aggregate normalized both attendance scores into a mean for each time the FIFA tournament occured so we can get a general idea of whether it's increasing or decreasing.

```python
#Function that will be applied to each row and give a consistent year that the last FIFA tournament occured for the year of attendance recorded. (This includes if the FIFA tournament occured in the same year)

def find_tournament(year):
    if(year < 2002):
        return 1998
    elif (year < 2006):
        return 2002
    elif(year < 2010):
        return 2006
    elif(year < 2014):
        return 2010
    elif(year < 2018):
        return 2014
    elif(year < 2022):
        return 2018
    elif(year < 2026):
        return 2022
```

#### Seeing the award distributions for each country over the entire time frame

```python
#List all the unique awards in order
unique_awards_for_distribution = ['Golden Ball', 'Silver Ball', 'Bronze Ball', 'Golden Boot', 'Silver Boot', 'Bronze Boot', 'Golden Glove', 'Best Young Player']
```

```python
#Create a function that will allow us to read each column for a dataframe in the important award columns and return a series of values that can be used to create a distribution of awards types.

def award_distribution(dataframe, country_name, award_types = unique_awards_for_distribution):
    return_dict = {}
    country_specific_df = dataframe[dataframe["Country"] == country_name]
    for award in unique_awards_for_distribution:
        return_dict[award] = [country_specific_df[award].sum()]
    #print(return_dict)
    return pd.DataFrame.from_dict(return_dict)

```

```python
fig, axes = plt.subplots(nrows=1, ncols=1, figsize=(10, 5), sharey = True)
axes = sns.barplot(data=award_distribution(df_1, "England"), color = palette_country["England"])
axes.set_title('Distribution of Award Types For England')
axes.set_xlabel("Award Types")
axes.set_ylabel("Amount of Times Country Received Type of Award")
axes.tick_params(axis='x', which='major', labelsize=8)
plt.show()
```

England has not won many awards over the course of FIFA world cups we chose for our period, with the only ones being Golden Boot and Best Young Player. As such, while they may not be affected by not winning any awards, the years where they did win the awards may show a signifigant increase in attendance if the awards won at the FIFA world cup acts as a legitimate confound predictor.

```python
fig, axes = plt.subplots(nrows=1, ncols=1, figsize=(10, 5), sharey = True)
axes = sns.barplot(data=award_distribution(df_1, "France"), color = palette_country["France"])
axes.set_title('Distribution of Award Types For France')
axes.set_xlabel("Award Types")
axes.set_ylabel("Amount of Times Country Received Type of Award")
axes.tick_params(axis='x', which='major', labelsize=8)
plt.show()
```

France has won a pretty equal distribution of awards over the time frame we chose and seems to have won Bronze Ball and Best Young Player the most. If awards are a legitimate confound predictor, then they will most likely see spikes in attendances when many or high ranking awards are won, but low spikes in attendances when not many or low-ranking awards are won in a year. Awards may not even have that much of an influence if they earn awards pretty consistently.

```python
fig, axes = plt.subplots(nrows=1, ncols=1, figsize=(10, 5), sharey = True)
axes = sns.barplot(data=award_distribution(df_1, "Germany"), color = palette_country["Germany"])
axes.set_title('Distribution of Award Types For Germany')
axes.set_xlabel("Award Types")
axes.set_ylabel("Amount of Times Country Received Type of Award")
axes.tick_params(axis='x', which='major', labelsize=8)
plt.show()
```

## Analysis of Linearity Between Predictors and Both the Normalized Attendance Trends Per 4 Years For All Countries Over Time Frame

While previous trends for both predictors don't show any clear linear positive relationship and that our hypothesis is disproven, it's still good to check how they fit and if forced into a linear relation, how much of an increase is caused if any at all.

We then see if the trend has a more positive or negative direction depending on either predictor and the likelihood of the effect not being due to random chance (used through Pearson's correlation coefficent and p-values in linear regression models).

```python
fig, axes = plt.subplots(nrows = 1, ncols = 1, figsize = (10,5))
sns.scatterplot(x='Placement (Encoded)', y='Award Amount', data = df_1, axes = axes);
axes.set_title('Encoded Placement and Award Amount Comparision')
axes.set_xlabel("Encoded Placement")
axes.set_ylabel("Amount of Awards Won")
print("")
```

These two predictors has a positive relation, where with a better placement comes an increase in the amount of awards won (which makes sense). This means in linear regression we must be wary that these two values will very likely be extremely eqqual to each other in chance of being a predictor. 

### BEFORE DOING THE NEXT SECTIONS, DATAFRAMES MUST BE COMBINED TO STUDY LINEARITY IN GRAPH AND HOW EFFECTIVE LINEAR REGRESSION IS

This means finding the mean z-score for both normalized attendance scores of each 4 year span of the tournament that can be found from df_2 so we can see whether it's generally increasing or decreasing more in that time frame and then combining it with df_1 to study effects 

```python
mean_average_attendance_tourn = df_2[["Tournament Year", "Country", "Normalized Average Attendance Per Match"]].groupby(by=["Tournament Year", "Country"]).mean(numeric_only=True).reset_index()
merged_average_linearity = df_1.merge(mean_average_attendance_tourn, left_on = ["Year", "Country"], right_on = ["Tournament Year", "Country"])
merged_average_linearity.head()
```

```python
mean_total_attendance_tourn = df_2[["Tournament Year", "Country", "Normalized Total Attendance"]].groupby(by=["Tournament Year", "Country"]).mean(numeric_only=True).reset_index()
merged_total_linearity = df_1.merge(mean_total_attendance_tourn, left_on = ["Year", "Country"], right_on = ["Tournament Year", "Country"])
merged_total_linearity.head()
```

### CHECK MAIN PREDICTOR, CONFOUND PREDICTOR, AND COMBINED EFFECT ON AVERAGE ATTENDANCE THROUGH LINEARITY

To use methods other than human judgement on whether an effect is actually present or not for average attendance per match across all stadiums in a league, we use the mean of the normalized average attendance per match scores over a period of 4 years after a FIFA world cup. This score acts as a representative for the trend itself and will then it's value will then be categorized in a scatter plot with types of placement are amounts of awards. After this, we see how linear the relation actually is and further prove it's due to random chance and that our hypothesis is incorrect (for at least average attendance per match)

#### Effect of Main Predictor Alone

```python
fig, axes = plt.subplots(nrows = 1, ncols = 1, figsize = (10,5))
sns.regplot(x="Placement (Encoded)", y="Normalized Average Attendance Per Match", data = merged_average_linearity)
axes.set_title('Linearity between Placement and Normalized Average Attendance')
axes.set_xlabel("Encoded Placement")
axes.set_ylabel("Mean of Normalized Average Attendance For a 4 Year Time Span Following a FIFA World Cup")
correlation_coefficient = np.corrcoef(merged_average_linearity["Placement (Encoded)"].to_numpy(), merged_average_linearity["Normalized Average Attendance Per Match"].to_numpy())[0][1]
print("Pearson's Correlation Coeeficient of Placement and Normalized Average Attendance:", correlation_coefficient)
```

The main predictor and all normalized average attendance scores obviously don't have a have a high fit and it's shown in the graph. Values are dispered in each placement and the amount of times all countries scored higher is severly less. The line seen is very unlikely to happen due to the Pearson correlation coefficient score being below the .7 range, and even if it was high, the amount of standard deviations the actual average attendance changes over the 4 years is minisculely positive.

If anything it seemed like our confound predictor (which represents player rather than the team performance at the FIFA World Cup) was more likely to affect either of them, although that was mostly due to chance still and may have just been due to the fact that a lot of attendance 4-year trends for either figure in any league had also had the corresponding country win 0 awards, with the few that didn't coincidentally being higher attendance.

There are also many other ways to both measure team performance at FIFA (total goals scored, total goals blocked, yards run for the entire team etc) and to measure popularity of the league back at home (ticket sales, news reports, social media posts) that could've given us a better idea of whether a country was supporting their top domestic league after a positive/negative FIFA world cup performance.

All in all, while our project raised an interesting idea of whether performance of a sport at a higher level can affect domestic leagues back home, the results disproved that this was the case. While not conclusive, it's more likely that instrinsic performance of the domestic leagues themselves affects their own spectator attendance and in turn, their own popularity.
