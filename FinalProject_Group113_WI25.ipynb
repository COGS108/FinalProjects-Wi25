### **2.1 Net Price Dataset**

- **Data Content**: This dataset contains the average net price for students across different income brackets at various U.S. institutions. Net price reflects the actual cost students pay after accounting for grants and scholarships.

- **Purpose:** College affordability is a major concern for students and families. Analyzing net price helps compare institutions in terms of financial accessibility and allows us to assess the real cost of attending different universities for students from diverse economic backgrounds.

- **Desired Output**:
A clean dataset containing:
  - Institution names
  - A computed overall average cost for each institution

**2.1.1. Import and Data Selection**
- We load the dataset from an Excel file.
- We select columns relevant to institution name and net price by income bracket for 2021-22, as this represents the most recent and standardized cost estimates.

```python
# Load the Institution Tuition data set in data frame
tuition_df = pd.read_excel("dataset/net-price.xlsx")

# See the headers
tuition_df.head()
```

```python
# Selected relevant columns:
# including institution name and the average net price for all income levels during the 2021-2022 academic year

tuition_columns = [
    'Institution Name',
    'Average net price-students awarded grant or scholarship aid  2021-22 (SFA2122)',
       'Average net price (income 0-30 000)-students awarded Title IV federal financial aid  2021-22 (SFA2122)',
       'Average net price (income 30 001-48 000)-students awarded Title IV federal financial aid  2021-22 (SFA2122)',
       'Average net price (income 48 001-75 000)-students awarded Title IV federal financial aid  2021-22 (SFA2122)',
       'Average net price (income 75 001-110 000)-students awarded Title IV federal financial aid  2021-22 (SFA2122)',
       'Average net price (income over 110 000)-students awarded Title IV federal financial aid  2021-22 (SFA2122)',
]

# create a new dataframe, called 'tuition_df' and put all the selscted columns in that dataset
tuition_df = tuition_df[tuition_columns]
tuition_df.head()
```

**2.1.2.Data Cleaning**
- **Compute the Average Cost:** Since tuition varies by income bracket, we calculate an institution-wide average cost to obtain a single affordability metric per school. This allows for institution-to-institution affordability comparisons using a single value.

```python
# Define the list of net price columns (excluding 'Institution Name')
price_columns = tuition_columns[1:]  # Excluding column 1: 'Institution Name'

# Create a new column 'average_cost' that computes the row-wise mean across these columns
tuition_df['average_cost'] = tuition_df[price_columns].mean(axis = 1)
```

- **Handle Missing Values:** Some institutions lack tuition data for certain income brackets, leading to incomplete records. We remove institutions without a valid overall average cost to maintain dataset quality. 

```python
# Drop institutions missing average cost data
tuition_df = tuition_df.dropna(subset=['average_cost'])
```

- **Standardize Institution Names:** Since we will merge multiple datasets, standardizing institution names ensures consistent matching across datasets. This could prevent mismatches due to capitalization differences adn avoids formatting inconsistencies that could cause errors during data merging.

```python
# Convert institution names to lowercase
tuition_df['Institution Name'] = tuition_df['Institution Name'].str.lower()

# Remove special characters (ex. punctuation, hyphens)
tuition_df['Institution Name'] = tuition_df['Institution Name'].str.replace(r'[^a-zA-Z0-9\s]', ' ', regex=True)

tuition_df = tuition_df[["Institution Name", "average_cost"]]
```

**2.1.3. Data Storage & Export:**
- To ensure efficient data management, we store the cleaned dataset in a dedicated folder for further analysis. This allows for easy reproducibility and prevents accidental data loss, while ensuring that all future analysis uses the cleaned dataset.

```python
# Save 'tuition_df' as a csv file in 'clean' folder
tuition_df.to_csv('dataset/clean/tuition_cleaned.csv', index=False)

# Show the final dataset
tuition_df.head()
```

- We load the 'Sticker Price' dataset from an Excel file.

```python
# Load the aticker Price data set in data frame
living_cost = pd.read_excel("dataset/sticker-price.xlsx")

# see the variables
living_cost.head()
```

```python
# Keep columns: Institution name, and total on and off campus cost for year of 22-23
living_cost_columns = [
    'Institution Name',
    'Total price for in-state students living off campus (not with family) 2022-23',
    'Total price for in-state students living on campus 2022-23'
]

# Create a new dataframe called 'living_cost', which stores only the columns saved in 'living_cost_columns'
living_cost = living_cost[living_cost_columns]
```

**2.3.2. Data Cleaning**
- **Renaming Columns for Simplicity:** Rename long column names 'to off_campus' and 'in_campus' for easier access during analysis, while making the dataset clear to see.

```python
# Rename columns for clarity: change the original price related columns to 'off_campus' and 'in_campus'
living_cost.rename(columns={
    "Total price for in-state students living off campus (not with family) 2022-23": "off_campus",
    "Total price for in-state students living on campus 2022-23": "in_campus"
},inplace=True)

living_cost.head()
```

- **Handling Missing Values:** From the 'living_cost' dataset, we have noticed multiple 'NaN' value in the price columns. We aim to replace "NaN" values with "0" to avoid calculation errors for average and identify institutions lacking cost data.

```python
# Fill missing values with zero in both 'off_campus' and 'in_campus' columns
living_cost['off_campus'] = living_cost['off_campus'].fillna(0)
living_cost['in_campus'] = living_cost['in_campus'].fillna(0)
```

- **Calculate Average Living Cost:** Compute 'the average_living_cost' as the maximum of 'off_campus' and 'in_campus costs'. This ensures the dataset reflects the higher cost a student might face either from on-campus or off-campus.

```python
# Compute the maximum living cost per institution
living_cost['average_living_cost'] = living_cost[['off_campus', 'in_campus']].max(axis=1)

living_cost.head()
```

- **Filtering Out Institutions Without Living Cost Data:** Some institutions still have 0 in the average_living_cost column, meaning they lack reliable cost data. We remove these rows.

```python
# Keeps only rows with all non-zero value
living_cost = living_cost[living_cost['average_living_cost'] != 0]

# Drop unnecessary columns: 'off_campus', 'on_campus'
living_cost.drop(columns=['off_campus', 'in_campus'], inplace=True)
living_cost.head()
```

- **Standardizing Institution Names:** Convert institution names to lowercase and remove special characters to facilitate consistent merging with other datasets.

```python
# Convert institution names to lowercase
living_cost['Institution Name'] = living_cost['Institution Name'].str.lower()

# Remove special characters (to match other datasets)
living_cost['Institution Name'] = living_cost['Institution Name'].str.replace(r'[^a-zA-Z0-9\s]', ' ', regex=True)
```

```python
import requests
from lxml import html

# List of URLs to scrape
urls = [
    "https://www.oedb.org/rankings/graduation-rate/#table-rankings",
    "https://www.oedb.org/rankings/graduation-rate/page/2/#table-rankings",
    "https://www.oedb.org/rankings/graduation-rate/page/3/#table-rankings",
    "https://www.oedb.org/rankings/graduation-rate/page/4/#table-rankings",
    "https://www.oedb.org/rankings/graduation-rate/page/5/#table-rankings",
    "https://www.oedb.org/rankings/graduation-rate/page/6/#table-rankings"
]

# Custom headers to mimic a real browser
headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/98.0.4758.102 Safari/537.36"
    )
}

all_data = []

for url in urls:
    # Make the request
    response = requests.get(url, headers=headers)
    response.raise_for_status()  # Raise an HTTPError if the request was not successful
    
    # Parse the HTML content
    tree = html.fromstring(response.content)
    
    # Select all table rows using XPath
    rows = tree.xpath('/html/body/div[1]/div/div[1]/table/tbody/tr')
    
    for row in rows:
        # Extract the institution name
        name = row.xpath('./td[2]/h3/text()')
        # Extract the graduation rate
        grad_rate = row.xpath('./td[4]/text()')
        
        # Clean and handle missing data
        name = name[0].strip() if name else ''
        grad_rate = grad_rate[0].strip() if grad_rate else ''
        
        # Append to our list
        all_data.append([name, grad_rate])

# Create a DataFrame from all_data
grad_df = pd.DataFrame(all_data, columns=['Institution Name', 'Graduation Rate'])

# Optionally, remove any duplicates
grad_df.drop_duplicates(inplace=True)

# Preview the first few rows
print(grad_df.head())

# Save to CSV if desired
# grad_df.to_csv('grad_df.csv', index=False)
```

## Re-Ranking Schools Using a Normalized Composite Score

To address potential biases and ensure ethical rigor in our ranking model, we developed a composite scoring system grounded in direct input from computer science (CS) students.

Recognizing that no universal weighting scheme exists for evaluating universities, we conducted research from fellow CS students across diverse demographics (including international students, first-generation students, and those from varying socioeconomic backgrounds), and composed the following weighting formula based on their feedback.

### Weighting Scheme

- **average_cost (10%)** + **living_cost (10%)**  
  - **Lower** cost is **better**; invert these values before scaling.
- **acceptance_rate (5%)** + **graduation_rate (5%)**  
  - **Higher** rates are **better**; no inversion needed.
- **safety_index (10%)**  
  - **Higher** is **better**; no inversion needed.
- **rank (20%)**  
  - **Better rank** = smaller numeric value, so we invert it to favor lower ranks.
- **Starting Median Salary (20%)**  
  - **Higher** is **better**; no inversion needed.
- **Mid-Career Median Salary (20%)**  
  - **Higher** is **better**; no inversion needed.

While the default weights reflect aggregate student preferences, we explicitly acknowledge that individual priorities vary. Therefore, we choose to leave the project open for public usage and encourage users to adjust weight based on their unique goals. For instance, a student prioritizing immediate job placement might increase career outcomes to 50% and reduce selectivity to 0%. A safety-conscious applicant might allocate 25% to safety and lower affordability to 20%, this makes the model more accessible for future researchers to manipulate with revised weightings to provide further insights.

#### Ethical Justification for Weightings
- Participatory Design:
By centering student voices, rather than relying on assumptions or institutional rankings, we mitigate researcher bias and align the model with the lived experiences of its intended users.
For example, one student emphasized, "I care more about graduating with minimal debt than getting into a 'top-ranked' school," explaining the high weight for affordability. Another noted, "A high starting salary matters, but I also want to feel safe on campus," justifying the balance between career outcomes and safety.
Below is an approach to create a new ranking for each institution, incorporating multiple factors with customized weights. By inverting certain columns (where lower values are better) and normalizing all metrics to a [0, 1] scale using min-max scaling, we ensure that each variable contributes fairly to the final score. We then compute a weighted sum of these scaled values and sort by the resulting composite score in descending order (best to worst).



### Steps to Compute the Composite Score

1. **Invert Columns Where Lower is Better**  
   - Multiply columns like `average_cost`, `living_cost`, and `rank` by -1 so that lower values become higher values.

2. **Min-Max Scale All Relevant Columns**  
   - Transform each (inverted or original) column to the range [0, 1].  
   - A value of 0 corresponds to the worst performance in that column, while 1 corresponds to the best.

3. **Compute the Weighted Sum**  
   - Multiply each scaled column by its respective weight.  
   - Sum these weighted values to produce a final `custom_score` in the range [0, 1] (assuming your total weights sum to 1).

4. **Sort by Composite Score**  
   - Sort the DataFrame in descending order of `custom_score`.  
   - Institutions with higher scores rank better under this custom scheme.

By **normalizing** the data, you prevent any single metric from dominating due to its scale. Inversion ensures that “lower is better” columns contribute positively once scaled. The final ranking thus offers a balanced view of each institution’s performance across cost, selectivity, safety, rank, and salary outcomes.


```python


# Make a copy of the original DataFrame to avoid altering it
df_rerank = df.copy()

# 1. Rename columns to ensure they match the code references
#    Make sure these columns actually exist in df. If your DataFrame uses different names,
#    adjust them here accordingly.
df_rerank.rename(columns={
    "Acceptance Rate": "acceptance_rate",
    "Graduation Rate": "graduation_rate",
    "Overall Safety Index": "safety_index"
}, inplace=True)

# 2. Invert columns where a lower value is better (average_cost, average_living_cost, rank)
df_rerank['avg_cost_inv'] = -df_rerank['average_cost']
df_rerank['living_cost_inv'] = -df_rerank['average_living_cost']
df_rerank['rank_inv'] = -df_rerank['rank']

# 3. Define the columns to scale
#    (For columns where "lower is better," we use the inverted versions;
#     for columns where "higher is better," we use them as-is.)
cols_to_scale = [
    'avg_cost_inv',              # inverted average_cost
    'living_cost_inv',           # inverted living_cost
    'acceptance_rate',           # higher is better
    'graduation_rate',           # higher is better
    'safety_index',              # higher is better
    'rank_inv',                  # inverted rank
    'Starting Median Salary',    # higher is better
    'Mid-Career Median Salary'   # higher is better
]

# 4. Apply min-max scaling to transform all selected columns into [0, 1]
scaler = MinMaxScaler()
df_rerank[cols_to_scale] = scaler.fit_transform(df_rerank[cols_to_scale])

# 5. Compute the weighted sum (custom_score)
#    The total weight is 1.0 if we sum all specified percentages:
#      10% + 10% + 5% + 5% + 10% + 20% + 20% + 20% = 100%
df_rerank['custom_score'] = (
    0.10 * df_rerank['avg_cost_inv'] +
    0.10 * df_rerank['living_cost_inv'] +
    0.05 * df_rerank['acceptance_rate'] +
    0.05 * df_rerank['graduation_rate'] +
    0.10 * df_rerank['safety_index'] +
    0.20 * df_rerank['rank_inv'] +
    0.20 * df_rerank['Starting Median Salary'] +
    0.20 * df_rerank['Mid-Career Median Salary']
)

# 6. Sort the DataFrame by this new composite score in descending order
df_rerank = df_rerank.sort_values(by='custom_score', ascending=False).reset_index(drop=True)

# After sorting df_rerank by custom_score in descending order, assign a new "rerank" column
df_rerank['rerank'] = df_rerank.index + 1

# Display the top 20 schools with the new rerank column
df_rerank[
    [
        'rerank',
        'Institution Name',
        'rank',
        'average_cost',
        'average_living_cost',
        'acceptance_rate',
        'graduation_rate',
        'safety_index',
        'Starting Median Salary',
        'Mid-Career Median Salary',
        'custom_score'
    ]
].head(20)
```
