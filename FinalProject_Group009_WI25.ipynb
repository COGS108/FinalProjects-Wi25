## Data Cleaning
We will be using pandas to obtain the data from our "data" directory where all our CSVs are. There are a total of 33 datasets that all have the same entries only seperated by years. Thus here will being the cleaning process. We will go through each year where there are 3 sets per year. Well will hold these DataFrames, modify their column names and merge all. For each year, we will only if the country shows in those three datasets. Then well will add them to the data list.

After we do this process to each year, we will merge the list into only large DataFrame which hold all the data we need for each country. Each country will have 33 columns in total. Any countries that to not appear in all the datasets will be dropped at the end of the cleaning process.

```python
import pandas as pd
print(pd.__version__)
```

```python
data = []
for i in range(2010, 2021):   
    df_alcohol = pd.read_csv(f"data/alcohol_consumption_{i}.csv")
    df_household_income = pd.read_csv(f"data/household_income_{i}.csv")
    df_life_expectancy = pd.read_csv(f"data/life_expectancy_{i}.csv")
    df_temp = df_alcohol.merge(df_household_income, on="Category", how="inner").merge(df_life_expectancy, on="Category", how="inner")
    df_temp = df_temp.rename(columns={"Gross disposable income per capita of households and NPISH": f"Disposable Income ($) {i}"})
    df_temp[f"Disposable Income ($) {i}"] = df_temp[f"Disposable Income ($) {i}"].round(2)
    df_temp = df_temp.rename(columns={"Category": "Country"})
    df_temp = df_temp.rename(columns={"Litres per person": f"Litres per person {i}"})
    df_temp = df_temp.rename(columns={"Life expectancy at birth": f"Life expectancy at birth {i}"})
    data.append(df_temp)
df = data[0]
for d in data[1:]:
    df = pd.merge(df, d, on="Country", how="outer")
df = df.dropna()
df = df.reset_index()
df = df.drop(columns=['index'])
new_df = pd.DataFrame(columns=['Country', 'Year', 'Alcohol Consumption (Litre/Person)', 'Disposable Income ($)', 'Life Expectancy (Years)'])

for index, row in df.iterrows():
    country = row['Country']
    for i in range(2010, 2021):
        entry = [country, i]
        entry.append(row[f'Litres per person {i}'])
        entry.append(row[f'Disposable Income ($) {i}'])
        entry.append(row[f'Life expectancy at birth {i}'])
        new_df.loc[len(new_df)] = entry
df = new_df
df.to_csv("cleandata.csv", index=False)
```

### Section 2 - What are the outliers?

```python

df = pd.read_csv("cleandata.csv")

# Set plot style

sns.set_style("whitegrid")

# Create subplots for distributions
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Boxplot Distribution
sns.boxplot(df["Alcohol Consumption (Litre/Person)"], ax=axes[0], color="blue")
axes[0].set_title("Alcohol Consumption Boxplot")

sns.boxplot(df["Disposable Income ($)"], ax=axes[1], color="green")
axes[1].set_title("Disposbale Income Boxplot")

sns.boxplot(df["Life Expectancy (Years)"], ax=axes[2], color="red")
axes[2].set_title("Life Expectancy Boxplot")

plt.tight_layout()
plt.show()

# Using IQR to find outliers
def find_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    iqr = Q3 - Q1
    lower = Q1 - 1.5 * iqr
    upper = Q3 + 1.5 * iqr
    return data[(data[column] < lower) | (data[column] > upper)]

alc_outliers = find_outliers(df, "Alcohol Consumption (Litre/Person)")
income_outliers = find_outliers(df, "Disposable Income ($)")
life_outliers = find_outliers(df, "Life Expectancy (Years)")

# print out IQR outliers
print("IQR Alcohol Outliers:\n", alc_outliers)
print("IQR Income Outliers:\n", income_outliers)
print("IQR Life Expectancy Outliers:\n", life_outliers)

# Using Z-Score to identify potential outliers
def zscore_outliers(data, column, threshold=3):
    mean_value = data[column].mean()
    std_deviation = data[column].std()
    z_scores = (data[column] - mean_value) / std_deviation
    return data[(z_scores < -threshold) | (z_scores > threshold)]


alc_z_outliers = zscore_outliers(df, "Alcohol Consumption (Litre/Person)")
income_z_outliers = zscore_outliers(df, "Disposable Income ($)")
life_z_outliers = zscore_outliers(df, "Life Expectancy (Years)")

# print out z-score outliers
print("Alcohol Z-Score Outliers:\n", alc_z_outliers)
print("Income Z-Score Outliers:\n", income_z_outliers)
print("Life Expectancy Z-Score Outliers:\n", life_z_outliers)


```

## Explanation
The boxplots show the distribution of every variable and also detect outliers. For Alcohol Consumption, the plot shows some values much higher or lower than most, implying extreme variations in drinking habits between nations or individuals. For Disposable Income, the boxplot shows some very high values, implying wide income differences. All values in Life Expectancy are relatively in a common range, and there are a few outliers toward the lower side, which suggest some populations are having lifespans significantly less, which could be due to socio-economic reasons or health factors.

We also used the IQR and Z-score methods to determine outliers. The IQR method confirmed that Alcohol Consumption and Disposable Income both have outliers at the high and low ends, whereas Life Expectancy has low-end outliers. The Z-score method also yielded the same results, which show that these extreme points are substantial. This tells us where the data may have unusual or extreme values which can affect our analysis.
