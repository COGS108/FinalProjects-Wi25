```python
import pandas as pd
import numpy as np

# import seaborn
import seaborn as sns
sns.set()
sns.set_context('talk')

# import matplotlib
import matplotlib.pyplot as plt 
import matplotlib.dates as mdates
import matplotlib as mpl 

import warnings
warnings.filterwarnings('ignore')

import patsy
import statsmodels.api as sm
import scipy.stats as stats
from scipy.stats import ttest_ind, chisquare, normaltest
```

```python
# function to read in CSV files and clean data for each month

def clean_wait_times(month_csv):

    # read in the CSV file per month
    df = pd.read_csv(month_csv)

    # set date/time column from the csv file into datetime format
    df['date'] = pd.to_datetime(df['Date/Time'])

    # get the average of each date, reset the index for the dataframe, and rename columns
    month_averages = df.groupby(df['date'].dt.date)['Wait Time'].mean().reset_index()
    month_averages.columns = ['date', 'average wait time']
    
    # make the column datatime format again and then use the year attribute to only look at dates from 2024
    month_averages['date'] = pd.to_datetime(month_averages['date'])
    month_averages = month_averages[month_averages['date'].dt.year == 2024]
    
    # add back ride column
    month_averages['ride'] = 'ride'
    month_averages = month_averages[['ride', 'date', 'average wait time']]
    return month_averages
```

### Overall Disneyland Trends

```python
# function to process data for Disney (general) data

def process_disney(csv_file):
    
    # read CSV file
    df = pd.read_csv(csv_file)
    
    # ensure 'Date/Time' is in datetime format
    df['Date/Time'] = pd.to_datetime(df['Date/Time'])
    df['Date/Time'] = df['Date/Time'].dt.strftime('%m/%d/%Y')
    df['Date/Time'] = pd.to_datetime(df['Date/Time'], format='%m/%d/%Y')
    
    # filter out only dates in 2024
    df = df[df['Date/Time'] <= pd.to_datetime('12/31/2024')]

    # group by date and calculate the average wait time
    avg_wait_time = df.groupby('Date/Time')['Wait Time'].mean().reset_index()

    # extract unique ride names for each date
    ride_info = df[['Park', 'Date/Time']].drop_duplicates().reset_index(drop=True)

    # merge the two tables to retain ride names
    processed_df = ride_info.merge(avg_wait_time, on='Date/Time')

    # rename columns
    processed_df = processed_df.rename(columns = {'Date/Time': 'Date', 'Wait Time': 'Avg. Wait Time'})

    return processed_df
```

### Events and Pricing

To give some context for individual ride analyses, we'll first briefly look at how ticket prices change on different holidays and how events also change pricing. 

```python
# function to replace ticket tiers with actual prices
def tier_price(num):
    if num == 0.0:
        num = 104
    if num == 1.0:
        num = 126
    if num == 2.0:
        num = 142
    if num == 3.0:
        num = 164
    if num == 4.0:
        num = 180
    if num == 5.0:
        num = 196
    if num == 6.0:
        num = 206    
    return (num)

disney_events['Park Ticket (1 Day)'] = disney_events['Park Ticket (1 Day)'].apply(tier_price)
```

```python
holidays = (disney_events.loc[disney_events['Holiday'].notna()])
```

```python
# function to convert prices for night events
def convert_event_price(value):
    if pd.isna(value): 
        return value
    
    value = str(value).replace('$', '') 
    try:
        return float(value)  
    except ValueError:
        return None 

disney_events['Event Price'] = disney_events['Event Price'].apply(convert_event_price)
```

```python
holiday_price = sns.barplot(x='Park Ticket (1 Day)', y='Holiday', data=holidays)
plt.xlabel('Average Ticket Price ($)', fontsize='15')
plt.ylabel('Holiday', fontsize='17')
plt.title('Ticket Prices on Different Holidays', fontsize='15')
holiday_price
```

Based on the barplot above looking at park ticket prices on different holidays in 2024, we can see that three holidays (President's Day, Memorial Day, and Labor Day) had tickets that fell in tiers two and three, while all of the other holidays were tier four and above.
```python
# linear regression for average wait times (Indiana Jones) and precipitation

disney_weather['date'] = disney_weather['datetime']
disney_weather['date'] = pd.to_datetime(disney_weather['date'])

df = pd.merge(indiana_jones_averages, disney_weather, on='date', how='inner')

df = df.reset_index()
df = df.drop('index', axis=1)
df = df[['date', 'average wait time', 'precip']]
df['average'] = df['average wait time']

outcome, predictor = patsy.dmatrices('average ~ precip', df)
model_1 = sm.OLS(outcome, predictor)
results_1 = model_1.fit()
print(results_1.summary())
```

As such, the relationship between Indiana Jones' average wait times and precipition is not the strongest. Instead, it seems that precipitation is mainly a factor in impacting wait times on days with a lot of rain, instead of simply days where it was sprinkling/raining intermittently throughout the day.
