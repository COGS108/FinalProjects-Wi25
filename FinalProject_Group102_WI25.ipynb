## Background and Prior Work


Anxiety disorders are among the most prominent and common types of mental health conditions. Based on research conducted by the National Institute of Mental Health (NIH), it was estimated that 19.1% of US adults had experienced an anxiety disorder, with an estimated 31.1% of US adults experiencing any anxiety disorder at some point in their lives.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)This high prevalence of anxiety symptoms across all adults led us to research more into what factors contribute to anxiety severity and explore potential academic and lifestyle influences.

Due to the widespread prevalence of anxiety disorders, researchers have increasingly studied factors contributing to anxiety symptoms over the past decade. In 2024, a study was conducted by ten researchers that aimed to find the global trends in anxiety disorders among adolescents and young adults aged 10-24 and how these trends vary by region, age, sex, socioeconomic status, etc., while also considering the impact of risk factors like bullying victimization. They found that regardless of age (given the 10-24 age bracket), region, and socioeconomic status, there has been a 52% increase in global incidence. Not only that, but this increase has predominantly occurred within the past 10 years.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) While there are baseline differences in the number of people who are affected by anxiety attacks based on socioeconomic status and region, there is an abundant amount of evidence that shows that recent global trends have caused a sharp increase in anxiety-related disorders.

With an ample amount of research that supports the recent spike in anxiety attacks, we aim to turn our focus and figure out which aspects of people’s lives contribute to different levels of anxiety attacks. In a paper published by the National Library of Medicine, Carlos Blanco and five other researchers conducted a study looking into risk factors for anxiety disorders within various groups of people. Using data drawn from a large, nationally representative sample, the researchers applied confirmatory factor analysis to identify the latent structure of anxiety disorders. Through this, they found that in general, those who have low self-esteem, a family history of depression, years of education, and/or traumatic experiences have a high correlation to getting an attack.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Through our analysis, we hope to align our results with this paper while providing deeper insight into unconventional factors.

This dataset is a large survey of college students across multiple universities. It contains a students gender, college, year, and cgpa, and asks 10 questions per category of depression, anxiety, and stress that ask the student to rate the severity out of 5. The dataset is mostly integers with the only string data being the qualities of the students. In order to preprocess the data, we will check null data and fill the null data for questions with the nearest average integer and remove any missing student data. Since the survey is bounded from 0-4 (0-Never, 1-Almost Never, 2-Sometimes, 3-Fairly Often, 4-Very Often), there should not be any outliers. The final data-preprocessing we will do is one-hot-encoding for gender, age, academic year, and other string data except college. 

The main part of the dataset uses established stress, anxiety, and depression surveys to establish levels of each category. These surveys are the Perceived Stress Scale (PSS) with 10 questions, the General Anxiety Disorder-7 (GAD) with 7 questions, and the Patient Health Questionnaire (PHQ) with 9 questions. All the questions from each survey were directly used with one tweak; all questions were put in the scope of school and academics. For example, the question "In the last month, how often have you been upset because of something that happened unexpectedly?" from the PSS became "In a semester, how often have you felt upset due to something that happened in your academic affairs?".

### Exploration of Anxiety, Stress, and Depression Surveys


```python
# helper function to plot histograms in a tight layout
def plot_histograms_wrapped(df, questions, figsize=(5, 4)):
    """Plots histograms in a wrapped layout."""
    num_questions = len(questions)
    num_rows = math.ceil(num_questions / 3)

    fig, axes = plt.subplots(num_rows, 3, figsize=(figsize[0] * 3, figsize[1] * num_rows))

    for i, question in enumerate(questions):
        r = i // 3
        c = i % 3

        if num_rows > 1:
          ax = axes[r, c] 
        else:
          ax = axes[c]

        sns.histplot(df[question], ax=ax, bins=range(int(df[question].min()), int(df[question].max()) + 2))
        ax.set_title(f"Histogram of {question}")
        ax.set_xlabel("Answer Value")
        ax.set_ylabel("Frequency")
        ax.set_xticks(range(int(df[question].min()), int(df[question].max()) + 1))

    if num_questions < num_rows * 3:
        for i in range(num_questions, num_rows * 3):
            r = i // 3
            c = i % 3
            if num_rows > 1:
              fig.delaxes(axes[r, c])
            else:
              fig.delaxes(axes[c])

    plt.tight_layout()
    plt.show()
```

```python
# Histograms of GAD questions
gad_questions = ['GAD1', 'GAD2', 'GAD3', 'GAD4', 'GAD5', 'GAD6', 'GAD7']

print("\nGAD Questions and Histograms")
plot_histograms_wrapped(df, gad_questions, figsize=(5, 4))
```

The set of plots above show the number of people that gave a specific rating per each of the seven questions in the GAD survey. Its not surprising that 0 has the lowest frequency in all 7 questions since "0" means they have never felt a afraid, nervous, anxious, etc in the past semester. It is surprising that the rest of the answers seem to be fairly uniform across all the questions. This means that the average person in this study tends to be anxious in one aspect or another. 

```python
# Histograms of PHQ questions
pss_questions = ['PSS1', 'PSS2', 'PSS3', 'PSS4', 'PSS5', 'PSS6', 'PSS7', 'PSS8', 'PSS9', 'PSS10']

print("\nPSS Questions and Histograms")
plot_histograms_wrapped(df, pss_questions, figsize=(5, 4))
```

The set of plots above show the number of people that gave a specific rating per each of the ten questions in the PSS survey. Most plots follow a standard bell curve shape with the majority of participants answering "2" or sometimes to questions like "how often have you been upset" or "how often have you been angered". However, one outlier is question #3: "In a semester, how often you felt nervous and stressed because of academic pressure?". This question had nearly equal peaks with students answering "2" and "4" meaning there are a large percent of students who have felt nervous and stressed due to academic pressure very often. 

```python
# Histograms of PHQ questions
phq_questions = ['PHQ1', 'PHQ2', 'PHQ3', 'PHQ4', 'PHQ5', 'PHQ6', 'PHQ7', 'PHQ8', 'PHQ9']

print("\nPHQ Questions and Histograms")
plot_histograms_wrapped(df, phq_questions, figsize=(5, 4))
```

The set of plots above show the number of people that gave a specific rating per each of the nine questions in the PHQ survey. Some plots follow a traditional bell curve but others share the same pattern as the plots from the GAD. Two questions of note are questions #4 and #9. In question #4, "In a semester, how often have you been feeling tired or having little energy?", students predominantly answered nearly every day. This question could have a correlation with question #3 from the PSS since pressures in academics could lead to feelings of tiredness. 

```python
# Finding the correlation between PSS3 and PHQ4
filtered_df = df[(df['PSS3'] == 4) & (df['PHQ4'] == 3)]
print(f"Number of people with PSS3=4 and PHQ4=3: {len(filtered_df)}")
print(f"Percent of people that have PSS3=4 and PHQ4=3: {round(len(filtered_df) / len(df) * 100, 2)}%")
print(f"Average Anxiety Value for these rows: {round(filtered_df['Anxiety Value'].mean(), 2)}")
print(f"Correlation between Stress and Anxiety values for these rows: {round(filtered_df['Stress Value'].corr(filtered_df['Anxiety Value']), 2)}")
```

While not very significant, there is some correlation between the people that answered "4" for PSS3 and "3" PHQ4. It is further supported by the fact that this subset of people make up 21.12% of the data which means the questions, at the very least, overlap with each other and trigger similar feelings across participants. 

```python
# Prep Data for Correlation Heatmap
columns_to_remove = ['University', 'Department', 'Stress Label', 'Anxiety Label', 'Depression Label', 'Gender'] # Remove non-numeric columns
filtered_df = df.drop(columns=columns_to_remove)
correlation_matrix = filtered_df.corr()

# Plot Correlation Heatmap
plt.figure(figsize=(25, 20)) 
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Heatmap of Features')
plt.show()
```

```python
# Plot Zoomed-In Correlation Heatmap
correlation_matrix = filtered_df.iloc[:,:37].corr()
plt.figure(figsize=(25, 20)) 
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Heatmap of Features (Zoomed-In)')
plt.show()
```

Here is the correlation map between all numeric features in the dataset. It can be seen that answers from many of the questions across all the surveys to have some level of correlation between each other. By looking at the value of anxiety, stress, or depression, it is also recognizable which questions more heavily dictate the value. An interesting thing to note is that PSS5-8 have a low correlation between all features except between themselves. While it is not obvious through the questions, it could point to overlap in evoked feelings caused by questions 5 to 8. 

```python
# Normalize anxiety, stress, and depression values
df['Normalized Stress Value'] = df['Stress Value'] / 40
df['Normalized Anxiety Value'] = df['Anxiety Value'] / 21
df['Normalized Depression Value'] = df['Depression Value'] / 27

# Melt the DataFrame to prepare for seaborn's barplot
melted_df = pd.melt(df, id_vars=['Gender'],
                    value_vars=['Normalized Stress Value', 'Normalized Anxiety Value', 'Normalized Depression Value'],
                    var_name='Value Type', value_name='Normalized Value')

# Create the grouped bar plot
plt.figure(figsize=(12, 6))
sns.barplot(x='Gender', y='Normalized Value', hue='Value Type', data=melted_df)
plt.title('Comparison of Normalized Anxiety, Stress, and Depression Values by Gender')
plt.xlabel('Gender')
plt.ylabel('Normalized Value')
plt.legend(title='Value Type')
plt.show()
```

An common thing to look at when dealing with anxiety, stress or depression are the differences between gender. Surprisingly, results seem to be fairly similar between those who answered male or female with normalized values of all survey hovering around 0.6. Those who answered "prefer not to say" seem to have a higher anxiety, stress, and depression value overall.

```python
# Get CGPA columns 
cgpa_columns = ['Current_CGPA_3.80 - 4.00', 'Current_CGPA_3.40 - 3.79', 'Current_CGPA_3.00 - 3.39', 'Current_CGPA_2.50 - 2.99', 'Current_CGPA_Below 2.50', 'Current_CGPA_Other']

for col in cgpa_columns:
    df.loc[df[col] == 1, 'CGPA'] = col.replace('Current_CGPA_', '')

# Specify the order of the CGPA categories
cgpa_order = ['3.80 - 4.00', '3.40 - 3.79', '3.00 - 3.39', '2.50 - 2.99', 'Below 2.50', 'Other']

# Convert 'CGPA' to a categorical type with the specified order
df['CGPA'] = pd.Categorical(df['CGPA'], categories=cgpa_order, ordered=True)

melted_df = pd.melt(df, id_vars=['CGPA'], value_vars=['Normalized Stress Value', 'Normalized Anxiety Value', 'Normalized Depression Value'],
                    var_name='Value Type', value_name='Normalized Value')

# Create Bar Plot
plt.figure(figsize=(14, 9))
ax = sns.barplot(x='CGPA', y='Normalized Value', hue='Value Type', data=melted_df)
plt.title('Comparison of Normalized Anxiety, Stress, and Depression Levels by CGPA (Ordered)')
plt.xlabel('CGPA')
plt.ylabel('Normalized Value')
plt.legend(title='Value Type')
plt.xticks(rotation=45, ha='right')
yticks = np.arange(0, 0.66, 0.05)
ax.set_yticks(yticks)
plt.tight_layout()
plt.show()

df.drop('CGPA', axis=1, inplace=True)
```

While there does not seem to be any increase or decrease in anxiety or depression levels as CGPA decreases, there is a slight increase in stress level as CGPA decreases. This may be caused by additional pressures of achieving a suitable CGPA to graduate or difficutly of classes. 

```python
# Melt the DataFrame to prepare for seaborn's barplot
melted_df = pd.melt(df, id_vars=['Department'], # Assuming 'Department' represents majors
value_vars=['Normalized Stress Value', 'Normalized Anxiety Value', 'Normalized Depression Value'],
var_name='Value Type', value_name='Normalized Value')

# Create the grouped bar plot
plt.figure(figsize=(14, 8))
ax = sns.barplot(x='Department', y='Normalized Value', hue='Value Type', data=melted_df)
plt.title('Comparison of Normalized Anxiety, Stress, and Depression Levels by Major')
plt.xlabel('Major')
plt.ylabel('Normalized Value')
plt.legend(title='Value Type')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

As we can see, there are definitely distinct associations with each individual major and their associated stress, anxiety, and depression values. All three values vary from major to major, as well as which value is the highest and lowest. This could be correlated with the type of people who tend to choose to go into that major, or it could be the result of a student pursuing that major, such as the courseload or topics that are studied.
