# Abstract

The undergraduate engineering students at UC San Diego had been subjected to severe mental health impacts, as well as impacts on their academic performance in the COVID-19 pandemic break. Our project is based on how student mental health trends have shifted between Fall 2015–Spring 2018 (pre-COVID) and Fall 2021–Spring 2024 (post-COVID) and correlated with students' cumulative GPA. We compared GPA data with self-reported mental health questionnaires and tried to determine if there is a link between student's academic performance and their psychological distress.

To track patterns in mental health service usage like reported anxiety, depression levels, and overall academic trends, our approach involved cleaning and merging datasets to track patterns in mental health service usage. The average GPA has remained stable or slightly increased post-COVID and students have reported higher levels of burnout, stress, and mental exhaustion. These results raised an important question—are students doing well after the COVID-19 pandemic's impact, or are they just pushing through despite the odds of worsening mental health caused by or related to the COVID-19 pandemic?

Our results suggest that a student's academic GPA alone isn’t a reliable indicator of a student's mental well-being. The increasing rate of anxiety and psychological distress compels a call for firmer institutional support, though whether the average student's academic work is assumed to remain unaffected. Our results confirm the necessity for mental assistance for students at UC San Diego in undergraduate engineering majors to succeed academically and for their mental well-being.


# Research Question

How has mental health among UCSD undergraduate engineering majors changed between Fall 2015-Spring 2018 (Pre-Covid) and Fall 2021-Spring 2024 (Post-Covid), and how does this correlate with students' cumulative GPA for each school year?

## Background and Prior Work

Impact of COVID-19 on UCSD Students: 

COVID-19 Impact on UCSD Students UCSD's Department of Public Health conducted research that assessed the COVID-19 lockdown's effect on self-reported quality of life, mental health, and physical health among students. Distributed between April and May 2020, The questionnaire reported that students' quality of life during the pandemic declined significantly. The mean quality of life score decreased from 3.75 before the pandemic to 2.75 during lockdown, indicating substantial impacts on physical and mental health.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) 
UCSD's Tritonlytics group also developed the COVID-19 Impact Survey to observe how the crisis affected their community. The survey was intended to gather information to guide connection, communication, productivity, and well-being among employees and students during the pandemic.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2)

Link Between Mental Health and Academic Performance:

Association Between Academic Achievement and Mental Well-being Research has uniformly established a relationship between academic achievement and mental health. A study conducted on college students with lower socioeconomic status revealed anxiety, depression, and internet addiction to be significantly inversely correlated with academic achievement, whereas self-esteem and social support were significantly correlated with GPA. Social support was the mediating variable and implies that the reinforcement of support systems could result in an increase in mental health as well as academic performance.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3)
Another study identified that depression and anxiety both had adverse effects on GPA and credit hours earned per semester, with depression having more significant effects. The study also showed that pass/fail grade policies implemented in the COVID-19 pandemic had adverse effects on future GPA, especially among students who experienced anxiety and depression.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4)
Furthermore, a longitudinal study explored the relationship between GPA and burnout and concluded that greater GPA is accompanied by greater burnout. The study established that 'grit-passion'—passion for academic work among students—can moderate this relationship, suggesting that developing passion can act as a buffer for high-achieving students from burnout.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5)
The findings in the study emphasize the importance of attending to mental health to keep college students well and performing well academically. In summary, the COVID-19 pandemic has greatly affected the quality of life of UCSD students, and has further proven a strong link between mental health and academic achievement.<a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6)

# Hypothesis


Building on existing research, undergraduate students' mental health declined significantly due to the COVID-19 quarantine, and these negative effects may persist due to prolonged academic disruptions, social isolation, and increased stress, impacting coping mechanisms and overall well-being. We hypothesize that UCSD undergraduate engineering majors will report worse mental health in Fall 2021–Spring 2024 (post-COVID) compared to Fall 2015–Spring 2018 (pre-COVID), and that poorer mental health will correlate with lower yearly GPAs. To test this, we will use a questionnaire to assess self-reported mental health across UCSD undergraduates and other findings, analyzing trends based on students’ field of study and academic performance.

```python
GPA_avg_2015_2018 = pd.read_csv('https://raw.githubusercontent.com/kbali1/COGS108CSV/5647e6e290e4c681b4187f9da0e6674bac17ef52/ucsd_mean_gpa.csv')
fall2017_spring2018 = GPA_avg_2015_2018.loc[0:2, ['Female', 'Male', 'Entire Population']].mean()
fall2016_spring2017 = GPA_avg_2015_2018.loc[3:5, ['Female', 'Male', 'Entire Population']].mean()
fall2015_spring2016 = GPA_avg_2015_2018.loc[6:8, ['Female', 'Male', 'Entire Population']].mean()
data = {
    'Year': ['2015-16', '2016-17', '2017-18'],
    'Average GPA at Graduation': [fall2015_spring2016['Entire Population'].round(decimals=1), 
                                  fall2016_spring2017['Entire Population'].round(decimals=1), 
                                  fall2017_spring2018['Entire Population'].round(decimals=1)]
}
GPA_avg_2015_2018 = pd.DataFrame(data)
GPA_avg_2015_2018
```

```python
#Methods to standardize the upcoming datasets
def dropPercent(string):
    string = string.strip()
    string = string.replace("%","")
    if string.isnumeric():
        output = float(string)
    else:
        output = string
    return output

def IntYear(string):
    return (int(string))
```

## Mental Health of UCSD Undergraduates for Year 2016

```python
mental_health2016 = pd.read_csv("https://raw.githubusercontent.com/MonicaMejia3/DatsetTrial_1/main/2016Datacheck1.csv")
#Flip the columns and rows so that we can get 1 row of percentages and columns of category names
mental_health2016.rename(columns={'Category':'Index','Percentage (%)': 0}, inplace = True)
mental_health2016.set_index('Index', inplace=True)
mental_health2016 = mental_health2016.transpose()
#Add year of when data was collected
mental_health2016.insert(0, "Year", '2016')
#Drop the percentage sign for the column names
mental_health2016.rename(columns={'Visited Psychiatry Office (%)': 'Visited Psychiatry Office', 
                                  'Had Counseling Sessions (%)':'Had Counseling Sessions',
                                  'Reported High Stress (%)':'Reported High Stress',
                                  'Diagnosed with Anxiety (%)':'Diagnosed with Anxiety',
                                  'Diagnosed with Depression (%)':'Diagnosed with Depression',
                                  'Seriously Considered Suicide (%)':'Seriously Considered Suicide',
                                  'Attempted Suicide (%)':'Attempted Suicide',
                                  'Experienced Severe Psychological Distress (%)':'Experienced Severe Psychological Distress'}, inplace=True)
#Change year type to int
mental_health2016['Year']=mental_health2016['Year'].apply(IntYear)
mental_health2016
```

```python
#Combine Average GPA of UCSD Undergraduates to include years of data we have
GPA = pd.DataFrame({
    'Year': ['2015', '2016','2017','2021','2022','2023'],
    'Percentage' : [
        GPA_avg_2015_2018['Average GPA at Graduation'].values[0],
        GPA_avg_2015_2018['Average GPA at Graduation'].values[1],
        GPA_avg_2015_2018['Average GPA at Graduation'].values[2],
        GPA_avg_2021_2024['Average GPA at Graduation'].values[0],
        GPA_avg_2021_2024['Average GPA at Graduation'].values[1],
        GPA_avg_2021_2024['Average GPA at Graduation'].values[2]
        ]
})

#Merge Average GPA and percentage of students who have visited a psychiatry office
GPA_Psychiatry = pd.merge(visit_psychiatry_office,GPA, on='Year', how='left')
GPA_Psychiatry = GPA_Psychiatry.dropna()
GPA_Psychiatry = GPA_Psychiatry.reset_index(drop=True)
GPA_Psychiatry
```

```python
fig, axes = plt.subplots(1,2, figsize=(12,5))
sns.lineplot(x='Year', y='Percentage_y', data=GPA_Psychiatry, color='red', ax=axes[0])
axes[0].set_title('Average GPA')
axes[0].set_xlabel('School Year')
axes[0].set_ylabel('Average GPA')
axes[0].set_yticks([3.0, 3.1, 3.2, 3.3, 3.4, 3.5, 3.6, 3.7])
sns.lineplot(x='Year', y='Percentage_x', data=GPA_Psychiatry, color='blue', ax=axes[1])
axes[1].set_title('Visited Psychiatry Office')
axes[1].set_xlabel('School Year')
axes[1].set_ylabel('Percentage')
axes[1].invert_yaxis()
plt.show()
```

Due to the limits of our dataset, we only have years 2016, 2021, and 2023 to plot and compare. However, we can see that while average GPA increased after COVID, so did the percentage of students who visited the psychiatry office.

# Ethics & Privacy

For our project, we recognize there are several ethical and privacy concerns. Since we are working with data on sensitive topics, privacy is a top priority, and we will do our best to ensure our study is both responsible and ethical. One challenge we have considered is the potential limitations in our dataset’s representation. Since these surveys rely on voluntary participation, there is a risk of self-selection bias. As students who experience more severe mental health issues may respond less, therefore making them underrepresented.

As previously mentioned, privacy is a major concern. We will make sure that no personally identifiable information is included in our dataset. Moreover, when accessing data that requires institutional approval, like UCSD CAPS reports, we will strictly adhere to the data usage agreements and university policies. To ensure we detect and work around biases in our analysis we will check the response rates across different groups to detect potential unequal participation; and acknowledge the limitations of self-reported data to ensure we don't generalize conclusions beyond the available data. Since the findings in our project could have real-world affects for students well-being policies at UCSD. We will make sure that our analysis does not stigmatize mental health conditions and avoid misleading interpretations; and make sure to frame our results in a responsible way.

The analysis may be helpful to learn more about UCSD students' mental health and initiatives to take while complying with ethical and privacy concerns.

# Discussion and Conclusion


Research on the mental health of UCSD undergraduate students between Fall 2015–Spring 2018 (pre-COVID) and Fall 2021–Spring 2024 (post-COVID) has revealed significant negative changes over time, with direct correlations to academic performance as reflected in cumulative GPA. Prior studies from UCSD’s Department of Public Health and Tritonlytics have demonstrated a significant decline in students’ quality of life and mental well-being during the pandemic, leading to increased stress, anxiety, and depression. These psychological stressors did not disappear but rather progressed after the pandemic, continuing to impact students’ academic motivation and performance in new ways.

By incorporating mental health datasets from 2016 to 2023 into a single comprehensive data frame our research indicated a clear correlation between mental health and GPA. This analysis includes students who visited psychiatry offices, attended counseling sessions, reported high stress levels, were diagnosed with anxiety and depression, or experienced severe psychological distress between 2016 and 2023. The data illustrates that mental health struggles intensified due to the isolation and disruptions caused by COVID-19, disproportionately affecting undergraduate students, particularly those in engineering majors.

Furthermore, an analysis of GPA trends alongside mental health data highlights that engineering students, who already face rigorous coursework, have experienced shifts in academic performance due to psychological distress. The finding that higher GPAs correlate with greater burnout establish the need for supportive academic practices that promote both excellence and well-being. Programs focused on resilience, self-efficacy, and community support may help reduce some of the pressures that contribute to mental health challenges among students.

Our research builds upon prior work by emphasizing how the pandemic has altered the landscape of academic success and mental health at UCSD. While previous studies examined the immediate impacts of COVID-19, this study expands upon those findings by analyzing long-term changes and their correlation with GPA. The findings reinforce the need for colleges to prioritize mental health services, implement flexible academic policies, and introduce interventions that address stress in demanding fields like engineering.

Ultimately, the decline in mental health among UCSD undergraduate engineering students and its impact on academic performance suggest that UCSD must continue adapting its policies and resources to support student well-being. Mental health and academic success are connected, and promoting an environment that prioritizes students’ well-being will contribute to a healthier and more productive society. Future research could explore how specific mental health interventions, such as mentorship programs or counseling services, influence GPA outcomes, ensuring that students not only succeed academically but also thrive personally beyond school.
