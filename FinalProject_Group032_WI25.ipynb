### Inferential Analysis with OLS regression

Linear Regression - Scatterplots

```python
#importing neccessary libraries and packages for OLS regression and scatterplots. 
import pandas as pd
import numpy as np

# Import seaborn and apply its plotting styles
import seaborn as sns
sns.set_theme(style="white", 
              font_scale=2, 
              rc={'axes.spines.right': False,'axes.spines.top': False}) # this last thing is equivalent to always having sns.despine()

# import matplotlib
import matplotlib as mpl
import matplotlib.pyplot as plt
# set plotting size parameter
plt.rcParams['figure.figsize'] = (8, 8)

# Statmodels & patsy
import patsy
import statsmodels.api as sm
import statsmodels.formula.api as smf
from scipy.stats import pearsonr, ttest_ind
```

```python
#merging the two datasets for OLS. 
df = pd.concat([housing_update_filtered, calfire_subset], axis=1)
df.head()
```

```python
df = df.rename(columns={'Estimated_Financial_Loss (Million $)': 'Estimated_Financial_Loss'})
#converting values in median listing price column into millions.
df['median_listing_price'] = (df['median_listing_price']/1000000)
df.head()
```

```python
# check linearity and homo/heteroscasticity
plt.figure(figsize = (1, 2))
sns.lmplot(x="Estimated_Financial_Loss", y="median_listing_price", data=df, height = 6, aspect = 1.2)
plt.show()
```

It seems that there is a potential Homoscedasticity issue as the spread of points widens at higher values of Estimated_Financial_Loss. The points are widely scattered around the regression line which shows that there is a lot of noise in the data. Therefore, this plot might indicate that this model is weak fit. Specifically this graph shows a weak positive relationship between the two variables. Below is a OLS regression model that will provide us with a better understanding of the relationship between these two variables.

```python
#Creating model for OLS regression. 
model = smf.ols(formula='median_listing_price ~ Estimated_Financial_Loss', data=df)
#fitting the model
results = model.fit()
## look at the results
print(results.summary())
```

The r squared value for the model, 0.029, indicating a very weak positive relationship between the two variables. The models indicates that for every 1 unit increase in Estimated_Financial_Loss the median listing price increases by $0.00001012 on average. The p value is 0.160 and so too high for results to be statistically significant. Strong multicollinearity issues are indicated, but true multicollinearity can't occur when there is only one independant variable. This is likely due to presence of outliers or due to difference in the range of values of the two variables. Below is scatterplot with a regression line that visually depicts this. 

```python
sns.scatterplot(x='Estimated_Financial_Loss', y='median_listing_price', data=df, s=100)

# Generate and plot the model fit line
xs = np.arange(df['Estimated_Financial_Loss'].min(), df['Estimated_Financial_Loss'].max())
ys = results.params.iloc[1] * xs + results.params.iloc[0]
plt.plot(xs, ys, '--k', linewidth=4, label='Model')

# make the title tell us model parameters
vals='$y={:.3f} + {:.3f}x, p={:.3f}, r={:.3f}, r^2={:.3f}$'.format(results.params.iloc[0], 
                                                                   results.params.iloc[1], 
                                                                   results.pvalues.iloc[1], 
                                                                   df[['Estimated_Financial_Loss','median_listing_price']].corr().iloc[0,1], 
                                                                   results.rsquared)
plt.title(vals)

plt.xlabel('Estimated_Financial_Loss')
plt.ylabel('median_listing_price')
plt.legend()
```

Now that we've done that we can add the housing_update_filtered dataset as well so that each row has each value with a one to one matching.

```python
df_combined['Date'] = pd.to_datetime(df_combined['Date'])
housing_update_filtered['Date'] = pd.to_datetime(housing_update_filtered['Date'], format='%Y%m')

# 🔹 Merge datasets
df = pd.merge(df_combined, housing_update_filtered[['Date', 'median_listing_price']], on='Date', how='left')

# 🔹 Handle missing values
df.fillna(0, inplace=True)  # Replace missing values with 0
```

Now we can dive straight into our linear regression model, we're not gonna look at all the factors as we believe the most important ones would be area burned, homes destroyed and the estimated financial loss.

```python
# 🔹 Define Multiple Linear Regression Model
mod = smf.ols(formula='median_listing_price ~ Q("Estimated_Financial_Loss (Million $)") + Q("Area_Burned (Acres)") + Q("Homes_Destroyed")', data=df)
res = mod.fit()
# 🔹 Print Model Summary
print(res.summary())
```

While our p-values for Estimated Financial Loss and Homes Destroyed on Median Listing Price (dependent variable) showcase lack of statistical significance, we can see that the p-value for Area_Burned in acres is statiscally significant. Let's look at this further. 

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
sns.regplot(x=df["Estimated_Financial_Loss (Million $)"], y=df["median_listing_price"], ax=axes[0])
axes[0].set_title("Median Listing Price vs. Financial Loss")

# Scatter plots with regression Lines
sns.regplot(x=df["Area_Burned (Acres)"], y=df["median_listing_price"], ax=axes[1])
axes[1].set_title("Median Listing Price vs. Area Burned")

sns.regplot(x=df["Homes_Destroyed"], y=df["median_listing_price"], ax=axes[2])
axes[2].set_title("Median Listing Price vs. Homes Destroyed")

#Show plots
plt.tight_layout()
plt.show()
```

After doing regression analysis on different variables that we thought may also contribute an impact to estimated financial loss throughout our observed years. We came to the conclusion earlier that due to the low R-squared value of Median Listing Price (dependent) and Estimated Financial Loss (independent), variation is not explained by looking at linear regression. But looking at the graph of Median Listing Price and Area Burned (acres) with its statistically significant p-value, as well as looking at the scatterplot and linear regression line (steeper slope compared to the other two variables we had measured above, and a bit more even scatter), there might be some correlation here. 
######
The results summary shows that the r squared value for this relationship is 0.074 which means the independant variable (area burned(Acres)) only explains 0.074 of the variation in the dependant variable (median_listing_price). The low p-value means the dependant variable is a significant predictor of the independant variable. The coefficient of the median listing price says that for every 1 acre burned the median listing price increases by 1.6036 units. We can see that there is minor, but weak correlation between acres burned and median housing price, and due to the low R-squared value, the effect of the independent on the dependent variable is to insignificant to be notable.
######
Lets do further analyses to prove this weak correlation we have discovered as the R-squared for the OLS regression we did is based on all three variables on the dependent variable we are measuring.

```python
#OLS regression of Area Burned on Median Listing Price
model2 = smf.ols(formula='median_listing_price ~ Q("Area_Burned (Acres)")', data=df)
#fitting the model
results2 = model2.fit()
## look at the results
print(results2.summary())
```

In conclusion, as we can see the R-squared value is extremely low and indicates that the amount of variation of the dependent variable (Median Listing Price) described by the independent variable (Area Burned) is deemed insignificant to be notable.
