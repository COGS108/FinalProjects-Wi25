### Prior Work
The gender wage gap remains a persistent issue worldwide, reflecting disparities in earnings between men and women across various industries and regions. Organizations and institutions continue to analyze this gap, shedding light on its extent and the factors contributing to wage inequality. Works that have been done before include reports from the OECD, California's Civil Rights Department, and UC Berkeley provide valuable insights into these disparities, particularly in California, where legislative efforts have aimed to narrow the gap.

According to the OECD's 2022 report, the gender wage gap is measured as the difference in median earnings between men and women relative to men's earnings. Their data visualization highlights wage disparities across different countries, emphasizing the ongoing global challenge of pay inequality.

A more localized analysis by California's Civil Rights Department, using 2021 data, found that women in California earn approximately 81 cents for every dollar earned by men. The gap widens in certain industries, with women earning 71 cents on the dollar in finance, 76 cents in construction, and 77 cents in trade. Additionally, racial disparities further exacerbate wage inequalities, as Latina women earn just 44 cents and Black women 58 cents for every dollar earned by white men.
A study conducted by UC Berkeley, analyzing U.S. Census American Community Survey (ACS) data from 2014–2018, found that California has one of the smallest gender pay gaps in the U.S., outperforming the national average with a 13% gap. The report credits California’s progressive legislative measures, such as the California Fair Pay Act of 2015 (SB 358), which broadened the definition of "equal work" to "substantially similar work" and mandated equal pay across different establishments. Subsequent amendments further expanded protections to cover race and ethnicity-based pay disparities, banned salary history inquiries, and extended pay equity laws to both public and private employees.

## Gender Pay Gap Dataset: Panel Study Income Dynamics

##### df is the data frame that only has data from the years 2000-2011, and only including people aged 25-40.

##### df_tech is the data frame that only has data from the years 2000-2011, and only including people aged 25-40 that have occupations in the techinical field.


```python
import pandas as pd
import seaborn as sns
import numpy as np
# import matplotlib
import matplotlib.pyplot as plt # Typical way of import MPL
import matplotlib as mpl # This line is used less frequently
import zipfile
zf = zipfile.ZipFile('PanelStudyIncomeDynamics.csv.zip')
df = pd.read_csv(zf.open('PanelStudyIncomeDynamics.csv'))
df = df[['wave','sex','region','age','annlabinc','yrsexp','occ','degree']]
df = df.rename(columns = {'wave':'year','annlabinc': 'annual income','yrsexp': 'experience','occ': 'occupation code'})
df = df.dropna()
df['sex'] = df['sex'].replace({1: 'Male', 2: 'Female'})
df['sex_male'] = (df['sex'] == 'Male').astype(int)
df['sex_female'] = (df['sex'] == 'Female').astype(int)
df['degree'] = (df['degree']).astype(int)
df['degree_new'] = df['degree'].replace({0: 'NoCollege', 1: 'Bach, no adv', 2: 'Adv'})
columns = list(df.columns)
new_order = columns[:columns.index('sex') + 1] + ['sex_male', 'sex_female'] + [col for col in columns if col not in ['sex', 'sex_male', 'sex_female']]
df = df[new_order]
df = df.loc[:, ~df.columns.duplicated()]
df = df[df['year'] >= 2000]
#only ages 25 to 40
df = df[df['age'] <= 40]
#change column order
tech_occ_codes = [11, 71, 100, 101, 102, 104, 106, 110, 111, 140, 141, 143, 172, 174, 263, 284, 341, 351, 580, 581, 582, 583, 790, 896, 890]
df_tech = df.copy()
df_tech = df_tech[df_tech['occupation code'].isin(tech_occ_codes)]
occupation_to_job = {
    580: 'Computer Operators',
    581: 'Data Entry Keyers',
    582: 'Word Processors and Typists',
    583: 'Desktop Publishers',
    790: 'Computer Control Programmers and Operators',
    896: 'Semiconductor Processors',
    890: 'Cooling and Freezing Equipment Operators and Tenders',
    351: 'Medical Records and Health Information Technicians',
    341: 'Health Diagnosing and Treating Practitioner Support Technicians',
    284: 'Technical Writers',
    263: 'Designers',
    172: 'Chemists and Materials Scientists',
    174: 'Environmental Scientists and Geoscientists',
    11: 'Computer and Information Systems Managers',
    71: 'Management Analysts',
    100: 'Computer Scientists and Systems Analysts',
    101: 'Computer Programmers',
    102: 'Computer Software Engineers',
    104: 'Computer Support Specialists',
    106: 'Database Administrators',
    110: 'Network and Computer Systems Administrators',
    111: 'Network Systems and Data Communication Analysts',
    140: 'Computer Hardware Engineers',
    141: 'Electrical and Electronics Engineers',
    143: 'Industrial Engineers, Including Health and Safety'
}
df_tech['job title'] = df_tech['occupation code'].replace(occupation_to_job)
print(df_tech.shape)
df_tech.head()
```

### Set up for EDA

The following was aided by chatGPT, we wanted to make the numbers more readable so instead of 1 1e6, we will say 1 mill
or instead of 0.1 1e6 we will say 100k

```python
#set up for EDA
#The followign was from chatGPT, we wanted to make the numbers more readable so instead of 1 1e6, we will say 1 mill

import matplotlib.ticker as ticker
def custom_formatter(x, pos):
    if x >= 1_000_000:
        return '{:.1f}mill'.format(x * 1e-6)  # Format numbers in millions
    elif x >= 1_000:
        return '{:.1f}k'.format(x * 1e-3)  # Format numbers in thousands
    else:
        return '{:.0f}'.format(x)  # For values less than 1,000, display as is
color_palette = {'Male': 'blue', 'Female': 'pink'}
```

### Boxplot for Income Distribution by Degree Level

```python
from matplotlib.ticker import MaxNLocator
plt.figure(figsize=(12.5, 5))
degree_order = ['NoCollege', 'Bach, no adv', 'Adv']
sns.boxplot(x='annual income', y='degree_new', data=df_tech,linewidth=2, order = degree_order)
plt.gca().xaxis.set_major_locator(MaxNLocator(nbins=25))  # Increase number of bins (ticks). From chatGPT, we asked how to make more bins so it is easily read
plt.gca().xaxis.set_major_formatter(ticker.FuncFormatter(custom_formatter))
plt.yticks(fontsize=14)
plt.ylabel("Degree Type",fontsize=15)
plt.xlabel("Annual Income",fontsize=15)
plt.show()
```

### Testing
#### No College Degree t-test

```python
from scipy.stats import ttest_ind

bach = df_tech[df_tech['degree_new'] == 'Bach, no adv']
Adv = df_tech[df_tech['degree_new'] == 'Adv']
NoCollege = df_tech[df_tech['degree_new'] == 'NoCollege']

male_income = NoCollege[NoCollege['sex'] == 'Male']['annual income']
female_income = NoCollege[NoCollege['sex'] == 'Female']['annual income']
t_stat, p_value = ttest_ind(male_income, female_income, equal_var=False)
t_stat, p_value
```

- t-statistic = 7.5377
- p-value = 1.03e-12 (which is very close to 0)
- Interpretation: Since p ≈ 0 is much smaller than 0.05, the income difference between males and females is statistically significant. The large positive t-statistic suggests that males earn significantly more than females among those without a college degree.


#### Bachelor Degree t-test

```python
# Bachelor degree t-test
male_income = bach[bach['sex'] == 'Male']['annual income']
female_income = bach[bach['sex'] == 'Female']['annual income']
t_stat, p_value = ttest_ind(male_income, female_income, equal_var=False)
t_stat, p_value
```

- t-statistic = 1.8093
- p-value = 0.0725
- Interpretation: Since p = 0.0725 is greater than 0.05, the difference in income between males and females is not statistically significant at the 5% level. This suggests that we do not have strong evidence to conclude that male and female income differs significantly for bachelor’s degree holders.


#### Advanced Degree t-test

```python
male_income = Adv[Adv['sex'] == 'Male']['annual income']
female_income = Adv[Adv['sex'] == 'Female']['annual income']
t_stat, p_value = ttest_ind(male_income, female_income, equal_var=False)
t_stat, p_value
```

- t-statistic = -1.4057
- p-value = 0.1705
- Interpretation: Since p = 0.1705 is greater than 0.05, the income difference between males and females is not statistically significant. The negative t-statistic indicates that, on average, females may earn slightly more than males, but this difference is not strong enough to be considered statistically significant.
