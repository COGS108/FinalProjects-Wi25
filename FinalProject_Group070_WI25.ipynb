# Abstract

This study investigates the combined effects of AI-driven workload automation and remote work conditions on layoff risks across various industries. The motivation stems from the increasing integration of AI in workplaces, which has significantly enhanced efficiency but raised concerns about job displacement, and the widespread adoption of remote work, which presents challenges such as stress and mental health issues. Both trends have reshaped the workforce landscape, making it essential to understand their combined impact on layoffs.

To address this, we utilized three datasets: one examining AI impacts on jobs, another assessing the effects of remote work on mental health, and a layoff dataset analyzing mass dismissals. These datasets were standardized and merged based on industry attributes, with data cleaning and transformation ensuring consistency. Statistical and analytical methods were applied to explore correlations between AI workload ratios, remote work factors, and layoff metrics.

Preliminary findings indicate that industries with higher AI automation experienced elevated layoff risks, particularly those reliant on repetitive tasks. Remote work stressors, such as mental health challenges and reduced productivity, were also correlated with increased layoffs. However, industries prioritizing work-life balance and mental health support demonstrated resilience, with lower layoff risks even under high AI integration.

Our results highlight the importance of balancing technological adoption with employee well-being initiatives. This study underscores the need for proactive measures, such as upskilling and mental health programs, to mitigate the adverse effects of automation and remote work on job security.

## **Background and Prior Work**


The integration of artificial intelligence (AI) into the workplace has significantly transformed operational dynamics across industries. AI's capability to automate routine tasks has led to increased efficiency but has also raised concerns about job displacement. In 2024, the technology sector experienced substantial layoffs, with over 74,000 workers dismissed by 255 tech companies, including major players like IBM, Amazon, Apple, and Meta. <a name="cite_ref-1"></a><sup>1</sup>.

IBM's strategic shift towards AI exemplifies this trend. In early 2023, IBM announced the elimination of 3,900 positions, followed by plans to replace 8,000 jobs with AI systems by August of the same year. Subsequently, CEO Arvind Krishna emphasized the company's commitment to "massively upskilling all of our employees on AI," highlighting a dual approach of workforce reduction and reskilling. <a name="cite_ref-2"></a><sup>2</sup>.

The World Economic Forum's 2020 report projected that AI could displace approximately 85 million jobs by 2025, while simultaneously creating 97 million new roles requiring advanced skill sets. <a name="cite_ref-3"></a><sup>3</sup>.

The COVID-19 pandemic has also accelerated the adoption of remote work, introducing challenges related to employee well-being. Studies have indicated that remote work can lead to increased stress, diminished work-life balance, and mental health concerns. For instance, research has shown that employees working remotely report higher levels of anxiety and burnout compared to their in-office counterparts. <a name="cite_ref-4"></a><sup>4</sup>.

While AI-driven automation and remote work have been individually studied, there is a paucity of research examining their combined impact on layoff risks across industries. This study aims to bridge this gap by analyzing how AI workload automation and remote work conditions collectively influence layoff trends.<a name="cite_ref-5"></a><sup>5</sup>.

## Cleaning Dataset to match the format of Remote Work & Mental Health Dataset

```python
from IPython.display import display
def display_df(df, rows=pd.options.display.max_rows, cols=pd.options.display.max_columns):
    """Displays n rows and cols from df"""
    with pd.option_context("display.max_rows", rows,
                           "display.max_columns", cols):
        display(df)
```

```python
# ✅ Convert industry names to lowercase and remove whitespace
ai_impact['Domain'] = ai_impact['Domain'].str.lower().str.strip()
layoffs['Industry'] = layoffs['Industry'].str.lower().str.strip()
remote_work_mental_health['Industry'] = remote_work_mental_health['Industry'].str.lower().str.strip()

# ✅ Define industry mapping
industry_mapping = {
    'communication & pr': 'IT', 'data & it': 'IT',
    'administrative & clerical': 'consulting', 'leadership & strategy': 'consulting',
    'medical & healthcare': 'healthcare', 'supply chain & logistics': 'manufacturing',
    'law enforcement': 'other', 'construction': 'manufacturing',
    'sales & marketing': 'retail', 'hospitality': 'retail',
    'food': 'retail', 'finance': 'finance', 'crypto': 'IT', 'marketing': 'retail',
    'other': 'other', 'healthcare': 'healthcare', 'consumer': 'retail', 'travel': 'retail',
    'infrastructure': 'manufacturing', 'sales': 'retail', 'transportation': 'manufacturing',
    'education': 'education', 'hr': 'consulting', 'manufacturing': 'manufacturing',
    'product': 'IT', 'recruiting': 'consulting', 'hardware': 'IT', 'energy': 'manufacturing',
    'support': 'consulting', 'fitness': 'retail', 'aerospace': 'manufacturing',
    'retail': 'retail', 'real estate': 'consulting', 'ai': 'IT', 'logistics': 'manufacturing',
    'security': 'IT', 'media': 'IT', 'data': 'IT', 'construction': 'manufacturing',
    'legal': 'consulting', 'unknown': 'other', 'it': 'IT'
}

# ✅ Apply industry mapping
ai_impact['Standardized_Industry'] = ai_impact['Domain'].map(industry_mapping)
layoffs['Standardized_Industry'] = layoffs['Industry'].map(industry_mapping)
remote_work_mental_health['Industry'] = remote_work_mental_health['Industry'].map(industry_mapping)

# ✅ Identify common industries
common_industries = set(ai_impact['Standardized_Industry']).intersection(set(layoffs['Standardized_Industry']))
print(f"Common Industries: {common_industries}")

# ✅ Filter datasets to only include industries present in all datasets
ai_impact_filtered = ai_impact[ai_impact['Standardized_Industry'].isin(common_industries)]
layoffs_filtered = layoffs[layoffs['Standardized_Industry'].isin(common_industries)]
remote_work_mental_health_filtered = remote_work_mental_health[remote_work_mental_health['Industry'].isin(common_industries)]

# Calculate counts for each industry in each dataset
ai_impact_counts = ai_impact_filtered.groupby('Standardized_Industry').size().to_dict()
layoffs_counts = layoffs_filtered.groupby('Standardized_Industry').size().to_dict()
remote_work_counts = remote_work_mental_health_filtered.groupby('Industry').size().to_dict()

# Find minimum counts across all datasets
industry_min_counts = {}
for industry in common_industries:
    ai_count = ai_impact_counts.get(industry, 0)
    layoffs_count = layoffs_counts.get(industry, 0)
    remote_count = remote_work_counts.get(industry, 0)
    industry_min_counts[industry] = min(ai_count, layoffs_count, remote_count)

# ✅ Function to limit dataset size per industry
def limit_rows_per_industry(df, industry_col, min_count_dict):
    return df.groupby(industry_col, group_keys=False).apply(
        lambda x: x.head(min_count_dict.get(x.name, len(x)))
    ).reset_index(drop=True)

# Now apply the limit_rows_per_industry function as before
remote_work_mental_health_filtered = limit_rows_per_industry(remote_work_mental_health_filtered, 'Industry', industry_min_counts)
ai_impact_filtered = limit_rows_per_industry(ai_impact_filtered, 'Standardized_Industry', industry_min_counts)
layoffs_filtered = limit_rows_per_industry(layoffs_filtered, 'Standardized_Industry', industry_min_counts)

# ✅ Reset index for proper concatenation
remote_work_mental_health_filtered = remote_work_mental_health_filtered.reset_index(drop=True)
ai_impact_filtered = ai_impact_filtered.reset_index(drop=True)
layoffs_filtered = layoffs_filtered.reset_index(drop=True)

# ✅ Rename industry column for consistency
ai_impact_filtered = ai_impact_filtered.rename(columns={'Standardized_Industry': 'Industry'})
layoffs_filtered = layoffs_filtered.rename(columns={'Standardized_Industry': 'Industry'})

# ✅ Drop redundant columns
ai_impact_filtered = ai_impact_filtered.drop(columns=['Domain'], errors='ignore')
layoffs_filtered = layoffs_filtered.drop(columns=['Industry_x', 'Industry_y'], errors='ignore')
```

## Merging Datasets

#### 1️⃣ Ensure Consistency in Industry Naming


```python
# ✅ Concatenate datasets column-wise
final_merged_data = pd.concat([
    remote_work_mental_health_filtered,
    ai_impact_filtered.drop(columns=['Industry']),
    layoffs_filtered.drop(columns=['Industry'])
], axis=1)

# ✅ Sort by Employee_ID if available
if 'Employee_ID' in final_merged_data.columns:
    final_merged_data = final_merged_data.sort_values(by="Employee_ID").reset_index(drop=True)

# ✅ Display final dataset shape
print(f"Final Merged Dataset Shape: {final_merged_data.shape}")

# ✅ Display dataset
display_df(final_merged_data.head(), cols=30)
```

### 2️⃣ AI Automation vs Layoff Risks
#### Investigating the Relationship Between AI Impact and Layoffs
To test the hypothesis that higher AI automation correlates with increased layoffs, we compare industries with high and low AI automation levels.

```python
# Categorize industries into high and low AI impact groups
ai_threshold = final_merged_data['AI Impact'].median()
final_merged_data['AI Category'] = final_merged_data['AI Impact'].apply(lambda x: 'High AI Impact' if x > ai_threshold else 'Low AI Impact')

# Mean layoffs in high vs. low AI impact industries
layoffs_high = final_merged_data[final_merged_data['AI Category'] == 'High AI Impact']['Laid_Off_Count'].mean()
layoffs_low = final_merged_data[final_merged_data['AI Category'] == 'Low AI Impact']['Laid_Off_Count'].mean()

print(f"Average Layoffs in High AI Impact Industries: {layoffs_high:.2f}")
print(f"Average Layoffs in Low AI Impact Industries: {layoffs_low:.2f}")

# Visualization
plt.figure(figsize=(8, 5))
sns.barplot(data=final_merged_data, x="AI Category", y="Laid_Off_Count", palette=["red", "blue"])
plt.title("Layoff Trends in High vs. Low AI Impact Industries")
plt.xlabel("AI Impact Category")
plt.ylabel("Average Layoffs")
plt.show()
```

# Description of the Layoff Trends Bar Chart

This bar chart visualizes the comparison of average layoff counts between industries categorized by their level of AI impact. The visualization directly addresses our research question about whether higher AI automation correlates with increased layoff risks.

## Key Findings:

- **Comparable Layoff Levels:** Surprisingly, high AI impact industries (red bar) and low AI impact industries (blue bar) show remarkably similar average layoff counts, both hovering around 290-295 employees.

- **Statistical Uncertainty:** The error bars (vertical black lines) indicate substantial variability within each category, suggesting that factors beyond AI impact may be influencing layoff decisions.

- **Challenging Initial Hypothesis:** This visualization challenges our initial hypothesis that increased AI automation necessarily correlates with higher layoff rates. The nearly identical heights of the bars suggest that the relationship between AI impact and layoffs is more nuanced than anticipated.

- **Further Investigation Needed:** The similarity in layoff counts despite differing AI impact levels suggests that companies may be using AI in ways that transform jobs rather than simply eliminating them, or that other factors (such as industry dynamics, company size, or economic conditions) have a stronger influence on layoff decisions.

This finding is particularly significant as it suggests that the narrative of "AI replacing jobs" may be oversimplified. Organizations appear to be managing workforce transitions in more complex ways than direct replacement, potentially integrating AI while maintaining similar staffing levels or redistributing human capital to different roles.

### 3️⃣ Remote Work, Mental Health, and Layoff Risks
#### Examining Work Location and Stress Levels
Remote work can introduce stress and mental health concerns, which may in turn affect job security. We analyze how different work locations (Onsite, Hybrid, Remote) relate to stress levels.

```python
# Create a cross-tabulation of Work_Location and Stress_Level
stress_location_counts = pd.crosstab(
    final_merged_data['Work_Location'], 
    final_merged_data['Stress_Level'],
    normalize='index'  # Normalize by row to get percentages
) * 100  # Convert to percentages

# Set up the plot with a clean style
plt.figure(figsize=(12, 8))
sns.set_style("whitegrid")

# Create the stacked bar chart
stress_location_counts.plot(kind='bar', stacked=True, 
                           colormap='viridis',
                           ax=plt.gca())

# Customize the plot
plt.title('Stress Level Distribution by Work Location', fontsize=16)
plt.xlabel('Work Location', fontsize=14)
plt.ylabel('Percentage of Employees (%)', fontsize=14)
plt.xticks(rotation=0)  # No rotation for better readability
plt.legend(title='Stress Level', bbox_to_anchor=(1.05, 1), loc='upper left')

# Add percentage labels on each segment
for i, (idx, row) in enumerate(stress_location_counts.iterrows()):
    cumulative = 0
    for j, value in enumerate(row):
        if value > 5:  # Only show labels for segments that are large enough
            plt.text(i, cumulative + value/2, 
                    f'{value:.1f}%', 
                    ha='center', 
                    va='center',
                    color='white',
                    fontweight='bold')
        cumulative += value

plt.tight_layout()
plt.show()
```

## Hypothesis Testing

### **Hypothesis 1**: AI-driven automation correlates with higher layoff rates

```python
# 1. Pearson correlation test for AI Impact and layoffs
subset_for_correlation = final_merged_data[['AI Impact', 'Laid_Off_Count']].dropna()
# Rename column to remove spaces
subset_for_correlation = subset_for_correlation.rename(columns={'AI Impact': 'AI_Impact'})
corr_ai_layoff, p_value_corr = stats.pearsonr(subset_for_correlation['AI_Impact'], 
                                             subset_for_correlation['Laid_Off_Count'])
print(f"Correlation between AI Impact and Layoff Count: {corr_ai_layoff:.4f}, p-value: {p_value_corr:.4f}")

# 2. Compare manufacturing and retail (routine task industries) vs others
routine_industries = ['manufacturing', 'retail']
routine_data = final_merged_data[final_merged_data['Industry'].isin(routine_industries)]
non_routine_data = final_merged_data[~final_merged_data['Industry'].isin(routine_industries)]

# Calculate correlation for routine industries
routine_subset = routine_data[['AI Impact', 'Laid_Off_Count']].dropna()
if len(routine_subset) > 1:
    corr_routine, p_routine = stats.pearsonr(routine_subset['AI Impact'], 
                                           routine_subset['Laid_Off_Count'])
    print(f"Correlation in routine industries: {corr_routine:.4f}, p-value: {p_routine:.4f}")
else:
    print("Insufficient data for routine industries correlation")

# Calculate correlation for non-routine industries
non_routine_subset = non_routine_data[['AI Impact', 'Laid_Off_Count']].dropna()
if len(non_routine_subset) > 1:
    corr_nonroutine, p_nonroutine = stats.pearsonr(non_routine_subset['AI Impact'], 
                                                 non_routine_subset['Laid_Off_Count'])
    print(f"Correlation in non-routine industries: {corr_nonroutine:.4f}, p-value: {p_nonroutine:.4f}")
else:
    print("Insufficient data for non-routine industries correlation")

# 3. Industry-specific analysis
industries = final_merged_data['Industry'].unique()
results = []

plt.figure(figsize=(12, 8))
subplot_count = 0
for i, industry in enumerate(industries):
    industry_data = final_merged_data[final_merged_data['Industry'] == industry]
    industry_subset = industry_data[['AI Impact', 'Laid_Off_Count']].dropna()
    
    if len(industry_subset) > 10:  # Only analyze industries with sufficient data
        subplot_count += 1
        if subplot_count <= 6:  # Limit to 6 subplots
            corr, p_val = stats.pearsonr(industry_subset['AI Impact'], 
                                       industry_subset['Laid_Off_Count'])
            results.append({
                'Industry': industry,
                'Correlation': corr,
                'P-Value': p_val,
                'Sample Size': len(industry_subset)
            })
            
            # Create subplot for each industry
            plt.subplot(2, 3, subplot_count)
            sns.regplot(x='AI Impact', y='Laid_Off_Count', data=industry_subset)
            plt.title(f'{industry}: r={corr:.2f}, p={p_val:.3f}')
            plt.tight_layout()

if subplot_count > 0:
    plt.suptitle('AI Impact vs Layoffs by Industry', fontsize=16)
    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.show()
else:
    print("No industries with sufficient data for plotting")

# 4. Linear regression model
if len(subset_for_correlation) > 1:
    model = ols('Laid_Off_Count ~ AI_Impact', data=subset_for_correlation).fit()
    print(model.summary().tables[1])  # Print coefficient table
else:
    print("Insufficient data for regression model")
```

### **Hypothesis 2**: Remote work conditions and stress levels

```python
# Create dummy variables for categorical variables
final_merged_data['is_remote'] = (final_merged_data['Work_Location'] == 'Remote').astype(int)

# Map stress levels to numeric values for analysis
stress_map = {'Low': 0, 'Medium': 1, 'High': 2}
final_merged_data['stress_numeric'] = final_merged_data['Stress_Level'].map(stress_map)

# Create clean subset for analysis
remote_analysis = final_merged_data[['Work_Location', 'Stress_Level', 'Laid_Off_Count', 
                                    'Productivity_Change', 'is_remote', 'stress_numeric']].dropna()

# Chi-square test for Work_Location vs. Stress_Level
contingency_table = pd.crosstab(remote_analysis['Work_Location'], remote_analysis['Stress_Level'])
chi2, p_value, dof, expected = stats.chi2_contingency(contingency_table)
print(f"Chi-square test for Work Location vs. Stress Level: chi2={chi2:.4f}, p-value={p_value:.4f}")

# Compare layoff counts by work location
locations = ['Remote', 'Hybrid', 'Onsite']
location_layoffs = []

for loc in locations:
    loc_layoffs = remote_analysis[remote_analysis['Work_Location'] == loc]['Laid_Off_Count']
    location_layoffs.append(loc_layoffs)
    print(f"Mean layoffs for {loc}: {loc_layoffs.mean():.2f}")

# ANOVA if we have data for all locations
if all(len(x) > 0 for x in location_layoffs):
    f_stat, p_value_anova = stats.f_oneway(*location_layoffs)
    print(f"ANOVA test for layoffs by work location: F={f_stat:.4f}, p-value={p_value_anova:.4f}")

# Visualize stress levels by work location
plt.figure(figsize=(10, 6))
sns.countplot(x='Work_Location', hue='Stress_Level', data=remote_analysis)
plt.title('Stress Levels by Work Location')
plt.xlabel('Work Location')
plt.ylabel('Count')
plt.legend(title='Stress Level')
plt.show()

# Regression analysis
model_remote = sm.OLS.from_formula('Laid_Off_Count ~ is_remote + stress_numeric', data=remote_analysis).fit()
print(model_remote.summary().tables[1])
```

### **Hypothesis 3**: Work-life balance and mental health support

```python
# Create numeric variables for analysis
final_merged_data['has_mental_health_support'] = (
    final_merged_data['Access_to_Mental_Health_Resources'] == 'Yes').astype(int)

# Clean subset for analysis
wellness_analysis = final_merged_data[['Work_Life_Balance_Rating', 'has_mental_health_support', 
                                     'Laid_Off_Count', 'AI Impact']].dropna()

# Rename column to avoid space issues
wellness_analysis = wellness_analysis.rename(columns={'AI Impact': 'AI_Impact'})

# Correlation between Work-Life Balance and layoffs
corr_wlb, p_value_wlb = stats.pearsonr(wellness_analysis['Work_Life_Balance_Rating'],
                                      wellness_analysis['Laid_Off_Count'])
print(f"Correlation between Work-Life Balance and Layoffs: {corr_wlb:.4f}, p-value: {p_value_wlb:.4f}")

# Compare layoffs between groups with and without mental health resources
with_support = wellness_analysis[wellness_analysis['has_mental_health_support'] == 1]['Laid_Off_Count']
without_support = wellness_analysis[wellness_analysis['has_mental_health_support'] == 0]['Laid_Off_Count']

t_stat, p_value_mh = stats.ttest_ind(with_support, without_support, equal_var=False)
print(f"T-test for mental health support: t={t_stat:.4f}, p-value={p_value_mh:.4f}")
print(f"Mean layoffs - With support: {with_support.mean():.2f}, Without: {without_support.mean():.2f}")

# Test interaction of wellness measures with AI impact
wellness_analysis['high_ai'] = wellness_analysis['AI_Impact'] > wellness_analysis['AI_Impact'].median()
wellness_analysis['high_wlb'] = wellness_analysis['Work_Life_Balance_Rating'] > 3

# Group statistics in high AI environments
high_ai_data = wellness_analysis[wellness_analysis['high_ai']]
print("\nIn high AI environments:")
print(f"Mean layoffs - High WLB: {high_ai_data[high_ai_data['high_wlb']]['Laid_Off_Count'].mean():.2f}")
print(f"Mean layoffs - Low WLB: {high_ai_data[~high_ai_data['high_wlb']]['Laid_Off_Count'].mean():.2f}")
print(f"Mean layoffs - With MH support: {high_ai_data[high_ai_data['has_mental_health_support'] == 1]['Laid_Off_Count'].mean():.2f}")
print(f"Mean layoffs - Without MH support: {high_ai_data[high_ai_data['has_mental_health_support'] == 0]['Laid_Off_Count'].mean():.2f}")

# Create interaction terms for regression
wellness_analysis['wlb_ai_interaction'] = wellness_analysis['Work_Life_Balance_Rating'] * wellness_analysis['AI_Impact']

# Regression model
model_wlb = sm.OLS.from_formula(
    'Laid_Off_Count ~ Work_Life_Balance_Rating + has_mental_health_support + AI_Impact + wlb_ai_interaction', 
    data=wellness_analysis).fit()
print(model_wlb.summary().tables[1])
```

### Results

#### Correlation Between Work-Life Balance and Layoff Counts

The Pearson correlation between Work-Life Balance Rating and layoff counts was very weak negative (r = -0.0212) and not statistically significant (p = 0.4381). This suggests that, across all companies in our dataset, there is no strong linear relationship between work-life balance and layoff risk.

#### Effects in High-AI Environments

In environments with high AI integration (above the median AI Impact score):

- **Mental Health Support Effect**: Contrary to expectations, companies offering mental health support in high-AI environments had higher mean layoffs (330.24) than those without such support (252.73)

#### Regression Analysis Results

The multiple regression model examining the relationships between work-life balance, mental health support, AI impact, and their interactions revealed:

None of the coefficients reached statistical significance at the conventional p < 0.05 level. The slight negative interaction term between work-life balance and AI impact (-0.4544) suggests that higher work-life balance may slightly mitigate the effect of AI on layoffs, but this effect was not statistically significant (p = 0.697).

### Interpretation and Conclusion

The results provide **limited and mixed support** for our third hypothesis:

2. **High-AI environments**: In companies with high AI integration, those with better work-life balance showed notably lower layoff counts. 3. **Mental health support**: Contrary to our expectations, companies providing mental health resources consistently showed higher average layoff counts. This counterintuitive finding might suggest that mental health programs are being implemented reactively in companies already experiencing higher stress and turnover, rather than serving as preventive measures.

4. **Interaction effects**: The regression model showed a weak protective interaction between work-life balance and AI impact, but it was not statistically significant.

In conclusion, while we observed some promising patterns suggesting that work-life balance may help mitigate layoff risks in high-AI environments, the overall statistical evidence does not strongly support our hypothesis that wellness measures broadly protect against layoffs during AI integration. These findings indicate that while work-life balance and mental health support remain important for employee well-being, they may not be sufficient on their own to protect against workforce reductions driven by strategic and technological factors. # Ethics & Privacy

Our dataset includes sensitive employment, mental health, and AI automation data, requiring strict privacy measures. To protect participant anonymity, all responses will be **de-identified** before analysis, ensuring that no personally identifiable information is stored or published.  

Biases in the dataset could arise due to **self-reporting, selection effects, and industry-specific variations** in AI adoption and remote work practices. Participants may underreport stress, burnout, or job insecurity due to social desirability or recall bias, and certain industries, such as **manufacturing or retail**, may be underrepresented due to lower remote work adoption. Additionally, perceptions of **AI-driven job displacement** may vary across different professional sectors.  

To address these concerns, we will **compare industry distributions** in our dataset against labor market benchmarks to identify **overrepresented or underrepresented sectors**. If imbalances are found, we will apply **post-stratification weighting** to adjust for sampling bias. Additionally, we will **cross-reference self-reported mental health and job security data** with objective indicators, such as **absenteeism logs, layoff records, and overtime trends**, to reduce self-report bias.  

Cultural and regional differences in mental health and job security perceptions may also introduce bias. To mitigate this, we will **normalize survey responses using standardized mental health scales** and segment data by industry and region to ensure a **balanced representation of workforce experiences**.  

Finally, to ensure **fairness and interpretability** in our predictive models, we will use **explainable AI techniques (e.g., SHAP values)** to analyze the impact of AI workload, work location, and workplace stress on layoffs. This will help us **detect potential biases in our predictions** and ensure that findings **reflect industry-wide trends rather than data artifacts**.  

By implementing these measures, we aim to **produce a privacy-conscious, unbiased, and representative analysis** of how AI automation and remote work conditions influence layoff risks across industries.
