# Abstract

Car accidents are among the most serious threats to people's safety, and the design of a vehicle is one of the most critical factors in preventing injuries to passengers. This project studies the possibility of making exact predictions about the impact of a crash on a driver based on the weight of the vehicle, its drivetrain type, body style, and model year. By using NHTSA’s SaferCar dataset, we examined various vehicle features to estimate how impact forces on the head, neck, and chest could differ by vehicle type and the resulting safety benefits.

As a primary step towards the realization of our objectives, we prepared the dataset by cleaning it, checking for missing values, and encoding the categorical variables while also normalizing the numerical features. Later on, we utilized different data visualization techniques to investigate the trends of crash forces through various vehicle characteristics. We did this by training some ML models (SVR and Random Forest) to estimate the impact forces from vehicle attributes. Our findings suggest that heavy vehicles, like SUVs and trucks, commonly show lower rib deflection and less severe head impact than light vehicles during side crashes. Nevertheless, neck and chest impact forces are not connected to vehicle weight remarkably; thus, overall crash safety cannot be determined just by the vehicle weight.

The subsequent analyses of body styles brought to light these safety patterns. For example, HIC-15 scores were generally lower on SUVs and passenger vans during frontal crashes, whereas pickups and smaller vehicles had wider ranges of HIC-15s with high-risk outliers. In a similar manner, side-impact HIC-36 scores were generally higher for sedans and coupes. As for SUVs and minivans, these results imply that the body style of a vehicle may make a difference in crash safety, showing lower injury risks for bigger vehicles during some types of crashes, while still holding valid. Nonetheless, drivetrain type (FWD, RWD, AWD) and model year were not statistically significant indicators of crash forces, so it can be concluded that the significant safety performance difference between distinct production years was negligible, which can be attributed to innovations in modern vehicle design.

In general, our research proves that vehicle weight and body style have an important influence on crash forces; at the same time, drivetrain and model year are not as impactful as to the severity of the crashes. Despite the fact that our models managed to identify the factors that contribute to vehicle accidents, it, however, is worth nothing that the crash data from actual accidents, crash test conditions, and biomechanical simulations should be incorporated to enhance the predictive accuracy. This can be achieved by widening the scope of this analysis to include other factors and variables that also contribute to vehicle safety assessments.

# Hypothesis


We believe that the following vehicle variables will likely affect the severity of crash impacts. In other words, we expect to see differences in the crash severity for cars with different body styles, drive trains, weights, and that are from different manufacturers and years. We define severity of crash impacts as a larger magnitude of the forces an occupant would experience on their head/neck/chest/leg during the crash test. Safer vehicles should result in less forces on these vital body parts.

In predicting impact forces, we expect to find that certain manufacturers, Volvo and Tesla, will produce safer vehicles, due to their renown for safety. Average impact forces will vary across brands.

Additionally, we expect to see that vehicles made more recently are safer, as manufacturers continuously upgrade safety features and structural integrity. As a result, newer vehicles should result in less impact forces on the driver/passenger.

Regarding drive train types, we predict that there will be a weak correlation between drive train types and crash forces. We assume that drive train mostly affects a vehicle's safety measure while tested on unsafe roads and weather conditions, in which AWD/4WD can improve traction and stability.

In terms of vehicle category, it is likely that larger vehicles (such as SUVs and pick up trucks) will protect passengers better. These vehicles have more volume and may result in better crumple effects, reducing the force transmitted to riders. 

Similarly, we hypothesize that heavier vehicles will result in lower impact forces on occupants. This expectation is based on observations that larger, heavier vehicles like buses often lack seatbelt requirements, possibly because of their mass and structure can reduce the forces transmitted during a crash.

These relationships will be explored in our EDA. 

Most importantly, since we expect to see differences in force magnitudes for vehicles with different characteristics, we hypothesize that it is possible to create a force prediction model from these 5 vehicle characteristics.

The NHTSA's ratings include data that describe vehicles and their results from crash tests that were performed in a controlled environment. The dataset has information on 128 variables. Some of these are categorical and some are numerical. Out of all the vehicle variables, we would like to focus on the brand, model year (2011-2025), category (SUV, 4 door, 2 door, pickup, van, station wagon, etc.), drive type (AWD, FWD, RWD, etc.), and weight. Most of these variables are recorded as chars (strings). 

Since this dataset only involves new vehicles, we do not have to worry about confounding factors when analyzing the model year. In other words, we will not expect to see that older vehicles are inherently less safe, since all of the cars tested will be new. 

Our variables of interest include make, model_yr, body_style, drive_train, curb_weight, hic15_driv, side_hic_36_driv, pole_hic_36_driv, nij_driv, neck_tens_driv, net_comp_driv, chest_defl_driv, and rib_deflection_driv.

All of the force variables are labeled with 'driv', indicating that they are the forces measured on the driver. 

HIC-15 stands for "Head Injury Criterion over 15 milliseconds". The hic15_driv measurement is taken from accelerometers inside of a dummy that is put through a crash scenerio. To find the exact number, the average acceleration is found over a 15 millisecond period, where the time period is determined as the one that maximizes the HIC. The average is taken to a power of 2.5, and is multipled with 15 milliseconds. 

A higher HIC-15 is more dangerous, and demonstrates a higher risk of injury. A value of 700 in test crashes is considered the maximum allowed score for cars in the US, established by the Federal Motor Vehicle Safety Standard in 2000.

The variables side_hic_36_driv and pole_hic_36_driv also measure Head Injurty Criterion, but in the side impact and side pole crash tests, respectively. Rather than over a 15 millisecond period, the measurements are taken over a 36 millisecond period. 

Some of the work that needs to be done on the data includes missing value cleanup, value groupings, and value type conversions. Although there are 11,000+ rows in the dataset, some columns of interest are missing 60% of their values. We will have to ensure there is sufficient diversity in the remaining 4-5k values. Moreover, some columns like 'drive_train' have entries that are essentially the same ('4x4' vs '4wd').

### Crash Test Data Background

Our data is taken from experiments performed on vehicles by the NHTSA. We analyze the forces on the driver in 3 different crash scenerios. 

The first one is a frontal crash. The NHTSA outlines the details of a frontal crash, in which an average-sized adult male is sitting in the driver's seat, and a small size adult female is in the passenger seat. In the experiments, both dummies are wearing their seatbelts, the vehicle is driving at 35mph head-on into a fixed barrier, and the results are meant to demonstrate what happens when 2 vehicles of a similar weight collide. To simplify our data testing, we focus on the forces affecting the driver.

The second crash scenerio is a side barrier crash, representing one car colliding into the driver side of another at an intersection. Again an average-sized adult male is in the driver seat, and a small-sized adult female is in the rear, on the side of the collision. Both are wearing sealtbelts. A 3,015 lb barrier moving at 38.5 mph is made to collide with the testing vehicle. 

Lastly, there is a side pole crash test, representing an out of control vehicle that crashes into a telephone pole. In this experiment, there is a small-sized adult female in the driver seat, wearing a seatbelt. The car is at a 75 degree angle, and is pulled at 25mph into a pole, which hits the driver's seat position.

## Model for Head Predictions

```python
# one hot encoding to encode categorical data (columns body_style, and drive_train)

#- "nij_driv" stands for “Frontal Impact Driver NIJ”
numerical_cols = ['curb_weight']
categorical_cols = ['drive_train', 'body_style']
X = safercar[numerical_cols + categorical_cols]
y = safercar['side_hic_36_driv'].dropna()

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=3)

# the column transformer combines numerical and categorical encoding into one step/object
col_transformer = ColumnTransformer(
    transformers=[
        ('numerical', Pipeline([
            ('imputer', SimpleImputer(strategy='mean')),  
            ('scaler', StandardScaler())  
        ]), numerical_cols),
        
        ('categorical', Pipeline([
            ('imputer', SimpleImputer(strategy='most_frequent')),  
            ('encoder', OneHotEncoder(sparse_output=False, drop='first'))  
        ]), categorical_cols)
    ]
)
rf_pipeline = Pipeline([
    ('preprocessor', col_transformer),
    ('rf', RandomForestRegressor(n_estimators=100, random_state=3))  
])

rf_pipeline.fit(X_train, y_train)
y_pred_rf = rf_pipeline.predict(X_test)

r2 = r2_score(y_test, y_pred_rf)
print(f"Random Forest R²: {r2:.4f}")
```

```python
mse = mean_squared_error(y_test, y_pred_rf)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test, y_pred_rf)

print(f"MSE: {mse:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"MAE: {mae:.4f}")
```

The model's performance is shown by several key metrics: MSE (Mean Squared Error), RMSE (Root Mean Squared Error), and MAE (Mean Absolute Error) and R². Our R² score evaluates how well the model fits the test data. The closer it is to 1, the better. A 1 means a perfectly fitted model. At 0.9504, our R² score indicates that our model is a pretty close fit.
```python
# Sklearn SVR with make, year, body style, drive train, and weight as predictors

# Drop manufacturers with too little samples (prevent data split issues later)
make_counts = safercar['make'].value_counts()
make_counts = make_counts[make_counts < 10]
makes_to_remove = make_counts.index.tolist()
safercar = safercar[~safercar['make'].isin(makes_to_remove)]

# Code below borrowed from Shanlin & Feiyi
numerical_cols_chest = ['model_yr', 'curb_weight']
categorical_cols_chest = ['make', 'drive_train', 'body_style']
X_chest = safercar[numerical_cols_chest + categorical_cols_chest]
Y_chest = safercar['chest_defl_driv'] # setting our target data to be chest deflection

X_train_chest, X_test_chest, y_train_chest, y_test_chest = train_test_split(X_chest, Y_chest, test_size=0.2, random_state=3)

# the column transformer combines numerical and categorical encoding into one step/object
col_transformer_chest = ColumnTransformer(
    transformers=[
        ('numerical', Pipeline([
            ('imputer', SimpleImputer(strategy='mean')),  
            ('scaler', StandardScaler())  
        ]), numerical_cols_chest),
        
        ('categorical', Pipeline([
            ('imputer', SimpleImputer(strategy='most_frequent')),  
            ('encoder', OneHotEncoder(sparse_output=False, drop='first', handle_unknown='ignore'))  # This website https://www.roelpeters.be/found-unknown-categories-in-column-sklearn/ solves a specific problem
        ]), categorical_cols_chest)
    ]
)

# the pipeline combines the encoding and random forest model creation into one object
rf_pipeline_chest = Pipeline([
    ('preprocessor', col_transformer_chest),
    ('rf', RandomForestRegressor(n_estimators=100, random_state=3)) 
])

rf_pipeline_chest.fit(X_train_chest, y_train_chest)
chest_defl_predictions = rf_pipeline_chest.predict(X_test_chest)
                 
feature_names_chest = numerical_cols_chest + list(rf_pipeline_chest.named_steps['preprocessor']
                                      .named_transformers_['categorical']
                                      .named_steps['encoder']
                                      .get_feature_names_out(categorical_cols_chest))
#print("features: ", feature_names_chest);

r2_chest = r2_score(y_test_chest, chest_defl_predictions)
print(f"R²: {r2_chest:.4f}")
```

For instance, as mentioned above, cars from certain brands may experience higher crash rates. However, this could partly be due to the driver. During our EDA and subsequently analysis, we tried to be clear about any unusual observations in the data. We made note of when values seemed incorrect. Instead of raw crash data tied to individuals, we’ll focus on overall trends.

In an attempt to be mindful of where our data comes from, we will rely primarily on crash tests performed in a laboratory. Measurements were obtained with different units and through different types of crash tests. Explaining the source of the measurements may help our audience understand how to interpret the results, avoiding misunderstandings. 

We primarily used our EDA to emphasize limitations in our data. Some values show up more often than others for certain variables. # Discussion and Conclusion

At the core of our project, we wanted to predict vehicle safety. Our main dataset comes from crash testing performed in a controlled environment by the National Highway Traffic Safety Administration (NHTSA).

Our EDA confirmed some of our background research and hypotheses by demonstrating that impact forces (and consequently vehicle safety) varied across vehicle brand, year, drive train type, style, and weight. 

For forces on the head, we explored frontal head-on crashes between a car and a vehicle of a similar weight, and side crashes with a barrier and a pole. Then, we created a random forest regression model to predict the relationship between the collision force of a side impact crash using the drive train, body style, and curb weight of certain cars. Overall, it was clear that the weight of the car being tested on was the biggest contributing factor to predicting the forces on the head. We could see that there was an inverse relationship between the safety of car in the crash and how heavy the car is. 

Apart from head forces, we also analyzed neck injuries. While prior studies primarily focused on fatality rates, our work directly links vehicle characteristics to biomechanical injury metrics like NIJ. This extension bridges the gap between vehicle design and occupant injury, offering a more precise understanding of how specific factors affect injury risk. We formed a model from our data with strong predictive ability with Random Forest regression, which achieved an R² score of 0.8528. This score shows that the selected vehicle characteristics (model year, curb weight, drive train, and body style) significantly impact the Frontal Impact Driver NIJ during a crash. The model's feature importance analysis highlights that vehicle types, particularly SUVs and 4-door sedans, played a crucial role in predicting crash forces. This aligns with previous research from IIHS, which found that larger vehicles tend to have lower fatality rates. While the prior study relied on descriptive statistics to assess vehicle safety, we used Random Forest regression to show the complex relationship.

Additionally, our analysis of Frontal Impact Driver NIJ reveals a downward trend in the mean value for newer model years, supporting the hypothesis that newer vehicles are safer. This aligns with findings from the NHTSA, which shows that older vehicles have a higher fatality risk. However, our analysis provides a quantitative link between vehicle model year and NIJ trends. The slightly increasing trend in crash forces for different drivetrain types (from FWD to AWD, RWD, 4WD, and finally 2WD) suggests that the drivetrain may play a role in occupant protection during a crash, which relates to 4WD vehicles tend to result in fewer deaths than their 2WD counterparts from IIHS's research. However, we include more drivetrain types in our analysis and find FWD, AWD, and RWD may be safer than 4WD and 2WD. This makes the understanding of how drivetrain types affect neck injury more comprehensive.

Heavier vehicles also generally provide better protection, as we found curb weight greatly impacts our prediction. This relates to IIHS's research again, in which the researchers suggest size and weight are often correlated and car safety differs for vehicles of different weights. However, our model isolates curb weight’s independent effect on NIJ, revealing that weight alone can greatly affect crash injury. Moreover, we found Hyundai, Cadillac, Volvo, and Mazda cause less mean NIJ during crashes, while Toyota, BMW, Lexus, and Volkswagen cause higher mean NIJ. As "Ram, Tesla, and Subaru Have the Worst Drivers" suggests, Tesla and BMW drivers have high rates of accidents, and "When Were Seat Belts Invented?" suggests Volvo has high safety values. However, we are more focused on the cars themselves, and we found Tesla and Subaru have a fair mean NIJ value, suggesting that their vehicles provide reasonable crash protection. Our analysis also only looked at the physical characteristics of the vehicle and did not take into account driver characteristics (such as driving habits etc.). The occupant's human parameters (such as height, weight, sitting position) also affect injuries, but our data set did not include these.
Another limitation of our analysis may be a result of using Random Forest for regression analysis. The Random Forest model is unable to capture complex causal relationships.

Overall, our original goal was to use certain vehicle characteristics to predict vehicle safety, and we consider our final models to be quite successful at that task.
