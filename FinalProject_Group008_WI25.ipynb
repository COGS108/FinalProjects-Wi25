## Background and Prior Work


Mental health and well-being are shaped by a range of social and environmental factors, including political context. As the U.S. has become increasingly politically polarized, individuals’ experiences of stress and mental health may be shaped by the ideological composition of their environment. According to research from the Pew Research Center, both major political parties have grown more ideologically homogeneous over time, with Republican-leaning voters shifting significantly toward conservative positions and Democratic-leaning voters moving further left.<a name='citeref1'></a><sup>1</sup> This growing ideological division is not just a national phenomenon but extends to regional and state-level political identities, potentially shaping social cohesion, perceived safety, and access to mental health resources.<a name='citeref1'></a><sup>1</sup>

Political polarization has further been linked to increased psychological stress, particularly in politically mixed or competitive environments. Research by Fraser et al. (2022) found that individuals living in politically homogeneous states reported higher levels of mental health distress in environments where they were of political minority.<a name='citeref2'></a><sup>2</sup> However, their study did not focus on politically divided or 'swing' states, where ideological conflict might create an even greater sense of uncertainty and instability. Other research into the neurological effects of the U.S.’ political climate on individuals has suggested that heightened polarization fosters an 'us vs. them' mentality, reinforced by neural synchrony among politically aligned individuals, which may exacerbate stress when individuals are exposed to opposing political ideologies.<a name='citeref3'></a><sup>3</sup> This aligns with findings from Jost et al. (2017), a study which demonstrated that political ideology correlates with cognitive and emotional responses to uncertainty, suggesting that individuals in politically volatile environments may experience heightened stress due to unpredictability in policy, governance, and social dynamics.<a name='citeref4'></a><sup>4</sup>

Another key factor in this relationship is the role of state-level policies and structural conditions. Finkelstein et al. (2022) highlights how social safety nets, healthcare accessibility, and economic stability contribute to mental health outcomes, with stronger social welfare systems generally associated with lower stress and improved well-being.<a name='citeref5'></a><sup>5</sup> These factors often align with state-level political leanings, as blue states tend to have more expansive public services compared to red states. However, existing research has not sufficiently explored how these structural variables mediate the relationship between state political affiliation and mental health, particularly, factoring in swing states, where political control—and thus policy stability—frequently shifts.

Our study aims to build on these findings by examining how the political leaning of a state (red, blue, or swing) correlates with residents’ suicide rates and reported depression. We will also investigate whether a state’s structural factors such as healthcare accessibility, economy, and affordability mediate this relationship. A key contribution of our research is the focus on swing states, where political division may amplify stress-related outcomes more than in solidly red or blue states. By incorporating economic and demographic variables such as HDI and safety, our study seeks to provide a more comprehensive understanding of how political environments shape mental health outcomes as a result of and beyond individuals’ political affiliation.

# Data Cleaning & Wrangling
First, we will clean our data. We have four datasets which we will import and merge them together to fit the purposes of our project.

## Dataset #1: 2024 Quality of Life by State
Here, we are importing the Quality of Life dataset. This dataset is already pretty "clean" (i.e. no null values, already tidy, and there aren't any variables that we need to drop). We made a minor change by renaming the 'stateFlagCode' column to 'state' in order to make it easier to merge with the other datasets.

```python
# Importing libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import geopandas as gpd
from scipy import stats

# Loading Dataset 1
qol_data = pd.read_csv('quality-of-life-by-state-2024.csv')

# Setting the state as the index for better usability
qol_data = qol_data.set_index('state')
qol_data = qol_data.drop(columns = ['stateFlagCode'])
qol_data.head()
```

## Dataset #2: 2024 US County Level Election Results
Next, we are importing the Election Results dataset. We made regrouped the data and added percentage columns to make data analysis easier. Like with the first dataset, the second dataset is pretty "clean" (i.e. no null values, already tidy, and there aren't any variables that we need to drop).

```python
# Loading in election data
election = pd.read_csv("https://raw.githubusercontent.com/tonmcg/US_County_Level_Election_Results_08-24/refs/heads/master/2024_US_County_Level_Presidential_Results.csv")
election.head()
```

```python
# Group by state and aggregate data by mean in order to more easily conduct analysis
election = election.groupby('state_name').agg({
    'votes_gop': 'sum',
    'votes_dem': 'sum',
    'total_votes': 'sum',
})

# Add percentage columns to election
election['per_gop'] = election['votes_gop']/election['total_votes']
election['per_dem'] = election['votes_dem']/election['total_votes']
election.head()
```

## Dataset #3: 2024 Suicide Rates By State
We then import the Suicide Rates dataset. We renamed the column that contains the state name for better usability. This was also a pretty clean dataset (i.e. no null values, already tidy, and there aren't any variables that we need to drop) so there wasn't much wrangling needed.

```python
# Loading Suicide Rates By State Data
suicide_rate = pd.read_csv('suicide-rates-by-state-2024.csv')

# Setting the state as the index for better usability
suicide_rate = suicide_rate.set_index("state")
suicide_rate = suicide_rate.drop(columns = ['stateFlagCode'])
suicide_rate.head()
```

## Dataset #4: 2024 Depression Rates By State
Finally, we import the Depression Rates dataset. We renamed the column that contains the state name to make it easier to merge the datasets. This dataset was also relatively clean, so there wasn't much wrangling needed.

```python
# Loading Depression Rates By State Data
depression_rate = pd.read_csv('depression-rates-by-state-2024.csv')
depression_rate = depression_rate.drop(columns = ['stateFlagCode'])
depression_rate

# Setting the state as the index for better usability
depression_rate = depression_rate.set_index("state")
depression_rate.head()
```

Next, we combined all of our datasets into one dataframe. This allows for easy data access and manipulation.

```python
combined_df = qol_data.join(election).join(suicide_rate).join(depression_rate)
combined_df.head()
```

Next, we will add a column that indicates if a state is a red, blue, or swing state. This is a critical part of our dataset as we need it in order to answer our research question. We determine that a state is a red state if >51% of the vote is Republican, a state is a blue state if >51% of the vote is Democrat, and a state is a swing state if it falls into neither category.

```python
# Method to categorize states as red or blue utilizing majority percentage for simplicity
def categorize_state(row):
    if row['per_gop'] > .51:
        return 'Red'
    elif row['per_dem'] > .51:
        return 'Blue'
    else:
        return 'Swing'

combined_df['political_leaning'] = combined_df.apply(categorize_state, axis=1)
```

```python
combined_df.head()
```

We are finished cleaning and wrangling our data! We will use this dataframe for our analysis.

Considering that swing states do not appear to have significantly higher rates of suicide and reported depression, for the rest of the project we plan to focus on red and blue states. Now, we will conduct some statistical tests to see if there is statistical significance. Note that Florida has a null value for the depression rates category, so we will use a new dataframe that excludes Florida in the analysis for depression rates.


```python
from scipy.stats import f_oneway, kruskal

# ANOVA for Suicide Rate across political categories
red_states = combined_df[combined_df['political_leaning'] == 'Red']['SuicideRate']
blue_states = combined_df[combined_df['political_leaning'] == 'Blue']['SuicideRate']
swing_states = combined_df[combined_df['political_leaning'] == 'Swing']['SuicideRate']

# Kruskal-Wallis test if data is not normally distributed
_, p_value_suicide = kruskal(red_states, blue_states, swing_states)
print(f'Kruskal-Wallis test p-value for Suicide Rates: {p_value_suicide}')

filtered_df = combined_df.dropna(subset=["DepressionRatesByStateAge-AdjustedDepressionRate"])

# Repeat the process for Depression Rates
red_states_depression = filtered_df[filtered_df['political_leaning'] == 'Red']['DepressionRatesByStateAge-AdjustedDepressionRate']
blue_states_depression = filtered_df[filtered_df['political_leaning'] == 'Blue']['DepressionRatesByStateAge-AdjustedDepressionRate']
swing_states_depression = filtered_df[filtered_df['political_leaning'] == 'Swing']['DepressionRatesByStateAge-AdjustedDepressionRate']

_, p_value_depression = kruskal(red_states_depression, blue_states_depression, swing_states_depression)
print(f'Kruskal-Wallis test p-value for Depression Rates: {p_value_depression}')
```

For suicide rates, the p-value is below the typical significance threshold (e.g., 0.05), indicating that there is a statistically significant difference in suicide rates between the political groups. This means that political affiliation (or political leaning) does have an effect on suicide rates across states. We can reject the null hypothesis (which states that there is no difference in suicide rates across political groups), supporting the alternative hypothesis that political affiliation influences suicide rates. However, for depression rates, the p-value is greater than 0.05, which suggests no statistically significant difference in depression rates between the political categories. This means that depression rates across red, blue, and swing states are not significantly different from one another.

## Linear Regression Analysis

We will create a linear regression model for suicide rates and one for depression rates. First, we will use OLS regression analysis.

```python
import statsmodels.api as sm

# Independent Variables
X = combined_df[['per_gop', 'QualityOfLifeTotalScore']]
X = sm.add_constant(X)  # Add intercept

# Model 1: Suicide Rate
y_suicide = combined_df['SuicideRate']
model_suicide = sm.OLS(y_suicide, X).fit()
print(model_suicide.summary())

# Model 2: Depression Rate. Note that we need to drop some states due to the data being null for depression rates.
filtered_df = combined_df.dropna(subset=["DepressionRatesByStateAge-AdjustedDepressionRate"])
X = filtered_df[['per_gop', 'QualityOfLifeTotalScore']]
X = sm.add_constant(X) 
y_depression = filtered_df['DepressionRatesByStateAge-AdjustedDepressionRate']
model_depression = sm.OLS(y_depression, X).fit()
print(model_depression.summary())
```

Above, we created two models. These are the insights from the models.

Suicide rate model: For the suicide rate model, the R-squared value is 0.283, meaning the model explains about 28.3% of the variance in suicide rates. The adjusted R-squared, which accounts for the number of predictors, is slightly lower at 0.252. The F-statistic is 9.270 with a p-value of 0.0004, indicating that the model as a whole is statistically significant. Looking at the coefficients, the intercept is 10.47 but is not statistically significant (p = 0.242). The per_gop variable, however, has a coefficient of 25.99 with a p-value of 0.001, meaning that as the percentage of GOP votes increases, suicide rates also increase significantly. Meanwhile, the Quality of Life Score has a small negative coefficient (-0.138), but it is not statistically significant (p = 0.293), suggesting that it does not strongly predict suicide rates. While the model shows a strong association between GOP vote share and suicide rates, it only explains about 28% of the variation, meaning other factors are at play.

Depression rate model: The depression rate model, on the other hand, has a lower R-squared value of 0.151, meaning it only explains 15.1% of the variance in depression rates. The adjusted R-squared is even lower at 0.115. Despite this, the F-statistic of 4.106 and its corresponding p-value of 0.0229 indicate that the model is still statistically significant. Examining the coefficients, the intercept is 14.86 and is statistically significant (p = 0.021), while the per_gop variable has a coefficient of 13.44 with a p-value of 0.012, again showing a significant positive relationship between GOP vote share and depression rates. The Quality of Life Score, similar to the suicide model, has a negligible coefficient (-0.011) and is not statistically significant (p = 0.905), suggesting it does not play a major role in predicting depression rates. This model explains even less of the variance compared to the suicide model, indicating that other factors are likely influencing depression rates beyond political affiliation and quality of life.

Overall, both models suggest that higher GOP vote percentages are significantly associated with higher suicide and depression rates, whereas the Quality of Life Score does not appear to be a strong predictor for either outcome. However, these models only explain a fraction of the variance, particularly in the case of depression rates, which suggests that many other factors—such as economic conditions, access to healthcare, cultural influences, or demographic differences—may be contributing to mental health outcomes across states.

## Feature Importance Analysis

Our first analysis suggested that higher GOP perecntages are correlated with higher suicide and depression rates, but the OLS models we used would only explain a fraction of the variance, there are other variables in play with this complicated issue. So, we want to look at other variables. We do this by using a Random Forest Model.

We will perform another analysis to see if another model can confirm our findings in order to be more rigorous with our data analysis. We will compare the R-squared values for every feature for predicting suicide/depression rates.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Define independent variables
independent_vars = [
    "QualityOfLifeTotalScore", "QualityOfLifeQualityOfLife", "QualityOfLifeAffordability",
    "QualityOfLifeEconomy", "QualityOfLifeEducationAndHealth", "QualityOfLifeSafety",
    "per_gop", "per_dem"
]

# Define dependent variables
combined_df.rename(columns={'DepressionRatesByStateAge-AdjustedDepressionRate': 'DepressionRateAdjusted'}, inplace=True)
dependent_vars = ["SuicideRate", "DepressionRateAdjusted"]

# Dictionary to store R-squared values
r2_scores = {dep_var: [] for dep_var in dependent_vars}

# Compute R-squared values for each independent variable
for dep_var in dependent_vars:
    for ind_var in independent_vars:
        df_filtered = combined_df[[ind_var, dep_var]].dropna()  # Drop NaN values
        if df_filtered.shape[0] > 1:  # Ensure there is enough data
            X = df_filtered[[ind_var]]
            y = df_filtered[dep_var]

            model = LinearRegression()
            model.fit(X, y)
            y_pred = model.predict(X)
            r2_scores[dep_var].append(r2_score(y, y_pred))
        else:
            r2_scores[dep_var].append(np.nan)  # Not enough data

# Plot bar chart
fig, ax = plt.subplots(figsize=(10, 6))
bar_width = 0.4
x = np.arange(len(independent_vars))

for i, dep_var in enumerate(dependent_vars):
    ax.bar(x + i * bar_width, r2_scores[dep_var], bar_width, label=dep_var)

ax.set_xticks(x + bar_width / 2)
ax.set_xticklabels(independent_vars, rotation=45, ha='right')
ax.set_ylabel("R-squared Value")
ax.set_title("R-squared Values for Predicting Suicide and Depression Rates")
ax.legend()
plt.tight_layout()
plt.show()
```

The bar chart shows the predictive power of various quality-of-life and political affiliation variables on suicide and depression rates, measured using R-squared values. Higher R-squared values indicate stronger explanatory power for each independent variable. Political affiliation appears to be the strongest predictor in the dataset. This graph overall aligns with the Random Forest models but there are some differences. Key similarities are that Percent Republican and Percent Democrat are influential features. For suicide rates, Quality of Life and Safety stand out as other influential features. This is slightly different with depression rates, as Affordability is a significant variable.

# Ethics & Privacy

When considering the data we will use in our project, we understand that many of the existing public mental health datasets rely on self-reported mental health conditions which may underrepresent certain groups. Likewise, individuals like older aged individuals who experienced negative stigmas around reporting mental health may be less likely to report symptoms accurately or unbiasedly ([link to evidence](https://pmc.ncbi.nlm.nih.gov/articles/PMC4074963/#sec3)). We recognize that user confidentiality, especially of mental health data, is important, so we aim to only prioritize using de-identified datasets to preserve participant anonymity.

We acknowledge in our datasets, there may be a lower willingness to report stress levels due to cultural stigma which leads to self-reporting bias, with rural areas potentially underreporting stress due to privacy concerns or perceived weakness. Underrepresentation of rural residents in stress surveys (e.g., due to limited internet access) may bias the data, potentially leading to the false conclusion that urban populations experience more stress. Reporting bias may also be skewed across political cultures due to mental health stigma and social norms, affecting help-seeking behavior. Furthermore, online stress datasets may overrepresent younger, digitally connected individuals, while underrepresenting older, rural, and/or lower-income populations with limited internet access. These biases in the dataset could then alter our results which may result in false conclusions or generalizations of various populations.

In order to detect these specific biases throughout our project, we collectively recognize the biases we may have when analyzing our data sets and are aware that mental health is influenced by all kinds of factors rather than just political/geographical affiliation. With that being said, we want to avoid overly deterministic conclusions that reinforce stereotypes or contribute to further stigmatization of mental health challenges within specific groups. We detect and analyze any biases in response rates across demographics and making note of that as we go through our project.

To preserve data privacy and ensure equitable impact, we used aggregated and anonymized data sets rather than raw patient records to stay compliant to mental health dataset standards. Using multiple datasets help to balance our perspectives and identify any potential gaps in representation. The datasets we use provide data aggregated at the state level. Therefore, our analysis and its conclusions operate at the state level, and may not be generalizable to individuals. The data itself is sourced from publicly available datasets on quality of life, election results, suicide rates, and depression rates. The metrics we use are optimized for the variables available to us in these datasets.
