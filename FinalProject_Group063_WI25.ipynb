## Student AI survey dataset

## Data Cleaning

The dataset originally stores multiple class responses from a single student in the same row. Our goal is to restructure it so that each row represents one class per student instead of grouping them together. There are three types of repeating columns in the dataset (Ai Policy Strictness, AI Usage, Grade). We find how many classes each student reported by counting how many times the stricness column appears. Here, we load our data in and rename the question columns into shorter names.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the CSV file
df = pd.read_csv('StudentAISurveyResponse.csv')

df.rename(columns={
    df.columns[0]: "Time",
    df.columns[1]: "Major",
    df.columns[2]: "Year",
    df.columns[3]: "AI_Familiarity",
    df.columns[4]: "Course_AI_Restrict_1",
    df.columns[5]: "Personal_AI_Use_1",
    df.columns[6]: "Grade_1",
    df.columns[7]: "Comments_1",
    df.columns[8]: "Course_AI_Restrict_2",
    df.columns[9]: "Personal_AI_Use_2",
    df.columns[10]: "Grade_2",
    df.columns[11]: "Comments_2",
    df.columns[12]: "Course_AI_Restrict_3",
    df.columns[13]: "Personal_AI_Use_3",
    df.columns[14]: "Grade_3",
    df.columns[15]: "Comments_3",
    df.columns[16]: "Course_AI_Restrict_4",
    df.columns[17]: "Personal_AI_Use_4",
    df.columns[18]: "Grade_4",
    df.columns[19]: "Comments_4",
    df.columns[20]: "Course_AI_Restrict_5",
    df.columns[21]: "Personal_AI_Use_5",
    df.columns[22]: "Grade_5",
    df.columns[23]: "Comments_5"
}, inplace=True)

df.head()
```

Before sorting the data, we begin by cleaning out any false or inconsistent information. Due to the structure of our survey, some questions allowed free-response answers, meaning participants could type any input. We identified a data point where the grades were set to all "F" and the open-ended responses contained inappropriate content that did not align with the rest of the data. As a result, we removed this row to ensure data integrity.

```python
# Remove row contating false information
df= df[~df.apply(lambda row: row.astype(str).str.contains('LeBron', case=False, na=False).any(), axis=1)]
```

To ensure column names are unique, we renamed specific class-related columns by appending a numerical suffix to duplicate occurrences. This was necessary because our dataset contained multiple columns with similar names, such as "Course_AI_Restrict", "Personal_AI_Use", and "Grade". By maintaining a count for each of these columns, we ensured that repeated column names were properly distinguished, preventing potential conflicts when analyzing the data.

```python
# Renaming columns to ensure unique class-related names
new_column_names = []
class_count = {"Course_AI_Restrict": 1, "Personal_AI_Use": 1, "Grade": 1}

for col in df.columns:
    if col in class_count:
        new_column_names.append(f"{col}_{class_count[col]}")  # Append number suffix
        class_count[col] += 1  # Increment count for the next occurrence
    else:
        new_column_names.append(col)  # Keep original name for non-class-related columns

df.columns = new_column_names  # Apply corrected column names
```

To restructure the dataset, we reshaped it so that each student has five rows, corresponding to their five classes. The original dataset stored class-related responses in separate columns, making it difficult to analyze individual class-level data. By transforming the data, we ensured that each row represents a single class while retaining student-specific details such as "Time", "Major", "Year", and "AI_Familiarity". However, in this process, we specifically selected only the columns "Course_AI_Restrict", "Personal_AI_Use", and "Grade" for reshaping, meaning that other columns, including "Comments", were not included in the transformed dataset. As a result, "Comments" was removed during this restructuring. The newly structured data is stored in "df_expanded", where each student now has five separate rows, each corresponding to one of their classes.

```python
# Reshaping the dataset so that each student has 5 rows (one per class)
id_vars = ["Time", "Major", "Year", "AI_Familiarity"]
value_vars = ["Course_AI_Restrict", "Personal_AI_Use", "Grade"]

reshaped_data = []

# Distribute each student's class responses into separate rows
for _, row in df.iterrows():
    for i in range(5):  # Each student has up to 5 classes
        reshaped_row = {col: row[col] for col in id_vars}  # Keep student info
        for var in value_vars:
            reshaped_row[var] = row.get(f"{var}_{i+1}", None)  # Assign class data
        reshaped_data.append(reshaped_row)

# Creating a new DataFrame with the reshaped data
df_expanded = pd.DataFrame(reshaped_data)
```

To refine the dataset, we first removed any rows where both "Course_AI_Restrict" and "Personal_AI_Use" were missing. This ensures that we only retain meaningful data for analysis.

Next, we filtered out rows where either "Course_AI_Restrict" or "Personal_AI_Use" contained non-numeric characters. This step was necessary to ensure consistency in our numerical analysis by eliminating any unexpected text responses or formatting errors in these columns. The resulting dataset now contains only valid numerical values for these fields.

```python
df_expanded = df_expanded.dropna(subset=["Course_AI_Restrict", "Personal_AI_Use"], how='all')

# Remove rows where "Course_AI_Restrict" or "Personal_AI_Use" contain any non-numeric characters
df_expanded = df_expanded[
    df_expanded["Course_AI_Restrict"].astype(str).str.match(r'^\d+(\.\d+)?$', na=True) &
    df_expanded["Personal_AI_Use"].astype(str).str.match(r'^\d+(\.\d+)?$', na=True)
]
```

To standardize the grading data, we mapped letter grades to their corresponding GPA values based on UCSD's numerical grading scale. This conversion allows for numerical analysis of grades while preserving any unmatched values. By applying the mapping to the "Grade" column, we ensured that all standard letter grades were converted to their respective GPA equivalents, while any non-matching values remained unchanged. This approach maintains data integrity while facilitating statistical operations on the grades. A portion of the data structure is shown below that displays the changes we made to organize and standardize the dataset.

```python
# Define the grade mapping
grade_mapping = {
    'A+': 4.0, 'A': 4.0, 'A-': 3.7, 'B+': 3.3, 'B': 3.0, 'B-': 2.7, 
    'C+': 2.3, 'C': 2.0, 'C-': 1.7, 'D+': 1.0, 'D': 1.0, 'D-': 1.0, 'F': 0.0
}

# Convert "Grade" column values using the mapping, ignoring unmatched values
df_expanded["Grade"] = df_expanded["Grade"].map(grade_mapping).fillna(df_expanded["Grade"])

df_expanded
```

## Data Analysis and Results

The first step is for us to explore the distributions of both AI policy strictness and AI tool usage. This will give us a better understanding of how the data is spread out and help us identify any patterns or outliers in each variable. Next, we’ll examine how AI familiarity could potentially explain differences in AI tool usage. Specifically, we want to see if students who are more familiar with AI are more likely to use AI tools, and whether this is influenced by the AI policy strictness. Once we've explored the distributions and understood the effect of AI familiarity, we can calculate the correlation between AI policy strictness and AI tool usage. This will give us insight into how strongly these two variables are related.

First, lets see the distribution of values (e.g., AI Policy Strictness or AI Tool Usage), using a histogram. A histogram will show the frequency of different levels of AI policy strictness, which can help you assess how strict (or lenient) the policies are overall. A histogram will show the frequency of different levels of AI tool usage and allow you to assess how frequently students report using AI tools.

```python
import seaborn as sns  # Import seaborn for plotting

# Convert 'Course_AI_Restrict' and 'Personal_AI_Use' columns to numeric, forcing errors to NaN
df_expanded['Course_AI_Restrict'] = pd.to_numeric(df_expanded['Course_AI_Restrict'], errors='coerce')
df_expanded['Personal_AI_Use'] = pd.to_numeric(df_expanded['Personal_AI_Use'], errors='coerce')

# Drop rows where the values in these columns are NaN
df_expanded = df_expanded.dropna(subset=['Course_AI_Restrict', 'Personal_AI_Use'])

# Plot histogram for 'Course_AI_Restrict' and 'Personal_AI_Use'
plt.figure(figsize=(12, 6))

# Histogram for AI Policy Strictness
plt.subplot(1, 2, 1)
sns.histplot(df_expanded['Course_AI_Restrict'], kde=True, color='blue')
plt.title('Distribution of AI Policy Strictness')
plt.xlabel('AI Policy Strictness')
plt.ylabel('Frequency')

# Histogram for AI Tool Usage
plt.subplot(1, 2, 2)
sns.histplot(df_expanded['Personal_AI_Use'], kde=True, color='orange')
plt.title('Distribution of AI Tool Usage')
plt.xlabel('AI Tool Usage')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()

# Calculate value counts for 'Course_AI_Restrict' and 'Personal_AI_Use'
ai_policy_counts = df_expanded['Course_AI_Restrict'].value_counts().sort_index()
ai_tool_usage_counts = df_expanded['Personal_AI_Use'].value_counts().sort_index()

# Display the results
print("AI Policy Strictness Counts:")
print(ai_policy_counts)

print("\nAI Tool Usage Counts:")
print(ai_tool_usage_counts)
```

Based on the histograms, we can identify which distributions had higher frequencies. For the AI policy strictness, the highest frequency occurred at a strictness level of 2.0. Similarly, for AI tool usage, the highest frequency was observed at a usage level of 1.0.

Next, we will find out if the familarity of AI has a correlation with AI tool usuage.

# Scatter Plot: AI Familiarity vs. AI Tool Usage
This scatter plot will allow us to see if AI familiarity has a positive correlation with AI tool usage. This is to explore the idea that students who are more familiar with AI are more likely to use AI tools.

```python
# Scatter Plot with Regression Line: AI Familiarity vs. AI Tool Usage
plt.figure(figsize=(10, 6))
sns.regplot(x='AI_Familiarity', y='Personal_AI_Use', data=df_expanded, scatter_kws={'s': 30}, line_kws={'color': 'red'})
plt.title('Regression Line: AI Familiarity vs. AI Tool Usage')
plt.xlabel('AI Familiarity')
plt.ylabel('AI Tool Usage')
plt.show()
```

We can tell from our data that the regression line is sloping upwards from left to right, it indicates a positive correlation—as AI familarity increases, AI tool usage also increases. However, since the points are more spread out and don’t follow the line closely, it suggests a weaker relationship. Lets further investigate their relationship.

# Correlation Coefficient Between AI Familiarity and AI Tool Usage:

```python
# Calculate the correlation between Course_AI_Restrict and Personal_AI_Use
correlation = df_expanded[['Course_AI_Restrict', 'Personal_AI_Use']].corr()

# Heatmap: Correlation Matrix
plt.figure(figsize=(6, 4))
sns.heatmap(correlation, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Heatmap: Correlation between AI Policy Strictness and AI Tool Usage')
plt.show()
```

As we can see, we have a rate of .47. This means there is a moderate positive relationship between student's usage with AI with their familarity with the tool itself.

# Heatmap for correlation of Course AI Strictness and AI tool usage
A correlation matrix will help us quantify the strength of the relationship between the variables. Specifically, we are interested in whether AI policy strictness and AI tool usage have a strong negative correlation.

# Correlation Coefficient Between AI Familiarity and AI Tool Usage and AI Course Strictness:

```python
# Convert relevant columns to numeric (if necessary), forcing errors to NaN
df_expanded['Course_AI_Restrict'] = pd.to_numeric(df_expanded['Course_AI_Restrict'], errors='coerce')
df_expanded['Personal_AI_Use'] = pd.to_numeric(df_expanded['Personal_AI_Use'], errors='coerce')
df_expanded['AI_Familiarity'] = pd.to_numeric(df_expanded['AI_Familiarity'], errors='coerce')

# Drop rows with NaN values in any of the relevant columns
df_expanded = df_expanded.dropna(subset=['Course_AI_Restrict', 'Personal_AI_Use', 'AI_Familiarity'])

# Calculate the correlation matrix for AI Usage, AI Policy Strictness, and AI Familiarity
correlation_matrix = df_expanded[['Personal_AI_Use', 'Course_AI_Restrict', 'AI_Familiarity']].corr()

# Display the correlation matrix
print("Correlation Matrix:")
print(correlation_matrix)

# Plot the correlation matrix as a heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1, center=0, fmt='.2f')
plt.title('Correlation Matrix: AI Tool Usage, AI Policy Strictness, and AI Familiarity')
plt.show()
```

The correlation matrix shows that AI familiarity remains the strongest predictor of AI usage (0.53), as students with more exposure to AI tools use them more frequently. The relationship between AI use and course restrictiveness (0.52) is now more neutral, suggesting that students may still engage with AI despite strict policies. The weakest correlation (0.29) indicates that rigid policies do not significantly impact AI familiarity, likely due to prior experience with its usage.
