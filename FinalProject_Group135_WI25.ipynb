# Abstract

In the recent years of 2023-2024, the number of boba shops have been increasing all over the U.S., and this rising trend is especially prominent in the San Francisco Bay Area. We chose to look into this because we were curious what factors were most correlated with such high boba shop densities in some Bay Area cities, and because we did not find any prior studies that specifically investigated the relationship between demographic data and boba shops in the U.S. 

To investigate which demographic factors, if any, best predict the boba shop density in the Bay Area, we used the U.S. Census Bureau demographic data and a dataset of all boba shops in the Bay Area as of 2024. We used both traditional data analysis methods as well as geospatial analysis to investigate our hypothesis that higher percentage of children and teenagers, larger population size, higher percentage of Asian race, more people per household, and higher median income would best predict boba shop density. 

Overall, we found that there generally wasn’t any extremely strong correlation between boba shop density and any of the demographic variables we investigated. However, we can also confidently reject the null hypothesis that boba shops are just randomly distributed through the Bay Area. Some of the demographic factors we chose—percentage of Asian race and population size—moderately predicted the location of boba shops in the Bay Area, while the rest—people per household, median income, and percentage of children and teenagers—had little to no correlation at all. Future research could involve testing other demographic factors we did not look into and using other statistical methods to investigate further.

## Background and Prior Work


Boba drinks, also known as bubble tea, are associated with a category of drinks that add boba as a topping to them. Boba is a small chewy sphere, usually black in color, and can also be referred to as tapioca or pearls.<sup><a href="#ref1">1</a></sup> The most well-known form of the drink is milk tea with boba, but in reality, it can range from slushies to fruit teas. Recently, boba has become so popular, especially with Asian Americans, that it can even be seen as a cultural and aesthetic phenomenon, often a subject of attention on social media.<sup><a href="#ref2">2</a></sup> Naturally, an increase in boba popularity would lead to an increase in boba establishments. In California, the trend of more and more boba shops can be clearly observed over the past few years, and this is especially true in the Bay Area.<sup><a href="#ref2">2</a></sup> As college students, with some of us from the Bay Area, we have frequented many of these establishments, as well as have personally experienced the rise of boba becoming the trendy drink it is today. Because of our personal experiences, we became curious about whether or not there is an underlying trend that can predict where these establishments are appearing.

# Hypothesis



We predict a strong positive correlation between the following variables and density of boba shops (number of boba shops/square mile) in a Bay Area city: higher percentage of children and teenagers, larger population size, higher percentage of Asian race, more people per household, higher median income.

We hypothesize younger ages to be a predictor because boba seems more popular with younger generations and recent trends. The reason for larger population size is that businesses often open up where there can be more customers. The reason for the higher Asian race percentage is that boba originated as an Asian drink, which may mean people of Asian race are more likely to purchase it. The reason for more people per household is similar to younger age: families are likely to have children and teenagers who enjoy boba. The reason for higher median income is that boba can often be expensive and a luxury consumable for people, so those with higher income would be able to afford it more, and in turn, businesses would be more likely to open up in places with customers with disposable income.


### Null Hypothesis (H₀):
There is no correlation between any of the selected demographic variables and boba shop density in Bay Area cities. In other words, boba shop density is randomly distributed and not influenced by population size, age, income, household size, or race.

### Alternative Hypothesis (H₁): (Stated above)
There is a positive correlation between one or more of the following demographic variables and boba shop density in Bay Area cities:
* Higher percentage of children and teenagers
* Larger population size
* Higher percentage of Asian residents
* More people per household
* Higher median household income

### Analysis: Monte Carlo with ANN

First, we used a Monte Carlo test with Average Nearest Neighbor (ANN) to investigate whether or not the distribution of boba shops in the Bay Area was a result of a CSR process.

For this part, we heavily referenced the textbook introduced in class: https://mgimond.github.io/Spatial/hypothesis-testing.html#irpcsr


```python
# Define function to plot Monte Carlo simulated boba shop locations
def plot_points(df, points_gdf, data, cmap):
    f,ax = plt.subplots(1,1, figsize=(15,10), 
    sharex=True, sharey=True, dpi=300)
    f.tight_layout()
    plt.title('Bay Area Cities - Variable = ' + data)
    ax.set_axis_off()
    divider = make_axes_locatable(ax)
    cax = divider.append_axes("right", size="3%", 
    pad=0.5,alpha=0.5)
    df.plot(data, ax=ax, alpha=0.5, cmap=cmap, 
    edgecolor='k', legend=True, cax=cax, linewidth=0.1)
    points_gdf.to_crs(epsg=3857).plot(ax=ax, marker='o', color='red', markersize=2, label="Boba Shop Locations")
    
    # Add city names
    for idx, row in df.iterrows():
        if "NAME" in row.index:
            ax.text(row.geometry.centroid.x, row.geometry.centroid.y, row["NAME"], 
                    fontsize=2, ha="center", color="black")
        else:
            ax.text(row.geometry.centroid.x, row.geometry.centroid.y, row["city"], 
                    fontsize=2, ha="center", color="black")

    plt.ylabel('Square meters', fontsize=12)
    plt.savefig('Bay_Area_Cities_' + data + '.png')
    plt.show()
```

```python
# Define function to compute ANN-Average Nearest Neighbor
def compute_ann(points):
    if len(points) < 2:
        return np.nan  # ANN is not defined for a single point
    
    tree = cKDTree(points)
    distances, _ = tree.query(points, k=2)  # k=2 to exclude self
    return np.mean(distances[:, 1])  # Get the mean nearest neighbor distance

# Compute observed ANN value
observed_ann_value = compute_ann(points)
print(f"Average Nearest Neighbor (ANN) for Boba Shops: {observed_ann_value:.2f} meters")
```

```python
# Define general function for getting demographic per city and cleaning data
def compute_demographic(row, demographic):
    if 'NAME' in row:
        census_row = census_boba_df[census_boba_df['City, Town, CDP'] == row['NAME']]
    elif 'city' in row:
        census_row = census_boba_df[census_boba_df['City, Town, CDP'].str.upper() == row['city']]
    if census_row.empty:
        print('Failed:', row['city'])
        return
    # Use 2020 population census if the 2023 population estimate is blank
    if "population" in demographic.lower() and census_row[demographic].iloc[0] == 'X':
        demographic_val = float(census_row['Population, Census, April 1, 2020'].iloc[0].replace(',', '').replace('$','').replace('+',''))
    elif census_row[demographic].dtype == float:
        demographic_val = float(census_row[demographic].iloc[0])
    else:
        demographic_val = float(census_row[demographic].iloc[0].replace(',', '').replace('$','').replace('+','').replace('%',''))
    return demographic_val
```

```python
def compute_boba_count(row):
    census_row = census_boba_df[census_boba_df['City, Town, CDP'] == row['NAME']]
    if census_row.empty:
        print('Failed')
        return
    return census_row['boba_shop_count'].iloc[0]

bay_area_gdf['boba_shop_count'] = bay_area_gdf.apply(compute_boba_count, axis=1)
bay_area_gdf.head()
```

### Monte Carlo ANN - Random

```python
num_simulations = 100 # Number of Monte Carlo simulations
simulated_anns = []

for i in range(num_simulations):
    random_points = []

    # Generate as many random points as there are boba shops in the bay area
    while len(random_points) < len(boba_df):
        # Randomly generate a point within the bounds of the GeoDataFrame, then use .contains to check if point is actually in the bay area
        p = Point(np.random.uniform(xmin, xmax), np.random.uniform(ymin, ymax)) 
        if bay_area_map_gdf.contains(p).any():
            random_points.append(p)
    random_points = np.array([[p.x, p.y] for p in random_points])
    simulated_anns.append(compute_ann(random_points)) # compute and store ANN of iteration
```

We formed our null hypothesis that boba shops are randomly distributed in the Bay Area, which would mean that the distribution is consistent with a completely random process (CSR). We computed the observed ANN value by using a k-d tree to calculate the nearest neighbor distances between the actual coordinate points of all the boba shops in the Bay Area. We then created a complete spatial randomness (CSR) distribution by randomly sampling a city in the Bay Area and then generating a random point within the boundaries of the selected city for as many times as there are observed boba shops in the Bay Area. We ran this simulation for a total of 100 times and then calculated the ANN value for each one. 

We found the number of simulated ANN values that were greater than our observed ANN value to compute the p-value, which we found to be 0.01. ### Monte Carlo ANN - Population

```python
# Get population per city and clean data
bay_area_map_gdf['population'] = bay_area_map_gdf.apply(compute_demographic, axis=1, args=("Population estimates, July 1, 2023, (V2023)",))
bay_area_gdf['population'] = bay_area_gdf.apply(compute_demographic, axis=1, args=("Population estimates, July 1, 2023, (V2023)",))

# Calculate weights for each city based on population
bay_area_map_gdf["p_weight"] = bay_area_map_gdf["population"] / bay_area_map_gdf["population"].sum()
bay_area_gdf["p_weight"] = bay_area_gdf["population"] / bay_area_gdf["population"].sum()

bay_area_map_gdf.head(1)
```

```python
# Average Nearest Neighbor simulation for cities weighted by population
pop_weighted_simulated_anns = []

for i in range(num_simulations):
    random_points = []
    
    while len(random_points) < len(boba_df):
        # Select a city weighted by population density
        selected_city = bay_area_map_gdf.sample(weights="p_weight").iloc[0]
        
        # Get city boundary
        city_boundary = selected_city.geometry
        
        # Generate a random point within the city's boundary
        while True:
            minx, miny, maxx, maxy = city_boundary.bounds
            p = Point(np.random.uniform(minx, maxx), np.random.uniform(miny, maxy))
            if city_boundary.contains(p):
                random_points.append(p)
                break
    
    # Convert to NumPy array
    random_points = np.array([[p.x, p.y] for p in random_points])
    
    # Compute ANN for this simulation
    pop_weighted_simulated_anns.append(compute_ann(random_points))
```

### Monte Carlo MAE

Aside from a distance-based approach, we also used Monte Carlo simulation with Mean Average Error (MAE) to determine the accuracy of the Monte Carlo simulation with sampling weighted based on a demographic factor, where the smaller the MAE, the more accurate the simulation was. Then, we compared its accuracy to a baseline Monte Carlo simulation with completely random sampling to see how well the weighted simulation is performing. With this methodology, we were able to determine whether or not any one specific variable independently had a strong influence on boba density. 

The decision to use MAE rather than Mean Squared Error (MSE) was because we noticed several outliers in our dataset early on in our exploratory data analysis, such as San Jose, and we did not want to penalize too heavily for these outliers. 

Furthermore, we believe a simulation with random sampling would be a good baseline to compare to because if a weighted simulation performed significantly, or even moderately, better than random simulation, that would indicate there is some degree correlation between boba shop density and the respective variable used to weight the simulation. 


```python
# However, we can't just randomly place boba shop locations within the city, as there are likely structural reasons, etc. for the city 
# that would result in boba shops clumping in certain locations. Therefore, we can do a simpler test where we only check the distribution
# for boba shop density for each city based on variables like population, median income, etc.

# Baseline model with random points for boba shop density
num_simulations = 100
boba_density_rand_simulation_list = []

for i in range(num_simulations):
    random_points = 0
    rand_simulated_boba_density = defaultdict(int)
    
    while random_points < len(boba_df):
        # Select a random city 
        selected_city = bay_area_gdf.sample().iloc[0]
        rand_simulated_boba_density[selected_city['NAME']] += 1
        random_points += 1
        
    for (key,value) in rand_simulated_boba_density.items():
            census_row = census_boba_df[census_boba_df['City, Town, CDP'] == key]
            if census_row.empty:
                print('Failed')
                break
            land_square_meters = census_row['Land area in square miles, 2020'].iloc[0] * 2.59e+6
            rand_simulated_boba_density[key] = value / land_square_meters
        
    boba_density_rand_simulation_list.append(rand_simulated_boba_density)
```

```python
# Get the observed density for each city
density_observed = defaultdict(int)
for b, n in zip(bay_area_gdf['b_density'].values, bay_area_gdf['NAME'].values):
        density_observed[n] = b

# Get the observed count for each city
count_observed = defaultdict(int)
for b, n in zip(bay_area_gdf['boba_shop_count'].values, bay_area_gdf['NAME'].values):
        count_observed[n] = b

sorted_cities = sorted(bay_area_gdf['NAME'].tolist())
num_cities = len(sorted_cities)
```

```python
# General function for running monte carlo simulation based on weighted sampling
def monte_carlo_weighted_sampling(demographic_name, demographic_col):
    num_simulations = 100
    simulation_list = [] # list of dictionaries
    count_simulated = defaultdict(int)

    for i in range(num_simulations):
        random_points = 0
        demographic_weighted_simulated_boba_count = defaultdict(int)

        while random_points < len(boba_df):
            # Select a city weighted by demographic
            selected_city = bay_area_gdf.sample(weights=(demographic_name + "_weight")).iloc[0]
            demographic_weighted_simulated_boba_count[selected_city['NAME']] += 1
            random_points += 1
        
        for (key,value) in demographic_weighted_simulated_boba_count.items():
            census_row = census_boba_df[census_boba_df['City, Town, CDP'] == key]
            if census_row.empty:
                print('Failed')
                return
            count_simulated[key] = value # For graphing later (not a part of calculation)
            land_square_meters = census_row['Land area in square miles, 2020'].iloc[0] * 2.59e+6
            demographic_weighted_simulated_boba_count[key] = value / land_square_meters
            
        count_simulated[key] /= num_simulations
        simulation_list.append(demographic_weighted_simulated_boba_count)
        
    return simulation_list, count_simulated
```

```python
def calculate_MAE(sim_list):
    mae_list = defaultdict(int)

    # Calculate the MAE of the simulations compared to the observed for each city
    for city in sorted_cities:
        total = 0
        for simulated in sim_list:
            total += abs(density_observed[city] - simulated[city])
        city_avg = total / num_simulations
        mae_list[city] = city_avg
        
    return mae_list
```

```python
# General function for calculating MAE and printing averaged MAE
def calculate_MAE_averages(name, simulation_list, count_simulated):
    mae_list = defaultdict(int)

    # Calculate the MAE of the simulations compared to the observed for each city
    mae_list = calculate_MAE(simulation_list)
    
    # Calculate the MAE for the baseline
    boba_density_rand_mae_list = calculate_MAE(boba_density_rand_simulation_list)
    
    # Calculate averaged MAE and compare to baseline
    count_range = max(bay_area_gdf['b_density'].values) - min(bay_area_gdf['b_density'].values)

    mae_average = sum(mae_list.values()) / num_cities
    print('Average MAE:', mae_average)

    relative_mae_average = (sum(mae_list.values()) / num_cities) / (count_range)
    print('Relative Average MAE:', relative_mae_average)

    baseline = sum(boba_density_rand_mae_list.values()) / num_cities
    print('Baseline MAE:', baseline)

    relative_baseline = (sum(boba_density_rand_mae_list.values()) / num_cities) / (count_range)
    print('Relative Average MAE:', relative_baseline)

    # Plot histogram
    hist = sns.histplot(data=mae_list.values(), bins=20, kde=True, label='Demographic Simulated MAE Distribution')
    hist.set_title("Distribution of Expected MAE Values")
    hist.set_xlabel("MAE Values")
    plt.savefig("monte_carlo_plots/" + name + "_hist.png")
    plt.show()

    # Plot bar graph
    values_obs = [count_observed[city] for city in sorted_cities]
    values_sim = [count_simulated[city] for city in sorted_cities]
    
    y = np.arange(len(sorted_cities)) 
    width = 0.35 
    
    plt.figure(figsize=(12, 20))
    plt.yticks(y, sorted_cities, fontsize=10)
    
    # Plot bars side by side by shifting x-coordinates
    plt.barh(y - width/2, values_obs, height=width, label='Actual Counts', color='#1f77b4')
    plt.barh(y + width/2, values_sim, height=width, label='Average Expected Counts', color='orange')
    
    plt.xlabel("Boba Shop Counts")
    plt.ylabel("City")
    plt.legend()
    plt.title("Actual vs. Expected Boba Shop Counts per City")
    plt.savefig("monte_carlo_plots/Actual_vs_Expected_" + name + "_Count.png")
    plt.show()
```

```python
# General function 
def calculate_monte_carlo_MAE(demographic_name, demographic_col):
    # Create new column in gdfs for new variable using compute_demographic
    bay_area_map_gdf[demographic_name] = bay_area_map_gdf.apply(compute_demographic, axis=1, args=(demographic_col,))
    bay_area_gdf[demographic_name] = bay_area_gdf.apply(compute_demographic, axis=1, args=(demographic_col,))
    
    # Create weight column in gdfs for weighting
    bay_area_map_gdf[demographic_name + "_weight"] = bay_area_map_gdf[demographic_name] / bay_area_map_gdf[demographic_name].sum()
    bay_area_gdf[demographic_name + "_weight"] = bay_area_gdf[demographic_name] / bay_area_gdf[demographic_name].sum()
    
    # Run monte carlo simulation based on weighted sampling
    simulation_list, count_simulated = monte_carlo_weighted_sampling(demographic_name, demographic_col)
    
    # MAE calculations and output
    calculate_MAE_averages(demographic_name, simulation_list, count_simulated)
    
```

Again, this number is extremely small, similar to the MAE difference for median income and children/teenager percentage, which were both under 0.020. In turn, this means there is little to no correlation between the average household size and the boba shop density for a city.

As both the histogram and the count bar graph look very similar to median income and children/teenager percentage, we can summarize that both of these figures indicate there is a lack of correlation. With the graph we can clearly see that most of the distinct features of the observed data were not really captured, especially for the outliers, and there was little variation in the simulated boba counts between each city.

# Discussion

First, using the Monte Carlo technique with ANN for hypothesis testing, we rejected our null hypothesis that the distribution of boba shops in the Bay Area is due to complete spatial randomness. This intuitively makes sense because, obviously boba shop owners do not decide the location of their shop completely randomly. This is why we then decided to use demographic data such as population and median income as weights; real-world processes do not follow CSR. When using either the estimated population size or the median income as the weight for generating a simulated distribution, we were able to reject the null hypothesis that either were the sole determinant of the boba shop distribution in the Bay Area. These results also make sense because it is unlikely for solely population size or solely median income to account for the distribution of boba shops in the Bay Area. 

Additionally, we can't just randomly place boba shop locations anywhere within the boundaries city, as there are likely structural reasons in the city and such that would result in boba shops clustering in certain locations (they have to be somewhere zoned for retail). Therefore, we decided to also do a non-distance based test, where we simply checked the distribution of boba shop density across each city based on demographic variables like population density, median income, Asian population percentage, children/teenager percentage, etc. In order to determine the accuracy of these non-distance based tests, we used the metric of Mean Average Error (MAE) and measured the loss of the simulation compared to the observed values. The smaller the loss, the more accurate the simulation was. Overall, we were able to obtain some meaningful data on the correlation between boba shop density and the demographics we tested on. The demographics with little to no correlation demonstrated a MAE value that was extremely similar to the baseline simulation of random count distribution, while those with some correlation had significantly smaller MAE values compared to the baseline.

A huge influence on our datasets were the existence of two primary outliers that definitely cannot be removed: San Francisco and San Jose. These two cities both have much higher boba shop counts compared to any other city, and from the count bar graphs we can see that if the distribution can account for these outliers to some extent, then they would already be a significant improvement over random ditributions. That's why we believe poopulation size has the strongest correlation out of all the demographics we tested, as both San Jose and San Francisco have very high populations. 

# Conclusion

Overall, we combined two data sets—one on boba shops in the Bay Area and one on census data on cities, towns and CDPs in the Bay Area to investigate what demographic would be the best predictor for boba shop density.

Among all predictors analyzed, the percentage of Asian residents showed the strongest positive correlation with boba shop density, regardless of total population or wealth, which supports the idea that boba, an Asian drink, would appeal more to Asian customers. Also as expected, population size showed a moderate positive correlation with boba density because higher population may mean more potential customers. Median household income showed a somewhat positive correlation with boba density. One possible reasoning could be that cities where people with an average higher income might be more likely to support a higher number of boba shops due to increased consumer spending, greater commercial infrastructure, or lifestyle preferences. However, boba is not restricted to wealthy areas as several medium-income cities also show high density, suggesting that boba has a cross-class appeal. 

From our inferential analysis with Monte Carlo simulations with ANN, we rejected the null hypothesis that boba shops are completely randomly distributed in the Bay Area, and we also reject the null hypothesis that boba shops are randomly distributed relative to population size or income. Also, further analysis with Monte Carlo simulations with MAE suggest that there appears to be correlations between boba shop density and two of the demographic variables that we had tested, the two variables being population size and Asian percentage. The remaining three variable were essentially no better at predicting boba shop density than pure random guessing. This was also mostly in-line with our assumptions made from the plots from the exploratory data analysis, which also showed the most correlation for Asian percentage and population. Median income unexpectedly showed very little to no correlation using Monte Carlo, contrary to our results in the exploratory data analysis. This may be due to more complicated patterns that Monte Carlo with MAE may not have captured; for instance, income may vary drastically across a city. 

To conclude, besides possibly Asian percentage, we did not find any strong evidence for any demographic factor being a predictor for boba shop density in the Bay Area. However, a limitation of our methodology was that we could not use Monte Carlo with ANN to exactly replicate our intended null hypothesis since boba shops are not randomly distributed throughout a city, such as due to structural reasons. As stated previously in the background, there is no prior study focusing on the relationship between demographic and boba shop density in the U.S., so we hope our work inspires future investigation in this subject. Further research should be conducted with other statistical methods (e.g. LLS, Poisson point process model) to investigate further, and perhaps consider some other demographic factors that we did not look into in this project.
