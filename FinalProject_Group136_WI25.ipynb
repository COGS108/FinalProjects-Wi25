# Abstract

Our study investigated the relationship between the emotional sentiment of lyrics across the ten most popular languages. Music serves as a language that transcends languages, cultures, and different political views and identities. Our objective with this project was to determine what the sentiment of different songs in the top ten languages in our dataset looked like. Using sentiment analysis, a natural language processing (NLP) model that assigns sentiment to text and TF-IDF (term Frequency-Inverse Document Frequency), a tool that unfilters insignificant words, our goal was to determine what kinds of lyrics or words in songs appear across all ten languages (English, French, Russian, Spanish, German, Italian, Turkish, Polish, Portuguese, and Korean) from a random sample of 50 songs from each language, forming a bigger sample of 500 songs. These top ten most popular languages were found by processing our 9GB dataset and identifying the languages based on view counts. 

The lyrics from our sample of 500 songs wer translated into English using GoogleTranslator to allow for Sentiment Analysis via VADER (Valence Aware Dictionary and sEntiment Reasoner). We cleaned and tidied this data to then apply TF-IDF. We found that Spanish, French, and Korean had high median positive compound scores (the polarity score for sentiment); English, Polish, Italian, and Portuguese had moderate median positive compound scores; and German, Russian, and Turkish had moderate median negative compound scores. We also found that positive sentiment correlated with higher popularity (more views) across most languages and songs with negative sentiment were less popular. These findings indicate that cultural preferences and linguistic characteristics shape a song’s sentiment and thus its popularity.

As shown in the lyrics above, there are several words/phrases that needs to be cleaned, such as \n, brackets, etc. To prevent translation error, we will first remove all the paratheses and brackets, as well as the words in them, which are just descriptions.

```python
import re
removed = []
for lyrics in song_samples['lyrics']:
    removed_lyrics = re.sub(r'\[.*?\]', '', lyrics)
    removed_lyrics = re.sub(r"\(.*?\)", "", removed_lyrics)
    removed.append(removed_lyrics)
song_samples['lyrics'] = removed
song_samples.lyrics[0]
```

Since we are only using an English lexicon, we need to translate all the non-English lyrics into English for sentiment analysis. We achieved this by importing GoogleTranslator. We do this before removing \n because we need \n to help split the lyrics into chunks for translation; some lyrics exceed the 5000-character limit per request for GoogleTranslator.

```python
# Import GoogleTranslator for translating lyrics
from deep_translator import GoogleTranslator
```

```python
# A function (written by ChatGPT) that splits long lyrics into chunks (each with at most 5000 char.)
def split_text(text, max_chars=5000):
    """Split text into chunks under max_chars, ideally at line breaks."""
    lines = text.split('\n')
    chunks = []
    current_chunk = ''
    
    for line in lines:
        if len(current_chunk) + len(line) + 1 <= max_chars:
            current_chunk += line + '\n'
        else:
            chunks.append(current_chunk.strip())
            current_chunk = line + '\n'
    
    if current_chunk:
        chunks.append(current_chunk.strip())
    
    return chunks

# A function that translates long lyrics by chunks
def translate_large_text(text):
    chunks = split_text(text)
    translated_chunks = []
    for chunk in chunks:
        translated = GoogleTranslator(source='auto', target='en').translate(chunk)
        translated_chunks.append(translated)
    return '\n'.join(translated_chunks)
```

```python
# Translate all the song lyrics in the song_samples dataframe
translated_lyrics = []
for lyrics in song_samples['lyrics']:
    translated = translate_large_text(lyrics)
    translated_lyrics.append(translated)
```

```python
# Take a look at one of the translated lyrics
translated_lyrics[0]
```

From the translated lyrics sample above, we can see that the translation contains labels in brackets, \n, and other punctuation marks. We will keep the lyrics' punctuations and capitalizations since Vader considers them in the sentiment score calculation. We will create a column with translated song lyrics with only bracketed text, \n, , and extra spaces removed. However, for word frequency analysis, we need to tokenize and stem the lyrics, which requires removing all the punctuations and transforming the lyrics to lowercase. Thus, we added another column for translated song lyrics with only lowercase letters and no punctuation.

```python
# Import packages to preprocess the translated song lyrics
import re
import string
```

```python
# Clean up and preprocess the translated song lyrics
lyrics_with_punc_list = []
lyrics_without_punc_list = []
for lyrics in translated_lyrics:
    lyrics = lyrics.replace('\n', ' ')
    lyrics = lyrics.replace('  ', ' ')
    lyrics = lyrics.replace('\\', '')
    lyrics = lyrics.strip()
    lyrics_with_punc_list.append(lyrics)
    lyrics = lyrics.lower()
    for punc in string.punctuation:
        lyrics = lyrics.replace(punc, '')
    lyrics_without_punc_list.append(lyrics)
```

```python
# Take a look at one of the preprocessed, translated song lyrics
print(f'Lyrics with punctuation:', lyrics_with_punc_list[0])
print(f'Lyrics without punctuation:', lyrics_without_punc_list[0])
```

```python
# Add the preprocessed, translated song lyrics in to the song_samples dataframe
song_samples['lyrics_with_punc'] = lyrics_with_punc_list
song_samples['lyrics_without_punc'] = lyrics_without_punc_list
song_samples
```

Lastly, we stem the words so that variants of the same word are reduced to a common root (ex. both 'babies' and 'baby' becomes the root 'babi'), enabling words with the same root to be recognized as the same word in frequency analysis.

### EDA: Sentiment Analysis of Song Lyrics From the Top 10 Most Popular Song Languages

Before performing TF-IDF, we will like to investigate the sentiment of the lyrics from the top 10 most popular song languages. Sentiment analysis is the process of computationally categorizing the attitude, opinions, or emotion expressed in a piece of text, which often contains three categories positive, negative, and neutral. To achieve this, words in a piece of text are matched to a lexicon, a table containing sentiment scores (often normalized to between -1 and 1) of many different words, to determine whether a word is positive, negative, or neutral. In our case, each word in a lyric will be associated with a sentiment score, and the overall sentiment of a lyric is obtained by considering sentiment scores of all words in the lyrics. 

As the first step, we first import the sentiment analyzer and download the vader lexicon (a package in nltk that contains english words/punctuations and their sentiment scores) for sentiment analysis.

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer 
nltk.download('vader_lexicon')
```

After contructing the sentiment analyzer, we preprocess the lyrics with punctuation (since Vader considers punctuation and exclamation marks) into word lists for each song. We also need to create a function that calculates the metrics (polarity scores) for sentiments (negative, positive, neutral, and the compound score). The negative, positive, and neutral polarity scores are just binary numbers with 1 indicating that the word has the corresponding sentiment and 0 indicating that it is not within the category. The compound score is a metric that measures the normalized sum of all lexicon ratings in a value between -1 and 1. A compound score >= 0.05 (the threshold) represents words/song lyrics with a positive sentiment, while a compound score <= -0.05 represents words/song lyrics with a negative sentiment. A compound score of -0.05 to 0.05, all inclusive, indicates that the song lyrics has neutral sentiment. After the calculation of metrics, we organize the obtained scores into a dataframe along with the titles, languages, and number of views for the songs. The resulting dataframe is shown below.

```python
analyzer = SentimentIntensityAnalyzer()
```

```python
language_lst = song_samples['language'].unique()
lyrics_lst = []
for lang in language_lst:
    lang_df = song_samples[song_samples['language'] == lang]
    lyrics_lang = list(lang_df['lyrics_with_punc'].values)
    lyrics_lst.append(lyrics_lang)
```

```python
def sentiment_scores(lst):
    sentiment_df = pd.DataFrame()
    for lyrics in lst:
        polarity = analyzer.polarity_scores(lyrics)
        polarity['lyrics'] = lyrics
        sentiment_df = pd.concat([sentiment_df, pd.DataFrame([polarity])], ignore_index=True)
    return sentiment_df
```

### EDA: Term Frequency - Inverse Document Frequency (TF-IDF) Analysis

After sentiment analysis, we will investigate which words are unique and important for song lyrics of different languages in this section using Term Frequency-Inverse Document Frequency (TF-IDF), a numerical statistic measuring how important a word is to a document relative to a set of documents. TF-IDF is the product of two metrics, term frequency (TF) and inverse document frequency (IDF). TF measures how frequent a term appears in a document, and IDF measures how rare a term appears across all documents, giving less weight to common words. A high TF-IDF score will indicate that the word is appears frequently in the document but is rare across all documents, implying that the word is highly relevant and important. Thus, through TF-IDF, we can discover what words are important and how relevant are they in song lyrics from each of the top ten most popular song languages. Important and relevant words with high TF-IDF in song lyrics of each language might be words that affects the sentiment of song lyrics of different languages the most, contributing to the difference in sentiment of song lyrics across the ten languages.

First, we import packages needed to conduct TF-IDF analysis. The TF-IDF scores of all words will be extracted using TfidfVectorizer.

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from nltk.tokenize import word_tokenize
```

Then, we extract the top ten most popular song langauges we used for analysis in this project.

Next, just to explore what words are important for each songs in our sample, we calculated the TF-IDF score for each word in each song and append the word with the highest TF-IDF score to the song sample dataframe. As shown below, the most relevant/important word for the song Sepia zu Gold is "fans", and the most important word for the song Was Geht is word "loud". Those words that has the highest importance for each song potentially convey what the songs are mainly about as well as the sentiment of the song.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize TF-IDF Vectorizer
tfidfUnique = TfidfVectorizer(
    sublinear_tf=True, 
    analyzer='word', 
    max_features=2000, 
    tokenizer=word_tokenize, 
    stop_words=stopwords.words("english")
)

# Convert lyrics to list
text = song_samples['lyrics_without_punc'].tolist()

# Fit TF-IDF and create DataFrame
song_tfidf = pd.DataFrame(
    tfidfUnique.fit_transform(text).toarray(), 
    index=song_samples['title'], 
    columns=tfidfUnique.get_feature_names_out()
)

# Find the most unique word per song
most_unique = song_tfidf.idxmax(axis=1)

# Add column to DataFrame
song_samples['most_unique'] = most_unique.tolist()

# Display the result
song_samples.head(5)
```

After confirming that our method for extracting TF-IDF scores works on each song lyric, we then use the same method to first extract the TF-IDF of words in song lyrics grouped by languages. Using the TF-IDF scores, we constructed a dataframe that contains the top 30 words with the highest TF-IDF scores for each language. These words have the highest importance and relevance for lyrics in the corresponding language.

```python
# Define any custom filler words if needed
custom_filler = [
    "im", "dont", "aint", "gonna", "wanna", "gotta",
    "cant", "couldnt", "shouldnt", "wouldnt", "didnt", 
    "doesnt", "havent", "isnt", "wasnt", "werent",
    "oh", "ooh", "uh", "ah", "ha", "hey", "ho", "woah", "whoa", 
    "la", "nah", "da", "huh", "mm", "mmm", "’", "go",
    "like", "yeah", "ya", "cause", "cause'", "aint", "uhh"
]
stop_words = stopwords.words("english") + custom_filler

# Store results
results = []

# Process for each language
languages = song_samples["language"].unique()
for lang in languages:
    # 1. Filter DataFrame by language
    df_lang = song_samples[song_samples["language"] == lang]
    
    # 2. Initialize TF-IDF Vectorizer
    tfidf = TfidfVectorizer(
        sublinear_tf=True,
        analyzer='word',
        max_features=3000,
        tokenizer=word_tokenize, 
        stop_words=stop_words
    )
    
    # 3. Compute TF-IDF
    tfidf_matrix = tfidf.fit_transform(df_lang['lyrics_without_punc'].astype(str))
    feature_names = tfidf.get_feature_names_out()
    
    # 4. Sum up TF-IDF scores across all songs in this language
    sums = tfidf_matrix.sum(axis=0).A1
    
    # 5. Identify top words
    sorted_indices = np.argsort(sums)[::-1]  # Sort in descending order
    top_indices = sorted_indices[:30]  # Get top 10 words

    # 6. Store results
    for rank, idx in enumerate(top_indices, start=1):
        results.append((lang, rank, feature_names[idx], sums[idx]))

# 7. Create DataFrame of top words per language
df_top_words_per_lang = pd.DataFrame(results, columns=["language", "rank", "word", "tfidf_sum"])
df_top_words_per_lang.sort_values(["language", "rank"], inplace=True)
df_top_words_per_lang
```

```python
# this code is from chatgpt to save time of process

languages = song_samples["language"].unique()
results = []

# Define any custom filler words if needed
custom_filler = [
    "im", "dont", "aint", "gonna", "wanna", "gotta",
    "cant", "couldnt", "shouldnt", "wouldnt", "didnt", 
    "doesnt", "havent", "isnt", "wasnt", "werent",
    "oh", "ooh", "uh", "ah", "ha", "hey", "ho", "woah", "whoa", 
    "la", "nah", "da", "huh", "mm", "mmm", "’", "go",
    "like", "yeah", "ya", "cause", "cause'", "aint", "uhh"
]

stop_words = stopwords.words("english") + custom_filler

for lang in languages:
    # 1. Subset the DataFrame to just this language
    df_lang = song_samples[song_samples["language"] == lang]
    
    # 2. Compute TF-IDF for these lyrics
    tfidf = TfidfVectorizer(
        sublinear_tf=True, 
        analyzer='word', 
        max_features=3000, 
        tokenizer=word_tokenize, 
        stop_words=stop_words
    )
    tfidf_matrix = tfidf.fit_transform(df_lang["lyrics_without_punc"].astype(str))
    
    # 3. Check if "love" is in the vocabulary
    feature_names = tfidf.get_feature_names_out()
    if "love" in feature_names:
        # 3A. Find the index of "love" and sum TF-IDF across all docs
        love_idx = np.where(feature_names == "love")[0][0]
        sums = tfidf_matrix.sum(axis=0).A1  # sum TF-IDF for each feature
        love_tfidf_sum = sums[love_idx]
    else:
        # 3B. If "love" not present, set TF-IDF to 0.0
        love_tfidf_sum = 0.0
    
    # 4. Store the result for this language
    results.append((lang, love_tfidf_sum))

# Create a DataFrame of TF-IDF for "love" per language
df_love = pd.DataFrame(results, columns=["language", "tfidf_love"]).sort_values("tfidf_love", ascending=False)

df_love
```

Our analysis focused on whether the sentiment of song lyrics differs across songs in the top nine most popular languages, including English, Spanish, French, Russian, etc. By analyzing over 500 songs and conducting sentiment analysis through Natural Language Processing (NLP), we wanted to uncover patterns in emotional tones, with the hypothesis that Western languages may exhibit more positive sentiments – due to these cultures’ emphasis on emotional expression, openness, and communication –, while Eastern languages might reflect more neutral or melancholic tones, as we hypothesized that these cultures might have a bigger interest in melancholic themes like resilience, duty, loyalty, and identity, which would give song lyrics more neutral or somber sentiments. Our project extends previous studies on sentiment analysis in music, deepening our understanding of the intersection of culture, language, and emotional expression. We used Google Translate to normalize lyrics into English and the VADER lexicon to assess their sentiment, which allowed us to have a robust foundation for cross-lingual comparison. The following sections discuss the results in relation to cultural expectations, linguistic differences, and previous research.

### Sentiment Compound Scores and Consistensy with Hypothesis
From our sentiment analysis and the boxplot of compound sentiment scores, we found that Spanish, French, and Korean lyrics had high median positive compound scores. This suggests that these languages used more positive emotional expression. English, Polish, Italian, and Portuguese lyrics showed a more moderate positive sentiment, as their median scores were around 0.5. German, Russian, and Turkish lyrics had moderate negative median compound scores, demonstrating a more negative sentiment in the lyrics. These results were generally consistent with our Hypothesis. Our Hypothesis stated that Western Languages would exhibit a more positive sentiment. This was confirmed by Spanish, French, English, Italian, and Polish having moderate to high median positive sentiment compound scores. Korean, surprisingly, had a high positive sentiment. We hypothesized that Eastern languages like Korean, Russian, and Turkish would have neutral to negative sentiment in their lyrics. This was true for Russian and Turkish. However, German also had more negative sentiment. 

Words or lyrics that had a positive sentiment attached included “love”, “heart”, and “life”. These words generally evoke positive emotions in song lyrics. They are emotions that the audience can afix their own meaning to and connect to the song that much more deeply. Frequently occurring lyrics with negative sentiment included “blood” and many swear words that elicit stronger negative emotions. These words evoke imagery of violence and pain which can trigger negative emotions in the audience. In the same vain, swear words are usually associated with anger or frustration, which emphasizes the intensity of the negative emotions in those lyrics. Songs with either a strong positive sentiment or strong negative sentiment showed a variation in the number of views; they exhibited compound scores near -1 or 1, which is pretty bimodal. This means that even if a song has an extreme emotion, it is usually up to the audience to decide if they resonate with it or not. Some emotionally intense songs can garner niche popularity, but might not be widely relatable. Neutral words like “will” and “want” appeared frequently across both positive and negative sentiment lyrics, indicating that they have manh functions in the way that they evoke emotions. Songs with neutral sentiment, contrastingly, had fewer views, suggesting that neutral lyrics are less relatable or engaging.    

Our analysis revealed a weak positive correlation between the prevalence of the word “love” and overall song sentiment, which is intuitive. Languages that emphasize love in their lyrics tend to have slightly more positive emotions. Conversely, a more substantial negative correlation existed between the use of the word “f***” and song sentiment, indicating that languages that have more occurrences of this word in their songs demonstrate more negative emotional tone. These findings highlight the complexity of emotional expression in music.

### TF-IDF and Sentiment Analysis Surprising Findings
Sentiment analysis is a natural language processing (NLP) technique that is used to categorize text into emotional tone into classes such as positive, negative, or neutral. It can assign this sentiment score to phrases of text or words, which helps to reveal emotional intent. However, when analyzing the word “love”, for example, sentiment analysis categorizes the word as strongly positive as it prompts positive emotions like warmth and affection. But, the word love can have different meanings depending on the context. For example, the sentence “I love this” has a positive connotation, but “I don’t love that anymore” has a mixed sentiment. So, TF-IDF (Term Frequency-Inverse Document Frequency) helps us to filter out contectually insignificant words like “the”, “of”, and “it” and measure how often a word occurs in a text. When TF-IDF is used in conjunction with Sentiment Analysis, it allows for a more meaningful understanding of words like “love” and how they are used, whether as a term of endearment or affection, or as a way of expressing frustration. TF-IDF emphasizes the significance of the lyric or word, in this case “love”, and sentiment analysis assesses whether it has positive, negative, or neutral sentiment. 

One unexpected finding was the negative sentiment of Turkish lyrics despite their high TF-IDF scores. The mean negative sentiment compound score for the Turkish lyrics was -0.221, versus the high TF-IDF score for “love” being 2.026. This relationship is weak and the reason for that may be that the sentiment analysis model struggled to capture the nuanced emotional tone that the word “love” brings. The Turkish language could also have some regional-specific linguistic/idiomatic expressions or cultural references that can’t be directly analyzed by our sentiment analysis model. The high TF-IDF score for Turkish could also be because it only accounts for the number of occurrences of the word without taking into account the emotional context or meaning of the word “love” within a set of lyrics or song. Moreover, sentiment models that are trained on primarily English may not translate as seamlessly in other languages and perform as accurately. This can cause oversimplification of some emotions in languages where sentiments associated to “love” are expressed through specific vernacular. 


### Limitations and Implications
Another limitation our analysis faced has to do with the normalization from other languages to English. Translating lyrics using Google Translate did not always produce accurate results, and in some cases the nuances and cultural context were inevitably lost. Some specific words did not get translated, such as the word “que” in Spanish, which remained unvaried in the translation despite having an English equivalent, and other translations did not make sense, like the word “you” being translated to the letter “u”. In those cases, whenever we noticed such errors, we added code to ensure that these words would get accurate translations. 

Future research could implement our analysis by employing multilingual sentiment models that are trained on diverse datasets in different languages. While the tool we used, VADER, is designed to work with English text – and English is the language that worked for the scope of our research – future projects could include analyses in different languages without needing to translate the lyrics beforehand, which would help maintain emotional accuracy and nuance for each language. Expanding and deepening the research can also allow for multiple sub-languages (dialects) to be considered in the analysis, providing different emotional patterns not only between languages, but also within them. 

### Main Takeaways
Our analysis provides a new understanding of how language and culture influence emotional expression in music. Our analysis partially confirmed our hypothesis that linked Western music with a more positive sentiment and Eastern and Central European music with a generally more negative or neutral one, with the exception of Korean music, which resulted in a more positive sentiment score, and German, which showed a more negative approach to music. These examples of unexpected results could be an indication of how industry-specific demands and cultural factors also play a significant role in shaping lyrical sentiment.
