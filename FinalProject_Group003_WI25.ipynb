# Abstract

This project explores the relationship between the economic state of the United States and the musical attributes of popular songs from 1975 to 2019. Specifically, we investigate how features like acousticness, danceability, instrumentalness, tempo, and valence change during periods of economic recession versus economic growth. By leveraging datasets from Kaggle for song attributes and economic indicators such as unemployment rates, oil prices, housing prices, and GDP, we aim to determine if economic conditions correlate with shifts in musical characteristics.

Our approach integrates song popularity metrics with historical economic data, enabling a comparative analysis across economic cycles. Through statistical and visual analysis, we examine trends and patterns to understand whether societal mood, influenced by economic hardship or prosperity, is reflected in the music people consume.

Ultimately, our findings show no clear correlation between economic conditions and changes in musical attributes. Despite previous works, the data suggests that popular music trends may evolve independently of economic factors, potentially indicating that other cultural or industry-driven influences likely play a more significant role.

# Hypothesis


We hypothesize that the audio features of generally popular songs during certain time periods will reflect shifts in the U.S. economy during those time periods. Specifically:

* During periods of economic prosperity and social stability, songs with upbeat characteristics such as high danceability (indicating a rhythm that encourages movement), low instrumentalness (suggesting a focus on vocals), higher tempo (faster beats per minute), lower acousticness (less singing, more beats, higher energy), and other factors will be more prevalent.

* Conversely, during economic downturns or times of social unrest, we expect songs with lower danceability (indicating slower rhythms), higher instrumentalness (suggesting a greater emphasis on background instruments or less focus on vocals), slower tempo (lower beats per minute), higher acousticness (more singing, less beats, lower energy), and other factors to be more prevalent.

This hypothesis is grounded in existing research on music and emotions, where faster tempos and major keys are associated with positive emotions, while slower tempos and minor keys often align with sadness or introspection. We predict that we will find a strong correlation between these audio features and U.S. economic conditions of the time periods under study.

Valence: A measure from 0.0 to 1.0 describing the musical positiveness conveyed by a track. Tracks with high valence sound more positive (e.g. happy, cheerful, euphoric), while tracks with low valence sound more negative (e.g. sad, depressed, angry).

Acousticness: A confidence measure from 0.0 to 1.0 of whether the track is acoustic. 1.0 represents high confidence the track is acoustic.

Danceability: Danceability describes how suitable a track is for dancing based on a combination of musical elements including tempo, rhythm stability, beat strength, and overall regularity. A value of 0.0 is least danceable and 1.0 is most danceable.

Energy: Energy is a measure from 0.0 to 1.0 and represents a perceptual measure of intensity and activity. Typically, energetic tracks feel fast, loud, and noisy. For example, death metal has high energy, while a Bach prelude scores low on the scale. Perceptual features contributing to this attribute include dynamic range, perceived loudness, timbre, onset rate, and general entropy.

Instrumentalness: Predicts whether a track contains no vocals. "Ooh" and "aah" sounds are treated as instrumental in this context. Rap or spoken word tracks are clearly "vocal". The closer the instrumentalness value is to 1.0, the greater likelihood the track contains no vocal content. Values above 0.5 are intended to represent instrumental tracks, but confidence is higher as the value approaches 1.0.

Loudness: The overall loudness of a track in decibels (dB). Loudness values are averaged across the entire track and are useful for comparing relative loudness of tracks. Loudness is the quality of a sound that is the primary psychological correlate of physical strength (amplitude) and the average loudness in decibels (dB) across the entire track is what was being recorded. A score of 1.0 resembles an overall "louder" song where a score of 0.0 resembles a "quieter" song.

Sadness: The sadness score was calculated on using a combination of Natural Language Processing techniques, Latent Dirichlet Allocation (LDA) topic modeling, and word embedddings on a song's lyrics. With these tools, the score was calculated on undergoing sentiment analysis and analysing how "sad" a song was by their word choice. A score of 1.0 resembles a higher pressence of "sadness" in the song whereas a score of 0.0 resembles a low/no presence of "sadness". (Externally calculated and provided within the dataset)

Feelings: The feelings score was calculated similarly to the sadness score but instead of only focusing on whether or not a song was just sad, it analyses a broader range of emotions. Each emotion had a weight and after undergoing sentiment analysis, the feeling score was calculated based on the percentage of pressence of each emotion in the song. A score of 1.0 resembles a larger quantity of "feelingness" in the song whereas a score of 0.0 represents the opposite. (Externally calculated and provided within the dataset)

```python
# Standardize Dates
housing['Year'] = pd.to_datetime(housing['Date'], errors='coerce')
housing['Year'] = housing['Year'].dt.year
```

```python
# Calcute Percentage Changes for remainin Columns
housing['House_Price_Index_Change'] = housing['House_Price_Index'].pct_change() * 100
housing['Real_GDP_Change'] = housing['Real_GDP'].pct_change() * 100
housing['Unemployment_Rate_Change'] = housing['Unemployment_Rate'].pct_change() * 100
housing['Mortgage_Rate_Change'] = housing['Mortgage_Rate'].pct_change() * 100
housing = housing[sorted(housing.columns)]

# Trim dataset
housing = housing[housing['Year'] < 2021]
housing.head()
```

```python
# Calculate Percentage Change
# Prevent automatic silent downcasting of data types in future versions (Previous warning).
pd.set_option('future.no_silent_downcasting', True)
yearly_sum = us_oil_prices.groupby('Year')['Oil_Price'].mean().reset_index()
yearly_sum['Oil_Price_Change'] = yearly_sum['Oil_Price'].pct_change() * 100
yearly_sum = yearly_sum[yearly_sum['Year'] < 2021]
yearly_sum.head()
```

## Combine U.S. Oil Prices with U.S. Housing Market Factors and compute economic misery index (EMI)

```python
economic_factors = pd.merge(yearly_sum, housing, on='Year', how='inner')
economic_factors = economic_factors.dropna()
economic_factors.head()
```

To more easily read and determine whether the U.S. economy was going through a turmoil or growth, we devise an ecnomic misery index based on the available data. Given our background research and prior knowledge, we devise weights for each column listed above to calculate our economic misery index score. 

First, we compute the correlation coefficients for each of the change columns with the GDP change

```python
correlation_vars = ['Oil_Price_Change', 'House_Price_Index_Change', 'Mortgage_Rate_Change', 'Unemployment_Rate_Change']

# Code was taken from Chat-GPT 4o
correlation_results = economic_factors[correlation_vars + ['Real_GDP_Change']].corr()

# Extract correlations with GDP
gdp_correlation = correlation_results['Real_GDP_Change'].drop('Real_GDP_Change')

# Sort by strength of correlation
print(gdp_correlation.sort_values(ascending=False))
```

```python
# Calculating the economic misery index with the correlations described above
economic_state = pd.DataFrame()
economic_state['Year'] = economic_factors['Year']

# Calculate the EMI as the sum of the product of correlations
# Calculate the weighted sum using dot product
economic_state['EMI'] = economic_factors[correlation_vars].dot(gdp_correlation[correlation_vars])
economic_state['EMI'] = economic_state['EMI'].astype(float)
# Reset index if needed
economic_state = economic_state.reset_index(drop=True)
economic_state.head()
```

## Section 1 of Modeling - Sentiment Analysis on Lyrics

```python
!pip3 install nltk
from nltk.sentiment import SentimentIntensityAnalyzer
import nltk
```

```python
nltk.download('vader_lexicon')
```

```python
analyser = SentimentIntensityAnalyzer()
```

```python
def get_sentiment_score(text):
    if isinstance(text, str):
        return analyser.polarity_scores(text)['compound']
    return None
```

```python
lyrics['sentiment_score'] = lyrics['lyrics'].apply(get_sentiment_score)
```

## Section 2 of Modeling: K-means Clustering

Another consideration could be that because we're only looking at each of the attributes one by one, we're missing an overall relationship between multiple musical attributes. In order to find a relationship, we will perform K-means clustering to try and group multiple songs with similar musical attributes, then compare the EMI values of that group to try and find some correlation with the EMI.

```python
cluster_music = cluster_music.merge(economic_state, on = 'Year', how = 'left') #Merge EMI values with individual songs on year
cluster_music = cluster_music.drop(['artist_name', 'track_name', 'Unnamed: 0', 'index'], axis = 1)
cluster_music.head()
```

Now that we have the dataset ready, we can plot two values features each other and see the clustering in action. However, due to the numerous variety of songs and also the fact that we're using 8 features for the k-means clustering, this graph isn't exactly very clustered.

```python
from sklearn.cluster import KMeans

#Selecting the features for clustering (you can include/exclude some as needed)
features = ["sadness", "feelings", "danceability", "loudness", 
            "acousticness", "instrumentalness", "valence", "energy"]

#Extracting the data for clustering
X = cluster_music[features]

#Number of clusters
kmeans = KMeans(n_clusters = 7, n_init = 10, random_state=42)
cluster_music['Cluster'] = kmeans.fit_predict(X)

#For visualization, but it only shows two features rather than all 8
plt.scatter(cluster_music['valence'], cluster_music['sadness'], c = cluster_music['Cluster'], cmap='viridis')
plt.xlabel('Valence')
plt.ylabel('Sadness')
plt.title('K-Means Clusters, k = 7')
plt.colorbar()
plt.show()
```

With help from ChatGPT, we used the elbow method to determine the optimal number of clusters to use. The elbow method is a plot of the WCSS (Within-Cluster Sum of Squares) for different numbers of clusters. WCSS measures how close the points in a cluster are to the center and when the WCSS stops changing significantly we aren't getting as much information by adding more clusters. The optimal point is when the steepness of the curve starts to flatten and we can see that would be at k = 4 clusters.

```python
# Code formulatted and written in colaboration with Chat GPT-4o
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, n_init = 10, random_state=42)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

plt.plot(range(1, 11), wcss)
plt.title('Elbow Method for Optimal k')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

We set the number of clusters to 4 and separate the dataset by the clusters. We can then print data for each individual cluster to see if any trends can be found. Just looking at the mean/median and the standard deviation for the 8 attributes and the EMI doesn't give us very much info. They all seem relatively similar.

```python
#Set k = number of clusters
k = 4
kmeans = KMeans(n_clusters = k, n_init = 10, random_state=42)
cluster_music['Cluster'] = kmeans.fit_predict(X)

#Separate data and describe, scroll to view data
clusters = list() #index corresponds to cluster
for i in range(k):
    clusters.append(cluster_music[cluster_music['Cluster'] == i])
    print(f"Cluster {i}: ")
    print(clusters[i].describe())
    print(clusters[i]['genre'].value_counts())
```

Since there's no way for us to visualize 8-dimensional similarities of the clusters, we'll use dimensionality reduction to bring it down to 2 dimensions which helps us interpret our clusters. With the help of PCA to perform dimensionality reduction, we are able to see all eight attributes condensed down into a plot. Here we can see a relatively clear separation between each of the clusters except for the Yellow cluster (3) which seems to coincide with the other clusters.

```python
# Code formulatted and written in colaboration with Chat GPT-4o
from sklearn.decomposition import PCA

# Perform PCA to reduce dimensions to 2
pca = PCA(n_components=2)
pca_result = pca.fit_transform(cluster_music[features])

# Add cluster labels to the PCA result for coloring
pca_result_df = pd.DataFrame(pca_result, columns=['PC1', 'PC2'])
pca_result_df['Cluster'] = cluster_music['Cluster']

# Plot the clusters
plt.figure(figsize=(8, 6))
plt.scatter(pca_result_df['PC1'], pca_result_df['PC2'], c=pca_result_df['Cluster'], cmap='viridis', alpha=0.7)
plt.colorbar(label='Cluster')
plt.title('PCA of Song Features')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```

By using the feature loadings of the axes, we're able to find out which features contributed most to each one. A high value means that there was a significant contribution whereas a small value means there was little to no contribution. Positive values mean that songs with that attribute are towards the positive end of the graph whereas negative values mean the opposite. Here we can see that Principal Component 1 is positively influenced mostly by acousticness and negatively influenced mostly by energy with some slight influence from valence. Songs that are high on this scale are most likely songs that are calm, acoustic and possibly even sad. Principal Component 2 is positively influenced by loudness and sadness with a lot of negative influence from valence and danceability. Songs that are high on this scale most likely very loud and sad and not very danceable. Now let's use this to interpret our clusters.

Cluster 0 (Green): This cluster is very negative on both scales which suggests these are songs that are loud and energetic but with sad/negative elements.

Cluster 1 (Purple): This cluster is fairly high on PC2 but low on PC1 which indicates these are songs that are loud and energetic and also positive and danceable.

Cluster 2 (Blue): This cluster is fairly high on PC1 but very spread out with repsect to PC2 which indicates that these songs are more acoustic and calm with varying levels of positivity/negativity.

Cluster 3 (Yellow): This cluster is very difficult to draw conclusions on as it's spread throughout the entire graph.

```python
feature_loadings = pd.DataFrame(pca.components_, columns=features, index=[f'PC{i+1}' for i in range(pca.n_components_)])
print(feature_loadings)
```

Now that we understand what the clusters seem to represent, we can try to determine the distribution of the EMI for each cluster. Although simply looking at the exact mean/median didn't show significant differences, perhaps the distributions will give us more insight. We expect according to our hypothesis that the largest difference in EMI would be between Cluster 0 and CLuster 1 because they have a large split of positive/negative songs whereas Cluster 3 would be somewhere in the middle due to its large spread into different categories. Our hypothesis stated that during period of economic downturns/times of social unrest, we expected songs with lower measured values on "happier" song attributes. (E.g. danceability, tempo, energy). We predicted the complement for times of econimci prosperity and social stability.
