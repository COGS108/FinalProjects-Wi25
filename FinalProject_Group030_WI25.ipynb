### D1a) Standardizing Ages

```python
# Check different age group input data
df1['ages'].value_counts()
```

```python
# Since AGE2 has the age 21 separated from the other groupings, combine values 10 and 11.
# Creating a function to standardize the ages. 
def standardize_ages(x):
    
    try:
        # Convert int values to string for string grouping.
        x = str(x)

        # Use a match-case statement to pattern match values.
        match x:
            case '10':
                output = x.replace('10', '21-23')
            case '11':
                output = x.replace('11', '21-23')
            case '12':
                output = x.replace('12','24-25')
            case '13':
                output = x.replace('13','26-29')
            case _:
                output = np.nan
                
    except:
        output = np.nan
        
    return output
```

```python
print(standardize_ages(10))
print(standardize_ages(11))
print(standardize_ages(12))
print(standardize_ages(13))
print(standardize_ages('hi'))
```

```python
# Apply transformation
df1['ages'] = df1['ages'].apply(standardize_ages)
# Check results
df1['ages'].unique()
```

### D1b) Standardizing Alcohol Consumption Responses

```python
# Many responses are not useful for our dataset since they don't really provide much information. 
def standardize_alc(x):

    try:
        match x:
            # 91 = Subject *never* used alcohol
            case 91:
                x = -1
                output = x
            # 93 = Subject did not use alcohol during range of days
            case 93:
                x = 0
                output = x
            # These responses are irrelevant for the purposes of our project
            case 85 | 89 | 94 | 97 | 98 | 99: 
                output = np.nan
            case x if 1 <= x <= 31:
                output = x
            case _:
                output = np.nan
        
    except Exception:
        output = np.nan
    return output
                
```

```python
print(standardize_alc(91))
print(standardize_alc(93))
print(standardize_alc(94))
print(standardize_alc(8))
print(standardize_alc('pop'))
```

```python
# Apply transformation
df1['alc_dayspm'] = df1['alc_dayspm'].apply(standardize_alc)
df1['alc_days'] = df1['alc_days'].apply(standardize_alc)
# Check Results
print(df1['alc_dayspm'].unique())
print(df1['alc_days'].unique())
```

```python
# Drop irrelevant responses
df1 = df1.dropna().reset_index(drop = True)
# Check results
print(df1['alc_dayspm'].unique())
print(df1['alc_days'].unique())
# Compare to original shape: (13605, 4)
print(df1.shape)
```

# Ethics & Privacy

The data we are gathering, does include some personal identifying information, date of birth, and gender, which could raise certain privacy concerns, although it does not include names that require especially cautious handling. We will handle this data responsibly by ensuring that it is anonymized and used solely for analytical purposes. If a dataset we use does have additional identifying information, we will remove it. We think it’s unlikely that there will be many biases in our datasets since we are looking at a large range of individuals across a large population. It is possible that biases could be present if the data does not appropriately sample a large range of individuals, therefore we must ensure our datasets are appropriately representative of the US population.

Since we will be using publicly available data, we aren’t worried about encryption or data security. The data we access comes from publicly available sources such as the NSDUH public use files (PUFs). These datasets have undergone confidentiality protections to remove identifying information, such as the exclusion of most geographic variables, to protect participants' privacy. While the PUFs allow for broad analysis, they use disclosure-avoidance methods. This means that estimates generated from them may not be identical to those officially by SAMHSA. However, they still provide valuable insights into alcohol use trends in youth.

If the data was collected through surveys it’s possible that participants weren’t fully honest about their responses to curb potential judgment, so we recognize that the data we use won’t be 100% accurate. There would be concern about response bias, especially for the results of responses from underage respondents among the younger generation, due to the relatively higher possibility of their underestimation of the frequency of drinking. However, since our focus is the 21-29 year olds and we use a sampling of responses to the frequency of drinking up to 21-29, the age at which one can legally drink, the risk of bias due to dishonest responses can be minimized.

Since our hypothesis is correct about young adults in 2023 drinking less than those in 2003, we recognize this information could be used to help acknowledge positive public health trends while avoiding stigmatization of different drinking behaviors. In addition, We understand that the data may not fully represent all young adults due to sample limitations; however, we did our best to uphold ethical responsibility to present findings accurately without exaggeration or misinterpretation.

# Discussion

The decline in alcohol consumption among young adults in 2023 compared to 2003 is a significant shift that has been influenced by a range of social, economic, and cultural factors. Our analysis supports previous studies that indicate young adults, ages 21 to 29, today not only drink less frequently but also have higher rates of sobriety than their counterparts two decades ago. One of the key contributing factors is the rise of the "hustle culture," which prioritizes productivity, career development, and self-optimization over social drinking. This cultural shift has led to a greater emphasis on health-conscious behaviors and a reduced association between alcohol consumption and social status​.

Another possible contributing factor is the impact of the COVID-19 pandemic, which disrupted traditional drinking environments such as bars, clubs, and social gatherings. During lockdowns, many young adults either reduced their alcohol consumption or stopped drinking altogether. Studies have shown that some of these individuals did not return to pre-pandemic drinking habits even after restrictions were lifted. The pandemic also reinforced a focus on mental and physical well-being, leading to increased awareness of the negative effects of alcohol such as anxiety and poor sleep​.

Generational differences in attitudes toward alcohol have also played a role in this trend as young adults in 2023 engage in fewer risky behaviors compared to previous generations, reflecting broader shifts in risk perception and social norms​. The normalization of alcohol-free socializing, combined with the availability of non-alcoholic alternatives, has made it easier for young people to abstain from drinking without feeling socially excluded.

In addition, Public health campaigns and increased awareness of alcohol’s risks may also be shaping drinking habits. The past two decades have seen a rise in harm reduction education, discouraging binge drinking and promoting healthier choices. Furthermore, changes in alcohol regulations, such as stricter enforcement of drinking age laws and higher taxes on alcoholic beverages, could be contributing to this broader societal shift.

# Conclusion 

The decline in alcohol consumption among young adults in 2023 compared to 2003 is a multifaceted phenomenon driven by cultural shifts, changing social norms, and the long-term effects of the COVID-19 pandemic. Our findings align with existing research that suggests young adults today are more health-conscious, career-focused, and less inclined to view alcohol as a central aspect of their social lives.

While the reduction in drinking can be seen as a positive development in terms of public health, it also raises new questions about how young people are socializing and coping with stress in an era of increasing pressures related to work, education, and financial stability. Future research should explore whether these changing behaviors are sustainable over time and how they may influence broader public health trends in the coming years.
