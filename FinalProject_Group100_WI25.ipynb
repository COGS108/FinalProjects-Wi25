# Abstract

Our group wanted to see the impact of having access to a car on UCSD students’ social wellness and academic achievement. Through the survey data (142 participants) we analyzed factors including car ownership, social satisfaction, GPA, and variables such as living situation, academic workload, and major. We found that vehicle access alone does not significantly affect social life satisfaction or GPA. Regression analysis showed that factors such as major, year in school, and extracurricular involvement in conjunction with access to a personal vehicle play a larger role in influencing both social and academic success. While access to a car provides greater flexibility and social opportunities, that factor does not define one's level of satisfaction with their social life or GPA alone.

# Research Question

At UCSD, does having regular access to a personal vehicle increase students’ own perception of social happiness and academic performance in the form of GPA?

## Survey Data

The following dataset includes the answers from the survey we conducted asking questions related to vehicle access, social life, and academics. The questions asked consist of questions essential to our analysis, such as GPA and how satisfied they are with their social life, but also additional information to serve as context or to check for potential confounding factors. Currently, all the data will be in the form of strings, and we will have the clean and transform the data into the proper data types. We would want the data containing information about hours, days, GPA, etc. to be in the form of floats so aggregation can be performed.

We gathered information on the following questions:

##### Do you own or have access to a car?
- Essential to our project: This lets us know if they have a car or not and will be essential for grouping the data in our analysis and allowing for comparisons between the two groups.
  
##### How satisfied are you with your own social life?
- Essential to our project: This lets us know how satisfied they are with their current social life.
  
##### What is your GPA?
- Essential to our project: This is how we will measure their academic performance.
  
##### What year are you?
- Possible confounding factor: Older students are more likely to have a car and the year a student is in may contribute to them having more or less responsibilities that detract from their ability to have a satisfactory social life or have good academic success.
  
##### On average, how many days per week do you use your car for leisure, not for commuting to school/work? 
- Even if they have access to a car, it may not always be used for “fun” activities. It would be interesting to explore within the group that has access to a vehicle how the amount of days per week they tend to spend on leisure affects GPA and their perception of their social life.
  
##### Do you live on campus?
- Possible confounding factor: Those who live on campus are less likely to have a car and the proximity to the community at school can have a large effect on your ability to socialize or attend classes.
  
##### How do you think your social life compares to the average UCSD student?
- This will allow us to see how students think they compare to the rest of the population and be another point to examine. There is a stigma around UCSD being socially dead, so this can be used to see if there was a bias in their answer and give us a reference point for how strong their social life really is.

##### What is your gender?
- Possible confounding factor: Gender may be an even more important factor to one's success socially and academically. We also collected this data so we could check if we have responses from a large variety of people and to see if our sample is actually representative.
  
##### How many hours each week do your academic extracurricular, athletic activities, or job commitments require?
- This gives us a measure of how much time they spend a week on activities that could not be considered as leisure.
  
##### What field is your primary major?
- Possible confounding factor: Certain majors may be more stressful or harder than others and this can have a larger impact on GPA and how fulfilling their social life is, moreso than whether they have access to a car

Currently, our data contains a column for the time the survey was taken, which is useless for our project. Additionally, the column names are very long and will be annoying to deal with when performing analysis later on, so we will rename the columns to be more efficient and also drop the time column.

```python
survey = survey.rename(columns={
    'Do you own or have access to a car?': 'has_car', 
    'How satisfied are you with your own social life?': 'social_life_satisfaction', 
    'What is your GPA?': 'gpa',
    'What year are you?': 'year',
    'On average, how many days per week do you use your car for leisure, not for commuting to school/work? ': 'days_per_week_for_leisure',
    'Do you live on campus?': 'on_campus',
    'How do you think your social life compares to the average UCSD student?': 'social_life_vs_avg',
    'What is your gender?': 'gender',
    'How many hours each week do your academic extracurricular, athletic activities, or job commitments require?': 'extracurricular_and_job_hours',
    'What field is your primary major?': 'major_type'}).drop(columns=['Timestamp'])
survey.head()
```

```python
survey.dtypes
```

The data types for certain columns are not what we want them to be. We want the 'days_per_week_for_leisure' and'extracurricular_and_job_hours' to be floats and 'social_life_satisfaction' to be ints so we can perform aggregations such as finding the mean for EDA and later analyses.

```python
def clean_days_per_week_for_leisure(string):
    if string == "I don't have access to a car.":
        return np.nan
    else:
        return string

def clean_extracurricular_and_job_hours(string):
    if type(string) != str:
        return np.nan
        
    string = string.strip()
    string = string.lower()
    string = string.replace('hours', '')
    string = string.replace('hrs', '')
    string = string.replace('hour', '')
    string = string.replace('hr', '')
    string = string.replace('~', '')
    string = string.replace('+', '')
    string = string.replace('for academic', '')
    string = string.replace('i spend 6 hours a week on athletic activities. i spend 5 hours on academic extracurricular.', '11')
    string = string.replace('N\A', '0')
    string = string.replace('above', '0')
    string = string.strip()

    if '-' in string:
        split = np.array(string.split('-')).astype(float)
        string = str(np.mean(split))

    try:
        return float(string)
    except:
        return np.nan

def satisfaction_to_numeric(string):
    if string == 'Greatly Satisfied':
        return 5
    elif string == 'Slightly Satisfied':
        return 4
    elif string == 'Neutral':
        return 3
    elif string == 'Slightly Unsatisfied':
        return 2
    else:
        return 1

survey['days_per_week_for_leisure'] = survey['days_per_week_for_leisure'].apply(clean_days_per_week_for_leisure).astype(float)
survey['extracurricular_and_job_hours'] = survey['extracurricular_and_job_hours'].apply(clean_extracurricular_and_job_hours)
survey['social_life_satisfaction'] = survey['social_life_satisfaction'].apply(satisfaction_to_numeric)

survey.head()
```

Some of the columns contain answers that are undesirable and not what we are looking to use in our calculations and analysis.

```python
survey['major_type'].value_counts()
```

The 'major_type' column includes undesirable responses that should not be their own group. The answers 'Biz' and 'Cognitive Science' are categories of other answers already present in the data. 'Biz' will be added to the 'Business/Finance' group and 'Cognitive Science' will be added to the 'Social Sciences' group.

```python
survey['major_type'] = survey['major_type'].apply(lambda x: 'Social Sciences' if x == 'Cognitive Science' else 'Business/Finance' if x == 'Biz' else x)
```

Our survey included "It's complicated" as a possible answer for whether or not you had a car. However, we now know that we do not want this to be something to be grouped on, so we will convert all instances of "It's complicated" to be "No" responses instead. We think this is reasonable to do because "It's complicated" suggests that there are some complications that make the vehicle not fully accessible at all times. 

```python
def standardize_car(string):
    # Basic string pre-processing
    string = string.strip()
    string = string.replace("It’s complicated", "No")
    string = string.replace("Not at school", "No")
    string = string.strip()
    return string

survey["has_car"] = survey["has_car"].apply(standardize_car)
```

We want the answers for the 'on_campus' column to be either 'On Campus' or 'Off Campus'. 

```python
survey['on_campus'] = survey['on_campus'].apply(lambda x: 'On Campus' if x == 'Yes' else 'Off Campus')
```

# Permutation Testing

To see if people with access to a car are generally more satisfied with their social lives or have better GPAs, we will run a permutation with difference in group means as a test statistic to if there is a significant difference in average social life satisfaction or GPA for people who do and do not have access to a personal vehicle.

## 'has_car' vs 'social_life_satisfaction'

```python
has_car_vs_sls
```

```python
has_car_vs_sls_mean_diffs = survey.groupby('has_car')[['social_life_satisfaction']].mean()
has_car_vs_sls_mean_diffs
```

```python
has_car_vs_sls_obs_diff = (has_car_vs_sls_mean_diffs.loc['Yes'] - has_car_vs_sls_mean_diffs.loc['No']).iloc[0]
has_car_vs_sls_obs_diff
```

Is a difference of 0.042 statistically significant with a significance value of 0.05?

**Null Hypothesis:** Whether one has access to vehicle or not has no relation with the person's satisfaction with their own social life and observed differences in the sample are due to random chance

**Alternative Hypothesis:** People who have access to a car are more satisfied with their social life than people without access and observed differences are not due to random chance

```python
has_car_vs_sls_diffs = []
for _ in range(1000):
    survey['shuffled_has_car'] = np.random.permutation(survey['has_car'])
    vs_sls_means = survey.groupby('shuffled_has_car')['social_life_satisfaction'].mean()
    has_car_vs_sls_diffs.append(vs_sls_means.loc['Yes'] - vs_sls_means.loc['No'])
```

```python
sls_diffs = sns.histplot(has_car_vs_sls_diffs, color='gray', stat='probability')
plt.axvline(has_car_vs_sls_obs_diff, 0, 1, linewidth=2, color='red')
plt.xlabel('Difference in Group Mean\n(Car Owner - Non-Car Owner)')
sls_diffs.set(title='Distribution of Differences in Group Mean\nFor Social Life Satisfaction')
sns.despine()
```

```python
(has_car_vs_sls_diffs >= has_car_vs_sls_obs_diff).mean()
```

Since this value is greater than our significance level of 0.05, we fail to reject the null hypothesis. We cannot say that people who have access to cars are more satisfied with their social lives than people without access to cars.

## 'has_car' vs 'gpa'

```python
has_car_vs_gpa
```

```python
has_car_vs_gpa_mean_diffs = survey.groupby('has_car')[['gpa']].mean()
has_car_vs_gpa_mean_diffs
```

```python
has_car_vs_gpa_obs_diff = (has_car_vs_gpa_mean_diffs.loc['Yes'] - has_car_vs_gpa_mean_diffs.loc['No']).iloc[0]
has_car_vs_gpa_obs_diff
```

Is a difference of -0.067 statistically significant with a significance value of 0.05?

**Null Hypothesis:** Whether one has access to vehicle or not has no relation with the person's overall GPA and observed differences in the sample are due to random chance

**Alternative Hypothesis:** People who do not have access to a vehicle have higher overall GPAs and observed differences are not due to random chance

```python
has_car_vs_gpa_diffs = []
for _ in range(1000):
    survey['shuffled_has_car'] = np.random.permutation(survey['has_car'])
    vs_gpa_means = survey.groupby('shuffled_has_car')['gpa'].mean()
    has_car_vs_gpa_diffs.append(vs_gpa_means.loc['Yes'] - vs_gpa_means.loc['No'])
```

```python
gpa_diffs = sns.histplot(has_car_vs_gpa_diffs, color='gray', stat='probability')
plt.axvline(has_car_vs_gpa_obs_diff, 0, 1, linewidth=2, color='red')
plt.xlabel('Difference in Group Mean\n(Car Owner - Non-Car Owner)')
gpa_diffs.set(title='Distribution of Differences in Group Mean\nFor GPA')
sns.despine()
```

```python
(has_car_vs_gpa_diffs <= has_car_vs_gpa_obs_diff).mean()
```

Since this value is greater than our significance level of 0.05, we fail to reject the null. We cannot say that people without access to a vehicle have higher overall GPAs than people who do have access to a vehicle.

## Linear Regression Models

Since we found that having a car does not necessarily mean you will be more satisfied with your social life or have a lower GPA than if you did not own a car, we thought that we should generate linear regression models with 'social_life_satisfaction' and 'gpa' as response variables and look at the coefficients to see which numerical and categorical features had the greatest overall weight on the outcome.

```python
import statsmodels.formula.api as smf
```

We need to clean and standardize the dataset we have for the purpose of regression.

```python
for_regression = survey.copy(deep=True)
```

We found in our EDA process that there was an especially large value in the 'extracurricular_and_job_hours' column. For our regression, we will consider it an outlier and remove it so that it does not skew our coefficients.

```python
for_regression['extracurricular_and_job_hours'].loc[52] = np.nan
for_regression.loc[52]
```

We also need to standardize our numerical columns so that the weights of those columns can be compared.

```python
def standardize(arr):
    mean = np.mean(arr)
    std = np.std(arr)
    standardized_arr = (arr - mean) / std
    return standardized_arr

for_regression['gpa'] = standardize(for_regression['gpa'])
for_regression['days_per_week_for_leisure'] = standardize(for_regression['days_per_week_for_leisure'])
for_regression['extracurricular_and_job_hours'] = standardize(for_regression['extracurricular_and_job_hours'])
```

In our survey, we included a question that was only intended to be answered by people who had access to a car: 'days_per_week_for_leisure'. Since we do not want the fact that all people who do not own a car to have missing values for that column, we will instead create two models each for 'social_life_satisfaction' and 'gpa'. In total, we will have 4 models to analyze.

From these models, we look at the most impactful features and draw conclusions about what that means for the overall population.

## Model 1 - Predicting 'social_life_satisfaction' given that you own a car

```python
mod_sls_car = smf.ols(formula='social_life_satisfaction ~ gpa + days_per_week_for_leisure + extracurricular_and_job_hours + C(year) + C(on_campus) + C(social_life_vs_avg) + C(gender) + C(major_type)', data=for_regression[for_regression['has_car'] == 'Yes'])
res_sls_car = mod_sls_car.fit()
print(res_sls_car.summary())
```

With an R-squared value of 0.563, the strength of the relationship between the variables and 'social_life_satisfaction' is at an acceptable level and the conclusions drawn from the coefficients should be usable. This section looks at the most significant weights and features.

For the categorical columns, it seems like being on campus has the strongest and most significant impact on your perception of the quality of your social life, dropping your perceived satisfaction rating 1.2 points down. People who aren't satisfied with their social lives are also more likely to feel that they lead below average social lives. It also looks like third years are the most unsatisfied with their social lives, lower than 1st, 2nd, and 4th years.

As for the numerical columns, GPA has the largest weight when it comes to defining how satisfied one is with their lives closely followed by the amount of hours you spend on extracurricular. The weight found for the effect of the number of days a week you utilize your car for leisure looks negligible and also looks to be statistically insignificant given its p-value.



### Model 2 - Predicting 'social_life_satisfaction' given that you do not own a car

```python
mod_sls_no_car = smf.ols(formula='social_life_satisfaction ~ gpa + extracurricular_and_job_hours + + C(has_car) + C(year) + C(on_campus) + C(social_life_vs_avg) + C(gender) + C(major_type)', data=for_regression[for_regression['has_car'] == 'No'])
res_sls_no_car = mod_sls_no_car.fit()
print(res_sls_no_car.summary())
```

The strength of relationship is slightly weaker than the model for vehicle owners, having a R-squared score of 0.465, so insights gained from this model will be very slightly less reliable. 

Similarly to the previous model, third years are the least satisfied with their social lives and having extra hours to spend on extracurriculars or non-academic work boosts how satisfied you are with your social life. However, lacking ownership of a car made a difference in that living on campus now positively affects your level of satisfaction and people with higher GPAs tend to be less satisfied as well. More extracurricular hours also meant more satisfaction.

## Model 3 - Predicting 'gpa' given that you own a car

```python
mod_gpa_car = smf.ols(formula='gpa ~ days_per_week_for_leisure + extracurricular_and_job_hours + C(social_life_satisfaction) + C(year) + C(on_campus) + C(social_life_vs_avg) + C(gender) + C(major_type)', data=for_regression[for_regression['has_car'] == 'Yes'])
res_gpa_car = mod_gpa_car.fit()
print(res_gpa_car.summary())
```

The R-squared value is 0.526 so it is relatively reliable. 

Those who were 'greatly unsatisfied' with their social life also tended to have lower GPAs than those who rated their social lives more favorably and third and fourth years seemingly have higher GPAs than first and second years.

The more hours one spent on extracurriculars or work, the worse their GPA generally was but the more days per week they spent for leisure purposes, the better their GPA was.

## Model 4 - Predicting 'gpa' given that you do not own a car

```python
mod_gpa_no_car = smf.ols(formula='gpa ~ extracurricular_and_job_hours + C(social_life_satisfaction) + C(year) + C(on_campus) + C(social_life_vs_avg) + C(gender) + C(major_type)', data=for_regression[for_regression['has_car'] == 'No'])
res_gpa_no_car = mod_gpa_no_car.fit()
print(res_gpa_no_car.summary())
```

This model is quite a bit more unreliable than the previous model for car owners, with an R-squared value of only 0.322.

Compared to the previous model, one's social life satisfaction was much less significant of a factor. Instead, it looks like major type is the more significant factor for people who do not own cars when it comes to the categorical features. 

For this model, having more extracurricular hours coincided with higher GPA, the opposite of what we saw in the previous model.

## Simple Summary

Across all models, different features have the most weight in determining social life satisfaction and GPA for car owners and non-car owners. It looks like there is no singular variable that ultimately determines your social life satisfaction or GPA, and the response variables are not simple to predict.

Notably, the weights of the numerical features for every model were found to be statistically insignificant given a 0.05 significance level, so differences between car owners and non-car owners pertaining to those numerical features are not absolute. It seems like the categorical features are of greater importance when it comes to both response variables.

In the case of our survey, the data collected from our questions are broad, and it will be unlikely that a person could identify an individual based off the combination of answers given. In fact, it would not be unexpected if some participants give the same combination of answers.

When surveying random people, it is also important to ensure that consent is given. This will keep particpants properly informed and leave them with the ultimate decision of whether or not to answer the survey questions.

Unfortunately, there is nothing we can do to circumvent this, and all we can do is keep this fact in mind while we are analyzing our data and drawing conclusions.

The 1-5 scale can be troublesome because it is relative for every person and an answer of 3 for one person can be very different from an answer 3 for another person. However, since we are looking to answer how access to a vehicle affects how one feels about their social life, this subjectivity is important and is what we are looking for. Additionally, we included a question asking about how one thinks their social life compares to the average student to address this subjectivity to an extent by having a standard to compare to.

Before our analysis, we predicted that access to a personal vehicle would inherently make you more satisfied with your social life than if you did not own a car. We also predicted that a car owner's GPA was moreso conditional on whether they lived on campus, with people who owned cars having better GPAs if they loved off campus and people who did not own cars having better GPAs if they lived on campus. The permutation tests we ran and the regression models we created were intended to verify if our predictions were correct and to draw possible conclusions about the UCSD student population.

We discovered from our permutation tests that having a car is not a significant factor that makes you more satisfied with your social life than those without cars and also is not a significant factor that makes your overall GPA worse than those without cars. We were wrong about having access to a vehicle inherently making you more satisfied than those without cars. Results may have been different if a more objective and comparable metric was used for this variable. Vehicle access not being a statistically significant factor when it comes to GPA matched what we expected, though it was still necessary to see what features had the largest weights, which is what we looked at in our regression models.

We found from our regression models for social life satisfaction that, for car owners, your satisfaction with your social life was positively correlated with GPA and and negatively correlated with the number of hours you spent on extracurriculars or work per week. People who lived on campus and people who were third years while owning a car tended to be more unsatisfied than those who lived off campus. For people who did not own a car, third years were similarly unsatisfied with their social lives, but being on campus and having a higher GPA had the opposite effect for this group, being negatively correlated with their satisfaction levels instead. 

These are some possible explanations for those insights on social life satisfaction:
- The reason third years are the most unsatsified for both groups may be due to the fact that they start taking upper-division classes and focusing on the future, doing things like networking.
- For car owners, having a higher GPA may make spending time with other people and relaxing more justifiable and rejuvenating, as thye do not feel burdened by their academic performance
- For car owners, extra extracurricular and work responsibilities may be more taxing, as those responsibilities might be more time-consuming and increase the need to drive, leaving less time for socialization.
- For car owners, living on campus means you need to pay for parking and that might compel you stay on campus more often, which can be quite mundane as the facilities on campus will eventually lose their novelty
- For non-car owners, a higher GPA might mean they are spending much more time studying, taking away from time for socialization
- For non-car owners, living on campus gives you more places to hang out with friends, rather than being isolated in the city with no means of moving other than public transportation
- For non-car owners, extracurricular hours are more likely to be on campus clubs, which frequently have meetings to interact with club members and have fun

We found from our regression models for GPA that, for car owners, the more unsatisfied with your social life you were, the worse your GPA tended to be. The year you were in was also a large factor, with third and fourth years tending to have higher GPAs. The more hours one spent on extracurriculars or work, the worse their GPA generally was but the more days per week they spent for leisure purposes, the better their GPA was. For non-car owners, major type was more significant than the year you were in. More extracurricular hours meant better GPAs for this group.

These are some possible explanations for those insights on GPA:
- For car owners, lower satisfaction levels correlating with lower GPA may be attributed to mental health concerns. There may be less resources to address these issues off campus, which the majority of car owners live.
- For car owners, additional extracurricular or non-academic work responsibilities may simply reduced the time they had to study. Driving a lot already takes time, and extracurricular responsibilities simply takes even more time away from studying.
- For car owners, having more days for relaxation and recreation can place you in a better state mentally, and that can contribute to better academic performance. The reason this improves GPA while extracurricular hours are detrimental to GPA may be that those extracurricular responsibilities are more stress-inducing than fun.
- For non-car owners, major might be a more significant factor because of the varying difficulty of classes for different major types and the decreased flexibility to move to get resources due to the lack of a vehicle.
- For non-car owners, extracurriculars are mostly centralized on campus, and these tend to be more fun and rejuvenating, improving mental health and subsequently GPA
