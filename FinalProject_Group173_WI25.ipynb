# Abstract
This study examines the relationship between a Major League Baseball (MLB) pitcher’s workload—measured by total pitches thrown and innings pitched—and subsequent declines in pitch velocity. Using Statcast data from 2015 onward, we analyzed whether higher workloads lead to significant decreases in fastball and overall pitch speed in the following season. Our findings indicate a moderate but statistically significant negative correlation (r = -0.1985, p = 0.0030) at the individual pitcher level and a much stronger correlation (r = -0.7600, p = 0.0175) at the league-wide level, suggesting that increased workload accelerates velocity decline. 

A key outlier, the 2021 season, revealed no velocity drop following the COVID-shortened 2020 season, further supporting the idea that reduced workload helps preserve velocity. These results have important implications for pitcher management, injury prevention, and career longevity, highlighting the need for workload monitoring and strategic rest periods in modern baseball.

These findings have critical implications for MLB teams and player management strategies. Monitoring workloads - through pitch count limits, increased recovery periods, and strategic bullpen usage – could help delay velocity decline, reduce injury risk, and extend pitcher careers. Furthermore, pitchers who rely heavily on high-velocity fastballs may benefit from diversifying their pitch repertoire, incorporating more breaking and offspeed pitches to maintain effectiveness as velocity diminishes. 

Future research could refine our understanding of workload thresholds beyond which velocity loss becomes more pronounced. Additionally, examining the interplay between workload, velocity decline, and injury risk could provide deeper insights into optimizing pitcher health and longevity. Understanding how different pitcher archetypes (e.g., power pitchers vs. finesse pitchers) respond to workload could further refine strategies for sustainable performance in professional baseball.

## Background and Prior Work

### **Understanding Pitching Velocity and Workload in the Statcast Era**  

In Major League Baseball (MLB), **pitching velocity** is one of the most critical factors in evaluating a pitcher’s effectiveness. With the **Statcast system in 2015**, teams now have access to highly accurate velocity-tracking data, allowing for a more precise understanding of how velocity correlates with performance and workload.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Over the past decade, there has been a clear increase in fastball velocities across the league, with more pitchers than ever reaching **95+ MPH** on a regular basis.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) However, while higher velocity is often associated with greater dominance on the mound, it also raises concerns about **fatigue, injury risk, and long-term sustainability**.  

A pitcher’s **workload**—defined by the total number of innings pitched, total pitches thrown, and batters faced—plays a crucial role in determining how long they can maintain peak performance. **Throwing harder often places greater stress on the arm**, particularly the elbow and shoulder, which can lead to long-term velocity decline. Additionally, pitchers often adjust their **pitch selection strategy** in response to changes in their velocity over time, increasing their reliance on breaking and offspeed pitches as their fastball declines.

This research will focus **exclusively on the Statcast era (2015–present)** to ensure consistency in tracking data. Instead of classifying pitchers as **starting pitchers (SPs) or relief pitchers (RPs)**—a distinction that would require manual classification for 1106 seasons—we will use **workload-based metrics (total pitches thrown and average pitch velocity) to assess long-term performance and velocity decline objectively**.  

### **Defining Key Baseball Metrics Used in This Study**  

To better define our research methodology, several key baseball terms must be clarified. **Pitching velocity (VELO)** refers to the maximum speed of a given pitch from its release to when it crosses home plate, measured in miles per hour (MPH). While fastball velocity is often the primary focus, we also analyze breaking balls (such as curveballs and sliders) and offspeed pitches (such as changeups and splitters), which are thrown with different intent and mechanics.  

Rather than distinguishing pitchers by role, we measure workload using **total pitches thrown and average pitch velocity**. These **workload indicators allow for a role-neutral assessment** of how pitching volume and intensity impact long-term velocity trends.  

### **Research Focus and Study Objectives**  

This study aims to determine whether **pitchers with higher workloads and higher fastball velocities experience greater velocity decline over time** and whether **these pitchers compensate by adjusting their pitch type usage**. Specifically, we seek to:  

- Track changes in velocity across **fastballs, breaking balls, and offspeed pitches** over multiple seasons.  
- Identify **pitchers with higher fastball percentage (e.g., >60% fastball threshold)** and measure their **velocity drop-off**.

First, let's group the data by the player_id and year so that the player data is organized. We can see below the data starts with Bartolo Colon, who pitched from 2015 to 2018.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
df_sorted = df.sort_values(by=["player_id", "year"])
df_sorted.head()
```

Now that our data is sorted, let's add a new measurement of "workload", which will measure the amount og work a pitcher put in in a given season. Workload is going to be the number of pitches thrown multiplied by the average speed of a given pitch. So, using the six columns which measure speed and percentage, let's figure out the average speed of any given pitch for the season and add it to the dataframe.

```python
# Calculate weighted average pitch speed using pitch type percentages
df_sorted['weighted_avg_pitch_speed'] = (
    df_sorted['fastball_avg_speed'] * (df_sorted['fastball_percentage'] / 100) +
    df_sorted['breaking_avg_speed'] * (df_sorted['breaking_percentage'] / 100) +
    df_sorted['offspeed_avg_speed'] * (df_sorted['offspeed_percentage'] / 100)
)
df_sorted.head()
```

Using the weighted_avg_pitch_speed, we can now calculate the workload.

```python
# Calculate total workload (total pitches thrown * weighted avg speed)
df_sorted['workload'] = df_sorted['pitch_count'] * df_sorted['weighted_avg_pitch_speed']
df_sorted.head()
```

Now that we have a measurement of the workload, we want to explore how workload relates to the speed of pitches the next season. So, using our weighted_avg_pitch_speed, we can measure the change in average pitch speed from season to season. We will place this data in a column named pitch_speed_pct_change, and it will measure the percent change in average pitch speed from the previous season.

```python
# Calculate percentage change in weighted average pitch speed from season to season
df_sorted['pitch_speed_pct_change'] = df_sorted.groupby('player_id')['weighted_avg_pitch_speed'].pct_change() * 100
df_sorted.head()
```

Now, there is two last things we need to filter out of our data: players who only pitched for 1-2 years and pitchers who don't throw many fastballs. Since we are analyzing how workload affects pitch speed over time, pitchers with only one or two seasons of data won’t provide meaningful insights. In fact, keeping them in the dataset could skew our results because they might have left the league for reasons unrelated to workload—such as poor performance or injuries. If a pitcher was cut after just one or two seasons, it’s possible that their pitch speed was already declining due to other factors, which could introduce bias in our analysis. To ensure we are measuring long-term trends, we will only include pitchers with at least three seasons of data.

Additionally, pitchers who throw fastballs less than 60% of the time are not as useful, since the effect of workload on pitch speed decline is likely less pronounced for them since, if they throw offspeed or breaking pitches, the pitch is by nature slower. These pitchers rely more on breaking balls and offspeed pitches, which are not as dependent on raw velocity. Because our analysis is focused on how workload impacts pitch speed over time, including pitchers who primarily throw other pitch types could introduce noise into the data and weaken our results. By filtering out these pitchers, we ensure that we are measuring workload effects on a group of pitchers whose performance is more directly tied to their fastball velocity.

```python
# Count the number of years each player has data for
player_season_counts = df_sorted.groupby('player_id')['year'].nunique()

# Filter out players with fewer than 3 seasons
valid_players = player_season_counts[player_season_counts >= 3].index
df_filtered = df_sorted[df_sorted['player_id'].isin(valid_players)]

# Filter out players who throw less than 60% fastballs
df_filtered = df_filtered[df_filtered['fastball_percentage'] >= 60]
```

We are nearly there, we just need to get the relevant data in the correct rows. Since we are measuring last season's workload to this season's percent change, we are going to want to shift all of our workload data last one, and store that in a column named "prev_workload". Then, we can drop all first-year pitchers, who won't have a percent change or prev_workload.

```python
df_filtered['prev_workload'] = df_filtered.groupby('player_id')['workload'].shift(1)


df_shifted = df_filtered.dropna(subset=['prev_workload'])
df_shifted.head()
```

Now that our data is all set up, we can start to graph it. First, let's graph a scatter plot of the data where the x-axis is "prev_workload" and the y-axis is "pitch_speed_pct_change". This will give us a general idea of how workload from the previous season affects the pitch speed the next season.

```python

# Create scatter plot
plt.figure(figsize=(8,6))
plt.scatter(df_shifted['prev_workload'], df_shifted['pitch_speed_pct_change'], alpha=0.5)

# Labels and title
plt.xlabel("Workload (Total Pitches * Avg Speed)")
plt.ylabel("Pitch Speed % Change (Next Season)")
plt.title("Impact of Workload on Pitch Speed Decline")
plt.axhline(0, color='gray', linestyle='--', linewidth=1)
```

This scatter plot represents the relationship between a pitcher's workload in the previous year and the percentage change in their fastball velocity in the following year. The x-axis represents the prior year's workload, calculated as the product of fastball speed and pitch count, while the y-axis denotes the percentage change in fastball velocity.

At first, the data doesn't seem to have much relation. However, we do notice that there is a group on the left separated from the rest of the data. This data is most likely from the 2021 season, where the prev_workload is low since in 2020, the season was shortened due to covid. Let's regraph the data with the 2021 marked red.

```python
# Create scatter plot
plt.figure(figsize=(8,6))
plt.scatter(df_shifted['prev_workload'], df_shifted['pitch_speed_pct_change'], alpha=0.5)

df_2021 = df_shifted[df_shifted['year'] == 2021]
plt.scatter(df_2021['prev_workload'], df_2021['pitch_speed_pct_change'], color='red', alpha=0.5, label="2021 Data")

# Labels and title
plt.xlabel("Previous Season Workload (Total Pitches * Avg Speed)")
plt.ylabel("Pitch Speed % Change From Last Season")
plt.title("Impact of Workload on Pitch Speed Decline")
plt.axhline(0, color='gray', linestyle='--', linewidth=1)
```

Data points are color-coded, with red representing the 2021 year and blue indicating all other years. A horizontal dashed line at y = 0 serves as a reference, distinguishing players who experienced an increase (above the line) or decrease (below the line) in fastball speed.

Notably, red points are clustered toward lower workloads, reflecting the impact of the shortened 2020 year, whereas blue points display a broader distribution. The clustering of 2021 data points at lower workloads suggests that the reduced opportunities to pitch during the COVID-19-affected 2020 year may have allowed pitchers more time for recovery, potentially mitigating velocity loss. This trend aligns with the overall negative correlation observed, supporting the idea that increased workload may contribute to a greater decline in fastball speed in subsequent years.

As we can see in the above graph, most of the data where the prev_workload is low is due to the 2020 season being shortened. Now, let's check how that season, where pitchers had less workload, affected the pitchers' speed the next season. To explore this, let's average all of the pitchers together by year, so that we can see how different 2021 is from the rest of the data.

```python
# Group by year and calculate the average prev_workload and pitch_speed_pct_change
df_yearly_avg = df_shifted.groupby('year', as_index=False).agg({
    'prev_workload': 'mean',
    'pitch_speed_pct_change': 'mean'
})
```

```python
# Create scatter plot
plt.figure(figsize=(8,6))
plt.scatter(df_yearly_avg['prev_workload'], df_yearly_avg['pitch_speed_pct_change'], alpha=0.5)

# Add year labels to each point
for i, row in df_yearly_avg.iterrows():
    plt.text(row['prev_workload'], row['pitch_speed_pct_change'], str(int(row['year'])), 
             fontsize=10, ha='center', va='bottom')
    
# Labels and title
plt.xlabel("Workload (Total Pitches * Avg Speed)")
plt.ylabel("Pitch Speed % Change (Next Season)")
plt.title("Impact of Workload on Pitch Speed Decline By Year")
plt.axhline(0, color='gray', linestyle='--', linewidth=1)
```

With this data, we can see a clear outlier in the 2021 season. In the 2021 season, pitchers' average pitch speed did not drop from 2020, unlike in every other season.

As we can see above, the line of best fit suggests a strong negative correlation between workload (total pitches * average speed) and the percentage change in pitch speed in the following season. Most data points align closely with the trend line, reinforcing the relationship.

Now, let's analyze the results using Pearson's correlation coefficient.

```python
plt.figure(figsize=(8,6))
plt.scatter(df_yearly_avg['prev_workload'], df_yearly_avg['pitch_speed_pct_change'], alpha=0.5)

for i, row in df_yearly_avg.iterrows():
    plt.text(row['prev_workload'], row['pitch_speed_pct_change'], str(int(row['year'])), 
             fontsize=10, ha='center', va='bottom')

x = df_yearly_avg['prev_workload']
y = df_yearly_avg['pitch_speed_pct_change']
slope, intercept = np.polyfit(x, y, 1)
regression_line = np.poly1d((slope, intercept))

plt.plot(x, regression_line(x), color='red', linestyle='-', linewidth=2, label=f"y = {slope:.2f}x + {intercept:.2f}")

plt.xlabel("Workload (Total Pitches * Avg Speed)")
plt.ylabel("Pitch Speed % Change (Next Season)")
plt.title("Impact of Workload on Pitch Speed Decline By Year")
plt.axhline(0, color='gray', linestyle='--', linewidth=1)
```

This real-world case study underscores the potential benefits of limiting excessive pitch counts or increasing rest periods to sustain a pitcher’s effectiveness over time.

As a result, developing a **more diverse pitch repertoire**—including effective **breaking and offspeed pitches**—could be a key strategy for maintaining long-term success. Pitchers who **adapt their approach before a noticeable decline sets in** may have **longer, more sustainable careers** than those who attempt to maintain velocity at all costs.
