## Background and Prior Work


Our project’s purpose is to evaluate which of the seven different highway types – 4+ UND, 4+ DIV, 4+ DIV EXP, 2 AND 3 LN, 2 AND 3 LN EXP, NON FWY, and FREEWAY – is the safest to drive in. Undivided conventional highways, which are referred to as “4+ UND” in our dataset, have lanes that are equal to or greater than 4, but have no barrier between the opposing roads. Divided conventional highways, which are seen as “4+ DIV” in our dataset, have lanes that are equal to or greater than 4 and have a barrier between the opposing roads. Expressways, denoted with “EXP” in our dataset, are highways that aim to maintain high-speed roads with controlled access, but also may have crossings and/or traffic signals. Expressways are also commonly known for high occupancy toll lanes like the “FasTrak” lanes seen throughout California. “2 AND 3 LN EXP” are expressways with 2-3 lanes, while “4+ DIV EXP” are expressways that have divided barriers with a number of lanes equal to or greater than 4. Additionally, in this dataset, “NON FWY” highways are calculated as the total of the undivided, divided, and express highways (4+ UND, 4+ DIV, 4+ DIV EXP, 2 AND 3 LN, and 2 AND 3 LN EXP). All “NON FWY” highways aim to maintain varying speeds (with expressways on the higher end of the speed limit) and may have crossings and/or traffic signals. The last and final category, “FREEWAY,” marks a specific type of highway that provides full-access control on the road with no direct access from adjacent roads, no crossings, and no traffic signals.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)

California being one of the most populated, yet beautiful states, has driven our motive to analyze car crashes on various highways, as an approach to enlighten our audience on the ways that safety regulations on these highways can be improved. We chose to analyze the data in urbanized areas of California, since these areas tend to have busier highways with the most traffic. A study done in 2020 used a machine learning-based approach to predict the short-term impacts of lane changing maneuvers on traffic oscillations<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2). They found that during traffic with the fastest speed and largest acceleration/deceleration rate in tight traffic, motorcycles were dangerously maneuvering by lane changing in smaller gaps and with larger speed differences. They also found that traffic flow was another factor in collisions during lane change, while truck lane changing showed restricted traffic flow. With this in mind, highways that require less lane changing would be the most optimal for safety. For instance, expressways pose very little danger in the severity of congestion from restricted traffic flow and allow vehicles to move at a relatively steady pace that is away from nearby freeway exits and freeway entrances. Another study done in 2020 revealed that expressway lanes improve traffic performance and mobility<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3). The study compared the performance of traffic flows between general-purpose lanes, which are defined as the ‘main’ highway, and expressway lanes. Performance was observed during various hours of the day, specifically peak a.m. and peak p.m. hours that are influenced by work or school commute. Expressway lanes were shown to be ‘more reliable’ in the flow of traffic heading to and from work or school, but on the rare occasion that they are closed due to construction, general-purpose lanes are utilized and become reliable. Overall, the expressway lanes mitigate traffic congestion, making them much more reliable in terms of efficient travel time and safety.

# **Chart 5: Comparing crash rate vs. travel volume (MVM)**

```python
crash_data = crash_data.reset_index(drop=True)
crash_data['TRAVEL_MVM'] = pd.to_numeric(crash_data['TRAVEL_MVM'], errors='coerce')
crash_data['TOTAL_CRASH_RATE'] = pd.to_numeric(crash_data['TOTAL_CRASH_RATE'], errors='coerce')
sns.set_style("whitegrid")

g = sns.FacetGrid(crash_data, col="LANE_TYPE", col_wrap=3, sharex=False, sharey=True, height=4, aspect=1.2)
g.map_dataframe(sns.scatterplot, x="TRAVEL_MVM", y="TOTAL_CRASH_RATE", alpha=0.7)
g.map_dataframe(sns.regplot, x="TRAVEL_MVM", y="TOTAL_CRASH_RATE", scatter=False, ci=None, line_kws={'color':'red'})

g.set_axis_labels("Travel Volume (MVM)", "Total Crash Rate")
g.fig.suptitle("Crash Rate vs. Travel Volume by Lane Type", fontsize=14)
plt.tight_layout(rect=[0, 0, 1, 0.95])  # Adjust layout to fit title
plt.show()
```

Overall, these graphs demostrate how there is not a strong relationship with the amount of travel volume causing a higher amount of total crash rates. There is a slight amount of increases in crashes in the freeway and expressway lanes with higher amount of travel volume; however, this could be due to higher amount of congestion within lanes. So, these graphs indicate how there is not a strong correlation with amount of travel volume causing a higher amount of total crash rates.
