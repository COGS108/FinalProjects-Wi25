## Cleaning the Datasets

Our analysis involved multiple datasets from different cities across California, each containing information on hate crimes. To ensure consistency and ease of analysis, we undertook a thorough data cleaning and standardization process:  

1. Merging Datasets: Since we had data from multiple cities, we combined all datasets into a single, unified dataset for comprehensive analysis.  

2. Selecting Relevant Variables: Each dataset contained numerous columns, many of which were not relevant to our research. We focused only on key variables: crime type, motivation, month, and year —the essential components for understanding trends in hate crimes.  

3. Standardizing Crime Categories: Crime types were reported differently across datasets, often with varying levels of detail. To ensure consistency, we grouped them into broader, standardized crime categories. 

4. Standardizing Motivation Categories: Similar to crime types, bias motivations were reported inconsistently across different datasets. We standardized them into broad categories, ensuring uniform classification of hate crime motives. 

This cleaning process ensured that our dataset was structured, consistent, and ready for exploratory data analysis.  

```python
import pandas as pd
```

```python
# Defining helper functions for standardization

def standardise_month(month_number):
    month_mapping = {
        1: "January", 2: "February", 3: "March", 4: "April", 5: "May", 6: "June",
        7: "July", 8: "August", 9: "September", 10: "October", 11: "November", 12: "December"
    }
    return month_mapping.get(month_number, "Unknown")

def standardize_bias_motivation(value):
    mapping = {
        "Race": "Race/Ethnicity",
        "Race ": "Race/Ethnicity",
        "Race/Ethnicity": "Race/Ethnicity",
        "Race/Ethnicity/Ancestry": "Race/Ethnicity",
        "Race/Religion": "Race/Ethnicity",
        "Race/Gender": "Race/Ethnicity",
        "Race/Sexual Orientation": "Race/Ethnicity",
        "Race/Disability": "Race/Ethnicity",
        "Sexual Orientation": "Sexual Orientation",
        "Sexuality": "Sexual Orientation",
        "Religion": "Religion",
        "Anti-Religion": "Religion",
        "Anti-Religious": "Religion",
        "Gender": "Gender",
        "Gender Nonconforming": "Gender",
        "Anti-Transgender": "Gender",
        "Religion/Sexual Orientation" : "Other",
        "Other": "Other",
        "Disability": "Disability"
    }
    
    return mapping.get(value, value) 

"""
Assault: "Aggravated Assault", "Simple Assault"
Property Crimes: "Destruction/Damage/Vandalism", "Burglary", "Arson", "Counterfeiting/Forgery"
Theft: "Robbery", "Larceny-Theft", "Motor Vehicle Theft"
Violent Crimes: "Murder", "Kidnapping/Abduction", "Extortion/Blackmail"
Sexual Crimes: "Rape", "Fondling", "Sodomy"
Fraud/Deception: "False Pretense/Swindle/Confidence Game"
Weapons Violations: "Weapons Law Violations"
Drug Crimes: "Drug/Narcotic Violations", "Drug Equipment Violations"
Pornography: "Pornography/Obscene Material"
Intimidation: "Intimidation"

"""

def standardize_crime(value):
    crime_mapping = {
        'Intimidation': 'Intimidation',
        'Threat': 'Intimidation',
        'Threats': 'Intimidation',
        'Southeastern': 'Intimidation',
          
        'Aggravated Assault': 'Assault',
        'Simple Assault': 'Assault',
        'Assault, No Weapon': 'Assault',
        'Assault, w/Weapon': 'Assault',
        'Assualt': 'Assault',
        'Battery': 'Assault',
        
        'Threat': 'Threat',
        'Threat, Phone Call': 'Threat',
        'Bomb Threat': 'Threat',
        
        'Robbery': 'Theft',
        'Larceny-Theft': 'Theft',
        'Motor Vehicle Theft': 'Theft',
        'Theft': 'Theft',
        
        'Burglary': 'Property Crime',
        'Destruction/Damage/Vandalism': 'Property Crime',
        'Vandalism': 'Property Crime',  
        'Counterfeiting/Forgery': 'Property Crime',
        'Arson': 'Property Crime',
        
        'Murder': 'Violent Crime',
        'Attempted Murder': 'Violent Crime',
        'Kidnapping/Abduction': 'Violent Crime',
        'Extortion/Blackmail': 'Violent Crime',
        
        'Rape': 'Sexual Assault',
        'Fondling': 'Sexual Assault',
        'Sodomy': 'Sexual Assault',

        'Pornography/Obscene Material': 'Pornography',
        
        'Weapons Law Violations': 'Weapons Violation',

        'Drug/Narcotic Violations': 'Drug Crime',
        'Drug Equipment Violations': 'Drug Crime',
        
        'False Pretense/Swindle/Confidence Game': 'Fraud',
        'Disorderly Conduct': 'Public Order Crime',       
        
        'Other': 'Other' 
        
    }
    
    return crime_mapping.get(value, value)

def reorder_columns(df):
    desired_order = ["County", "Year", "Month", "Crime", "Motivation"]
    return df[desired_order]

def sort_by_year_month(df):
    month_order = [
        "January", "February", "March", "April", "May", "June", 
        "July", "August", "September", "October", "November", "December"
    ]
    
    df["Month"] = pd.Categorical(df["Month"], categories=month_order, ordered=True)
    df = df.sort_values(by=["Year", "Month"]).reset_index(drop=True)
    
    return df  
```

## San Diego Hate Crimes

```python
sd = pd.read_csv("Datasets/SD.csv")
sd
```

Lots of info! We're mostly wanting the dates and the motivations, though, so we're going to only keep the columns that pertain to that!

```python
columns_to_keep = ["month", "year","motivation","crime"]

sd_cleaned = sd[columns_to_keep].dropna().reset_index(drop=True)

sd_cleaned = sd_cleaned.rename(columns={
    "month": "Month",
    "year": "Year",
    "motivation": "Motivation", 
    "crime": "Crime"
})

sd_cleaned["County"] = "SAN DIEGO"

sd_cleaned ["Month"] = sd_cleaned ["Month"].apply(standardise_month)

sd_cleaned = sd_cleaned[(sd_cleaned["Year"] >= 2016) & (sd_cleaned["Year"] <= 2023)]

sd_cleaned = sd_cleaned.sort_values(by="Year", ascending=True).reset_index(drop=True)

sd_cleaned["Motivation"] = sd_cleaned["Motivation"].map(standardize_bias_motivation)

sd_cleaned["Crime"] = sd_cleaned["Crime"].map(standardize_crime)

sd_cleaned = reorder_columns(sd_cleaned)

sd_cleaned = sort_by_year_month(sd_cleaned)

sd_cleaned
```

## Los Angeles Crimes 2010-2019

```python
la1 = pd.read_csv("Datasets/LA2016to2019.csv")
la1
```

As you can see, this is super messy. We also need to check which ones are actually hate crimes, since this lists all crimes. Luckily, the MOCODES are provided in a pdf (https://data.lacity.org/api/views/d5tf-ez2w/files/8957b3b1-771a-4686-8f19-281d23a11f1b?download=true&filename=MO_CODES_Numerical_20180627.pdf), so we can use that to filter through and see which crimes are hatecrimes! As such, we are only keeping the date and MOCODES.

```python
columns_to_keep = ["DATE OCC", "Crm Cd Desc", "Mocodes"]
la1_cleaned = la1[columns_to_keep]

hate_crime_mocodes = {"0903", "0921", "1907", "2035", "2036"}

def haters(mo_codes):
    if pd.isna(mo_codes):
        return False
    mo_list = set(mo_codes.split()) 
    return bool(hate_crime_mocodes & mo_list)  

la1_cleaned = la1_cleaned[la1_cleaned["Mocodes"].apply(haters)].reset_index(drop=True)

la1_cleaned
```

Now we can see how many hate crimes there are! Wow, that's a lot, and that's super depressing. Now we need to know how they're motivated, so we'll go through again and label them according to the motivation. We also want to rename the columns so they'll align with all the other datasets.

```python
hate_crime_categories = {
    "Race/Ethnicity": {"0903", "2035"},  # Hatred/Prejudice, Racial Slurs
    "Gender": {"2036"},  # Hate-related language
    "Religion": {"2191"},  # Victim's religion
    "National Origin": {"2192"},  # Victim's national origin
    "Sexual Orientation": {"1222", "1235", "1239"},  # Homosexual, Bisexual, Transgender victims
    "Other": {"0921", "1907"}  # Hate Incident, Hate Crime materials
}

def define_haters(mo_codes):
    if pd.isna(mo_codes):
        return "Unknown"
    
    mo_list = set(mo_codes.split())  
    
    for category, codes in hate_crime_categories.items():
        if mo_list & codes:  
            return category
    
    return "Other"  

la1_cleaned["Hate Crime Type"] = la1_cleaned["Mocodes"].apply(define_haters)

la1_cleaned = la1_cleaned.rename(columns={
    "DATE OCC": "date",
    "Hate Crime Type": "Motivation"
})

la1_cleaned
```

Now we've labelled all the motivations, and we don't need the MOCODES anymore! We also want to standardize the dates (without time) and the motivations so that it will go nicely with our other datasets!

```python
def categorize_crime(crime):
    categories = {
        "Assault": ["ASSAULT", "BATTERY", "RESISTING ARREST", "KIDNAPPING", "LYNCHING"],
        "Property Crime": ["VANDALISM", "BURGLARY", "ARSON", "TRESPASSING", "ILLEGAL DUMPING", "THROWING"],
        "Theft": ["THEFT", "SHOPLIFTING", "PURSE SNATCHING", "PICKPOCKET", "BURGLARY FROM VEHICLE", "TILL TAP", "EMBEZZLEMENT", "EXTORTION"],
        "Violent Crime": ["HOMICIDE", "MANSLAUGHTER", "ROBBERY", "ATTEMPTED ROBBERY", "EXTORTION"],
        "Sexual Assault": ["RAPE", "SEXUAL", "SODOMY", "LEWD", "INDECENT EXPOSURE", "PANDERING", "PROSTITUTION", "ORAL", "TRAFFICKING"],
        "Fraud": ["FRAUD", "DOCUMENT FORGERY", "CREDIT CARDS", "BUNCO", "DEFRAUDING INNKEEPER", "CONSPIRACY", "COUNTERFEIT"],
        "Weapons Violation": ["WEAPON", "DISCHARGE FIREARMS", "BOMB", "SHOTS FIRED", "FIREARMS"],
        "Drug Crime": ["DRUGS", "CONTROLLED SUBSTANCE"],
        "Pornography": ["CHILD PORNOGRAPHY", "OBSCENE MATERIAL"],
        "Intimidation": ["THREATS", "STALKING", "THREATENING PHONE CALLS", "BRANDISH WEAPON", "BOMB SCARE"],
    }
    
    for category, keywords in categories.items():
        if any(keyword in crime for keyword in keywords):
            return category
    return "Other"

la1_cleaned["Crime"] = la1_cleaned["Crm Cd Desc"].apply(categorize_crime)
la1_cleaned = la1_cleaned.drop(columns=["Crm Cd Desc", "Mocodes"])

la1_cleaned
```

```python
la1_cleaned["Year"] = pd.to_datetime(la1_cleaned["date"]).dt.year
la1_cleaned["Month"] = pd.to_datetime(la1_cleaned["date"]).dt.strftime('%B')
la1_cleaned = la1_cleaned.drop(columns=["date"])

la1_cleaned['County'] = "LOS ANGELES"
la1_cleaned = la1_cleaned[['County', 'Year', 'Month', 'Crime', 'Motivation']]
la1_cleaned
```

```python
print(la1_cleaned.isnull().sum())
```

## Exploring trends in hate crime frequency by year and month

##### Interpretation

Comparing the proportion of hate crimes by year, we see that starting from 2016, reports of hate crimes becomes more and more frequent as time goes on. Despite a small lull in 2019, reports skyrocket after 2020. A majority of hatecrimes occur after 2020 - approximately 53%. By our hypothesis, we were hoping to see peaks on election years (particularly, 2016 and 2020), but there are no peaks on this chart, only an increasing slope. We could suggest that the jump in reports in 2021 was caused by fallout after that particular election, but since 2017 didn't see an intense jump, **we can't definitively say that election years are a cause of increased hate crime frequency.**

Next, we noticed that by month, there is a significant increase in hate crime frequency from the start of summer to mid-fall before dramatically falling for the holiday season. On average, 10% of hate crimes occur in June, while a mere 6% occur in January. There are many reasons this could be. For one, people in the summer are more likely to be out and about and taking vacation from work than other seasons. This could correlate for people having more free time to either commit or be victimized by hate crimes. On the flip side, hate crimes could decrease in winter due to the shorter days and colder weather making people less likely to stay out. Ultimately though, **our hypothesis regarding hate crimes being more frequent during election months has also been proven false.**

Now, let's relate it back to hate crimes. Are there correlations between the types of crimes people want to commit and the motivation? For example, are racial hate crimes more likely to be assault (and thus more violent) than other hate crimes? Or are the percentage of assault crime types consistent between all the crime types (i.e. all hate crimes are equally violent)?

```python
crime_motivation_counts = final_dataset.groupby(["Crime", "Motivation"]).size().reset_index(name="Count")
crime_motivation_counts = crime_motivation_counts.sort_values(by="Count", ascending=False)

plt.figure(figsize=(12, 6))

sns.barplot(data=crime_motivation_counts, x="Crime", y="Count", hue="Motivation")

plt.xticks(rotation=90, ha="right")  
plt.xlabel("Crime Type")
plt.ylabel("Count")
plt.title("Hate Crimes by Type and Motivation")
plt.legend(title="Motivation")

plt.show()
```

That bar graph shows the total counts distribution of the motivation given the crime. But we wanna know if one specific type of crime is more common with a certain hate crime. Additionally, we things like Fraud, Vandalism, or Drug Crime that are basically nonexistant in in comparison to the other crimes. So, let's normalize! We will be dividing by the total number of that specific motivation hate crimes to see the rates of the types of crimes. We will also be putting all negligably small crimes into "Other" so that it makes the charts a little cleaner.   

```python
custom_palette = [
    'lightcoral', 'gold', 'skyblue', 'green', 'orange','cyan', 'magenta', 
    'lightgreen', 'plum', 'salmon', 'darkblue', 'darkgreen'
]

total_motivation_counts = final_dataset["Motivation"].value_counts().to_dict()

crime_motivation_counts = final_dataset.groupby(["Crime", "Motivation"]).size().reset_index(name="Count")

crime_motivation_counts["Rate"] = crime_motivation_counts.apply(
    lambda row: row["Count"] / total_motivation_counts[row["Motivation"]], axis=1
)

unique_motivations = crime_motivation_counts["Motivation"].unique()
threshold = 0.03

for motivation in unique_motivations:
    motivation_data = crime_motivation_counts[crime_motivation_counts["Motivation"] == motivation]

    if motivation_data.empty:
        continue

    motivation_data = motivation_data.sort_values(by="Count", ascending=False)
    percentages = motivation_data["Rate"].values
    labels = motivation_data["Crime"].values

    large_slices = percentages >= threshold
    small_slices = ~large_slices

    if np.any(small_slices):
        large_labels = labels[large_slices].tolist()
        large_percentages = percentages[large_slices].tolist()

        small_sum = percentages[small_slices].sum()
        large_labels.append("Miscellaneous")
        large_percentages.append(small_sum)
    else:
        large_labels = labels.tolist()
        large_percentages = percentages.tolist()

    plt.figure(figsize=(7, 7))
    wedges, texts, autotexts = plt.pie(
        large_percentages,
        labels=large_labels,
        autopct=lambda p: f"{p:.1f}%" if p > 2 else "",
        wedgeprops={"edgecolor": "black"},
        colors=custom_palette[:len(large_percentages)]
    ) 

    plt.title(f"Breakdown of {motivation} Hate Crimes by Crime Type (Normalized)")
    plt.show()
```

### Analysis

So, we now have the percentage breakdowns of the type of crimes committed given a certain motivation.

- 62.7% of Disability Hate Crimes were done via assault, 13.7 via intimidation, 13.7% via property crime, and 9.8% via theft. 

- 50% of the Other Hate Crimes (which are the ones that were classified as Other in some of the datasets), had "Other" as its type of crime. However, we see that assault is the most common type of crime after that, at 12.8%, then 12.4% for sexual assault (so 25.2% of assault in total), then 10.3% weapons crime, and finally, 9.7% of weapons violations, with 4.8% of miscillaneous means.

- 52.4% of Gender Hate Crimes were done via assault, 14.9% via intimidation, 9.6% via weapons violation, 8.4% via property crime, and the remaining crimes are miscillaneous. 

- 44.4% of Race/Ethnicity Hate Crimes were done via assault, 24.6% via property crime, 11.7% via weapon violations, 11% via intimidation, and 8.3% via other means.

- 64.6% of Religion Hate Crimes were done via property crime, 18.8% via intimidation, 12.8% via assault, and 3.9% via other means. 

- 45.5% of Sexual Orientation Hate Crimes were done via assault, 23.3% via property crime, 21.6% via intimidation, 7.0% via theft, and 2.6% via other means. 

### Interpretation

Disability (62.7%) has the highest assault rate, followed by Gender (52.4%), and then Sexual Orientation (45.5%), then Race/Ethnicity(44.4%), and then Religion. Of the crime types, we are classifying assault as the most violent, and from this, we see that Race/Ethnicity is only the 4th most violent, close to the bottom.  **Thus, we can say that our hypothesis of Race/Ethnicity Hate Crimes being the most violent has been proven false.** We cannot definitively say that Disability is the most violent, however, as the number of total Disability Hate Crimes occurences is very small - being 51 total occurences. Thus, it could be that Disability Hate Crime reports are unstable due to that small sample size, and may not be fully representative of the actual trends. However, we can see that Religious Hate Crimes are likely to be less violent, at 12.8% assault rates, compared to the other rates which range from 40-60%. **Thus, our hypothesis of Religious Hate Crimes being the least violent has been supported by this data.** Religion has the highest property crime rates, at almost 2/3 of the Religious Hate Crimes being property crimes. This makes sense on some levels, given the amount of reports of vandalism of places of worship (churches, mosques, temples, etc.). 

Overall, we see that assault is the most common type of crime when committed with the motivation of hate. However, we see that certain demographics may be at more risk for this violence than others (Disability and Gender), at an over 50% chance of encountering violence should a hate crime be carried against those two groups. That's not to say Sexual Orientation and Race/Ethnicity groups are safe from this violence either, though, with 44.5% and 44.4% rate of assault. Anyone in these four groups should be aware of the risk of violence if they are in counties that have assault as the most common crime. On the other hand, if you are in the Bay Area, you are more likely to experience property crime, so it's not as violent, but still frustrating to have happen, nonetheless.

```python
relation = pd.get_dummies(data=final_dataset, columns=['Motivation'])
relation = relation.rename(columns={'Motivation_Race/Ethnicity' : 'Motivation_Race', 'Motivation_Sexual Orientation' : 'Motivation_Sexual_Orientation'})
relation['Motivation_Race'] = relation['Motivation_Race'].astype(int)
relation['Motivation_Other'] = relation['Motivation_Other'].astype(int)
relation['Motivation_Sexual_Orientation'] = relation['Motivation_Sexual_Orientation'].astype(int)
relation['Motivation_Disability'] = relation['Motivation_Disability'].astype(int)
relation['Motivation_Religion'] = relation['Motivation_Religion'].astype(int)
relation['Motivation_Gender'] = relation['Motivation_Gender'].astype(int)

relation['is_assault'] = (relation['Crime'] == 'Assault').astype(int)
relation['is_intimidation'] = (relation['Crime'] == 'Intimidation').astype(int)

model = smf.ols(formula='is_assault ~ Motivation_Race + Motivation_Sexual_Orientation', data=relation)
results = model.fit()

print(results.summary())
```

Next we will see how motivations impact the crime type being intimidation.

```python
model = smf.ols(formula='is_intimidation ~ Motivation_Religion + Motivation_Gender', data=relation)
results = model.fit()

print(results.summary())
```
