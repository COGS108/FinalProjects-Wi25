# Abstract

In this project, we looked at Amazon review data to determine if correlation between price and customer satisfaction of clothing changed before and after COVID-19. We analyzed “customer satisfaction” as a star rating (out of five stars) and the sentiment of text reviews on an item. General consumer trends changed around the time of the COVID-19 pandemic, as inflation worsened and fast fashion (cheaply and rapidly produced clothing) became more popular. We hypothesized that these economic changes would affect how contented customers were when online shopping. Higher priced items which were high-quality pre-COVID would potentially decrease in quality since more companies adopted fast fashion practices after COVID, resulting in lower ratings post-2020. Items that cost less before COVID may have increased in price after pandemic inflation but would not have improved in quality, also leading to low ratings from customers.  

We began with two datasets. The first included metadata of clothing items listed on Amazon, and the second included review data of these items. Before analysis, we cleaned the two datasets and split them into two time periods: 2016-2019 ("pre-COVID") and 2020-2023 ("post-COVID"). We also limited our analysis to only shirts so that the prices of different items were comparable, and filtered out unverified purchases to limit bots or fake reviews. After wrangling the data, we started exploring its shape. We looked at the distributions of price and star rating and also looked at central tendencies across the two time periods. Overall, the pre- and post-COVID time periods were similar in distribution, with pre-COVID having a higher average price and post-COVID containing slightly more negative reviews.

For our main analysis, we plotted the relationship between price and star rating and carried out an OLS regression model as well as a t-test. Next, we carried out text analysis for the reviews, using TF-IDF to vectorize the text and VADER to examine the sentiment. We looked at word frequency distribution and frequency of different keywords across the time periods. We also used an OLS model and t-test to compare the price and the sentiment of the reviews. Based on the results of our analysis, we concluded that there has been a significant difference in the effect of price on customer satisfaction pre- versus post-COVID.

# Hypothesis



Before 2020, we predict that customer satisfaction with Amazon-listed shirts was generally consistent across price ranges, except for exceptionally low-quality items. However, from 2020 onward, we expect a positive correlation between price and satisfaction, as inflation and supply chain disruptions led to higher production costs and potential declines in quality for lower-priced shirts. Customers who previously bought higher-quality shirts may have opted for cheaper alternatives, leading to dissatisfaction and lower ratings. We will carry out OLS to discover more about the relationship. 

Next, we will perform Ordinary Least Squares (OLS) regression using the Patsy library to analyze the relationship between price and rating for two different time periods (2016-2019 and 2020-2023).
H0 = The effect of price on ratings has not changed between the two time periods.
H1 = The effect of price on ratings has changed significantly between the two time periods.

```python
# Define the formula for the regression model using patsy
formula = "rating ~ price"

# Create design matrices using patsy for 2016-2019 data
y_2019, X_2019 = patsy.dmatrices(formula, data=shirt_reviews2019)

# Create design matrices using patsy for 2020-2023 data
y_2020, X_2020 = patsy.dmatrices(formula, data=shirt_reviews2020)

# Fit OLS (Ordinary Least Squares) models for both time periods
model_2019 = sm.OLS(y_2019, X_2019).fit()
model_2020 = sm.OLS(y_2020, X_2020).fit()

print("OLS Regression for the relationship between price and rating for reviews between 2016-2019\n")
print(model_2019.summary())
```

```python
print("OLS Regression for the relationship between price and rating for reviews between 2020-2023\n")
print(model_2020.summary())
```

The results of this regression test indicate a notable shift in the effect of price on customer ratings between the two time periods. For pre-COVID, the estimated price coefficient is 0.0011, suggesting a very weak positive relationship between price and rating. However, this effect is not statistically significant (p-value = 0.284), meaning we cannot conclude that price had a meaningful impact on customer ratings before COVID-19. The R-squared value is 0, indicating that price explains virtually none of the variance in ratings. For post-COVID, the estimated price coefficient is 0.0116, which, while still small, is an order of magnitude larger than in the pre-COVID period. Unlike the pre-COVID data, this effect is statistically significant (p-value ≈ 0.00), suggesting that price had a measurable impact on customer ratings after COVID-19. The R-squared value is 0.014, meaning that while price explains some variation in ratings, the overall impact remains limited. 

Now we will perform a t test, which works well for our data because this test is designed for situations where variance is estimated from a sample (which our data is).

```python
price_coef_2019 = model_2019.params[1]  # Price coefficient for 2016-2019
price_coef_2020 = model_2020.params[1]  # Price coefficient for 2020-2023

# Compute the standard errors of the coefficients
se_2019 = model_2019.bse[1]
se_2020 = model_2020.bse[1]

# Compute degrees of freedom
df_2019 = model_2019.df_resid
df_2020 = model_2020.df_resid
df_pooled = df_2019 + df_2020

#Compute pooled standard error (assuming equal variances)
sp = np.sqrt(((df_2019 * se_2019**2) + (df_2020 * se_2020**2)) / (df_2019 + df_2020))

# Compute t-statistic
t_stat = (price_coef_2019 - price_coef_2020) / (sp * np.sqrt(1 / (df_2019 + 1) + 1 / (df_2020 + 1)))

# Display results
print(f"Price Coefficient (2016-2019): {price_coef_2019:.4f}")
print(f"Price Coefficient (2020-2023): {price_coef_2020:.4f}")
print(f"Standard Error for 2019 price coefficient: {se_2019:.4f}")
print(f"Standard Error for 2020 price coefficient: {se_2020:.4f}")
print(f"t statistic: {t_stat:.4f}")
```

```python
import scipy.stats as stats

#Calculate two-tailed p-value
p_value = 2 * stats.t.sf(abs(t_stat), df=df_pooled)
print(f"P-value: {p_value:.4f}")
```

The negative t-statistic of -403.1403 means that the price coefficient in the pre-COVID period was meaningfully lower than in the post-COVID period. The p-value was essentially 0, meaning that the difference between the two coefficients is statistically significant at a significance level of alpha = 0.05. There is strong statistical evidence to reject the null hypothesis (p < 0.05), indicating that the effect of price on ratings has changed significantly between the two time periods.

We define a function that finds the average length of a string by word count and character count. This will allow us to analyze the length of the reviews themselves, and their titles. 

```python
def check_average_length(col):
    i = 0
    sum_char = 0
    sum_word = 0
    for item in col:
        sum_word = sum_word + len(item.split())
        sum_char = sum_char + len(item)
        i=i+1

    average_char = round(sum_char/i)
    average_word = round(sum_word/i)
    print('The average amount of characters is '+str(average_char))
    print('The average amount of words is '+str(average_word))
```

Next, we define a function that finds the frequency of a word across reviews. This will allow us to analyze if there are particularly positive or negative words that show up frequently: 

```python
def check_word_freq(df, word):

    word = df[df['text'].str.contains(word, case=False, na=False)]
    proportion = word.groupby('rating').size() / df.groupby('rating').size()
    
    out = combined_df = pd.concat([proportion, proportion], keys=['text', 'title'], axis=1)

    return out
```

We will now analyze customer reviews from 2016-2019 and 2020-2023 to determine how often specific topics (ex: Price, Quality, Complaints) are mentioned:

```python
# Sets up keyword categories, with 10 words in each category
keyword_categories = {
    "Customer Service": ["customer service", "support", "helpful", "rude", "fast shipping", "slow shipping", "delayed", "lost package", "tracking", "response"],
    "Sustainability": ["sustainable", "organic", "eco-friendly", "recycled", "ethical", "fair trade", "fast fashion", "environment", "green", "biodegradable"],
    "Durability": ["durable", "lasted", "worn", "wear", "tear", "shrunk", "faded", "pilled", "strong", "fragile"],
    "Complaints": ["disappointed", "horrible", "bad", "poor", "refund", "return", "ripoff", "broken", "not worth", "would not buy"],
    "Positive Feedback": ["amazing", "great", "excellent", "love", "best", "high quality", "perfect", "worth it", "fantastic", "comfortable"],
    "Price": ["expensive", "cheap", "affordable", "price", "cost", "worth", "overpriced", "budget", "value", "deal"],
    "Quality": ["quality", "durable", "sturdy", "cheap", "poor", "excellent", "well-made", "flimsy", "premium", "inferior"],
    "Material": ["cotton", "polyester", "fabric", "soft", "rough", "material", "silk", "denim", "thread", "texture"],
    "Social Media": ["Instagram", "Facebook", "Twitter", "TikTok", "viral", "influencer", "trend", "hashtag", "review", "recommend"]
}

# Function to count keyword occurrences and compute proportions
def count_keyword_proportion(text_series, keywords):
    text_combined = " ".join(text_series.dropna()).lower()
    word_counts = Counter(text_combined.split())
    total_words = sum(word_counts.values())  # Total word count in reviews
    keyword_count = sum(word_counts[word] for word in keywords if word in word_counts)
    return keyword_count / total_words if total_words > 0 else 0  # Proportion of keyword mentions

# Set dictionaries
proportions_2019 = {}
proportions_2020 = {}

# Compute proportions by looping through keywords
for category, keywords in keyword_categories.items():
    proportions_2019[category] = count_keyword_proportion(shirt_reviews2019['text'], keywords)
    proportions_2020[category] = count_keyword_proportion(shirt_reviews2020['text'], keywords)

# Convert results to DataFrame
proportions_df = pd.DataFrame({
    "Year": ["2016-2019", "2020-2023"],
    **{f"{category}": [proportions_2019[category], proportions_2020[category]]
       for category in keyword_categories}
})

print(proportions_df)
```

```python
#Visualizing proportions_df:

# Convert proportions_df format for easier visualization
proportions_long_df = proportions_df.melt(id_vars=["Year"], var_name="Keyword Category", value_name="Proportion")

# Create bar plot
plt.figure(figsize=(12,6))
sns.barplot(data=proportions_long_df, x="Keyword Category", y="Proportion", hue="Year", palette=["blue", "#ff7300"])

# Customize plot
plt.xlabel("Keyword Category")
plt.ylabel("Proportion of Mentions")
plt.title("Proportion of Keyword Mentions Over Time")
plt.xticks(rotation=30)
plt.legend(title="Time Period")

plt.show()
```

The proportions between time periods are similar, although there are some slight differences in certain keywords. Material is also more frequent post-COVID, which could indicate that consumers are more aware or care more about the quality of material of their shirts post-COVID.

Now that we understand that line of points and have visualized price and review sentiment, it's time to perform an OLS regression test on this relationship. 

```python
# Define the formula for the regression model using patsy
formula = "sentiment ~ price"

# Create design matrices using patsy for 2016-2019 data
y_2019, X_2019 = patsy.dmatrices(formula, data=shirt_reviews2019)

# Create design matrices using patsy for 2020-2023 data
y_2020, X_2020 = patsy.dmatrices(formula, data=shirt_reviews2020)

# Fit OLS (Ordinary Least Squares) models for both time periods
model_2019 = sm.OLS(y_2019, X_2019).fit()
model_2020 = sm.OLS(y_2020, X_2020).fit()

print("OLS Regression for the relationship between price and sentiment of reviews between 2016-2019\n")
print(model_2019.summary())
```

```python
print("OLS Regression for the relationship between price and sentiment of reviews between 2020-2023\n")
print(model_2020.summary())
```

 These results indicate a notable shift in the effect of price on customer sentiment from pre- to post- pandemic. For pre-COVID, the estimated price coefficient is -0.0003, suggesting a very weak negative relationship between price and sentiment. This effect is not statistically significant (p-value = 0.344) and the R-squared value is 0. This means we cannot conclude that price had a meaningful impact on customer sentiment, since price explains none of the variance in ratings. For post-COVID, the estimated price coefficient is 0.0019, which is slightly larger than in the pre-COVID period. Unlike the pre-COVID data, this effect is statistically significant (p-value ≈ 0.00), suggesting that price had a measurable impact on customer sentiment after COVID-19. The R-squared value is 0.004, again meaning that price explains some of the variation in ratings, but the impact remains limited. 

We will also perform a t test in the same way we did for price and star rating earlier. 

```python
price_coef_2019 = model_2019.params[1]  # Price coefficient for 2016-2019
price_coef_2020 = model_2020.params[1]  # Price coefficient for 2020-2023

# Compute the standard errors of the coefficients
se_2019 = model_2019.bse[1]
se_2020 = model_2020.bse[1]

# Compute degrees of freedom
df_2019 = model_2019.df_resid
df_2020 = model_2020.df_resid
df_pooled = df_2019 + df_2020

#Compute pooled standard error (assuming equal variances)
sp = np.sqrt(((df_2019 * se_2019**2) + (df_2020 * se_2020**2)) / (df_2019 + df_2020))

# Compute t-statistic
t_stat = (price_coef_2019 - price_coef_2020) / (sp * np.sqrt(1 / (df_2019 + 1) + 1 / (df_2020 + 1)))

# Display results
print(f"Price Coefficient (2016-2019): {price_coef_2019:.4f}")
print(f"Price Coefficient (2020-2023): {price_coef_2020:.4f}")
print(f"Standard Error for 2019 price coefficient: {se_2019:.4f}")
print(f"Standard Error for 2020 price coefficient: {se_2020:.4f}")
print(f"t statistic: {t_stat:.4f}")
```

```python
#Calculate two-tailed p-value
p_value = 2 * stats.t.sf(abs(t_stat), df=df_pooled)
print(f"P-value: {p_value:.4f}")
```

This t test produced similar findings to the price~star rating t test. Comparing the price coefficients before and after COVID strongly supports our hypothesis that the relationship between price and sentiment changed significantly. The negative t-statistic of -268.2615 means that the price coefficient in the pre-COVID period was significantly lower than in the post-COVID period. Although there isn’t as much of a change for our second set of OLS models (t of -403 vs -268), the p-value remains essentially 0, meaning that the difference between the two coefficients is statistically significant at a significance level of alpha = 0.05. We can reject the null hypothesis, which states that there is no difference in the effect of price on review sentiment before and after COVID-19. Thus, we conclude that the relationship between price and review sentiment changed significantly post-COVID.
