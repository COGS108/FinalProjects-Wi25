# Abstract

This project examines the relationship between the VIX index and the S&P 500 during 2000–2020. We investigate whether fluctuations in the VIX correlate with S&P 500 volatility and returns, and if the VIX can predict trading volumes. Using publicly available data, we calculate logarithmic returns and a 21-day rolling standard deviation for volatility. Our analysis reveals clear evidence of volatility clustering, a statistically significant negative correlation (–0.1394) between volatility and returns, and a modest positive correlation (0.2314) between VIX levels and log-transformed trading volumes. These findings highlight the potential of the VIX as a risk management tool in predicting market behavior.

## Background and Prior Work

Financial markets are inherently volatile and understanding the dynamics of volatility is crucial for investors, traders, and policymakers. Market volatility influences investment decisions, risk management strategies, and economic policy responses (Whaley, 2000).<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) The VIX Index, often referred to as the "fear gauge," measures the market's expectation of 21-day volatility based on S&P 500 index options(Cboe, 2023).<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) It is widely used as a barometer of investor sentiment and market stress. Higher VIX values typically indicate increased investor uncertainty and potential market downturns, while lower values suggest stability (Carr & Wu, 2006).<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) It is crucial, however, to differentiate volatility and VIX.  

Volatility is the degree of flux in the index over time. There are three kinds of volatility - implied volatility (extracted from pricing options from VIX index), realized volatility (calculated from same day shares), and historical volatility (Standard deviation).  

This study examines the relationship between fluctuations in the VIX and in the S&P 500, specifically looking at volatility and its relation to S&P500 trading volumes and returns. This could help us predict drawdowns in the S&P500, which are generally observed in major market stress, not just including:

- The 2000-2002 Dot-Com Crash, when speculative technology stocks collapsed, leading to prolonged declines.

- The 2008 Global Financial Crisis, where the failure of major financial institutions caused the S&P 500 to drop over 50%.

- The Great Depression, Watergate

- The 2020 COVID-19 Crash, which saw the fastest recorded market decline due to economic shutdowns and uncertainty.

Volatility clustering is a phenomenon in financial markets where periods of high volatility follows other periods of high volatility, effectively predicting market turbulence over time. Spikes in the VIX are also heightened, despite one-month lagbacks, coinciding with increase trading volume as hedge funds and investors react to volatile conditions on the fly. This will help is predict whether volatility spikes serve as an indicator of market stress, predicting market participation.

## Analysis 1 - Volatility Clustering and Returns


Initially we cleaned the data for SPX - the S&P500 index and the VIX - Volatility index. We first calculate **return** - defined as the logarithm of the ratio between the Current Price and the Previous day's Price.

log returns formula.png

This formula generates for simple returns.
We are using logartithms here for additivity - this factors in compounding (where you invest money earned from previous investments).

```python
# Logarithms of prices
df["Returns"] = np.log(df["Close"] / df["Close"].shift(1))
df["Volatility"] = df["Returns"].rolling(window = 21).std() # We take one trading month(21days) as a window size for taxonomical purposes
df = df.dropna()
df
```

Plotting for returns vs. date:

```python
# Ten years of S&P 5000 daily return
plt.figure(figsize=(12,6))
plt.plot(df["Returns"], label="Daily Returns", color="blue", alpha=0.6)
plt.axhline(y=0, color="black", linestyle="--")
plt.xlabel("Date")
plt.ylabel("Returns")
plt.title("S&P 500 Daily Returns")
plt.legend()
plt.show()
```

Plotting the relationship between volatility and date:

```python
# Ten years of S&P 5000 volatility
plt.figure(figsize=(12,6))
plt.plot(df["Volatility"], label="21-day Rolling Volatility", color="red")
plt.xlabel("Date")
plt.ylabel("Volatility")
plt.title("S&P 500")
plt.legend()
plt.show()
```

We see that high and low return values almost "cluster" together, with nascent successive and previous values closely mirroring present returns. This is an indication of volatility clustering.


To verify whether the volatility truly exhibits clustering, the Autocorrelation Function (ACF) is used. The ACF (picture below) measures the correlation between current volatility and volatility after three days. If volatility clustering exists, ACF should remain relatively high in the short term, instead of random fluctuations. We wil eventually plot a volatility time series to observe clustering effects.

acf .png

If the ACF is significantly greater than zero, this is a prime indicator for volatility clustering. Volatility would be random if the ACF is close to zero.

```python
# Calculate the Autocorrelation Function, ACF）
acf_values = acf(df["Volatility"].dropna(), nlags=50) # We take 50-days as lagging
acf_values
```

```python
# Plot
plt.figure(figsize=(10,5))
plt.bar(range(1, 21), acf_values[1:21], color="blue", alpha=0.7)
plt.xlabel("Lag")
plt.ylabel("ACFk")
plt.title("Volatility Clustering")
plt.show()
```

We can see that the greater the time has passed, the lower the correlation!
This confirms our hypotheses.

#Part 2 - Plotting the relationship between volatility and returns

Now we are curious regarding the relationship between Volatility and return. If the market declines while volatility increases, it should satisfy the correlation(daily returns, volatility)<0.

```python
vix_series = vix_df['VIX_Close'].rename("VIX")
df = df.join(vix_series, how="inner")
# Inital renaming and prepping
```

```python
from scipy.stats import pearsonr
corr_volatility_returns, _ = pearsonr(df["VIX"].dropna(), df["Returns"].dropna())
print(f"correlation between volatility(VIX) and returns: {corr_volatility_returns:.4f}")

```

A negative correlation of -0.1394 suggests a weak relationship between volatility and returns. Higher values of VIX tend to associate with lower returns.

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x=df["Returns"], y=df["VIX"], alpha=0.3)
plt.xlabel("Returns")
plt.ylabel("Volatility")
plt.title("return vs. volatility")
plt.axvline(x=0, color="black", linestyle="--")
plt.axhline(y=df["VIX"].mean(), color="red", linestyle="--", label="Mean Volatility")
plt.legend()
plt.show()
```

We see that volatility is asymmetric with returns, reinforcing weak correlation between volatility and returns.
We now proceed to calcluate statistical significance using P-value.

```python
corr_vix, p_value = pearsonr(df["VIX"].dropna(), df["Returns"].dropna())

print(f"correlation between VIX and returns: {corr_vix:.4f}")
print(f"p-value: {p_value:.4f}")
```

p = 0 indicates that this is statistically significant, and that the relationship is NOT due to random chance.

```python
# （Returns > 0）
df_up = df[df["Returns"] > 0]
mean_vix_up = df_up["VIX"].mean()
std_vix_up = df_up["VIX"].std()
corr_vix_up, _ = pearsonr(df_up["VIX"], df_up["Returns"])

# （Returns < 0）
df_down = df[df["Returns"] < 0]
mean_vix_down = df_down["VIX"].mean()
std_vix_down = df_down["VIX"].std()
corr_vix_down, _ = pearsonr(df_down["VIX"], df_down["Returns"])

print(f"when the market goes up the mean value is: {mean_vix_up:.4f}, sd: {std_vix_up:.4f}, correlation: {corr_vix_up:.4f}")
print(f"when the market goes down the mean value is: {mean_vix_down:.4f}, sd: {std_vix_down:.4f}, correlation: {corr_vix_down:.4f}")
```

#Part 3 - finding the correlation between trading volumne and voaltility(vix)

```python
# Calculate the log_form of volume
df["Log_Volume"] = np.log(df["Volume"])

# Plot a comparison between normal volume distribution and log_form volume distribution
plt.figure(figsize=(12, 6))
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# NORMAL FORM
sns.histplot(df["Volume"], bins=50, kde=True, color="blue", ax=axes[0])
axes[0].set_xlabel("Volume")
axes[0].set_ylabel("Density")
axes[0].set_title("Distribution of Trading Volume (Raw)")

# LOGFORM
sns.histplot(df["Log_Volume"], bins=50, kde=True, color="red", ax=axes[1])
axes[1].set_xlabel("Log Volume")
axes[1].set_ylabel("Density")
axes[1].set_title("Distribution of Log Trading Volume")

plt.tight_layout()
plt.show()
```

### Observation1: The distribution of trading volume is often right-skewed (long right tail), meaning that most trading days have relatively low volumes, but occasionally, there are extreme spikes in volume. Using raw trading volume in regression analysis can be problematic, as extreme values may bias the model, making it more sensitive to these outliers.

### Obervation2: The distribution has two distinct peaks, indicating the presence of two different regimes or market states. This could suggest that trading volume behaves differently under varying market conditions, which is something we are interested in.

```python
# We plot the trading volume change overtime to see how the bimodal distribution
# forms to see if there are any outliers we need to remove from the dataset
plt.figure(figsize=(12, 5))
plt.plot(df.index, df["Log_Volume"], color='purple', alpha=0.7)
plt.xlabel("Date")
plt.ylabel("Log Volume")
plt.title("Log Trading Volume Over Time")
plt.show()
```

### observation: we found out there's a drift up from 2004. This could be resulted from a variety reasons ranging from the improvement of the trading market to the revolution of online trading technology due to the development of computer hardware and software. Since it's a natural process of human development and we are focusing on the correlation between trading volume and volatility, we will keep working with the full data without digging into the reason behind the growth in trading volumne.

```python
merged_df = df[['Log_Volume']].merge(df[['VIX']], left_index=True, right_index=True, how='inner')

corr_coeff, p_value = pearsonr(merged_df["VIX"], merged_df["Log_Volume"])
print(f"Pearson Correlation: {corr_coeff:.4f}, p-value: {p_value:.4g}")
```

# Discussion
Our analysis explored the relationship between **volatility, trading volume, and market returns** using historical data from the S&P 500 and the VIX index. The key findings are as follows:

### 1. Volatility Clustering  
- Our ACF analysis confirmed the presence of **volatility clustering**, indicating that periods of high volatility tend to be followed by high volatility, and low volatility tends to persist.  
- This aligns with well-documented financial market behaviors.

### 2. Volatility and Returns Correlation  
- We found that volatility and returns exhibit a **negative correlation(-0.1394)**, meaning that higher volatility is often associated with declining market returns.  
- This is consistent with the widely accepted **leverage effect**, where negative shocks increase market risk and cause further uncertainty.
- Volatility is more likely to increase when the stock market falls,

### 3. Volatility and Trading Volume Relationship  
- Our analysis suggests that **volatility and trading volume are positively correlated**, meaning that during highly volatile periods, trading volume tends to increase.  
- However, **this correlation does not imply causation**—both volatility and trading volume may be driven by external macroeconomic factors, investor sentiment, or market-wide liquidity constraints.

### 4. Statistical Insights on Trading Volume  
- The raw trading volume distribution was highly **right-skewed**, with extreme values distorting direct interpretations.  
- After applying a **log transformation**, the distribution became more **symmetrical**, reducing the influence of outliers and improving statistical robustness.  
- However, the log-transformed volume distribution exhibited **bimodal characteristics**, suggesting the presence of multiple market states (e.g., high-activity vs. low-activity periods).

### Conclusion
- **Causal Analysis**: To better understand the **cause-and-effect relationship** between trading volume and volatility, methods like **Granger causality tests** or **VAR models** could be applied.  
- **Market Regime Identification**: Given the **bimodal distribution** in trading volume, clustering techniques (e.g., K-Means, Gaussian Mixture Models) could be used to **segment market conditions** and study how volume dynamics differ.  
- **Event-Based Volatility Analysis**: Analyzing specific market events (e.g., financial crises, policy changes) in relation to volatility and volume could provide deeper insights into **what drives extreme market behaviors**.

### Final Remarks
While our findings support well-established theories in financial econometrics, they also highlight potential areas for further exploration.
