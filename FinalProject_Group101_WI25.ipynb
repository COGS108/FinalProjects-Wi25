## Background and Prior Work


Urban air pollution, particularly nitrogen dioxide (NO₂), remains a critical public health concern due to its direct impact on respiratory health and its role in forming secondary pollutants like ozone and nitrate particulates[<sup>1</sup>](#cite_ref-1). Data from California's air quality monitoring networks consistently show that NO₂ concentrations are highest near major freeways and heavy traffic corridors, creating localized "hot-spot" areas where exposure is significantly elevated[<sup>2</sup>](#cite_ref-2). These findings highlight the ongoing challenge of mitigating traffic-related air pollution, especially in urban environments where vehicle emissions are a dominant source of NO₂.

Research from the University of Southern California has provided compelling evidence that children living near high-traffic corridors face a significantly increased risk of developing asthma. Studies indicate that residing within 500 meters of such roads is associated with asthma risk elevations of 25% to nearly 50% compared to children living farther away[<sup>3</sup>](#cite_ref-3) [<sup>4</sup>](#cite_ref-4). Additionally, environmental justice concerns have emerged, as lower-income neighborhoods experience about 28% higher ambient NO₂ levels than wealthier areas, exacerbating existing health disparities[<sup>5</sup>](#cite_ref-5). These findings underscore the need for targeted air quality interventions and policies aimed at reducing exposure in vulnerable communities.

```python
def load_and_combine_csvs(folder_path, file_pattern = '*.csv'):
    """
    Loads all CSV files and returns a single DataFrame.
    """
    
    all_files = glob.glob(os.path.join(folder_path, file_pattern))
    df_list = []
    
    for file in all_files:
        temp_df = pd.read_csv(file)
        df_list.append(temp_df)
        
    return pd.concat(df_list, ignore_index = True) if df_list else pd.DataFrame()
```

## Dataset #2 Asthma-related deaths across California counties from 2014-2022

```python
import pandas as pd
import numpy as np
```

```python
def clean_asthma_data(filepath):
    # Read the CSV file
    df = pd.read_csv(filepath, encoding='cp1252')
    
    # 1. Fix column names - remove any potential whitespace
    df.columns = df.columns.str.strip()
    
    # 2. Clean the YEARS column - replace all possible dash characters
    df['YEARS'] = (df['YEARS']
                   .str.replace('–', '-')  # en dash
                   .str.replace('—', '-')  # em dash
                   .str.replace('−', '-')  # minus sign
                   .str.replace('‐', '-')  # hyphen
                   .str.replace('�', '-'))  # corrupted character
    
    # 3. Clean the NUMBER OF DEATHS column
    df['NUMBER OF DEATHS'] = (df['NUMBER OF DEATHS']
                             .str.replace(',', '')  # Remove commas
                             .replace('', np.nan))  # Replace empty strings with NaN
    # Convert to numeric, coercing errors to NaN
    df['NUMBER OF DEATHS'] = pd.to_numeric(df['NUMBER OF DEATHS'], errors='coerce')
    
    # 4. Clean the AGE-ADJUSTED MORTALITY RATE column
    df['AGE-ADJUSTED MORTALITY RATE'] = pd.to_numeric(df['AGE-ADJUSTED MORTALITY RATE'], errors='coerce')
    
    # 5. Clean the AGE GROUP column - standardize the dash character
    df['AGE GROUP'] = (df['AGE GROUP']
                      .str.replace('–', '-')
                      .str.replace('—', '-')
                      .str.replace('−', '-')
                      .str.replace('‐', '-')
                      .str.replace('�', '-'))
    
    # 6. Create a data quality column
    df['DATA_QUALITY'] = 'complete'
    mask_suppressed = df['COMMENT'].str.contains('suppressed', na=False)
    mask_unstable = df['COMMENT'].str.contains('instability', na=False)
    
    df.loc[mask_suppressed, 'DATA_QUALITY'] = 'suppressed'
    df.loc[mask_unstable, 'DATA_QUALITY'] = 'unstable'
    
    # 7. Extract year ranges - modified to handle the year ranges correctly
    df[['START_YEAR', 'END_YEAR']] = (df['YEARS']
                                     .str.extract(r'(\d{4})-(\d{4})'))
    
    # Convert to numeric after extraction
    df['START_YEAR'] = pd.to_numeric(df['START_YEAR'])
    df['END_YEAR'] = pd.to_numeric(df['END_YEAR'])
    
    # 8. Create a flag for state-level vs county-level data
    df['GEOGRAPHIC_LEVEL'] = np.where(df['COUNTY'] == 'California', 
                                    'State', 'County')
    
    # 9. Sort the dataframe
    df = df.sort_values(['START_YEAR', 'GEOGRAPHIC_LEVEL', 'COUNTY', 'STRATA', 'AGE GROUP'])
    
    return df
```

```python
def load_and_combine_datasets(filepath_2010_2019, filepath_2020_present):
    """
    Load and combine the two split datasets with robust error handling
    
    Parameters:
    filepath_2010_2019 (str): Path to the 2010-2019 crime data CSV
    filepath_2020_present (str): Path to the 2020-present crime data CSV
    
    Returns:
    pandas.DataFrame: Combined dataframe from both files
    """
    print("Loading and combining datasets...")
    
    # Load the 2010-2019 dataset with advanced error handling
    # This historical dataset contains crime records from 2010 through 2019
    try:
        # Use pandas read_csv with additional parameters to handle malformed data
        import csv
        df_2010_2019 = pd.read_csv(
            filepath_2010_2019,
            encoding='utf-8',
            on_bad_lines='skip',  # Skip lines with too many fields
            quoting=csv.QUOTE_MINIMAL,  # Try to be flexible with quotes
            escapechar='\\',  # Use backslash as escape character
            low_memory=False  # Avoid dtype guessing issues
        )
        # Print the dimensions (rows, columns) to verify successful loading
        print(f"2010-2019 dataset shape: {df_2010_2019.shape}")
    except Exception as e:
        # If loading with still fails try another way
        print(f"First attempt at loading 2010-2019 dataset failed: {e}")
        try:
            # Try with Python engine - note: we remove low_memory as it's not compatible with Python engine
            df_2010_2019 = pd.read_csv(
                filepath_2010_2019,
                encoding='utf-8',
                engine='python',  # Switch to Python engine which is more forgiving
                on_bad_lines='skip',
                quoting=csv.QUOTE_NONE  # Ignore quotes completely
            )
            print(f"2010-2019 dataset loaded with fallback method. Shape: {df_2010_2019.shape}")
        except Exception as e2:
            print(f"All attempts to load 2010-2019 dataset failed: {e2}")
            
            # Third attempt - try reading in smaller chunks
            try:
                print("Trying to read file in chunks...")
                # Initialize an empty list to store chunks
                chunks = []
                # Read the file in chunks of 100,000 rows
                for chunk in pd.read_csv(
                    filepath_2010_2019,
                    encoding='utf-8', 
                    chunksize=100000,  # Process 100,000 rows at a time
                    on_bad_lines='skip',
                    escapechar='\\'
                ):
                    chunks.append(chunk)
                
                # Concatenate all chunks
                df_2010_2019 = pd.concat(chunks, ignore_index=True)
                print(f"Successfully loaded 2010-2019 dataset in chunks. Shape: {df_2010_2019.shape}")
            except Exception as e3:
                print(f"All attempts to load 2010-2019 dataset failed: {e3}")
                print("Creating an empty DataFrame for 2010-2019 data.")
                df_2010_2019 = pd.DataFrame()
    
    # Load the 2020-present dataset with error handling
    # This recent dataset contains crime records from 2020 to present day
    try:
        # Use pandas read_csv with the same robust parameters
        import csv
        df_2020_present = pd.read_csv(
            filepath_2020_present,
            encoding='utf-8',
            on_bad_lines='skip',
            quoting=csv.QUOTE_MINIMAL,
            escapechar='\\',
            low_memory=False
        )
        # Print the dimensions (rows, columns) to verify successful loading
        print(f"2020-present dataset shape: {df_2020_present.shape}")
    except Exception as e:
        # If the first attempt fails, try the more robust approach
        print(f"First attempt at loading 2020-present dataset failed: {e}")
        try:
            # Try with Python engine - removing low_memory
            df_2020_present = pd.read_csv(
                filepath_2020_present,
                encoding='utf-8',
                engine='python',
                on_bad_lines='skip',
                quoting=csv.QUOTE_NONE
            )
            print(f"2020-present dataset loaded with fallback method. Shape: {df_2020_present.shape}")
        except Exception as e2:
            # Third attempt - try reading in chunks
            try:
                print("Trying to read 2020-present file in chunks...")
                chunks = []
                for chunk in pd.read_csv(
                    filepath_2020_present,
                    encoding='utf-8', 
                    chunksize=100000,
                    on_bad_lines='skip',
                    escapechar='\\'
                ):
                    chunks.append(chunk)
                
                df_2020_present = pd.concat(chunks, ignore_index=True)
                print(f"Successfully loaded 2020-present dataset in chunks. Shape: {df_2020_present.shape}")
            except Exception as e3:
                print(f"All attempts to load 2020-present dataset failed: {e3}")
                print("Creating an empty DataFrame for 2020-present data.")
                df_2020_present = pd.DataFrame()
    
    # Check if at least one dataset was loaded successfully
    if df_2010_2019.empty and df_2020_present.empty:
        raise ValueError("Both datasets failed to load. Cannot proceed with analysis.")
    
    # Only proceed with combining datasets if both were loaded successfully
    if not df_2010_2019.empty and not df_2020_present.empty:
        # Find columns that exist in both datasets
        # Using set intersection to identify common columns efficiently
        common_columns = list(set(df_2010_2019.columns).intersection(set(df_2020_present.columns)))
        print(f"Number of common columns: {len(common_columns)}")
        
        # Filter both datasets to include only columns that exist in both
        # This prevents errors when combining datasets with different structures
        df_2010_2019 = df_2010_2019[common_columns]
        df_2020_present = df_2020_present[common_columns]
        
        # Combine the filtered datasets using vertical concatenation
        # ignore_index=True resets the index to avoid duplicate indices
        combined_df = pd.concat([df_2010_2019, df_2020_present], ignore_index=True)
        # Print the dimensions of the combined dataset to verify the operation
        print(f"Combined dataset shape: {combined_df.shape}")
    elif not df_2010_2019.empty:
        # If only 2010-2019 data loaded, use just that
        print("Only using 2010-2019 dataset since 2020-present failed to load.")
        combined_df = df_2010_2019
    else:
        # If only 2020-present data loaded, use just that
        print("Only using 2020-present dataset since 2010-2019 failed to load.")
        combined_df = df_2020_present
    
    # Return the combined dataset for further processing
    return combined_df
```

## Crime Rate Analysis by Location

Now we'll calculate crime rates for each area in Los Angeles. Since raw crime counts can be misleading due to population differences, we'll normalize by population to get crime rates per 1,000 people. This allows for fair comparison between areas.

```python
def analyze_crime_rates(df_cleaned):
    """
    Analyze crime rates by location and year
    
    Parameters:
    df_cleaned (pandas.DataFrame): Cleaned crime dataframe
    
    Returns:
    pandas.DataFrame: DataFrame with crime rates by area and year
    """
    # 1. Group by location and year to count crimes
    print("Grouping data by area and year...")
    # This aggregation counts the number of crime incidents in each area for each year
    # The .size() counts occurrences, and reset_index() converts the result to a DataFrame
    # The resulting DataFrame has columns: 'AREA NAME', 'YEAR', and 'CRIME_COUNT'
    crime_by_area_year = df_cleaned.groupby(['AREA NAME', 'YEAR']).size().reset_index(name='CRIME_COUNT')
    
    # 2. Load/create population data for each LAPD area
    print("Creating population estimates for rate calculations...")
    # Dictionary mapping each LAPD area to its estimated population
    # These are approximate values for demonstration purposes only
    area_population = {
        'Central': 40000,        # Downtown LA business district
        'Rampart': 80000,        # Dense residential area near downtown
        'Southwest': 120000,     # University area with mixed density
        'Hollenbeck': 110000,    # East LA neighborhood 
        'Harbor': 90000,         # Port area and surrounding neighborhoods
        'Hollywood': 160000,     # Tourist and entertainment district
        'Wilshire': 180000,      # High density mixed commercial/residential
        'West LA': 140000,       # Affluent westside neighborhoods
        'Van Nuys': 130000,      # San Fernando Valley community
        'West Valley': 150000,   # Suburban valley area
        'Northeast': 120000,     # Eagle Rock and surrounding areas
        'Pacific': 170000,       # Coastal areas including Venice
        'N Hollywood': 140000,   # Entertainment industry hub
        'Newton': 130000,        # South central industrial area
        '77th Street': 150000,   # South LA residential neighborhoods
        'Foothill': 110000,      # Northeast valley communities
        'Devonshire': 130000,    # Northwest valley suburban area
        'Southeast': 120000,     # Watts and surrounding communities
        'Mission': 120000,       # Northeast valley area
        'Olympic': 130000,       # Koreatown and Mid-Wilshire
        'Topanga': 120000        # Western valley shopping district
    }
    
    # Convert the population dictionary to a DataFrame for easier merging
    # This creates a DataFrame with two columns: 'AREA NAME' and 'POPULATION'
    population_df = pd.DataFrame([
        {'AREA NAME': area, 'POPULATION': pop}
        for area, pop in area_population.items()
    ])
    
    # 3. Merge crime counts with population data
    print("Calculating crime rates...")
    # This left join combines crime counts with population data
    # 'left' join keeps all crime areas, even if we don't have population data
    crime_with_population = pd.merge(crime_by_area_year, population_df, on='AREA NAME', how='left')
    
    # 4. Calculate crime rate per 1000 people
    # Standard formula: (number of crimes / population) * 1000
    # This creates a normalized measure that allows comparison between areas of different sizes
    crime_with_population['CRIME_RATE'] = (crime_with_population['CRIME_COUNT'] / 
                                          crime_with_population['POPULATION']) * 1000
    
    # 5. Handle any areas with missing population data
    # Sometimes there may be mismatches between crime areas and population data
    if crime_with_population['POPULATION'].isna().any():
        print("Warning: Some areas have missing population data. Using median population for these areas.")
        # Calculate the median population as a reasonable substitute value
        median_pop = population_df['POPULATION'].median()
        # Fill in missing population values with the median
        crime_with_population.loc[crime_with_population['POPULATION'].isna(), 'POPULATION'] = median_pop
        # Recalculate crime rate with the imputed population values
        crime_with_population['CRIME_RATE'] = (crime_with_population['CRIME_COUNT'] / 
                                              crime_with_population['POPULATION']) * 1000
    
    # Return the final DataFrame with crime counts, populations, and rates
    return crime_with_population
```

2. Data Cleaning Function
This function cleans the raw asthma deaths data to prepare it for analysis.

```python
def clean_asthma_data(filepath):
    """
    Clean and prepare asthma deaths data for analysis
    
    Parameters:
    filepath (str): Path to the asthma deaths CSV file
    
    Returns:
    pandas.DataFrame: Cleaned dataframe with standardized formats
    """
    # Read the CSV file
    df = pd.read_csv(filepath, encoding='cp1252')
    
    # 1. Fix column names - remove any potential whitespace
    df.columns = df.columns.str.strip()
    
    # 2. Clean the YEARS column - replace all possible dash characters
    df['YEARS'] = (df['YEARS']
                   .str.replace('–', '-')  # en dash
                   .str.replace('—', '-')  # em dash
                   .str.replace('−', '-')  # minus sign
                   .str.replace('‐', '-')  # hyphen
                   .str.replace('�', '-'))  # corrupted character
    
    # 3. Clean the NUMBER OF DEATHS column
    df['NUMBER OF DEATHS'] = (df['NUMBER OF DEATHS']
                             .str.replace(',', '')  # Remove commas
                             .replace('', np.nan))  # Replace empty strings with NaN
    # Convert to numeric, coercing errors to NaN
    df['NUMBER OF DEATHS'] = pd.to_numeric(df['NUMBER OF DEATHS'], errors='coerce')
    
    # 4. Clean the AGE-ADJUSTED MORTALITY RATE column
    df['AGE-ADJUSTED MORTALITY RATE'] = pd.to_numeric(df['AGE-ADJUSTED MORTALITY RATE'], errors='coerce')
    
    # 5. Clean the AGE GROUP column - standardize the dash character
    df['AGE GROUP'] = (df['AGE GROUP']
                      .str.replace('–', '-')
                      .str.replace('—', '-')
                      .str.replace('−', '-')
                      .str.replace('‐', '-')
                      .str.replace('�', '-'))
    
    # 6. Create a data quality column
    df['DATA_QUALITY'] = 'complete'
    mask_suppressed = df['COMMENT'].str.contains('suppressed', na=False)
    mask_unstable = df['COMMENT'].str.contains('instability', na=False)
    
    df.loc[mask_suppressed, 'DATA_QUALITY'] = 'suppressed'
    df.loc[mask_unstable, 'DATA_QUALITY'] = 'unstable'
    
    # 7. Extract year ranges - modified to handle the year ranges correctly
    df[['START_YEAR', 'END_YEAR']] = (df['YEARS']
                                     .str.extract(r'(\d{4})-(\d{4})'))
    
    # Convert to numeric after extraction
    df['START_YEAR'] = pd.to_numeric(df['START_YEAR'])
    df['END_YEAR'] = pd.to_numeric(df['END_YEAR'])
    
    # 8. Create a flag for state-level vs county-level data
    df['GEOGRAPHIC_LEVEL'] = np.where(df['COUNTY'] == 'California', 
                                    'State', 'County')
    
    # 9. Sort the dataframe
    df = df.sort_values(['START_YEAR', 'GEOGRAPHIC_LEVEL', 'COUNTY', 'STRATA', 'AGE GROUP'])
    
    return df
```

This function visualizes the relationship between carbon monoxide (CO) and ozone (O3) levels across different LAPD districts. The goal is to explore whether higher crime areas, often correlated with lower socioeconomic status, experience higher exposure to certain pollutants, which may increase the risk of asthma and other respiratory conditions.

```python
def visualize_co_vs_o3_by_district(comparison_df=None):
    # Edge case, if not computed,calculate it
    if comparison_df is None:
        # Ensure required data exists
        if 'co_daily' not in globals() or 'ozone_daily' not in globals() or 'crime_rates' not in globals():
            print("Error: Required pollution and crime data not found. Run earlier cells first.")
            return None
        
        print("Analyzing pollution station data across LA districts...")
        
        # Clean station names by stripping whitespace
        co_daily['name'] = co_daily['name'].str.strip()
        ozone_daily['name'] = ozone_daily['name'].str.strip()
        
        # Define LAPD districts to nearby air quality monitoring stations mapping
        area_pollution_mapping = {
            # High crime areas
            'Central': ['Los Angeles-North Main Street', 'Los Angeles-651 Mott Street'],
            'Rampart': ['West Los Angeles-VA Hospital', 'Los Angeles-Westchester Parkway'],
            'Southwest': ['Compton-700 North Bullis Road', 'Long Beach-2425 Webster Street'],
            'Harbor': ['Wilmington-1115 N.Mahar Avenue', 'Long Beach-Signal Hill'],
            '77th Street': ['Lynwood', 'Compton-700 North Bullis Road'],
            # Medium crime areas
            'Newton': ['Los Angeles-North Main Street', 'Pico Rivera'],
            'Hollywood': ['West Los Angeles-VA Hospital', 'Burbank-W Palm Avenue'],
            'Southeast': ['Compton-700 North Bullis Road', 'Long Beach-Signal Hill'],
            'Olympic': ['Los Angeles-North Main Street', 'Los Angeles-Westchester Parkway'],
            # Lower crime areas
            'West LA': ['West Los Angeles-VA Hospital', 'Los Angeles-Westchester Parkway'],
            'Wilshire': ['West Los Angeles-VA Hospital', 'Los Angeles-Westchester Parkway'],
            'Pacific': ['Los Angeles-Westchester Parkway', 'Long Beach-2425 Webster Street'],
            'West Valley': ['Reseda', 'Burbank-W Palm Avenue'],
            'Devonshire': ['Reseda', 'Santa Clarita']
        }
        
        # Collect pollution and crime data for all areas
        pollutant_data = []
        
        # Loop through all areas and collect available data
        for area in area_pollution_mapping.keys():
            # Check if this area appears in our crime data
            area_crime_data = crime_rates[(crime_rates['AREA NAME'] == area) & 
                                        (crime_rates['YEAR'].between(2015, 2020))]
            
            if len(area_crime_data) == 0:
                print(f"Warning: No crime data for {area}, skipping")
                continue
                
            # Calculate average crime rate
            crime_rate = area_crime_data['CRIME_RATE'].mean()
            
            # Get valid CO stations for this area
            co_valid_stations = [
                s for s in area_pollution_mapping[area] 
                if s in co_daily['name'].values
            ]
            
            # Get valid O3 stations for this area
            o3_valid_stations = [
                s for s in area_pollution_mapping[area] 
                if s in ozone_daily['name'].values
            ]
            
            if not co_valid_stations or not o3_valid_stations:
                print(f"Warning: No matching pollution stations for {area}, skipping")
                continue
                
            # Get pollution data (2015-2020)
            co_values = co_daily[
                (co_daily['name'].isin(co_valid_stations)) &
                (co_daily['date'].between('2015-01-01', '2020-12-31'))
            ]['value'].mean()
            
            o3_values = ozone_daily[
                (ozone_daily['name'].isin(o3_valid_stations)) &
                (ozone_daily['date'].between('2015-01-01', '2020-12-31'))
            ]['value'].mean()
            
            # Store all data
            pollutant_data.append({
                'Area': area,
                'Crime Rate': crime_rate,
                'CO (ppm)': co_values,
                'O3 (ppm)': o3_values
            })
        
        # Create DataFrame and sort by crime rate
        comparison_df = pd.DataFrame(pollutant_data)
        
        if len(comparison_df) < 5:
            print("Error: Not enough areas with matching pollution data. The visualization may not be meaningful.")
            
        # Sort by crime rate for better visualization
        comparison_df = comparison_df.sort_values('Crime Rate', ascending=False).reset_index(drop=True)
        
        print("\nFinal comparison data (sorted by crime rate, high to low):")
        print(comparison_df)
    
    # color function for pollution visualizations
    def get_color(rate):
        if rate > 150:
            return '#e74c3c'  # High crime - red
        elif rate > 75:
            return '#f39c12'  # Medium crime - orange
        else:
            return '#2ecc71'  # Low crime - green
    
    # Variety of markers based on crime rate
    def marker_style(rate):
        if rate > 150:
            return 'o', 120  # Circle, larger
        elif rate > 75:
            return 's', 80   # Square, medium
        else:
            return 'D', 60   # Diamond, smaller
    
    # Create figure with appropriate size
    fig, ax = plt.subplots(figsize=(12, 10))
    
    # Plot the points first without labels
    for i, row in comparison_df.iterrows():
        marker, size = marker_style(row['Crime Rate'])
        ax.scatter(row['CO (ppm)'], row['O3 (ppm)'], 
                 s=size, alpha=0.7, 
                 color=get_color(row['Crime Rate']),
                 marker=marker, edgecolors='black', linewidth=1)
    
    # Group points that are close together to avoid label overlap
    from collections import defaultdict
    import numpy as np
    
    # Define threshold for considering points as "same location"
    co_threshold = 0.0005  # Threshold for CO in ppm
    o3_threshold = 0.0005  # Threshold for O3 in ppm
    
    # Create a grid to group nearby points
    point_groups = defaultdict(list)
    
    # Round points to group them
    for i, row in comparison_df.iterrows():
        # Round to nearest threshold multiple to create groups
        co_rounded = round(row['CO (ppm)'] / co_threshold) * co_threshold
        o3_rounded = round(row['O3 (ppm)'] / o3_threshold) * o3_threshold
        
        # Rounded values as keys to group points
        key = (co_rounded, o3_rounded)
        point_groups[key].append((row['Area'], row['CO (ppm)'], row['O3 (ppm)'], row['Crime Rate']))
    
    # Add labels for each group
    for key, points in point_groups.items():
        # Calculate average position for the label
        avg_co = sum(p[1] for p in points) / len(points)
        avg_o3 = sum(p[2] for p in points) / len(points)
        
        # Create combined label
        areas = [p[0] for p in points]
        if len(areas) == 1:
            label = areas[0]
        else:
            label = ", ".join(areas)
        
        # Add annotation with combined label
        ax.annotate(label, 
                  (avg_co, avg_o3),
                  xytext=(5, 5), textcoords='offset points',
                  fontsize=10)
    
    # Add legend for crime rate categories
    high_crime = plt.Line2D([], [], color='#e74c3c', marker='o', linestyle='None', 
                         markersize=12, label='High Crime > 150 per 100k')
    med_crime = plt.Line2D([], [], color='#f39c12', marker='s', linestyle='None', 
                        markersize=10, label='Medium Crime 75-150 per 100k')
    low_crime = plt.Line2D([], [], color='#2ecc71', marker='D', linestyle='None', 
                        markersize=8, label='Low Crime < 75 per 100k')
    
    ax.legend(handles=[high_crime, med_crime, low_crime], loc='best')
    
    ax.set_title('CO vs O3 Levels by District Crime Rate (Mean) 2015-2020', fontsize=16, fontweight='bold')
    ax.set_xlabel('CO Concentration (ppm)', fontsize=14)
    ax.set_ylabel('O3 Concentration (ppm)', fontsize=14)
    ax.grid(True, linestyle='--', alpha=0.3)
    
    context_text = (
        "Analysis Context:\n"
        "• Crime rates serve as a proxy for socioeconomic status\n"
        "• CO primarily comes from vehicle emissions and industrial activity\n"
        "• O3 forms through chemical reactions with other pollutants\n"
    )
    
    plt.figtext(0.5, 0.01, context_text, ha='center', fontsize=10, 
               bbox=dict(facecolor='white', alpha=0.8, boxstyle='round,pad=0.7', 
                       edgecolor='gray'))
    
    plt.tight_layout(rect=[0, 0.1, 1, 0.98])
    
    return fig
```

```python
# Display the visualization
fig = visualize_co_vs_o3_by_district()
plt.show()
```

- Medium-crime districts such as Hollywood, Olympic, and Harbor tend to have moderate CO and O3 levels, reflecting a mix of urban density and emissions.

Our study examined asthma-related deaths in Los Angeles County, revealing significant differences based on age, geography, and environmental pollution. This aligns with prior research showing that long-term exposure to air pollution, especially nitrogen dioxide (NO₂) and other traffic-related pollutants, can worsen asthma over time, leading to more severe outcomes in adults. Additionally, disparities in healthcare access and socioeconomic factors likely contribute to the increased mortality risk among older populations. 

A key aspect of our study was the geographic distribution of asthma deaths within Los Angeles County. While the county as a whole recorded the highest absolute number of deaths (approximately 300 every two years), certain areas exhibited disproportionately high mortality rates when adjusted for population size. For example, San Bernardino had a notably high mortality rate, suggesting that asthma severity is not strictly an urban issue. Limited healthcare access in less densely populated areas may play a role in these elevated fatality rates. The relationship between air pollution and asthma deaths is complex, particularly when comparing different pollutants such as carbon monoxide (CO) and ozone (O₃). Our final analysis mapped CO and O₃ levels alongside asthma mortality rates, revealing notable patterns. CO, which is emitted from vehicle exhaust and industry combustion sources, was more concentrated in high-traffic areas, aligning closely with regions that reported higher asthma death rates. This suggests that CO exposure may have a more immediate and severe impact on asthma-related fatalities, potentially due to its ability to reduce oxygen delivery in the bloodstream and increase respiratory distress. In contrast, O₃ is a secondary pollutant formed through photochemical reactions, meaning its highest concentrations do not always align with vehicle emissions but rather depend on factors such as sunlight and atmospheric conditions. While O₃ is known to cause airway inflammation and worsen asthma symptoms, its direct link to asthma deaths may be less immediate than CO. These findings suggest that pollution reduction strategies should prioritize controlling emissions from vehicles and other combustion sources to mitigate the most immediate risks to asthma patients.

A major challenge in our analysis was the nature of the data available. Initially, we sought to examine overall asthma cases rather than just deaths, but data limitations forced us to focus on mortality alone. This significantly shaped our conclusions, as asthma-related deaths represent only the most severe cases, not the full burden of the disease. If general asthma incidence data were available, we could have provided a more comprehensive analysis of how pollution contributes to asthma severity at different levels rather than just fatal outcomes. Another methodological consideration was our use of crime data as a proxy for socioeconomic conditions. Research has consistently shown that lower-income communities experience higher levels of air pollution, and prior studies have found that disadvantaged neighborhoods in Los Angeles have roughly 28% higher NO₂ levels than wealthier areas. In the absence of direct socioeconomic data, we used crime levels as an indicator of socioeconomic status, with higher crime rates representing lower-income areas. While this approach is not without limitations, it allowed us to explore how pollution and asthma mortality intersect with economic disparities. Our findings suggest that regions with both high pollution exposure and lower socioeconomic status may face compounded risks, reinforcing the need for targeted public health interventions in these communities.
