# Abstract

This project investigates the impact of the COVID-19 pandemic on deaths of despair in the United States. We aimed to understand how these mortality trends shifted during the pandemic and whether these effects varied across different demographic groups such as age, gender, race, and Hispanic origin.

To achieve this, our analysis integrated general population data and mortality trend records, along with a more specialized drug overdose dataset. Our approach involved meticulous data cleaning and preprocessing steps: dropping extraneous columns, renaming variables for clarity, filtering for relevant years, and eliminating null target values to ensure robust analysis. We merged these datasets to compute crude mortality rates per 100,000 residents, and then conducted exploratory data analysis using pivot tables and visualizations such as line plots, bar charts, and scatter plots. We further applied a variety of statistical tests—including t-tests, Shapiro-Wilk tests, ANOVA, and Mann-Whitney U tests—to assess differences in mortality rates across demographics, and utilized time series analysis with an ARIMA model to capture temporal patterns and seasonality.

Our analysis revealed a significant increase in drug overdose mortality during the pandemic, with notable racial disparities emerging in the rates of change; however, gender differences were not statistically significant. The time series forecasting also highlighted seasonal trends in the crude mortality rates. These findings underscore the importance of targeted public health interventions and suggest that the pandemic exacerbated existing disparities in deaths of despair.

## Background and Prior Work

Diseases of despair are a group of chronic conditions that are associated with feelings of hopelessness, hopelessness, and social isolation. These conditions are often linked to socioeconomic factors such as poverty, unemployment, and lack of access to healthcare. Deaths of despair—suicides, drug overdoses, and alcohol-related fatalities—have been rising in the U.S. for decades<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5). These trends have been attributed to a variety of factors, including the opioid crisis, as well as declines in both physical and mental health among affected populations<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5). The COVID-19 pandemic, which emerged in early 2020, exacerbated these issues, amplifying social isolation, economic hardship, and disruptions to healthcare services. 

During the first year of the pandemic, the CDC reported a nearly 30% increase in drug overdose deaths, surpassing 100,000 deaths<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1). While suicide rates generally remained stable or declined in most areas, other deaths of despair, such as those linked to substance use and alcohol, increased significantly<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4).  

Previous studies on deaths of despair have often been conducted separately from those analyzing the effects of the pandemic. However, many of these studies found trends similar to those observed during economic downturns. While suicide rates tended to stay the same or decrease in most countries, other deaths of despair, particularly those related to alcohol, opioids, and stimulants, rose during the pandemic<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4). Some researchers suggest that factors such as the inaccessibility of safe prescription drugs and difficulties in obtaining proper medical care contributed to the rise in these deaths<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3). Demographic analyses have revealed that women, ethnic minorities, and younger people were disproportionately affected by the increases in deaths of despair<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3).
Despite these findings, several issues persist in the literature. Many studies were conducted in high-income countries with strong welfare systems, leaving it unclear whether similar trends were observed in lower-income nations. Additionally, most studies were conducted during or shortly after the pandemic, meaning they do not account for the impact of rolling back COVID-19 aid benefits, which may have influenced the later rise in deaths of despair. Furthermore, there is little analysis on the specific aspects of the pandemic—such as the economic downturn or lockdowns—that may have contributed to the increase in deaths of despair<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3).  

This project seeks to address these gaps by examining how the emergence of the COVID-19 pandemic impacted rates of deaths of despair and how these effects varied across different demographic groups.

### Overdose Deaths Over Time and How They Differ Based on Age, Race, and Gender

This section focuses on determining whether there are any significant demographic differences in the changes in drug overdose mortality rates during the COVID-19 pandemic. Our results show that there is no significant difference between difference age groups and gender, but that there is a significant difference in race. The exact reasons for the difference in change in mortality rates amongst racial groups are unclear, however.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
import scipy.stats as stats
overdose_df
```

```python
overdose_df['SUBGROUP'] = overdose_df['SUBGROUP'].astype(str).str.strip()

pivot_df = overdose_df.pivot_table(index='TIME_PERIOD', values='ESTIMATE', aggfunc='mean')

# Plotting
pivot_df.plot(marker='o', figsize=(12, 6))

# Customize the plot
plt.xticks(pivot_df.index.astype(int))
plt.title('Overdose Deaths Over Time')
plt.xlabel('Year')
plt.ylabel('Estimate (deaths per 100,000)')
plt.grid(True)
```

This graph shows the change in drug overdose mortality from 2017 to 2021. From this graph, we saw that there was a noticeable increase in the drug overdose mortality starting in 2020.

```python
# Split the data into two groups
recent_mortality = overdose_df[overdose_df["TIME_PERIOD"].isin([2020, 2021])]["ESTIMATE"]
previous_mortality = overdose_df[overdose_df["TIME_PERIOD"] < 2020]["ESTIMATE"]

# Perform an independent t-test
t_stat, p_value = stats.ttest_ind(recent_mortality, previous_mortality, equal_var=False)

# Print results
print(f"T-test statistic: {t_stat}")
print(f"P-value: {p_value}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("Significant difference in average mortality between 2020-2021 and earlier years.")
else:
    print("No significant difference in average mortality between 2020-2021 and earlier years.")
```

We conducted a t-test on the average mortality from before the COVID-19 pandemic and during the pandemic. The null hypothesis is that there is no significant difference in drug overdose mortality from before and after the start of the pandemic. Our t-test yielded a p-value of 55.8 * 10^-12, which is below our alpha of 0.05, meaning there is a statistically significant difference in drug overdose mortality during the two time periods.

We ran Shapiro-Wilk tests to determine if we could treat the change in mortality over time as a normal distribution. After running the tests, we decided to treat the change in overdose mortality over time as a normal distribution for all age ranges.

We ran an ANOVA test to see if there was a significan difference between all of the age groups as a whole. Our null hypothesis was that there would not be any significant difference. Our test returned a p-value of 0.08, which is greater than our alpha value of 0.05, meaning there is no statistically significant difference between age groups overall.

```python
age_group_1 = [
    'Younger than 15 years',
    '75-84 years',
    '85 years and older'
]

# Filter the dataset for years 2019 and onward
age_df_filtered = age_df[age_df['TIME_PERIOD'] >= 2019]

# Add the custom age group column
age_df_filtered['Custom_Age_Group'] = age_df_filtered['SUBGROUP'].apply(
    lambda x: 'Specific Age Groups' if x in age_group_1 else 'Other Age Groups'
)

# Step 2: Calculate the change in overdose deaths over time for each group
# Group by Custom_Age_Group and TIME_PERIOD, and calculate the sum of overdose deaths
change_df = age_df_filtered.groupby(['Custom_Age_Group', 'TIME_PERIOD'])['ESTIMATE'].sum().reset_index()

# Pivot the data to calculate the change (subtract current from previous year)
change_df['Change'] = change_df.groupby('Custom_Age_Group')['ESTIMATE'].diff()

# Step 3: Perform a t-test to compare the changes in overdose deaths between the two groups
specific_age_group_change = change_df[change_df['Custom_Age_Group'] == 'Specific Age Groups']['Change'].dropna()
other_age_group_change = change_df[change_df['Custom_Age_Group'] == 'Other Age Groups']['Change'].dropna()

# Perform the t-test (assuming unequal variances)
t_stat, p_value = stats.ttest_ind(specific_age_group_change, other_age_group_change, equal_var=False)

# Output the results
print(f"T-test statistic: {t_stat}")
print(f"P-value: {p_value}")
```

We then singled out the  "Younger than 15 years", "75-84 years", and "85 years and older" groups that we previously identified as possibly following a different trend from the other age groups. # Ethics & Privacy

### Potential Biases in Data
- Dataset may reflect biases in reporting and classification of deaths such as some causes being underreported or misclassified due to stigma or even inconsistent documentation.
- Preferably not, but data may be disproportionately representing certain demographics of people based on access to healthcare and other medical services.

### Potential Privacy Concerns

# Discusison and Conclusion

Our analysis clearly indicates that the COVID-19 pandemic coincided with a substantial rise in drug overdose mortality in the United States. A t-test comparing pre-pandemic (2017–2019) and pandemic-era (2020–2021) overdose mortality yielded a p-value of approximately 5.58×10⁻¹¹, indicating a statistically significant jump in deaths after the onset of the pandemic. However, when breaking down the data by demographic factors, the effects varied:

* Age Groups: An ANOVA comparing changes in overdose mortality across multiple age brackets gave a p-value of 0.08, failing to show a statistically significant difference in how overdose rates shifted among these groups. A focused t-test (p = 0.18) comparing specific age groups likewise suggested that age did not drive distinctly different pandemic-related increases in overdose deaths.

* Gender: The Mann-Whitney U test comparing the change in overdose mortality rates for males and females produced a p-value of 0.52, implying no significant gender-based discrepancy in how the pandemic affected overdose deaths. While overall levels of overdose mortality differed by gender, the magnitude of the pandemic-related increase did not.

* Race: Racial disparities emerged as particularly noteworthy. An ANOVA on the rate of change among different racial categories showed a p-value of 0.04, indicating that some groups experienced significantly larger increases in overdose mortality than others. However, a narrower t-test comparing “Asian only” to all other groups returned a p-value of 0.068, suggesting that within this specific comparison, the difference was not statistically significant. A similar test for Hispanic vs. non-Hispanic mortality changes yielded a p-value of 0.09, again not crossing the threshold for significance.

In addition to these demographic analyses, we performed a time series examination of monthly crude mortality rates for all deaths of despair (suicide, overdose, and alcohol-related death) using an ARIMA model. After confirming non-stationarity through the augmented Dickey-Fuller test and differencing the data, we observed that the pandemic period mortality rates exceeded what pre-pandemic trends alone would have predicted. The ARIMA forecasts also hinted at seasonal fluctuations, suggesting that broader factors—such as economic stress, social isolation, and healthcare access—may influence overdose mortality patterns throughout the year.

Overall, our findings underscore two major points. First, the pandemic period saw a statistically significant surge in overdose deaths. Second, this surge was not uniform across all demographic categories; rather, it was especially pronounced among certain racial groups. While age and gender did not appear to produce significantly different pandemic-related changes, the presence of racial disparities calls for targeted public health interventions and policy measures to address structural inequities. Furthermore, the time series results highlight the importance of ongoing surveillance to capture seasonal and long-term trends in mortality. By combining demographic-specific statistical tests with predictive modeling, we gain a clearer picture of the pandemic’s impact and can better inform strategies aimed at reducing deaths of despair.

Despite the insights gained, our analysis has several limitations that could affect the robustness of our conclusions. Due to time limitations and overly-complex datasets, we decided to examine the CDC WONDER overdose dataset independently from the population/mortality trends dataset. This may have limited our ability to capture more nuanced relationships across all variables, and led to our prioritizing the analysis of overdose data over other types of deaths of despair. Additionally, our analysis focused on the early to mid-pandemic period (2020–2021) due to non-availabity of more recent data. Long-term or delayed effects of the pandemic on deaths of despair may only become evident over time. Addressing these flaws in future work—such as integrating the datasets more thoroughly, capturing a longer post-pandemic timeframe, and incorporating more detailed socioeconomic variables—could yield deeper insights into how the pandemic influenced different forms of mortality across diverse populations.
