## Background and Prior Work

Cardiovascular diseases (CVD) are a leading cause of mortality worldwide, with various lifestyle and health-related risk factors affecting them. There have been prior studies and datasets analyzing the relationship between certain health attributes and heart disease likeliness; this includes diets and health attributes such as cholesterol levels and blood pressure.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)

In fact, studies have shown that relatively healthy lower income people have higher health risks compared to relatively healthy higher income people<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5). Furthermore, in a study done by the UCSF Department of Epidemiology and Biostatistics, they found that poor Americans had almost twice the chance of getting cardiovascular diseases compared to higher income Americans, and that "Erasing the effects of traditional risk factors including smoking and obesity revealed that 70% of the excess burden of heart disease on this poorer cohort is due to poverty and its associated risks"<a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6). From this, we can analyze the effects of poverty on heart dieseases, as well as the associated health risks that contribute to heart diseases such as obesity and blood pressure. Income disparities could also affect the lifestyle of an individual, which also has a detrimental impact on disease development.

Multiple studies show the impact of how people’s lifestyle choices can affect their risk with CVD. For example, those with diets that contain large amounts of saturated fats, sodium, added sugar, and alcohol may develop hypertension, obesity, or high cholesterol, which are known to lead to heart disease.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Along with poor dietary habits, without proper exercise or the addition of smoking, these will raise the probability of developing CVD even higher. <a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2)

There are large-scale studies that have shown high cholesterol and elevated blood pressure increase the risk of heart disease. <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Aside from the risk of CVD through lifestyle choices, another factor of heart disease is age, which shows an increase in CVD risk as people grow older, especially with those already showing poor health patterns. Through the reports from the World Health Organization (WHO), their data shows how aging has a positive correlation with the likelihood of experiencing CVD-related symptoms, which could eventually evolve into CVD. <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) As global life expectancy continues to increase, the pressure on the risks of cardiovascular diseases also increases with hopes of the development of preventative healthcare measures. <a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4)

With the above plot, we can see how many individuals within each age group have had heart disease or attack, but since it is just a countplot it is biased towards age groups with more individuals in its representation. In other words, it does not proportionately show the relation between age and heart disease or attack - we want proportions of individuals within each age group to make a fair comparison.

```python
# function to turn counts into proportions within their groups, returns a dataframe 
def turn_to_prop(num, df, column_name, feature, index_padding=1):
    results = []
    groups = []
    yes_no = []

    for i in range(num):
        groups.append(i+index_padding)
        groups.append(i+index_padding)

        yes_no.append('Yes')
        yes_no.append('No')

    for i in range(num):
        group_i_yes = sum(df[df[column_name] == i+index_padding][feature] == 1)
        group_i_no = sum(df[df[column_name] == i+index_padding][feature] == 0)

        group_i_yes_prop = round((group_i_yes / sum(df[column_name] == i+index_padding)), 2)
        group_i_no_prop = round((group_i_no / sum(df[column_name] == i+index_padding)), 2)

        results.append(group_i_yes_prop)
        results.append(group_i_no_prop)

    prop_df = pd.DataFrame({column_name: groups, feature: yes_no, 'Proportion': results})
    return prop_df
```

```python
def turn_to_prop_diabetes(num, df, column_name, feature, index_padding=1):
    results = []
    groups = []
    yes_no = []

    for i in range(num):
        groups.append(i+index_padding)
        groups.append(i+index_padding)

        yes_no.append('Yes or Borderline')
        yes_no.append('No')

    for i in range(num):
        group_i_yes = sum(df[df[column_name] == i+index_padding][feature] == 2)
        group_i_borderline = sum(df[df[column_name] == i+index_padding][feature] == 1)
        group_i_no = sum(df[df[column_name] == i+index_padding][feature] == 0)

        group_i_yes_borderline_prop = round(((group_i_yes + group_i_borderline) / sum(df[column_name] == i+index_padding)), 2)
        group_i_no_prop = round((group_i_no / sum(df[column_name] == i+index_padding)), 2)

        results.append(group_i_yes_borderline_prop)
        results.append(group_i_no_prop)

    prop_df = pd.DataFrame({column_name: groups, feature: yes_no, 'Proportion': results})
    return prop_df
```

```python
def turn_to_prop_genhealth(num, df, column_name, feature):
    results = []
    groups = []
    yes_no = []

    for i in range(num):
        groups.append(i+1)
        groups.append(i+1)

        yes_no.append('Good')
        yes_no.append('Bad')

    for i in range(num):
        group_i_1 = sum(df[df[column_name] == i+1][feature] == 1)
        group_i_2 = sum(df[df[column_name] == i+1][feature] == 2)
        group_i_3 = sum(df[df[column_name] == i+1][feature] == 3)
        group_i_4 = sum(df[df[column_name] == i+1][feature] == 4)
        group_i_5 = sum(df[df[column_name] == i+1][feature] == 5)

        group_i_yes_prop = round(((group_i_1 + group_i_2 + group_i_3) / sum(df[column_name] == i+1)), 2)
        group_i_no_prop = round(((group_i_4 + group_i_5) / sum(df[column_name] == i+1)), 2)

        results.append(group_i_yes_prop)
        results.append(group_i_no_prop)

    prop_df = pd.DataFrame({column_name: groups, feature: yes_no, 'Proportion': results})
    return prop_df
```

### Confounders/Mediators

Now that we've established a relationship between Income and various variables, there are a couple that have similar relationships - they all have an inverse relation with income. These variables are HighBP, HighChol, and Diabetes, variables that are known factors of heart disease. We also know from the above plot that income has a generally inverse relation with heart disease, meaning we can likely deduce that higher proportions of HighBP, HighChol, and Diabetes leads to higher proportions of Heart Disease or attack, and lower proportions leads to the opposite. We can apply the same approach as above and look at the proportions of individuals with heart disease within each of the categories of HighBP, HighChol, and Diabetes.

```python
# function to turn individuals with HeartDiseaseorAttack within Diabetes into proportions
def turn_to_prop_diabetes_c(num, df, column_name, feature, index_padding=1):
    results = []
    groups = []
    yes_no = []

    for i in range(num):
        groups.append(i+index_padding)
        groups.append(i+index_padding)

        yes_no.append('Yes')
        yes_no.append('No')
    t = 0
    for i in range(num):
        if i == 0:
            group_i_yes = sum(df[df[column_name] == i+index_padding][feature] == 1)
            group_i_no = sum(df[df[column_name] == i+index_padding][feature] == 0)
        else:
            group_i_yes = sum(df[df[column_name] == i+index_padding][feature] == 1) + sum(df[df[column_name] == i+1][feature] == 1)
            group_i_no = sum(df[df[column_name] == i+index_padding][feature] == 0) + sum(df[df[column_name] == i+1][feature] == 0)
            t = sum(df[column_name] == i+1)
            
        group_i_yes_prop = round((group_i_yes / (sum(df[column_name] == i+index_padding)+t)), 2)
        group_i_no_prop = round((group_i_no / (sum(df[column_name] == i+index_padding)+t)), 2)

        results.append(group_i_yes_prop)
        results.append(group_i_no_prop)

    prop_df = pd.DataFrame({column_name: groups, feature: yes_no, 'Proportion': results})
    return prop_df
```

```python
mediators_subplots = plt.gcf()
# YOUR CODE HERE
mediators_subplots, axes = plt.subplots(1, 3, figsize=(17, 8))

ax1 = make_barplot('HighBP', 'Proportion', 'HeartDiseaseorAttack', turn_to_prop(2, features_we_want, 'HighBP', 'HeartDiseaseorAttack', index_padding=0), axes[0])
ax1.set_title('HighBP vs. HeartDiseaseorAttack')

ax2 = make_barplot('HighChol', 'Proportion', 'HeartDiseaseorAttack', turn_to_prop(2, features_we_want, 'HighChol', 'HeartDiseaseorAttack', index_padding=0), axes[1])
ax2.set_title('HighChol vs. HeartDiseaseorAttack')

ax3 = make_barplot('Diabetes', 'Proportion', 'HeartDiseaseorAttack', turn_to_prop_diabetes_c(2, features_we_want, 'Diabetes', 'HeartDiseaseorAttack', index_padding=0), axes[2])
ax3.set_title('Diabetes vs. HeartDiseaseorAttack')
```

Indeed, there are higher proportions of individuals with heart disease or attack among those with HighBP, HighChol, and Diabetes. In contrast, there are much lower proportions of individuals with heart disease or attack among those without HighBP, HighChol, and Diabetes. This indicates that these 3 variables should be taken into account as mediators - income levels affect the proportions of HighBP, HighChol, and Diabetes, which in turn affects rates of Heart disease or attack. In other words, income levels indirectly correlate to heart disease or attack rates through other factors like HighBP, HighChol, and Diabetes - lower income levels have a higher rate of HighBP, HighChol, and Diabetes, which leads to higher rates of heart disease or attack, while higher income levels have a lower rate of HighBP, HighChol, and Diabetes, which leads to lower rates of heart disease or attack. We can now assess whether age is a confounder.

```python
def turn_to_prop_age(num, df, column_name, feature, index_padding=1, f_range=0):
    results = []
    groups = []
    yes_no = []

    for i in range(num):
        for j in range(f_range):
            groups.append(i+index_padding)
            yes_no.append(j+index_padding)

    for i in range(num):
        group_i_j = 0
        for j in range(f_range):
            group_i_j = sum(df[df[column_name] == i+index_padding][feature] == j+index_padding)
            group_i_j_prop = round((group_i_j / sum(df[column_name] == i+index_padding)), 2)
            results.append(group_i_j_prop)

    prop_df = pd.DataFrame({column_name: groups, feature: yes_no, 'Proportion': results})
    return prop_df
```

However, since we saw earlier that proportions of heart disease does increase with age, it is safer to conclude that Age should be considered a confounding variable that affects both the dependent variable (heart disease or attack) and potentially the predictor (income) as well. 

### Logistic Regression - Income, HighBP, HighChol, Diabetes vs. Heart Disease or Attack

Since our outcome (dependent variable) HeartDiseaseorAttack has binary values 0/1, and our predictors (independent variables) Income, HighBP, HighChol, Cholesterol, and age are either ordinal or binary, it's a natural and obvious choice to use logistic regression to model our data. The coefficients of the resulting model allows us to assess the effect of each variable on heart disease. Note that HighBP, HighChol, Diabetes, and age are included as predictors to model a more accurate relationship between income and heart disease.

```python
X = features_we_want[['Income', 'HighBP', 'HighChol', 'Diabetes', 'Age']]
y = features_we_want['HeartDiseaseorAttack']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=108)

log_reg = LogisticRegression(fit_intercept=True, class_weight='balanced')
log_reg.fit(X_train, y_train)

class_0_proportion = np.sum(y_train == 0) / len(y_train)
class_1_proportion = np.sum(y_train == 1) / len(y_train)

# Calculate the log-odds for class 1 based on the class proportions
log_odds_class_1 = np.log(class_1_proportion / class_0_proportion)

# Adjust the intercept of the logistic regression model
log_reg.intercept_ = np.array([log_odds_class_1])

y_pred = log_reg.predict_proba(X_test)[:, 1]
threshold = 0.62
y_pred_custom = (y_pred >= threshold).astype(int)

#y_pred = log_reg.predict(X_test)
print("Classification Report:")
print(classification_report(y_test, y_pred_custom))
print('coef: ', log_reg.coef_)
print('accuracy: ', accuracy_score(y_test, y_pred_custom))
```

From the report, we can see the coefficients of each predictor, which we can then transform into odds ratios by exponentiation. The odds ratios tell us the relationship between our predictors and outcome. Specifically, odds ratios < 1 signifies an inverse relationship, while odds ratios > 1 signify a positive relationship. 

```python
odds_ratios_sklearn = np.exp(log_reg.coef_[0])
print(odds_ratios_sklearn)
```

As we can see above, the predictor Income has odds ratios < 1, meaning as income increases, the rate of heart disease or attack decreases. All the other odds ratios for HighBP, HighChol, Cholesterol, and Age, respectively, are > 1, meaning as those variables increase in value (or go from 0 to 1), the rate of heart disease increases. 

However, the precision for class 1 is quite low for our regression model. This is likely due to the heavy class imbalance of the HeartDiseaseorAttack feature, where there are much more 0's (no heart disease or attack) compared to 1's (yes heart disease or attack), which tends to make the model biased towards the majority class, in this case 0. We tried techniques such as oversampling or undersampling to try to correct for class imbalance, but the performance of the model still remained subpar, having around a 0.2 precision. As such, we tried to get the best balance of precision and recall we could for class 1, having a precision of 0.23 and recall of 0.7. This means that although 70% of positive cases are identified, only 23% were true positives. This is obviously not ideal, and the model or dataset would have to be further tuned in some way, but does still provide us with insights as to how each variable impacts the outcome. However, we can try other models to hopefully find a better fit for our data.

### XGBoost - Income, HighBP, HighChol, Diabetes, GenHlth, PhysHlth vs. Heart Disease or Attack

Although Logistic Regression works great for the data that we are working with, we wanted to find a model that had a higher precision rate for class 1. This is where XGBoost comes in as it effectively handles complex relationships between multiple features, allowing us to consider more predictors for heart disease such as GenHlth and PhysHlth; overall, it's capable of analyzing multiple features in order to produce results on complex relationships. Lastly, in this case, XGB has a higher focus on precision as a prediction model.

```python
# XGBoost model :)
X = features_we_want[['Income', 'HighBP', 'HighChol', 'Diabetes', 'Age', 'GenHlth', 'PhysHlth']]
y = features_we_want['HeartDiseaseorAttack']

# train and test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=108)

# model
model = xgb.XGBClassifier(objective='binary:logistic', random_state=42)

# fitting
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("Classification Report:")
print(classification_report(y_test, y_pred))
print('Accuracy: ', accuracy_score(y_test, y_pred))
```

As you can see in the report above, our accuracy was able to improve and our precision for class 1 also improved. However, with these improvements came the decrease in our recall of class 1 with a value of 0.05. This means our true positives are far less likely to be identified. Overall, we would need to resample the data in order to produce balance results and get a better recall and precision rate for our prediction model. As stated before, this is due to the imbalance of our data set with a far larger majority data (no heart disease) compared to our minority data (yes heart disease).

The important takeaway is that there is indeed a relation between income levels and health issues, which ultimately leads to higher rates of heart disease, as the logistic regression results suggest. The logical next step after this result is to build a robust prediction model that can model this relationship effectively, which we have attempted to accomplish above with different models.
