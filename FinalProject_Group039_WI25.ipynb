The goal is to predict future natural hazard vulnerability and observe the safety levels of various geographic locations. ## Distribution of Disaster Per Counties

#### First let's look at which counties face the most disaster delcarations
- This allows us to understand which areas are more prone to be identified as "high risk"

```python
#Number of disaster declarations for each county
county_disasters = newCA["designatedArea"].value_counts().reset_index()
county_disasters.columns = ["County", "DisasterCount"]
#DataFrame showing the counties with the most Disaster Counts
county_disasters_sorted = county_disasters.sort_values("DisasterCount", ascending=False)
county_disasters_sorted.head(10)
```

```python
import matplotlib.pyplot as plt

#The top 10 counties with the highest disaster declarations plotted
top10_county = county_disasters.head(10)
sns.barplot(x = "County", y = "DisasterCount", data = top10_county)
plt.xticks(rotation = 80)
plt.title("Top 10 Counties by Number of Disaster Declarations")
plt.xlabel("County")
plt.ylabel("Number of Disasters")
```

## Disaster Duration per County

Knowing the amount of Disaster Declarations is only one side of the story. It is important to understand that the amount of days that these disasters take is also a very important part of confirming "high risk" areas.

```python
#Calculate the disaster durations in days
newCA["disasterDuration"] = (newCA["incidentEndDate"] - newCA["incidentBeginDate"]).dt.days
#Groups area and the duration of disasters
county_disaster_duration = newCA.groupby("designatedArea")["disasterDuration"].sum().reset_index()

#Updated Plot of Counties who suffer for a longer combined total
top_10_duration = county_disaster_duration.sort_values("disasterDuration", ascending = False).head(10)
sns.barplot(x = "designatedArea", y = "disasterDuration", data=top_10_duration)
plt.title("Top 10 Counties by Total Disaster Duration")
plt.xlabel("County")
plt.ylabel("Total Duration (Days)")
plt.xticks(rotation = 80)
```

#### Key Takeaways
- Los Angeles tops the charts in both categories
- San Bernardino also maintains a top 3 position

```python
import statsmodels.api as sm

X = merged_df['distance_to_fault']
y = merged_df['InjuriesPerCapita']
X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
print('Results For Injuries Per Capita and Distance_to_Fault')
print(model.summary())

X2 = merged_df['distance_to_fault']
y2 = merged_df['FatalitiesPerCapita']
X2 = sm.add_constant(X2)
model2 = sm.OLS(y2, X2).fit()
print('\nResults For Fatalities Per Capita and Distance_to_Fault')
print(model2.summary())

X3 = merged_df['distance_to_fault']
y3 = merged_df['CropDmgPerCapita(ADJ 2023)']
X3 = sm.add_constant(X3)
model3 = sm.OLS(y3, X3).fit()
print('\nResults For Crop Damage Per Capita and Distance_to_Fault')
print(model3.summary())

X4 = merged_df['distance_to_fault']
y4 = merged_df['PropertyDmgPerCapita(ADJ 2023)']
X4 = sm.add_constant(X4)
model4 = sm.OLS(y4, X4).fit()
print('\nResults For Property Damage Per Capita and Distance_to_Fault')
print(model4.summary())
```

# Explanation of OLS Results
Here we can see that all the OLS tests except the one between Crop Damage Per Capita and distance to fault had statistically significant results. Thus our hypothesis of proximity to a fault line should have a positive correlation for all types of damages is correct (yay!). # Ethics & Privacy

### Overview
The focus of our project involves analyzing natural disasters that have occurred in California using the SHELDUS dataset, with a refined time span of 1990-2023. By ensuring that we have a specific time frame, this will allow us to make use of our data consistently and reliably. Below are some of the ethical concerns that can arise during the project and how we might address these biases and privacy concerns. 

### Bias in Data Collection & Representation
#### Historical Bias:
Concern: 
The dataset from SHELDUS can include records as early as 1960 to 2023. Since not all of the reports are recent, there could be changes in how data is collected and they are being reported, leading to having some inconsistencies in the data. Biases can appear in the analyses because earlier records might provide inaccurate or incomplete form.

Addressing: 
Concerning data from 1990 to 2023, we will be depending on reflections that are more reliable and uniform. We will also be cross-validating the SHELDUS data with FEMA records to ensure accuracy.
#### Geographic Bias:
Concern: 
The reporting of casualties and damage may vary because some counties receive more attention due to their population density, economic importance or access to government resources. Rural areas with lower economic resources might report fewer casualty and damage events than actual occurrences.
By underreporting in rural areas with lower economic resources can provide some misrepresentation to the true impact of natural disasters.

Addressing: 
In order to avoid putting too much emphasis on sparsely covered areas, we will lump data from smaller communities into larger regional groups and apply minimum reporting thresholds.
#### Socioeconomic Bias:
Concern: 
Property damage cost may not accurately represent the low-income population compared to wealthier communities.

Addressing:
We recognize this limitation and intend to include the results of sensitivity analyses, which would assess and discuss transparently how inequalities may affect our results.

### Data Privacy & Ethical Use
#### Risks of Breach of Privacy:
Concern: 
The SHELDUS data groups information based on the county level, but smaller communities are at greater risk of being identified. Which may allow for the exposure of sensitive information that may have negative impacts on the community.

Addressing: 
Data from very small communities will be combined into larger regional groups to reduce the risk of identification.
Minimum reporting thresholds will be invoked to suppress information that may lead to the identification of particulars about small communities.
All potentially identifying details will be deleted or obscured to prevent the possibility of re-identification.

### Addressing Bias and Ethical Communication
#### Validation of Data:
The SHELDUS data will be verified against other sources such as FEMA records in order to increase the accuracy of this data and to identify—and if possible fill—data gaps.
#### Transparency:
Our final report will explicitly present all restrictions of our dataset and analysis results.
#### Ethical Communication
When sharing results we will include a disclaimer to ensure that the results aren’t misinterpreted.
