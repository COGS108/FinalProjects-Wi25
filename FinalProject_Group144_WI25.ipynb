# Abstract

This project investigates the relationship between crime rates, weather conditions, and temperature in Chicago from 2019 to 2024. In general, we were curious about if weather had an impact on crime rates. We chose Chicago because it is a city with a relatively high crime rate and it has a diverse climate. The primary research question explores how differences in daily weather patterns influence reported crime incidents in the city. While previous studies have linked extreme weather events to changes in crime patterns, there is limited research on how daily weather conditions affect urban crime rates. Using publicly available crime and weather data, we conducted exploratory data analysis and a regression analysis using a merged dataset to examine the relationship between weather patterns and crime in Chicago.

We used a linear regression model to analyze the relationship between temperature and crime rates. The results show that both the intercept and temperature have a p-value of 0, which means there is a significant relationship between them. The temperature coefficient (2.7055) suggests that crime rates increase as temperature rises, but this does not imply causation. Next, we examined weather conditions by grouping them into larger categories to avoid multicollinearity issues. In our first model, snow had a significant effect on crime rates (p = 0.003), while rain had a weaker effect (p = 0.062). Overcast conditions were not significant (p = 0.847). In the second model, neither overcast (p = 0.851) nor the combined snow and rain category (p = 0.195) had a significant effect. Overall, temperature strongly impacts crime, but the role of specific weather conditions is unclear.

Our hypothesis predicted that crime rates would increase with rising temperatures and be lower on sunny days. Our findings support the first part of the hypothesis, as the regression analysis shows a significant relationship between temperature and crime rates (p = 0). However, the effect of weather conditions is less clear. While snow had a significant impact on lowering crime rates (p = 0.003), rain and overcast conditions were not strongly significant. This suggests that while temperature plays an important part in crime trends, the influence of specific weather conditions remains uncertain. These insights could help inform public safety policies and law enforcement strategies in urban areas.

```python
#load data
crime_dfs = []
for i in range(15):
    crime_dfs.append(pd.read_csv('crime{}.csv'.format(i)))
crime = pd.concat(crime_dfs).reset_index(drop=True)
crime
```

It's pretty clear there needs to be a way to better format this. We could remove some unecessary columns for our data as mentioned earlier and we could sort and filter out the years from 2019-2024.
```python
#Converts 'Date' to datetime format and Removes Time 
crime['Date'] = pd.to_datetime(crime['Date'], format = "%m/%d/%Y %I:%M:%S %p", errors = "coerce").dt.date

#Convert 'Arrest' to boolean 
crime['Arrest'] = crime['Arrest'].astype(bool) 

#Filter for crimes that only occured between 2019 and 2024 
crime = crime[(crime['Year'] >= 2019) & (crime['Year'] <= 2024)]

#Sorts 'Year' and 'Date' from 2019-2024 
crime = crime.sort_values(by = ["Year", "Date"], ascending = [True, True])

#Handles Outliers (for 'Year')
crime['Year'] = np.where(crime['Year']< 2019, 2019, crime ['Year'])
crime['Year'] = np.where(crime['Year']> 2024, 2024, crime ['Year'])

#Keep the necessary columns and drop the unecessary columns
columns_to_keep = ["Date", "Primary Type", "Description", "Location Description", "Arrest", "Year"] 
crime = crime[columns_to_keep]

#Resets index 
crime.reset_index(drop = True, inplace = True)
crime.index = range(1, len(crime) + 1)
print(crime.shape)
```

## Machine Learning- OLS Regression



We will use linear regression to observe the relationship between temperature and crime rate. From the model summary, we can see that both the intercept and temperature have a p-value of 0, indicating a relationship between temperature and crime rate (this does not imply a causal relationship). Another point to note is that temperature has a coefficient of 2.7055, a positive number, which shows that the crime rate increases as the temperature rises.

```python
import statsmodels.formula.api as smf
#A model with temp only
model1 = smf.ols(formula='crime_rate ~ temp', data=temp_rate)
result1 = model1.fit()
print(result1.summary())
```

To make the model more understandable, we draw a plot for it.\
x-axis is temperature and y-axis is crime rate.\
The line on the plot is the line we fit.

```python
sns.scatterplot(x='temp', y='crime_rate', data=temp_rate)
xs = np.arange(temp_rate['temp'].min(), temp_rate['temp'].max())
ys = result1.params.iloc[1] * xs + result1.params.iloc[0]
plt.plot(xs, ys, '--k', linewidth=4, label='Model')
vals='$y={:.3f} + {:.3f}x, p={:.3f}, r={:.3f}, r^2={:.3f}$'.format(result1.params.iloc[0], 
                                                                   result1.params.iloc[1], 
                                                                   result1.pvalues.iloc[1], 
                                                                   temp_rate[['crime_rate','temp']].corr().iloc[0,1], 
                                                                   result1.rsquared)
plt.title(vals);

plt.xlabel('temp')
plt.ylabel('crime rate')
plt.legend();
```

Then we want to take conditions into consideration. However, different categories in conditions share a many similarities. For example, Partially cloudy and Overcast are different categories but they are actually quite similar. Using the categories directly into linear regression has a large chance of causing multicollinearity. Therefore, we group the conditions into larger groups:
- Clear
- Overcast: Partially cloudy/Overcast
- Snow: Snow, Partially cloudy/Snow, Overcast
- Rain: Rain, Partially cloudy/Rain, Overcast/Rain
- Snow, Rain: Snow, Rain, Overcast/Snow, Rain, Partially cloudy

But one may notice that there is also a potential of multicollinearity. 'Snow', 'Rain' and 'Snow, Rain' share something in common. Thus, we create two models in which one doesn't contain 'Snow, Rain' and one doesn't contain 'Snow' and 'Rain'.

```python
#Here we create a DataFrame using new conditions category.
#We create a new DataFrame 'condition_bins_combine' which is similar to combined DataFrame but use new categories instead of old ones.
condition_bins_combine = temp_bins_combine.copy()
condition_bins_combine['conditions'] = temp_bins_combine['conditions'].replace('Partially cloudy', 'Overcast')\
    .replace('Snow, Partially cloudy', 'Snow').replace('Snow, Overcast', 'Snow')\
    .replace('Snow, Rain, Overcast', 'Snow, Rain').replace('Snow, Rain, Partially cloudy', 'Snow, Rain')\
    .replace('Rain, Partially cloudy', 'Rain').replace('Rain, Overcast', 'Rain')
#We create a new DataFrame 'condition_bins_weather' which is similar to weather DataFrame but use new categories instead of old ones.
condition_bins_weather = temp_bins_weather.copy()
condition_bins_weather['conditions'] = temp_bins_weather['conditions'].replace('Partially cloudy', 'Overcast')\
    .replace('Snow, Partially cloudy', 'Snow').replace('Snow, Overcast', 'Snow')\
    .replace('Snow, Rain, Overcast', 'Snow, Rain').replace('Snow, Rain, Partially cloudy', 'Snow, Rain')\
    .replace('Rain, Partially cloudy', 'Rain').replace('Rain, Overcast', 'Rain')

#Now we can calculate the crime rate for each combination of temp and conditions by dividing crime count using weather count
temp_5c_rate=((condition_bins_combine.groupby(['temp', 'conditions'])['Date'].count())/\
    (condition_bins_weather.groupby(['temp', 'conditions'])['datetime'].count())).reset_index().rename(columns={0:'crime_rate'})
```

From the model without 'Snow, Rain,' we can see that the p-value for C(conditions)[T.Snow] is only 0.003, meaning that crime rates differ between Snow and Clear conditions. The coefficient for C(conditions)[T.Snow] is -45.6310, a negative number, meaning the crime rate in snowy conditions is lower than the crime rate in clear conditions. Meanwhile, the p-values for C(conditions)[T.Overcast] and C(conditions)[T.Rain] are both greater than 0.05, meaning we don't have enough confidence to conclude that the values are not due to randomness.

```python
#Using the data left after removing conditions 'Snow, Rain' to build a model
model2 = smf.ols(formula='crime_rate ~ temp+C(conditions)', data=temp_5c_rate[temp_5c_rate['conditions']!='Snow, Rain'])
result2 = model2.fit()
print(result2.summary())
```

The model is shown as a plot.

```python
#Draw a plot to see the model visually
sns.scatterplot(data=temp_5c_rate[temp_5c_rate['conditions']!='Snow, Rain'], hue='conditions', x='temp', y='crime_rate',
                palette={'Clear': 'orange', 'Snow': 'brown', 'Overcast': 'grey', 'Rain': 'blue'})

xs = np.arange(temp_5c_rate[temp_5c_rate['conditions']!='Snow, Rain']['temp'].min(),
               temp_5c_rate[temp_5c_rate['conditions']!='Snow, Rain']['temp'].max())
ys = result2.params.iloc[4] * xs + result2.params.iloc[0]
plt.plot(xs, ys, '--', linewidth=2, label='Clear', color='orange')

ys = result2.params.iloc[4] * xs + result2.params.iloc[0] + result2.params.iloc[1]
plt.plot(xs, ys, '--', linewidth=2, label='Overcast', color='grey')

ys = result2.params.iloc[4] * xs + result2.params.iloc[0] + result2.params.iloc[2]
plt.plot(xs, ys, '--', linewidth=2, label='Rain', color='blue')

ys = result2.params.iloc[4] * xs + result2.params.iloc[0] + result2.params.iloc[3]
plt.plot(xs, ys, '--', linewidth=2, label='Snow', color='brown')
plt.title("OLS Regression Plot (Without Snow, Rain)")
```

Below is a model without 'Snow' and 'Rain'. We can't draw any useful conclusion about conditions because the p-value for both C(conditions)[T.Overcast] and C(conditions)[T.Snow, Rain] are larger than 0.1. 

```python
#Using the data left after removing conditions 'Snow' and 'Rain' to build a model
model3 = smf.ols(formula='crime_rate ~ temp+C(conditions)', data=temp_5c_rate[(temp_5c_rate['conditions']!='Snow')& (temp_5c_rate['conditions']!='Rain')])
result3 = model3.fit()
print(result3.summary())
```

The model is shown as a plot.

```python
#Draw a plot to see the model visually
sns.scatterplot(data=temp_5c_rate[(temp_5c_rate['conditions']!='Snow')& (temp_5c_rate['conditions']!='Rain')],
                hue='conditions', x='temp', y='crime_rate',
                palette={'Clear': 'orange', 'Snow, Rain': 'blue', 'Overcast': 'grey'})

xs = np.arange(temp_5c_rate[(temp_5c_rate['conditions']!='Snow')& (temp_5c_rate['conditions']!='Rain')]['temp'].min(),
               temp_5c_rate[(temp_5c_rate['conditions']!='Snow')& (temp_5c_rate['conditions']!='Rain')]['temp'].max())
ys = result3.params.iloc[3] * xs + result3.params.iloc[0]
plt.plot(xs, ys, '--', linewidth=2, label='Clear', color='orange')

ys = result3.params.iloc[3] * xs + result3.params.iloc[0] + result3.params.iloc[1]
plt.plot(xs, ys, '--', linewidth=2, label='Overcast', color='grey')

ys = result3.params.iloc[3] * xs + result3.params.iloc[0] + result3.params.iloc[2]
plt.plot(xs, ys, '--', linewidth=2, label='Snow, Rain', color='blue')
plt.title("OLS Regression Plot (Without Snow and Rain)")
```
