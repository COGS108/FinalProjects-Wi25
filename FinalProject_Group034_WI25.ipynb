## NBA dataset

```python
## converting the csv file to pandas dataframe

df_team = pd.read_csv("team_history.csv")
df_game = pd.read_csv("game_summary.csv")
df_game2 = pd.read_csv("game.csv")
```

```python
# Data cleaning for team data

df_team = df_team[df_team['year_active_till'] >= 2001]
df_team = df_team.drop_duplicates(subset=['team_id'])

# speical cases/cities name doesn't match with the city name in the game data
df_team.loc[df_team["city"] == "Golden State", "city"] = "San Francisco"
df_team.loc[df_team["city"] == "Brooklyn", "city"] = "New York"
df_team.loc[df_team["city"] == "Utah", "city"] = "Salt Lake City"
df_team.loc[df_team["city"] == "Minnesota", "city"] = "Minneapolis"

df_team
```

```python
# change seasson to int for easier filtering and comparing
df_game["season"] = df_game["season"].astype(int)

# data cleaning for game_summary data
df_game = df_game[df_game['season'] >= 2001]
df_game_filter = df_game[["game_id", "home_team_id", "visitor_team_id", "season"]]
```

```python
# data cleaning for game.csv data
df_game2_filter = df_game2[["game_id", "wl_home"]]
```

```python
# merging the two game dataset together
dfGame = pd.merge(df_game_filter, df_game2_filter, on='game_id')
```

```python
# group bying the team id and season to get the number of wins and losses for each team
df_win_loss = dfGame.groupby(["season", "home_team_id"])["wl_home"].value_counts().unstack(fill_value=0)
df_win_loss = df_win_loss.rename(columns={"W": "Wins at Home", "L": "Losses at Home"})
```

```python
# merging the team data with the win loss data
df_win_loss = df_win_loss.reset_index()  # Reset index to make 'home_team_id' a column
df_win_loss = df_win_loss.merge(df_team, left_on="home_team_id", right_on="team_id", how="left")
```

```python
# Cleaning the columns names, only selecting the columns we need, and dropping the rows with missing values
df_win_loss = df_win_loss[["season", "city", "nickname", "Wins at Home", "Losses at Home"]]
df_win_loss = df_win_loss.rename(columns={"city": "Team City", "nickname": "Team Name"})
df_win_loss = df_win_loss.dropna()
df_win_loss
```

```python
# data cleaning for NBAFinals data
nbaFinals = df_Final[["Year", "NBA Champion", "NBA Vice-Champion"]]
nbaFinals = nbaFinals[nbaFinals["Year"] >= 2001]
```

```python
# adding the NBA champ column to the win loss data, based on who won the NBA finals that year
# create a set of tuples for the NBA champions
champions_set = set(zip(nbaFinals["Year"], nbaFinals["NBA Champion"]))

df_win_loss["NBA Champs"] = df_win_loss.apply(lambda row: (row["season"], row["Team Name"]) in champions_set, axis=1)
df_win_loss
```

```python
#Helper method used for the GDP data, got from ChatGPT 
import requests

API_KEY = "CE12ACE7-E414-4FBB-8DAA-EE0D52330037"
BASE_URL = "https://apps.bea.gov/api/data"

def list_tables():
    """Retrieve and return available table names from the BEA API."""
    params = {
        "UserID": API_KEY,
        "method": "GetParameterValues",
        "datasetname": "Regional",
        "ParameterName": "TableName",
        "ResultFormat": "json"
    }
    response = requests.get(BASE_URL, params=params)
    data = response.json()
    
    if "BEAAPI" in data and "Results" in data["BEAAPI"] and "ParamValue" in data["BEAAPI"]["Results"]:
        tables = pd.DataFrame(data["BEAAPI"]["Results"]["ParamValue"])
        return tables
    else:
        print("Error retrieving tables:", data)
        return None

def search_table(keyword):
    """Search for tables containing a specific keyword."""
    tables = list_tables()
    if tables is not None:
        filtered_tables = tables[tables["Key"].str.contains(keyword, case=False, na=False)]
        return filtered_tables
    return None

def list_linecodes(table_name):
    """Retrieve available LineCodes for a given table."""
    params = {
        "UserID": API_KEY,
        "method": "GetParameterValues",
        "datasetname": "Regional",
        "ParameterName": "LineCode",
        "TableName": table_name,
        "ResultFormat": "json"
    }
    response = requests.get(BASE_URL, params=params)
    data = response.json()
    
    if "BEAAPI" in data and "Results" in data["BEAAPI"] and "ParamValue" in data["BEAAPI"]["Results"]:
        linecodes = pd.DataFrame(data["BEAAPI"]["Results"]["ParamValue"])
        return linecodes
    else:
        print("Error retrieving LineCodes:", data)
        return None

def fetch_data(table_name, line_code, geo_fips="STATE", year="ALL"):
    """Fetch BEA data for a specific table and LineCode."""
    params = {
        "UserID": API_KEY,
        "method": "GetData",
        "datasetname": "Regional",
        "TableName": table_name,
        "LineCode": line_code,
        "GeoFIPS": geo_fips,
        "Year": year,
        "ResultFormat": "json"
    }
    response = requests.get(BASE_URL, params=params)
    data = response.json()
    
    if "BEAAPI" in data and "Results" in data["BEAAPI"] and "Data" in data["BEAAPI"]["Results"]:
        df = pd.DataFrame(data["BEAAPI"]["Results"]["Data"])
        return df
    else:
        print("Error retrieving data:", data)
        return None
```

### Analysis of GDP and NBA Team Success

Now we look to find relationships between NBA cities success (or lack thereof) and the city GDP. First we get merged data for each teams season and the respective city's percentage increase in GDP that year. We use percentage to hopefully mitigate the effects of inflation

```python
# Remove rows with 'Cleveland-Elyria' and 'Portland-Vancouver-Hillsboro'
dfGDP_filtered = dfGDP[~dfGDP['City'].isin(['Cleveland-Elyria', 'Portland-Vancouver-Hillsboro'])]

# Calculate GDP Growth Rate
dfGDP = dfGDP_filtered.sort_values(by=['Matched City', 'TimePeriod'])  
dfGDP['GDP Growth (%)'] = dfGDP.groupby('Matched City')['GDP (Millions)'].pct_change() * 100

# Merge GDP Growth and NBA Team Data (dfNBA)
dfNBA_merged = dfNBA.merge(dfGDP[['Matched City', 'TimePeriod', 'GDP (Millions)', 'GDP Growth (%)']], 
                     left_on=['Team City', 'season'], 
                     right_on=['Matched City', 'TimePeriod'], 
                     how='left')


# Add Home Winning Percentage
dfNBA_merged["Total Games"] = dfNBA_merged["Wins at Home"] + dfNBA_merged["Losses at Home"]
dfNBA_merged["Winning Percentage"] = dfNBA_merged["Wins at Home"] / dfNBA_merged["Total Games"] 
dfNBA_merged
```

```python
plt.figure(figsize=(10, 6))


sns.regplot(x='GDP Growth (%)', y='Winning Percentage', data=dfNBA_merged, 
            scatter_kws={'s': 50, 'color': 'red'}, 
            line_kws={'color': 'red', 'linewidth': 2}) 

plt.title('Relationship between GDP Growth and NBA Home Winning Percentage', fontsize=16)
plt.xlabel('GDP Growth (%)', fontsize=12)
plt.ylabel('Winning Percentage', fontsize=12)
plt.grid(True)

plt.show()
```

If we plot home winning percentage and GDP growth percentage, it seems like there may be almost no correlation. Perhaps for some teams they are positively correlated and for others they are negatively correlated.

After looking at averages, we want to see the trends for the top 5 cities with the highest GDP. The top 5 cities are New York, Los Angelos, Chicago, San Francisco, and Dallas. We will compare their home city's GDP (in millions) to that team's winning percentage in a scatterplot below.


```python
top_gdp_cities_list = [
    "New York",
    "Los Angeles",
    "Chicago",
    "San Francisco",
    "Dallas"
]

# Filter for only the top 5 GDP cities
dfNBA_topGDP = dfNBA_merged[dfNBA_merged['Team City'].isin(top_gdp_cities_list)]

plt.figure(figsize=(10, 6))

# Create scatter plot with separate linear regressions for each city
sns.lmplot(data=dfNBA_topGDP, 
           x='GDP (Millions)', 
           y='Winning Percentage', 
           hue='Team City', 
           scatter_kws={'s': 50, 'alpha': 0.7}, 
           line_kws={'linewidth': 2},  
           aspect=1.5)

# Titles and labels with styling
plt.title("GDP vs Winning Percentage in Top 5 GDP Cities", fontsize=16)
plt.xlabel("GDP (Millions)", fontsize=12)
plt.ylabel("Winning Percentage", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)  

plt.show()
```

This scatterplot visualizes the relationship between GDP (in millions) and winning percentage for NBA teams in the top 5 GDP cities. Each city has it's specific color and each dot represents a data point for a team in a given season, and the regression lines show trends in how GDP correlates with the winning percentage in each of these cities. We see negative trends for New York (in purple),Los Angeles (in red), Chicago (in blue), and Dallas (in orange). The only city with a positive regression line is San Francisco (in green).

The scatterplot is very dispersed, interpreting that there may be no strong correlation between GDP and winning percentage across these cities like we initailly thought.

### Championship effects on City GDP

Next, we’ll investigate whether winning an NBA championship has a measurable effect on a city’s GDP. Specifically, we’ll compare each city's GDP in the three years leading up to the championship against the three years following their title. This will help us see if there’s a significant economic change tied to the championship win.



```python
# Collect GDP trends ±3 years around each championship event
years_window = 3
champ_gdp_trends = []

for idx, row in df_win.iterrows():
    city = row['Team City']
    year = row['season']
    
    # Get GDP data around the championship year
    city_gdp = dfGDP[
        (dfGDP['Matched City'] == city) &
        (dfGDP['TimePeriod'] >= year - years_window) &
        (dfGDP['TimePeriod'] <= year + years_window)
    ].copy()
    
    city_gdp['Champ Year'] = year
    city_gdp['Champ City'] = city
    champ_gdp_trends.append(city_gdp)

# Combine into a single dataframe
champ_gdp_df = pd.concat(champ_gdp_trends)
champ_gdp_df['Years from Champ'] = champ_gdp_df['TimePeriod'] - champ_gdp_df['Champ Year']

# Plot: one graph at a time for each city with multiple lines for each championship year
champion_cities = champ_gdp_df['Champ City'].unique()

for city in champion_cities:
    city_data = champ_gdp_df[champ_gdp_df['Champ City'] == city]
    
    plt.figure(figsize=(8, 5))
    sns.lineplot(
        data=city_data,
        x='Years from Champ',
        y='GDP (Millions)',
        hue='Champ Year',  # One line per championship year
        marker='o'
    )
    
    plt.axvline(0, color='purple', linestyle='--', label='Championship Year')
    plt.title(f"GDP Trend ±{years_window} Years for {city}", fontsize=14)
    plt.xlabel("Years from NBA Championship")
    plt.ylabel("GDP (Millions)")
    plt.legend(title="Champ Year")
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```
