# Abstract

Education is one of the many partisan issues of the United States. Policies implemented by politcal leaders can alter a student's education through changes in funding, academic requirements, available resources, and more. In this study, we looked at how the party of the governor, the majority party of Congress members (referred to as legislative control), and the combination of the two (referred to as state control) affects the state's average SAT and ACT scores. Our hypothesis was that states under Republican control would be lower than states under Democratic control.  
  
Taking offical exam score data from College Board and ACT Inc, along with government data from the National Conference of State Legislatures, we compared the change in average exam score to changes in the dominant political party. While initial observations show Democratic controlled states have higher ACT score and Republican controlled states have higher ACT scores, running statistical analyses indicate low correlation between exam scores and party, with total variance captured being <10% for most regressions. This is further proved when filtering for states with frequently changing controlling parties, or swing states.  
  
Comparing exam scores to urban and rural composition showed no particular favorability to both exams. Comparing exams scores against which exam the state preferred indicated a strong correlation. For an increase in 1% of the student body taking the SAT exam, the average score drops by 2.25 (p<0.001).

## Background and Prior Work

Education is a domain that shapes our future by providing better opportunities and higher-paying jobs based on the amount of time invested and the field of study chosen. In the United States, education policy plays a major role in the quality and accessibility of schools throughout the country from preschool all the way up to university (higher education).<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) In recent years, the republican party and democratic party have had opposing views regarding education and how issues such as funding, curriculum, and public schools should be handled.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) For example, the President of the United States Donald Trump recently ordered the Secretary of Education to expand school choice for each state and increase educational freedom.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2)  This directly reflects the republican ideology of education being a state choice compared to the democratic ideology of strengthened public education systems.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2)

The divide between political parties on education policy has been a key issue throughout the U.S. In the past, the two parties broke themselves down into two descriptions: The Democratic Party was for the “common man” while the Republic Party was for “big business” and the wealthy.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) These descriptions display very clear views on how members of the party may view education and its association with social mobility. In fact, each state is typically aligned with either the Democratic or Republican Party, which reflects the predominant political leanings and beliefs of that state.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) As elections come around, this political affiliation is re-defined by the House of Representatives and the Senate within each state. This directly results in influencing policy decisions at the state level.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3)

Previous research on education and political engagement worldwide has explored the direct causal link between education and political behavior, with evidence drawn from schooling reforms across 39 countries.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) This study found a positive correlation with higher levels of education leading to increased interest in politics and domains associated with it.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) Building on this idea, our group became curious about the downstream effect of how education policies influenced by political control might impact test score performance in the U.S. specifically compared to worldwide.

This leads to our main research question: How do changes in the political party of elected state officials correlate with SAT/ACT performance from 2009 to 2024? We hypothesize that if a state’s control is definitively Republican during a given year, then that state’s SAT/ACT scores for that year will be lower compared to Democratic-controlled states. In the U.S., scholastic assessment tests (SAT) and American college testing (ACT) are widely used standardized tests that evaluate high school students on their reading, writing, and math skills.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) These scores provide a valuable metric to compare student performance across states and have been implemented since the 1960s, making them an efficient measure to assess the impact of state-level policies on education outcomes.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) In terms of factors that directly influence a state’s political affiliation, our project will focus on legislature control, governor control, and state control. 

When a party has legislative control, this means that they hold the majority of seats in both chambers of the legislature.<a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6) This control allows the given party to pass laws, influence budget allocation, and implement new state policies.<a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6) Governor control refers to the power of a state by the governor who is the chief executive officer of that state.<a name="cite_ref-7"></a>[<sup>7</sup>](#cite_note-7) The governor of that state can veto bills passed by the legislature, issue executive orders, propose state budgets, and influence policies.<a name="cite_ref-7"></a>[<sup>7</sup>](#cite_note-7) Typically, legislative control holds more power over policy decisions than governor control. State control is when a party holds both legislative chambers and governorship.<a name="cite_ref-8"></a>[<sup>8</sup>](#cite_note-8) This allows the state to create and enforce a variety of laws based on the needs of the people in that state without interference from the federal government.<a name="cite_ref-8"></a>[<sup>8</sup>](#cite_note-8) 

When analyzing the impact of state control on policy making for educational matters, it is important to consider confounding variables that can influence this relationship and skew the results. The first variable to point out is rural versus urban population density per state.  
Rural areas tend to have a smaller population compared to urban areas and can in turn influence standardized test scores.<a name="cite_ref-9"></a>[<sup>9</sup>](#cite_note-9)  The next variable to consider is trends for educational policies across democratic majority and republican majority states. Political control of a state is influenced by that state’s needs and its demographics. Democratic-majority states may focus on policies that address challenges in urban areas while Republican-majority states are more likely to prioritize policies for rural populations.<a name="cite_ref-10"></a>[<sup>10</sup>](#cite_note-10) These different priorities can influence our measurement of standardized test scores as urban areas may have more educational opportunities compared to rural areas.<a name="cite_ref-10"></a>[<sup>10</sup>](#cite_note-10) The last confounding variable to account for is a preference for one test measure over the other.<a name="cite_ref-11"></a>[<sup>11</sup>](#cite_note-11) In coastal states, there is a preference to take the only SAT compared to southern and midwest states that prefer the ACT.<a name="cite_ref-11"></a>[<sup>11</sup>](#cite_note-11) This regional variation in standardized tests may affect our utilization of standardized test scores for measurement and interpretation.

### SAT Scores

Next we will wrangle our SAT dataset. Our SAT dataset has scores from two different eras of the test where one is scored on a scale of 2400 points and the other on a scale of 1600. The datasets also include information that is not relavent to our research question like columns that display the average score for specific subject categories. We will address both of these issue in our data cleaning.

```python
path = "./data/excel/sat/"
files = listdir(path)
```

```python
dfs = []
for file in files:
    dfs.append(pd.read_excel(path + file))
    dfs[-1]["Year"] = int(file[:4])

sat_scores = pd.concat(dfs)
```

```python
sat_scores
```

```python
# Clean up state values
sat_scores["State"] = sat_scores["State"].str.replace(r"\\\d\\", "", regex=True)
sat_scores["State"] = sat_scores["State"].str.replace(".", "").str.strip()
sat_scores = sat_scores[sat_scores["State"].notnull() & (sat_scores["State"] != "")]
```

```python
sat_scores["State"].unique()
```

```python
bad_states = ["District of Columbia", "District of", "Columbia"]
sat_scores = sat_scores[~sat_scores["State"].isin(bad_states)]
```

```python
# Verify only 50 states are left
len(sat_scores["State"].unique())
```

```python
sat_scores
```

Since the SAT had a redesign in 2016, we will convert older years to the newer format by simply taking the average old reading and writing sections.

```python
cond = sat_scores["ERW"].isnull()
sat_scores.loc[cond, "ERW"] = sat_scores.loc[
    cond, ["Critical reading", "Writing"]
].mean(axis=1)
sat_scores["Total"] = sat_scores["Mathematics"] + sat_scores["ERW"]
```

```python
sat_scores.isnull().any()
```

```python
# Get ACT equivalent
sat_scores["ACT Equivalent"] = pd.cut(
    x=sat_scores["Total"], bins=([0] + conv["SAT High"].tolist()), labels=conv["ACT"]
)
```

```python
sat_scores = sat_scores[
    ["State", "Year", "Percent taking SAT", "Total", "ACT Equivalent"]
]
sat_scores.columns = ["state", "year", "percent_taking_sat", "total", "act_equivalent"]
sat_scores = sat_scores.sort_values(["year", "state"], ignore_index=True)
```

### Exam Scores vs State Control

#### Analyzing Swing States Based on State Control

When attempting to answer the question of how does government control impact average testing scores, its important to identify how specific parties are connected to various scores. Based on the boxplots and lineplots, states that had divided control have average scores that fall between the averages for Republican and Democratic states. This finding is interesting because it suggests that there is some sort of middle ground where Republican control will result in one extreme and Democratic in another. States that find themselves divided will be influenced by both ends of the spectrum and may explain why their scores are in the middle. Because of this observation, we want to analyze those states that fall in the middle. Observing how scores fluctuate in conjection with switches in party control will grant us the opportunity to pinpoint trends that can be associated with each party.

#### Identifying Swing States

Swing states can be described as states who frequently switch between Democratic and Republican control. These regular switches can lead to vastly different policy proposals that will have downstream effects on the economy, education, and welfare of the state. Swing states are influenced by both Democratic and Republican ideals. These influences may coincide with how scores change and could be a reflection of these parties' policies. First, we want to establish which states we will label as "swing" states. To do this, we've developed a metric where every time a state experiences complete Democratic control, they will receive a score of +1. Everytime a state experiences complete Republican control, they will receive a score of -1. Lastly, everytime a state experiences divided control, they will receive a score of 0. The average state control score will be calculated and states that have averages between +0.5 and -0.5 will be classified as swing states.

```python
# Assign numerical values
control_values = {"Rep": -1, "Dem": 1, "Divided": 0}
sat_merged_data["state_control_numeric"] = sat_merged_data["state_control"].map(
    control_values
)

# Compute the average state control value for each state
state_control_avg = (
    sat_merged_data.groupby("state")["state_control_numeric"].mean().reset_index()
)

state_control_avg.head()
```

```python
# Swing States Dataframe
swing_states = state_control_avg[
    (state_control_avg["state_control_numeric"] >= -0.5)
    & (state_control_avg["state_control_numeric"] <= 0.5)
]
swing_states
```

#### State Control Influence on SAT and ACT Scores

Now that we have identified which states we'll classify as having mixed control, we will analyze how their average SAT scores change overtime, while also taking into account how party control changes at the state level.

```python
# Merge SAT data with swing states
swing_sat_data = sat_merged_data[sat_merged_data["state"].isin(swing_states["state"])]

# Calculate the average SAT score for each swing state over time
avg_sat_swing = swing_sat_data.groupby(["state", "year"])["total"].mean().reset_index()

# Plot the SAT scores over time for each swing state
plt.figure(figsize=(10, 6))
sns.lineplot(data=avg_sat_swing, x="year", y="total", hue="state")

plt.title("Average SAT Scores Over Time for Swing States")
plt.xlabel("Year")
plt.ylabel("Average SAT Score")
plt.legend(title="State", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.grid(True, linestyle="--", alpha=0.6)

plt.show()
```

## Addressing Confounding Variables

In order to address the variance in standardized test scores for students in the U.S. we want to pivot from looking at government as a contributing factor to assessing the influence of population type and test preference. We hypothesize that factors such as urban vs. rural populations and preferences for students to take a certain exam in a state have a significant influence on test scores. This is because different population subtypes will have access to different educational opportunities that could be reflected in their scores. Test preference may also influence score averages as higher volumes of students taking a certain exam will lead to more variables impacting the overall average.

### Urban and Rural Populations vs Exam Scores

In addition to SAT and ACT scores, we found data on urban and rural population distributions of each state from 2010-2020 which may influence standardized test performance. We suspect that a state's level of urbanization could act as a confounding variable that affects exam scores due to factors such as socioeconomic conditions, educational resources, and funding for schools. When investigating this, we need to account for the fact that our urban and rural population dataset was collected from the decennial census and does not align perfectly with our yearly SAT and SAT datasets. To address this, we will calculate the average urban and rural percentages for each state and merge this with our SAT and ACT score data. This will allow us to analyze whether urbanization and rural populations impact test scores and determine if this is a confounding variable that can influence our results.

First we calculate the urban and rural percentages.

```python
# Convert populations into percentages
population_comp["urban_percentage"] = (
    population_comp["urban"] / population_comp["total"]
) * 100
population_comp["rural_percentage"] = (
    population_comp["rural"] / population_comp["total"]
) * 100
```

```python
# get average of urban and rural percentages
population_comp_avg = (
    population_comp.groupby("state")[["urban_percentage", "rural_percentage"]]
    .mean()
    .reset_index()
)

# merge population_comp_avg with our SAT and ACT score data
sat_with_population = pd.merge(sat_scores, population_comp_avg, on="state", how="inner")
act_with_population = pd.merge(act_scores, population_comp_avg, on="state", how="inner")
```

Next, we will use a boxplot to visualize and compare the distribution of SAT and ACT scores between urban or rural states to see any noticeable differences. If there are any visual differences, we will conduct a t-test to verify if the differences are meaningful.

From this visualization, there appears to be a noticeable difference in SAT and ACT scores between urban and rural populations. It appears that urban students tend to score a higher SAT score than rural students. However, for the ACT, rural students have a slightly higher score then urban students.

To verify if these differences are statistically significant, we will conduct a t-test.

```python
# Separate SAT scores
urban_sat = sat_with_population[sat_with_population["Urban/Rural"] == "Urban"][
    "total"
]  # Updated to 'total'
rural_sat = sat_with_population[sat_with_population["Urban/Rural"] == "Rural"][
    "total"
]  # Updated to 'total'

# Separate ACT scores
urban_act = act_with_population[act_with_population["Urban/Rural"] == "Urban"][
    "total"
]  # Updated to 'total'
rural_act = act_with_population[act_with_population["Urban/Rural"] == "Rural"][
    "total"
]  # Updated to 'total'

# t-test
sat_ttest = stats.ttest_ind(urban_sat, rural_sat, equal_var=False)
act_ttest = stats.ttest_ind(urban_act, rural_act, equal_var=False)

print("SAT T-test results:", sat_ttest)
print("ACT T-test results:", act_ttest)
```

The following t-test indicates a statistically significant difference in SAT scores, with a p-value of 0.0017. This suggests that urban students tend to score higher on the SAT than rural students. However, for the ACT the difference is not statistically significant, as the p-value is 0.533 (greater than 0.05). This means there is no strong evidence that either urban or rural students perform better on the ACT. 

Based on our analysis, we can conclude that urbanization could be a confounding variable that influences SAT scores. However, since ACT scores do not show a statistical difference, future analysis could be done to get a better understanding of why SAT scores are effected by urbanization then ACT scores.

### Exam Scores vs State Exam Preference

To simplify our categorization, we will say that states with subsequent averages greater than or equal to 50 as a SAT preferred state and averages less than 50 as an ACT preferred state. ## Limitations  
  
When investigating our research question, we must recognize the limitations of our study. A significant limitation of this study is the vastly different population sizes for each state. When focusing on government control, we are looking at how a small portion of individuals influence the outcome of millions of students with some government officials being responsible for bigger portions of that population. Because the population sizes of each state are so different, states such as California will have more students contributing to the overall average than a state like Alaska. This difference will most likely also be reflected in our findings. Another limitation of this study was handling differences in exams. When doing conversions between the two SATs or between the SAT and ACT, chances are nuances that explain the specific scores were lost to generate larger generalizable datasets. Lastly, political affiliation in this study is being treated as an immediate and bisected contributor to factors like policy and budget that will impact education and therefore influence test scores. We were not looking at political values for each official on a state-by-state level meaning we have no way of determining what similarities and differences Republican and Democratic officials share that will influence education specifically. We also do not take into account when specific educational policies go into effect. We cannot assume that the election of an official will have an immediate impact which means we are losing some of the nuance associated with the relationship between time and governmental procedures that dictate how the education system is altered.
