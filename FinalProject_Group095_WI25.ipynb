# Abstract

These results challenge the notion of "superhero fatigue" and highlight the importance of maintaining production standards to sustain audience engagement.

## Background and Prior Work

For over a decade, superhero films based on comic book characters have dominated the U.S. box office. Since the release of Iron Man in 2008, which marked the beginning of the Marvel Cinematic Universe (MCU), and the release of the highly anticipated DC films like The Dark Knight in the same year, the genre has experienced explosive growth in both output and popularity. However, as the number of superhero films released each year has significantly increased, spanning both the MCU and the DC Extended Universe (DCEU), so too has the debate over whether audiences are experiencing "superhero fatigue", a decline in interest due to the oversaturation of the genre. 

The term "superhero fatigue" refers to the hypothesis that audiences are growing tired of the genre due to its overwhelming presence in popular culture, repetitive storytelling, and an overreliance on shared universes and interconnected stories. This phenomenon has been identified through observable trends such as declining box office returns, diminishing critical and audience review scores, and anecdotal evidence from viewers expressing disinterest or frustration with the genre. For example, recent MCU releases like Ant-Man and the Wasp: Quantumania (2023) and The Marvels (2023) underperformed at the box office compared to earlier installments, while DC films like Shazam! Fury of the Gods (2023) and The Flash (2023) faced similar struggles. Additionally, aggregated review scores on platforms like Rotten Tomatoes and Metacritic have shown a downward trend for many recent superhero films, suggesting a potential shift in audience and critic sentiment. This phenomenon raises important questions about the sustainability of the genre and the factors that drive its success. Are declining box office returns and review scores from critics and audiences evidence of audience fatigue, or do they reflect the varying quality of individual films?

Previous work on this topic has explored various aspects of superhero film performance, though only a few have directly addressed the question of superhero fatigue. For example, an analysis from Stat Significant examined box office trends from the early 2000s to 2023 and found that while superhero films experienced significant growth through 2019, the average box office revenues have declined in recent years, suggesting some possible shift in audience engagement<a name="cite_ref-1"></a><sup>1</sup>. Conversely, some argue against the notion of superhero fatigue. An article from Collider argues that the perceived decline in audience reception to these films may be more related to the quality of recent releases rather than an oversaturation of the genre. This suggests that audiences are reacting to subpar storytelling and production quality in those films, leading to underperforming at the box office, while films that offer a fresh story or maintain high production standards continue to perform well, indicating that audience demands persist when expectations are met<a name="cite_ref-2"></a><sup>2</sup>. However, these studies do not directly quantify whether superhero movie fatigue exists as a measurable phenomenon. By analyzing domestic box office performance and critic/audience reviews scores from 2008 to the present, this project aims to determine whether there is a significant decline in the genre's performance over the years and to explore the role of storytelling quality in sustaining audience engagement.

## Dataset 1 : Box Officof DC and Marvel Superhero Movies

```python
# Set up pandas, matplotlib, and seaborn
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib as mpl
import matplotlib.pyplot as plt
```

```python
# Read the CSV file from the dataset folder
df = pd.read_csv('dataset/dc_marvel_movie_performance.csv')

# Remove empty rows and filter for films since 2008
df = df.dropna(how='all')
df = df[df['Year'] >= 2008]

# Standardize column names
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')

# Remove '$' and commas, then convert money columns to float
money_columns = [
    'box_office_gross_domestic_(u.s._and_canada_)',
    'box_office_gross_other_territories',
    'box_office_gross_worldwide',
    'budget', 
    'inflation_adjusted_worldwide_gross',
    'inflation_adjusted_budget', 
    '2.5x_prod'
]
for col in money_columns:
    df[col] = df[col].replace(r'[\$,]', '', regex=True).astype(float)

# Fill missing values in 'box_office_gross_other_territories' with 0
df['box_office_gross_other_territories'] = df['box_office_gross_other_territories'].fillna(0)

# Drop rows with missing Rotten Tomatoes Critic Score
df = df.dropna(subset=['rotten_tomatoes_critic_score'])

# Convert release date from string to datetime
df['release_date'] = pd.to_datetime(df['u.s._release_date'], format='%d/%m/%Y')

# Calculate profit margin
df['profit_margin'] = (df['box_office_gross_worldwide'] - df['budget']) / df['budget']

# Convert year to integer to remove decimals (e.g., 2008.0 becomes 2008)
df['year'] = df['year'].astype(int)
# Rename the 'year' column to 'release_year'
df = df.rename(columns={'year': 'release_year'})

# Select the columns to keep – note that we now use 'release_year' instead of 'year'
keep_columns = [
    'film', 'release_year', 'release_date', 'franchise',
    'box_office_gross_domestic_(u.s._and_canada_)',
    'rotten_tomatoes_critic_score', 'budget', 'profit_margin',
    'male/female-led', 'break_even', 'domestic_%', 'gross_to_budget'
]

clean_df = df[keep_columns]

# Display the first few rows of the cleaned DataFrame
clean_df.head()
```

## Dataset 2 : Marvel vs DC

```python
# Read the CSV file from the dataset folder 
df2 = pd.read_csv('dataset/dbDCvsMarvel.csv', encoding='latin1')

# Standardize column names: lowercase, replace all non-alphanumeric with underscore, and strip underscores
df2.columns = df2.columns.str.lower().str.replace('[^a-z0-9]', '_', regex=True).str.strip('_')

# Trim whitespace from specific columns
df2['original_title'] = df2['original_title'].str.strip()
df2['company'] = df2['company'].str.strip()

# Fix typo in dataset
df2['original_title'] = df2['original_title'].replace('Captain Marve', 'Captain Marvel')

# Rename 'original_title' to 'film'
df2 = df2.rename(columns={'original_title': 'film'})

# Remove non-numeric characters from specific columns and convert them to float
numeric_cols = ['minutes', 'budget', 'opening_weekend_usa', 'gross_usa', 'gross_worldwide']
for col in numeric_cols:
    df2[col] = df2[col].astype(str).str.replace('[^0-9]', '', regex=True).astype(float)

# Rename columns to be more descriptive
df2 = df2.rename(columns={
    'minutes': 'runtime_minutes',
    'release': 'release_year',
    'gross_usa': 'domestic_gross',
    'gross_worldwide': 'worldwide_gross'
})

# Drop rows with missing data in critical fields
df2 = df2.dropna(subset=['budget', 'worldwide_gross'])

# Select desired columns 
keep_columns = [
    'film', 'company', 'release_year', 'runtime_minutes', 'rate', 'metascore',
    'budget', 'opening_weekend_usa', 'domestic_gross', 'worldwide_gross'
]
clean_df2 = df2[keep_columns]

# Display the first few rows of the cleaned DataFrame
clean_df2.head()
```

## Final Dataset : Merging both Datasets

Now that we have two dataframes, we will merge these dataframes together to create a comprehensive dataset for our analysis.

```python
# First, merge the dataframes
merged_df = pd.merge(
    clean_df, 
    clean_df2, 
    on=['film', 'release_year'], 
    how='outer',
    suffixes=('', '_df2')
)

# Function to fill missing values in the merged dataframe
def fill_missing_values(row):
    # For budget, use df2 value if df1 value is missing
    if pd.isna(row['budget']):
        row['budget'] = row['budget_df2']
    
    # For domestic gross, use df2 value if df1 value is missing
    if pd.isna(row['box_office_gross_domestic_(u.s._and_canada_)']):
        row['box_office_gross_domestic_(u.s._and_canada_)'] = row['domestic_gross']
    
    # Fill franchise information for newer movies
    if pd.isna(row['franchise']):
        if 'marvel' in str(row['company']).lower():
            row['franchise'] = 'Marvel'
        elif 'dc' in str(row['company']).lower():
            row['franchise'] = 'DC'
    
    # Calculate profit_margin if missing
    if pd.isna(row['profit_margin']) and not pd.isna(row['worldwide_gross']) and not pd.isna(row['budget_df2']):
        row['profit_margin'] = (row['worldwide_gross'] - row['budget_df2']) / row['budget_df2']
    
    # For newer movies, use metascore as a proxy for rotten_tomatoes if missing
    if pd.isna(row['rotten_tomatoes_critic_score']) and not pd.isna(row['metascore']):
        # Convert metascore (0-100) to approximate RT score (also 0-100)
        row['rotten_tomatoes_critic_score'] = row['metascore']
    
    return row

# Apply the function to fill missing values
complete_df = merged_df.apply(fill_missing_values, axis=1)

# Drop any rows that still have critical missing values
critical_columns = ['film', 'release_year', 'franchise', 'budget', 
                   'rotten_tomatoes_critic_score', 'profit_margin']
complete_df = complete_df.dropna(subset=critical_columns)

# Clean up release dates by removing the time portion
if 'release_date' in complete_df.columns:
    # Convert to string first to handle different date formats
    complete_df['release_date'] = complete_df['release_date'].astype(str)
    # Remove the time portion (everything after and including space + digits)
    complete_df['release_date'] = complete_df['release_date'].str.replace(r' 00:00:00.*$', '', regex=True)

# Select only the columns we want for our final dataset
final_columns = [
    'film', 'release_year', 'release_date', 'franchise',
    'box_office_gross_domestic_(u.s._and_canada_)',
    'rotten_tomatoes_critic_score', 'budget', 'profit_margin',
    'break_even', 'male/female-led', 'runtime_minutes'
]

# Keep only columns that exist in complete_df
final_columns = [col for col in final_columns if col in complete_df.columns]
final_df = complete_df[final_columns]

# Add missing films 
new_films = pd.DataFrame({
    'film': [
        'Joker: Folie à Deux',
        'Deadpool & Wolverine',
        'Kraven the Hunter', 
        'Venom: The Last Dance'
    ],
    'release_year': [2024, 2024, 2024, 2024],
    'release_date': ['2024-10-04', '2024-07-26', '2024-12-13', '2024-10-25'], 
    'franchise': ['DC', 'Marvel', 'Marvel', 'Marvel'],
    'rotten_tomatoes_critic_score': [31, 78, 15, 40],
    'budget': [200000000, 200000000, 110000000, 110000000],
    'box_office_gross_domestic_(u.s._and_canada_)': [58300287, 636745858, 61989100, 139755882],
    'male/female-led': ['Costarring', 'Male', 'Male', 'Male'],  
    'runtime_minutes': [138, 138, 127, 109]
})

# Calculate profit margin for the new films
new_films['profit_margin'] = (new_films['box_office_gross_domestic_(u.s._and_canada_)'] - new_films['budget']) / new_films['budget']

# Set break_even values to "Success" or "Flop" 
new_films['break_even'] = new_films['box_office_gross_domestic_(u.s._and_canada_)'] > new_films['budget']
new_films['break_even'] = new_films['break_even'].map({True: 'Success', False: 'Flop'})

# Select only columns that are in final_df
new_films = new_films[[col for col in new_films.columns if col in final_df.columns]]

# Concatenate with existing dataframe
final_df = pd.concat([final_df, new_films], ignore_index=True)

final_df.head()
```

## Success Rate Trends Over Time

Next, we will determine the commercial viability of superhero films over time by calculating a "success rate" 
for each year—defined as the percentage of films that achieved domestic box office revenues exceeding their production budgets. While this represents a simplified measure that doesn't account for marketing costs or additional revenue streams (international box office, streaming, merchandising), it provides a consistent metric to track commercial performance trends over our study period.

The success rate is calculated as follows:
- **Success**: Films where domestic box office > production budget
- **Flop**: Films where domestic box office <= production budget
- **Annual Success Rate**: Percentage of films released in a given year categorized as "Success"

This metric helps us identify periods of stronger and weaker financial performance in the superhero genre, providing evidence for or against the "superhero fatigue" hypothesis.

```python
# SUCCESS RATE ANALYSIS: Success Rate Trends Over Time
print("\n--- Success Rate Trends ---")
yearly_success = final_df.groupby('release_year')['break_even'].apply(
    lambda x: (x == 'Success').mean() * 100
).round(1)
print("Success Rate by Year (%):")
print(yearly_success)

plt.figure(figsize=(14, 6))
yearly_success.plot(marker='o', linestyle='-', color='purple')
plt.title('Superhero Film Success Rate by Year', fontsize=15)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Success Rate (%)', fontsize=12)
plt.grid(linestyle='--', alpha=0.7)
plt.xticks(yearly_success.index.astype(int))
plt.tight_layout()
plt.show()
```

### Success Rate Trends Over Time: Explanation

The graph above shows that the genre experiences a challenging period with a 0% success rate from 2004-2006. However, success rates began to improve but remained inconsistent, fluctuating between 0% and 50% between 2008-2011. In 2012-2029, superhero films had remarkably consistently high success rates above 65% and peaked at 100% in 2014, 2016, and 2018. During these eight years, the average success rate was about 84%, suggesting excellent commercial consistency. However, in 2020, the success rate was 0%, reflecting the impact of COVID-19. However, post-pandemic success rates showed more variability, ranging from 25% to 57%, with an average of approximately 39%. 

The extraordinary success rates of 2012-2019 (84% average) represented an unprecedented golden era that was likely unsustainable long-term for any genre. While lower, the current post-pandemic success rates are more comparable to the genre's emergence period (2008-2011) and still demonstrate that a significant portion of superhero films remain financially viable.

Rather than indicating audience rejection of the genre, these trends suggest a market normalization following an exceptional period of success, compounded by broader post-pandemic changes in theatrical attendance patterns affecting all genres. The data points to increased selectivity rather than wholesale rejection, suggesting audiences still support superhero films, but perhaps with greater quality expectations than during the genre's peak period.



## Profit Margins of Individual Films and Yearly Averages

We will now examine profit margins across the entire dataset. Unlike the binary success/failure metric, profit margins quantify the degree of success or failure, revealing how efficiently a film converts its production budget into box office returns

The profit margin is calculated by (Domestic Box Office - Budget) / Budget

```python
# PROFITABILITY ANALYSIS: Profit Margins of Individual Films and Yearly Averages
print("\n--- Profit Margin Trends ---")
yearly_profit_margin = final_df.groupby('release_year')['profit_margin'].mean()
print("Average Profit Margin by Year:")
print(yearly_profit_margin)

# Create a figure with two plots: individual films and yearly averages
fig, ax = plt.subplots(figsize=(16, 8))

# Plot individual films as scatter points
sns.scatterplot(
    data=final_df, 
    x='release_year', 
    y='profit_margin', 
    hue='franchise',
    palette='viridis', 
    s=100, 
    alpha=0.7,
    ax=ax
)

# Plot yearly averages as a connected line
yearly_profit_margin.plot(marker='o', linestyle='-', linewidth=3, color='red', 
                         label='Yearly Average', ax=ax)

# Add break-even line
plt.axhline(y=0, color='black', linestyle='--', alpha=0.7, label='Break-even point')

# Add reference lines for "blockbuster success" and "significant loss"
plt.axhline(y=1, color='green', linestyle=':', alpha=0.5, label='100% return')
plt.axhline(y=-0.5, color='red', linestyle=':', alpha=0.5, label='50% loss')

plt.title('Superhero Film Profit Margins: Individual Films & Yearly Averages', fontsize=15)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Profit Margin', fontsize=12)
plt.grid(linestyle='--', alpha=0.5)
plt.legend(title="", loc='upper left')

# Add annotations for notable outliers
# Find top 3 and bottom 3 films by profit margin
top_films = final_df.nlargest(3, 'profit_margin')
bottom_films = final_df.nsmallest(3, 'profit_margin')

for _, film in pd.concat([top_films, bottom_films]).iterrows():
    plt.annotate(
        film['film'],
        xy=(film['release_year'], film['profit_margin']),
        xytext=(5, 0),
        textcoords='offset points',
        fontsize=9,
        bbox=dict(boxstyle='round,pad=0.3', fc='yellow', alpha=0.3)
    )

plt.xticks(yearly_profit_margin.index.astype(int))
plt.tight_layout()
plt.show()
```

### Profit Margins of Individual Films and Yearly Averages : Explanation

In the graph above, the profitability of superhero films has generally increased over time, with notable peaks around 2016–2019, suggesting a golden era for the genre before a sharp decline in 2020, due to the COVID-19 pandemic. Marvel films (blue points) consistently show higher profit margins than DC films (green points), with outliers like Deadpool and Spider-Man: No Way Home demonstrating extreme profitability. While average profit margins have decreased from their 2016-2019 peaks, they remain positive (around 1.4-2.4), indicating that the genre continues to generate returns despite increased variance in individual film outcomes. This represents a normalization rather than collapse.

We will now look at the direct relationship between film budgets and domestic box office returns.

Higher-budget films such as those with greater than 200 million budget demonstrate greater volatility in outcomes as they can achieve extraordinary returns but also face substantial risks of significant losses. The scatter plot shows several high-budget films falling well below the break-even line. Whereas films in the 55-200 million range show a relatively favorable success-to-failure ratio, with many achieving substantial returns relative to their investment. This supports our hypothesis that this budget range represents an optimal balance between production quality and financial risk.

# Ethics & Privacy

There are several ethical and privacy concerns to consider in our analysis. First, if we use audience review scores from platforms like IMDb or Rotten Tomatoes, we must be mindful of potential biases such as review bombing, where certain groups manipulate scores based on external factors unrelated to film quality. Additionally, these platforms may not be demographically representative, as online reviewers tend to skew toward younger, male audiences, potentially introducing bias in how audience sentiment is measured. Box office data also has limitations, as it does not account for changing viewing habits, such as the shift to streaming, which could lead to misleading conclusions if not properly contextualized. Moreover, scraping data from review platforms may violate terms of use, requiring us to rely on publicly available APIs or licensed datasets.

To detect and mitigate these biases, we will assess data sources for representativeness, compare multiple review platforms to identify anomalies and analyze trends with consideration of external factors like streaming’s impact on box office revenue. During analysis, we will apply statistical methods to identify outliers, such as sudden score drops from review bombing. When presenting our findings, we will transparently acknowledge dataset limitations and potential biases, ensuring that our conclusions are framed appropriately. By prioritizing ethical data collection and clear communication of biases, we aim to provide an equitable and responsible analysis of superhero film trends.
