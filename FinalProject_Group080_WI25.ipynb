# Abstract

Climate change is a major problem in today's society, with CO2 emissions from ground transportation being one of the leading factors for climate change to occur. Public ground transportation aims to provide cheap and easy access to transportation, as well as pool in potential individual riders, effectively mitigating CO2 emissions. In this project, we will explore and find a relationship between public transit ridership and CO2 emissions by populous cities. We predict that cities with more public transit ridership will have less CO2 emissions than cities with less public transit ridership.  Using data from the Federal Transit Administration and the U.S. Census, we analyzed ridership and CO₂ emissions, both in total and on a per capita basis, across 22 metropolitan areas. Through gathering datasets on public transit ridership and CO2 emissions, we cleaned the data such that the cities specified match up with both datasets.  We created multiple graphs as points of visualization such as a scatterplot, heatmap, bubble map, and histogram. In the end, the result of the data analysis came out to be that there is a positive correlation between ridership of public transportation and CO2 emissions. With this in mind, correlation does not equal causation, and although there is a correlation clearly in sight, with a pretty strong strength, there are third variables to keep in mind. When adjusting for population, cities with high transit ridership per capita, such as New York and San Francisco, tend to have lower per capita CO₂ emissions, suggesting that well-integrated public transportation can help reduce individual carbon footprints. Despite this, high ridership does not always correspond with lower emissions, as factors like urban sprawl, car dependency, and transit system efficiency also play significant roles. Although public transportation may not be a direct causal factor for a decrease in CO2 emissions it is a part of a larger strategy that will help with the issues of climate change that our society alongside other urban areas deal with and suffer from.

The first dataset that we decided to use is named "Near-real-time daily estimates of CO2 emissions from 1500 cities worldwide." The dataset contains emissions from power generation, residential (buildings), industry, ground transportation, and aviation from January 2019 through December 2021 for 1500 cities in 46 countries.

3. Using the date variable, we only included data from 2019 since that year was not affected by the COVID-19 pandemic. We also added the date variable to go by the year, instead of daily.
4. We kept all the datapoints that are in the top 25 most populated cities in U.S and dropped the rest of the rows. This is because we are focused on looking at areas with the most people. To match the ridership dataset, we added the CO2 ground transportation emissions from cities that are grouped under one metro area.
5. Lastly, we dropped a few columns like country and timestamp as we felt they were irrelvant to our research question/hypothesis. Once we cleared the sector to only include Ground Transportation and the dates to be a sum of 2019, we dropped the sector and date column.

Purpose: Using this dataset will allow us to see if more ridership influences the overall CO2 emissions for cities in the United States.

# **Complete Monthly Ridership**

This dataset provides ridership in cities in the US. The ridership that this dataset tracks includes bus, rail, ferry, and other. Ridership is defined as someone using public transit one way. This means that one person can use ridership 3 times a day, and it will be accounted as 3 times. It has data on multiple years of ridership in multiple different cities, as well as giving data on which system of transportation, mode, agency, and more. Some key variables are the city name, mode, and ridership.

**Process of Wrangling and Cleaning the Data:**

1. **It is important to note that for our whole project, we had to base our focus regions based on this dataset. This dataset has a variable, "UZA Name," which stands for Urbanized Areas. These areas group together multiple cities that transportation agencies serve. For example, the cities New York City, Jersey City, and Newark are all served by MTA New York City Transit, so their ridership is reported under the UZA Name "New York--Jersey City--Newark, NY--NJ".**  
2. The overall wrangling process involved removing legacy IDs as we want to keep our IDs between datasets consistent, and the other dataset uses 'next generation IDs' which is the ID we chose to keep.
3. We dropped totals, years besides 2019, reporter type, and also the cities besides the 25 chosen (most populous) cities.

Los Angeles and New York are an outlier as they have significantly higher CO₂ ground transport emissions and high ridership while the other metro areas have lower values for both. # **Scatterplot Ridership vs CO2 per Capita**



---



When observing the results from the Ridership vs CO2 Emissions by Metro Area, we saw that the most populous cities, New York and Los Angeles, were the top CO2 emitters. To ensure that the results of the scatterplot was not just because of population, we created a new scatter plot that takes into consideration the population of the cities. To do this, we divided each ridership and CO₂ value by the population of the city and plotted those points.

```python
#Rename columns for easier access
top_pop_df = top_pop_df.rename(columns={'Mapped_City': 'metro_area'})
df_co2['metro_area'] = df_co2['metro_area'].str.strip().str.lower()
df_ridership['metro_area'] = df_ridership['metro_area'].str.strip().str.lower()
top_pop_df['metro_area'] = top_pop_df['metro_area'].str.strip().str.lower()
df_co2["metro_area"] = df_co2["metro_area"].replace({"oklahoma, ok": "oklahoma city, ok"})

#Merge the datasets (inner)
df_merged_capita = pd.merge(df_co2, df_ridership, on='metro_area', how='inner')
df_merged_capita = pd.merge(df_merged, top_pop_df, on='metro_area', how='inner')

#Convert ktCO2 to tons
df_merged_capita['ktco2_in_2019'] *= 1000

#Compute per capita values
df_merged_capita['ridership_per_capita'] = df_merged_capita['ridership'] / df_merged_capita['POPESTIMATE2019']
df_merged_capita['co2_per_capita'] = df_merged_capita['ktco2_in_2019'] / df_merged_capita['POPESTIMATE2019']

#Remove zeros before log transformation to avoid errors
df_merged_capita = df_merged_capita[(df_merged_capita['ridership_per_capita'] > 0) & (df_merged_capita['co2_per_capita'] > 0)]
plt.figure(figsize=(13, 8))

#Scatterplot
sns.scatterplot(data=df_merged_capita, x='ridership_per_capita', y='co2_per_capita', s=100, label="Data")

#Regression Line
sns.regplot(data=df_merged_capita, x='ridership_per_capita', y='co2_per_capita', scatter=False, color='red', line_kws={"linewidth": 2}, label="Regression Line", ci=None, logx = True)

#Format so that each point on the graph labels the city it is representing (again only first few letters)
for i, row in df_merged_capita.iterrows():
    label = row['metro_area'][:7].upper() + '.'
    plt.text(row['ridership_per_capita'], row['co2_per_capita'], label, fontsize=6.5, ha='right', va='bottom')

plt.xlabel('Transit Ridership per Capita (2019)',fontsize=10, fontweight='bold')
plt.ylabel('CO₂ Ground Transport Emissions per Capita (tons per person)',fontsize=10, fontweight='bold')
plt.title('Transit Ridership vs. CO₂ Emissions Ground Transport per Capita', fontsize=14, fontweight='bold', pad=10)
plt.xscale('log')
plt.legend()
plt.grid(True)
plt.show()
```

**Interpretation and Results**

Comparing the two scatter plots, we can see that there is a difference when we factor in population to the data. Compared to the Ridership vs CO₂ Emissions scatterplot, there is a weak positive correlation between Ridership vs CO₂ Emissions per Capita as the slope is less steep.

The Washington DC, Boston, and New York metropolitan areas are outliers in this graph.
The New York, Jersey, and Newark metro area demonstrates that very high transit ridership per capita can lead to lower per capita emissions which suggests that effective public transit systems can contribute to emission reductions. New York is one of the US cities that are known for their subways, so many individuals must use public transit which could correlate with individuals having low CO₂ ground transport emissions. Cities like Washington DC and Boston have high per capita emissions despite having one of the highest transit ridership which indicates that some other factors like urban sprawl and more car usage plays a role in emissions.

From both the scatterplots, we learn that CO₂ emissions do correlate with transit ridership, and this could be due to larger cities having more transportation overall.

# Ethics & Privacy

##Data Collection and Privacy
In these datasets, there is no direct personal information held available since they are mostly datasets on the topic of public transportation and CO2 emissions. There may still be the inclusion of sensitive information such as private city government information, transportation networks, and private data about people that may use the service regularly. As a team we are making sure that the data used from public and private sources are complying with the “Data Science Ethics Checklist" from Deon’s Ethics Checklist.

##Bias in the Data
There is an importance in the potential bias in data that we are collecting as used for our team’s research. With this in mind, there can be ridership data that are coming from different agencies from public transportation that are private companies. The definition of "usage" may differ based on the city and therefore possibly impact the comparisons we as researchers are able to make. When there is over representation of a certain demographic in certain cities, this may also become a bias as there are cities that are definitely more populated than others. There also may be areas in certain cities that have more of a public transportation culture such as college towns whereas other areas of the city are the opposite. The reason behind using public transportation may also vary and therefore may bias the results of cities that use more public transportation as maybe there are areas with the socioeconomic status of its residents that are lower than those that do not use public transportation as much. Keeping in mind these potential biases, it is vital that we as a team are able to be aware of these potential biases as we collect and analyze the data. If there is skewed data, it is important that we are tentative and check whether the findings are accurate and trustworthy. An example of this may be cities that have higher public transportation usage may also have a bigger demographic in terms of diversity and therefore more people that aren’t able to afford driving a car that takes up a lot of gas (which they have to pay for), and therefore affect the emissions results.

##Equitable Analysis and Representation
We must make sure that all datasets that are collected and analyzed are aligned with privacy standards. For example, when obtaining data from sources such as the Bureau of Transportation Statistics, or US Census data, we must ensure that the datasets do not violate the privacy of individuals, and are used according to their terms of use (found on the datasets). Any personal or sensitive information should not be shared or used for unethical purposes such as publicity or mishandling.

- Carefully evaluate sources of data
- Be transparent about the limitations and potential biases in analysis
- Clearly communicate uncertainties or assumptions made.
