# Abstract

Academic pressure is a common issue among students, and it often comes up as one factor that negatively impacts students' mental health and academic performance. This study explores the impact of academic course pressure, which is measured by course load and GPA expectations, and how it impacts a student's mental health, which is measured through self-reported anxiety levels (GAD-7), stress levels, and sleep schedules. We used multiple forms of data analysis. We used kernel data plotting and mean and standard deviation statistics to see the correlation between said variables. Then, we used a regression plot to see potential trends between the variables, where both posed no relation. Lastly, we used Ordinary Least Squares (OLS)  regression analysis to inspect the relationship between academic performance and previously listed mental health factors. Even though there is a wide belief that there is a strong correlation between academic course pressure and academic performance, our data failed to find statistically significant relationships. The only predictor near to being significant was the depression score ( p = 0.062), which could mean there is a weak negative association with GPA. Identical results were found when we looked at the relationship between the number of units someone takes and mental health, with the only score being close to significance being depression again ( p = 0.053). 

While our data concluded no correlation between academic course pressure and academic performance, it should not downplay the importance of mental health as essential for students' success. Furthermore, self-reported measures could be biased, which could be why we got the results we did. Future research should be done to test more metrics than we already did to understand better the complex relationship between academic performance and course pressure.

# Ordinary Least Squares (OLS) regression tests

To determine further if there is a significant relation between the scores and so on we will use OLS regression test for both GPA and Semester Unit Load to each mental health score. 

## GPA and Stress Level

```python
## OLS Regression between GPA and Stress_Level
y, X = patsy.dmatrices('GPA ~ Stress_Level',data = mental_health)
model = sm.OLS(y, X).fit()
print(model.summary())
```

```python
## OLS Regression between GPA and Depression Score
y, X = patsy.dmatrices('GPA ~ Depression_Score',data = mental_health)
model = sm.OLS(y, X).fit()
print(model.summary())
```

# Discussion and Conclusion

Our study examines the relationship between mental health factors and students' academic performance (GPA) and the potential association between semester unit load and mental health variables. Previous research has indicated that high levels of stress, anxiety, depression, and poor sleep quality can negatively impact academic performance (Pascoe et al., 2020; Richardson et al., 2012). However, the findings from this study did not support a statistically significant relationship between these mental health factors and GPA.

The kernel density estimation plot illustrates the distribution of semester unit load across different mental health scores. Since the lines overlap significantly, the semester unit load is similar across various mental health levels. This means that there isn’t a strong relationship between taking more courses and experiencing higher or lower mental health levels. The scatter matrix shows that the plots don’t show a strong correlation between the two variables since there’s a clear upward or downward trend. 
The OLS regression results showed that none of the predictors—stress level, depression score, anxiety score, sleep quality, financial stress, or semester unit load—significantly affected GPA. The only near-significant predictor was the depression score (p = 0.062), which suggested a weak negative association, though not at a level that allows for firm conclusions. Similarly, when analyzing the semester unit load about mental health, no significant relationships were found, with the depression score again being the closest to significance (p = 0.053).

These findings go against a lot of existing research that suggests mental health plays a significant role in academic success. One reason for this could be that GPA alone doesn’t fully capture the impact of mental health struggles straightforwardly. Academic performance is shaped by a mix of factors—time management, study habits, family support, and access to campus resources—all of which can influence grades just as much, if not more, than stress or anxiety. On top of that, some students might develop coping strategies, like seeking academic accommodations or adjusting their study routines, that help them manage the effects of stress without significantly affecting their GPA.

Another factor to consider is the reliability of self-reported data, which can be subjective and prone to bias. Students may not always have an accurate sense of their stress, anxiety, or depression levels, or they might downplay them. Plus, GPA is a broad measure of academic performance—it doesn’t necessarily reflect struggles with focus, motivation, or long-term material retention, all areas where mental health could have a more significant impact. This could be why our findings showed no statistically significant correlation between stress and academic pressure. 

Even though this study didn’t find a clear statistical link between mental health and GPA, that doesn’t mean mental health isn’t a crucial part of student success. Colleges and universities should still prioritize mental health resources because student well-being affects far more than grades—it influences overall academic and personal growth. Future research should take a more in-depth approach, looking at how different stressors interact, using qualitative insights from students, and tracking mental health and academic performance over time.
