## Linear Regression Analysis

```python
## YOUR CODE HERE
## FEEL FREE TO ADD MULTIPLE CELLS PER SECTION
diesel_emissions = df[df['Fuel Type'] == 'Diesel']['PM2.5 Emissions']
petrol_emissions = df[df['Fuel Type'] == 'Petrol']['PM2.5 Emissions']

t_stat, p_val = stats.ttest_ind(diesel_emissions, petrol_emissions, equal_var=False)
print("T-statistic:", t_stat)
print("p-value:", p_val)

if p_val < 0.05:
    print("There is a statistically significant difference between Diesel and Petrol PM2.5 emissions.")
else:
    print("No statistically significant difference was found between Diesel and Petrol PM2.5 emissions.")
```

The Welch’s t-test shows that the difference in mean PM2.5 emissions between diesel and petrol vehicles is not statistically significant (p ≥ 0.05). In other words, based on this dataset, we cannot conclude that one fuel type consistently emits more PM2.5 than the other under low wind speed conditions.

```python
model_fuel = smf.ols("Q('PM2.5 Emissions') ~ C(Q('Fuel Type'))", data=df).fit()
print("\nSimple OLS with Fuel Type only:")
print(model_fuel.summary())
```

This simple regression model indicates that Fuel Type (Diesel vs. Petrol) explains only about 0.2% of the variance in PM2.5 emissions (R² ≈ 0.002). The p-value for the Petrol coefficient (≈ 0.157) suggests there is no statistically significant difference in mean PM2.5 emissions between Petrol and Diesel vehicles at the 5% significance level. Thus, when examined in isolation, Fuel Type alone does not appear to be a strong predictor of PM2.5 emissions under these conditions.

```python
model_vehicle = smf.ols("Q('PM2.5 Emissions') ~ C(Q('Vehicle Type'))", data=df).fit()
print("\nSimple OLS with Vehicle Type only:")
print(model_vehicle.summary())
```

This simple OLS model shows that Vehicle Type alone explains less than 1% of the variance in PM2.5 emissions (R² ≈ 0.008). While motorcycles exhibit a small positive coefficient (p ≈ 0.036), the overall F-statistic (p ≈ 0.0681) suggests no strong evidence that vehicle type alone significantly differentiates mean PM2.5 emissions under these conditions.

```python
model_engine = smf.ols("Q('PM2.5 Emissions') ~ Q('Engine Size')", data=df).fit()
print("\nSimple OLS with Engine Size only:")
print(model_engine.summary())
```

This simple OLS model indicates that Engine Size alone explains essentially none of the variation in PM2.5 emissions (R² ≈ 0.000). With a non-significant p-value (≈ 0.525), there is no strong evidence that engine size by itself is a major driver of PM2.5 emissions under these conditions.

```python
model_traffic = smf.ols("Q('PM2.5 Emissions') ~ C(Q('Traffic Conditions'))", data=df).fit()
print("\nSimple OLS with Traffic Conditions only:")
print(model_traffic.summary())
```

This simple OLS model shows that Traffic Conditions (Heavy, Moderate, or Free flow) explain virtually none of the variation in PM2.5 emissions (R² ≈ 0.000). The high p-value (≈ 0.914) indicates no statistically significant differences in mean PM2.5 emissions across the three traffic condition categories in this dataset.

```python
model_wind = smf.ols("Q('PM2.5 Emissions') ~ Q('Wind Speed')", data=df).fit()
print("\nSimple OLS with Wind Speed only:")
print(model_wind.summary())
```

This simple OLS model suggests that Wind Speed alone explains only about 0.5% of the variation in PM2.5 emissions (R² ≈ 0.005). Although the p-value (≈ 0.0402) indicates a marginally significant relationship, the effect size appears quite small, implying that wind speed alone is not a major predictor of PM2.5 emissions in this dataset.

```python
model = smf.ols(
    "Q('PM2.5 Emissions') ~ C(Q('Fuel Type')) + Q('Engine Size') + C(Q('Vehicle Type')) + C(Q('Traffic Conditions')) + Q('Wind Speed')",
    data=df
)

results = model.fit()
print(results.summary())
```

This multiple regression model indicates that Fuel Type, Engine Size, Vehicle Type, Traffic Conditions, and Wind Speed together explain only about 1.5% of the variance in PM2.5 emissions (R² ≈ 0.015). None of these predictors are statistically significant at the 5% level, suggesting that under these conditions, no single factor strongly drives PM2.5 emissions or that other, unmeasured variables may have a larger influence.

```python
model_interact = smf.ols(
    "Q('PM2.5 Emissions') ~ C(Q('Fuel Type'))*Q('Engine Size') + C(Q('Vehicle Type')) + C(Q('Traffic Conditions')) + Q('Wind Speed')",
    data=df
).fit()
print("\nMultiple Regression with Fuel Type × Engine Size interaction:")
print(model_interact.summary())
```

In this model, we introduced an interaction term between Fuel Type and Engine Size to see if the effect of Engine Size on PM2.5 emissions differs by fuel type. The R² remains around 1.5%, indicating a very low explanatory power. None of the interaction or main effects are statistically significant at the 5% level, suggesting that neither Fuel Type nor Engine Size—nor their combination—robustly explains PM2.5 emission variability under these conditions.

# Permutation analysis
Based on the graphs we created in the EDA checkpoint, we can see that the amount of PM2.5 emissions from petrol and diesel vehicles is not significantly different. The mean PM2.5 emissions of diesel vehicles is slightly below 0.100, while the mean for petrol vehicles is above 0.110. Even though the difference is not too large, we can see that petrol vehicles contribute more to emissions problems than diesel vehicles.

```python
## YOUR CODE HERE
diesel_pm25 = df_heavy[df_heavy["Fuel Type"] == "Diesel"]["PM2.5 Emissions"].values
petrol_pm25 = df_heavy[df_heavy["Fuel Type"] == "Petrol"]["PM2.5 Emissions"].values

print("Number of Diesel samples:", len(diesel_pm25))
print("Number of Petrol samples:", len(petrol_pm25))
```

```python
observed_diff = diesel_pm25.mean() - petrol_pm25.mean()

pooled_data = np.concatenate((diesel_pm25, petrol_pm25))

n_permutations = 10_000

count_extreme = 0

for _ in range(n_permutations):
    permuted = np.random.permutation(pooled_data)

    perm_diesel = permuted[:len(diesel_pm25)]
    perm_petrol = permuted[len(diesel_pm25):]

    diff_perm = perm_diesel.mean() - perm_petrol.mean()

    if abs(diff_perm) >= abs(observed_diff):
        count_extreme += 1
p_value_perm = count_extreme / n_permutations

print(f"Observed difference in means (Diesel - Petrol): {observed_diff:.4f}")
print(f"Permutation p-value: {p_value_perm:.4f}")
```

The p-value of 0.3230 in the above cell shows that there is no statistically significant difference between PM2.5 emissions of diesel and petrol vehicles based on my permutation test.

This is different from our hypothesis, which thinks that diesel vehicles will statistically significantly generate more emissions than petrol vehicles. However, from our analysis above, the difference is not too significant. So under this dataset, my analysis does not confirm the hypothesis.
