# Abstract

Our project examined the potential relationships between the federal availability of four food categories (dairy, fruit, vegetables, and red meat) and rates of 22 high-incidence cancers. The cancer dataset was initially organized by state, so we had to clean it to be organized by year and specific cancer. Our analysis used food availability as a substitute metric for average food consumption in the US from 1999-2021 to determine whether the consumption of certain nutrients is correlated with the incidence of specific cancers. To determine the nature of the relationship, we found the five least and five most correlated cancers to food group availability, and we calculated Spearman’s correlation coefficient for each. The five least correlated cancers to food groups were ovarian, prostate, larynx, cervical, and lung. This suggests that these cancers are likely driven by other factors outside of diet such as genetics, environmental characteristics, or lifestyle.


On the other hand, kidney, uterine, melanoma, oral, and pancreatic cancers revealed stronger relationships with the food groups overall. Red meat usually undergoes heavy processing and high-temperature cooking that could result in carcinogenic effects consistent with the above cancers. Conversely, fruits and vegetables have negative correlations to cancer rates that could be attributed to their antioxidant and fibrous nature preventing cell damage and supporting digestive health. Overall, some cancers exhibit weak relationships to cancer while others appear more strongly associated, highlighting the potential for differences in nutrients being the root of cancer.

```python
## YOUR CODE TO LOAD/CLEAN/TIDY/WRANGLE THE DATA GOES HERE
## FEEL FREE TO ADD MULTIPLE CELLS PER SECTION 

# Clean Cancer Data
cancer = pd.read_csv('cancerData_USA.csv')
cancer = cancer.drop(labels= ['Notes', 'States Code', 'Year Code', 'Leading Cancer Sites Code'], axis= 1)
cancer = cancer.rename(columns= {"Leading Cancer Sites" : "Sites"})

# Convert Count column to floats
def convert(value):
    try:
        return float(value)
    except:
        return np.nan
    
cancer['Count'] = cancer['Count'].apply(convert)

#Filter out rows that contain no information
cancer = cancer[cancer.isna().all(axis=1) == False]
```

```python
# Map site names to short names
site_names = ['Brain and Other Nervous System', 'Breast', 'Cervix Uteri',
 'Colon and Rectum', 'Corpus Uteri', 'Esophagus', 'Gallbladder',
 'Kidney and Renal Pelvis', 'Larynx', 'Leukemias', 'Liver',
 'Lung and Bronchus' ,'Melanoma of the Skin' ,'Myeloma',
 'Non-Hodgkin Lymphoma', 'Oral Cavity and Pharynx', 'Ovary', 'Pancreas',
 'Prostate', 'Stomach', 'Thyroid', 'Urinary Bladder, invasive and in situ']

site_short = ['Brain', 'Breast', 'Cervical', 
 'Colon', 'Uterine', 'Esophagus', 'Gallbladder', 
 'Kidney', 'Larynx', 'Leukemias', 'Liver', 'Lung', 
 'Melanoma', 'Myeloma', 'Lymphoma', 'Oral', 
 'Ovary', 'Pancreas', 'Prostate', 'Stomach', 'Thyroid', 'Urinary']

site_dict = dict(zip(site_names + site_short, site_short + site_names))
```

```python
# Convert site names in data frame
cancer["Sites"] = [site_dict[site] for site in cancer["Sites"]]
```

In order to combine these two datasets, we will merge them using an inner join on the "Year" column for both dataframes. We only want to keep the data that is in the years that overlap the two datasets, hence why we are using an inner join. This will then allow us to perform EDA and other analyses on our data.

```python
# Create dictionary for missing information
missing_info = cancer[cancer.isna().any(axis=1) == True]

def missing(info):
    dict = {}
    for year in info["Year"].unique():
        dict[year] = {}
        year_info = missing_info[info["Year"] == year]
        for state in year_info["States"].unique():
            missing_sites = year_info[year_info["States"] == state]["Sites"]
            s = [site for site in missing_sites]
            dict[year][state] = s
    return dict

missing_dict = missing(missing_info)
```

# Missing_data:

```python
# Print/explain missing results
def what_missing(dict):
    what_missing = ""
    for year in dict:
        pre_str = f"{year}:\n"
        for state in dict[year]:
            missing_num = ''
            if len(dict[year][state]) == 22:
                missing_num = "all"
            pre_str += f"{state} is missing data for {missing_num} cancers\n"
        what_missing += pre_str + "\n"
    print(what_missing)

what_missing(missing_dict)
```

```python
# Graph the cumulative site data
def cancer_plots():
    fig, axes = plt.subplots(8, 3, figsize=(15, 25))
    axes = axes.flatten()
    label_size = 9
    i = 0
    for site in sites:
        site_data = site_df[site_df["Sites"] == site]
        site_data["Count"] = site_df["Count"] / 1000
        sns.regplot(data= site_data, x="Year", y="Count", ax=axes[i])
        axes[i].set_title(f"{site_dict[site]}", fontsize=label_size)
        axes[i].set_xlabel("Year", fontsize=label_size)
        axes[i].set_ylabel("Count (thousands)", fontsize=label_size)
        axes[i].tick_params(axis="both", labelsize=label_size)
        i += 1
    fig.delaxes(axes[-1])
    fig.delaxes(axes[-2])
    plt.tight_layout()

cancer_plots()
```

## Regression Data
We clearly see that for many of the 22 most prevalent cancers, incidence rate follows a positive linear progression. So, we analyze these regressions:

```python
# Create an empty dataframe with the necessary indices and column names to store regression information
cancer_regression_stats = pd.DataFrame(index= sites, columns=['intercept', 'slope', 'R-squared', 'p-value'])

# Define a function to systematically input data
def compare_site_data():
    for site in sites:
        site_data = site_df[site_df["Sites"] == site]
        site_data = site_data.assign(Year= site_data["Year"] - 1999)    # Subtract 1999 to lower arbitrarily high intercept values
        
        # Create response (y) and predictor (x) matrices for linear regression --> Fit OLS regression model
        y, x = patsy.dmatrices("Count ~ Year", data=site_data, return_type="dataframe")
        model = sm.OLS(y, x).fit()

        cancer_regression_stats.loc[site] = [model.params[0], model.params[1], model.rsquared, model.pvalues[1]]

compare_site_data()
```

```python
#Create dataframe for spaerman correlations: food vs. cancer

# Create dataframe for spearman correlations
spearman_corr = cancer_food.corr(method= 'spearman').iloc[:-4][food_df.columns].drop(labels= "Year", axis= 1)
spearman_corr["Absolute Correlation"] = pd.Series(dtype= 'float64')

# Add column for absolute correlation
for row in spearman_corr.index:
    spearman_corr["Absolute Correlation"].loc[row] = spearman_corr.loc[row].abs().sum()/4

# Find cancers most/least correlated to food availability
least_corr = spearman_corr.sort_values(by= "Absolute Correlation").index[:5]
most_corr = spearman_corr.sort_values(by= "Absolute Correlation").index[-6:-1]
```

```python
# Create two dataframes for most and least correlated cancers
least_corr_df = cancer_food[list(food_df.columns) + list(least_corr)]
most_corr_df = cancer_food[list(food_df.columns) + list(most_corr)]
least_corr_df["LC Total Count"] = least_corr_df[least_corr].sum(axis=1)
most_corr_df["MC Total Count"] = most_corr_df[most_corr].sum(axis=1)
```

The data frame ‘LM_spearman_df  stores correlation values for different cancers and types of food categorizing them as most or least correlated. T-values are then calculated so we can assess the strength of the relationships. Then using statistical tests, they are converted into p values that help find out whether observed correlations are statistically significant. Lower values will hint at stronger evidence for a significant relationship between food consumption and cancer frequency. 


```python
# T-values for all the correlations
LM_tvalues_df = LM_spearman_df.set_index("Cancer")[['Dairy Pounds', 'Vegetables Pounds', 'Fruit Pounds', 'Red Meat Pounds', 'Absolute Correlation']].apply(
    lambda r: r * np.sqrt(22 - 2) / np.sqrt(1 - r**2))

LM_tvalues_df
```

```python
# converting the t-values to p-values
import scipy.stats as stats

p_val_df = LM_tvalues_df.apply(lambda x: stats.t.sf(np.abs(x), 21)*2)
p_val_df
```

```python
# Function for printing which p-values are significant
def significant_pval(data, group):
    output = f"Significant P-values for {group}\n\n"
    threshold = 0.05
    sig_data = data.apply(lambda p: p <= threshold)
    for row in data.index:
        for col in data.columns:
            if sig_data.loc[row, col] == True:
                output += f"{row} vs. {col}:\t\t P-value: {data.loc[row, col]:10f}\n"
    return output
```
