This study investigates the relationship between video game summary text and user review ratings using Natural Language Processing (NLP) and machine learning techniques. By analyzing a dataset of 18,800 video game reviews, we aim to identify the words and features in game summaries that are most predictive of high or low ratings. The project first involved preprocessing the text data by removing punctuation, converting words to lowercase, and performing tokenization. The Term Frequency-Inverse Document Frequency (TF-IDF) approach was then applied to extract important terms from the summaries, and a linear regression model was built to assess how these terms relate to user ratings.

The regression model achieved a reasonable R² score, indicating a moderately successful relationship between summary text and user reviews. This research demonstrates that features of a game’s promotional summary, such as specific keywords and language choices, can offer valuable insights for predicting user satisfaction. These results could be useful for game developers and marketers seeking to optimize game descriptions to increase player engagement and ratings.

### Text Analysis of Summary Data

Now we want to use NLP and machine learning techniques to find which words are most indicative of a highly-rated game. We will do this by extracting word tokens for each summary, and calculating the average rating for each summary, and then creating a linear regression model to relate a summary's word tokens to the average rating.

First, let's group by each summary and average their user ratings, then clean up the text by removing punctuation, evenly spacing words, and making every word lowercase.

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
## YOUR CODE HERE
summaries = videogames[['summary','user_review']].groupby('summary').mean().reset_index()
summaries['summary'] = summaries['summary'].apply(lambda s: ' '.join(re.sub(r'[^\w\s]', '', s).split()).lower())
summaries
```

Next, we'll calculate the TF-IDF for words across all our game summaries.

```python
tfidf = TfidfVectorizer(
    sublinear_tf=True,
    analyzer='word',
    max_features=1000,
    tokenizer=word_tokenize,
    stop_words=stopwords.words("english")
)
```

```python
summaries_tfidf = pd.DataFrame(tfidf.fit_transform(summaries['summary']).toarray(), 
                             columns=tfidf.get_feature_names_out())
summaries_tfidf['mean_user_review'] = summaries['user_review']
```

```python
summaries_tfidf
```

Great! Now, just to get a sense of our data, why don't we find out which is our most unique word in each of our game summaries?

```python
summaries_tfidf.drop(columns=['mean_user_review']).idxmax(axis=1)
```

Now let's create a simple linear regression model to relate our TF-IDF data to user reviews! (The code for the following model and the subsequent visualizations were made with the assistance of ChatGPT.)

```python
# Step 1: Split the data into features (X) and target (y)
X = summaries_tfidf.drop(columns=['mean_user_review'])
y = summaries_tfidf['mean_user_review']

# Step 2: Split data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Step 3: Create and train the linear regression model
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)

# Step 4: Extract and sort the coefficients to see which words are most associated with high/low reviews
coefficients = pd.Series(lr_model.coef_, index=X.columns)
top_positive = coefficients.sort_values(ascending=False).head(10)
top_negative = coefficients.sort_values(ascending=True).head(10)

print("Top words associated with higher reviews:")
print(top_positive)
print("\nTop words associated with lower reviews:")
print(top_negative)

# Step 5: Evaluate the model on the test set
y_pred = lr_model.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("\nModel Evaluation")
print(f"Mean Squared Error: {mse}")
print(f"R^2 Score: {r2}")
```

Looks like our linear regression model's performance is a bit poor. Our MSE is roughly 1.8, which is significant considering that reviews only range from 0 to 10. Moreover, our $r^2$ score is very close to 0, and slightly negative. This suggests that our model does a poor job of explaining variance in our user review data, and performs slightly worse than simply guessing the mean, so video game summary text might not be predictive of our user review data.

Let's look at the distribution of our coefficients for each of our tokens, and observe the coefficients for the top tokens influencing user revoews.

```python
plt.figure(figsize=(8, 6))
sns.histplot(coefficients, bins=30, kde=True)
plt.xlabel('Coefficient Value')
plt.title('Distribution of TF-IDF Coefficients')
plt.show()
```

```python
top_positive = coefficients.sort_values(ascending=False).head(10)
top_negative = coefficients.sort_values(ascending=True).head(10)
top_coef = pd.concat([top_negative, top_positive]).sort_values()

plt.figure(figsize=(10, 8))
colors = ['red' if c < 0 else 'green' for c in top_coef]
top_coef.plot(kind='barh', color=colors)
plt.xlabel('Coefficient Value')
plt.title('Top Tokens Influencing User Reviews')
plt.show()
```

Our token coefficients seem evenly distributed. Notably among some of the top tokens are words like 'link', 'sega', and 'mario', which are all associated with established, sucessful, famous franchises (although the former is ambiguous, since it could double as the word in addition to the protagonist from the Legend of Zelda games). The most negative coefficients, on the other hand, seem fairly abstract and difficult to derive interesting observations from. 

Finally, let's look at how our model's predictions compare with the actual user review data.

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x=y_test, y=lr_model.predict(X_test))
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r', label='Perfect Prediction')
plt.xlabel('Actual User Reviews')
plt.ylabel('Predicted User Reviews')
plt.title('Predicted vs. Actual User Reviews')
plt.legend()
plt.show()
```

Not very encouraging; it looks like our linear model simply guessed roughly average user reviews for nearly all of our tokenized summary data. As we can see, there isn't a clear crrelation between 'Actual User Reviews' and 'Predicted User Reviews'. It seems like video game summary text may not be very predictive of user reviews.

While this dataset primarily contains game-related attributes and aggregated user ratings, we have still identified several important ethical and privacy considerations:

1. <b>Personal Data Exclusion</b>

- The dataset we have does not include any personally identifiable information (PII) about individual reviewers. This greatly reduces privacy concerns related to doxxing or unintended disclosure of sensitive data.
- Because it’s aggregated at the game level, there is no direct link to any single user’s identity or demographic profile.

2. <b>Review Manipulation & Bias</b>

- User review scores, even when anonymized or aggregated, can be influenced by “review bombing” (when a large group coordinates to downvote or upvote a game for reasons beyond its content). If such events occurred, it could skew the data and yield misleading conclusions about a game’s true reception.
- Similarly, some publishers might encourage positive reviews or distribute free copies in exchange for higher ratings, introducing potential bias in user scores. We acknowledge that such bias may affect our interpretation of the results.

3. <b>Terms of Use & Platform Policies</b>

- It is critical to ensure that using the dataset’s user review information (if it originally comes from a review aggregator) aligns with that platform’s terms of service. Since we only have aggregated/summary data in our file, we are likely within fair use. However, we will still be attentive to any constraints on how these data can be redistributed or analyzed.

4. <b>Limitations on Data Interpretation</b>

- Without demographic breakdowns, we cannot make definitive statements about how different types of players (e.g., based on age, location, or gaming experience) feel about a particular title.
- Since the data does not include developer/publisher info or marketing budgets, concluding that one platform or release date “causes” higher/lower ratings could be overly simplistic. We must ethically represent our findings as correlational, not strictly causal.

5. <b>Communication of Findings</b>

- In reporting our results, we will be careful to avoid overgeneralizing about a particular platform, release date, or game summary style. We will emphasize that these patterns represent associations in our dataset, which may or may not hold universally across all gaming communities.
- If we detect possible anomalies (e.g., suspiciously high or low ratings for certain games), we will note that these outliers might be tied to external factors not captured in the data, preventing unfair mischaracterizations of game quality or fan reception.

By acknowledging these considerations and transparently disclosing them in our final analysis, we aim to handle the data ethically, respect user privacy, and communicate our findings responsibly.

# Discusison and Conclusion

Our analyses offer several insights into how user review ratings vary by platform, release year, and the text-based summaries of video games. First, we observed that older titles (particularly those released before 2005) tend to have higher average user ratings than more recent releases. One plausible explanation is that many classic games in the late 1990s are iconic or nostalgic favorites, leading to consistently enthusiastic reviews, whereas modern games face ever-evolving player expectations and a much broader market where not every title rises to the same critical acclaim. Moreover, the declining linear trend in ratings over the years—albeit modest—reinforces the idea that as the video game industry expanded, it inevitably included more varied or average-quality titles, resulting in lower aggregated ratings. These findings suggest that developers and publishers could benefit from understanding how nostalgia, platform-specific user bases, and evolving industry trends influence reception. Modern studios may need to focus not only on innovation but also on retaining elements that evoke the emotional connections seen in classic games.

Regarding platform differences, our findings show that systems with smaller, curated libraries or a strong nostalgia factor (like the Nintendo 64 and Dreamcast) hold some of the highest mean user ratings. By contrast, newer and more expansive platforms (PlayStation 4, Xbox One, and even PC) have so many games—ranging from top-tier AAA blockbusters to niche indie projects—that their overall averages skew lower. This platform contrast aligns with prior studies that note how user bases, hardware capabilities, and sheer volume of titles can impact overall reception. Modern platforms face unique challenges that older systems did not, such as live-service models, microtransactions, and post-launch updates, all of which can significantly influence user perception. A game that initially launches with issues may receive lower ratings early on, even if later patches improve it. This is particularly evident on platforms like PC and PlayStation 4, where frequent updates and downloadable content (DLC) can alter the user experience over time. In contrast, older platforms primarily featured complete, standalone games upon release, meaning their ratings are often reflective of a finished product.

Interestingly, when we examined the text summaries for these games, we found no strong correlation between summary length and average user rating (a correlation coefficient close to zero). This suggests that a short, punchy game description can be just as compelling to potential players as a longer, more detailed summary. In other words, textual length alone doesn’t capture the essence of quality or user satisfaction. Similarly, after transforming our summary data using TF-IDF (Term Frequency-Inverse Document Frequency) for our linear regression model, we found that our model had overall low predictive power. This suggests that simple keyword-based models may not capture the complexity of user satisfaction. If textual features are to help predict or explain user ratings, more nuanced NLP methods for keyword extraction or even techniques like sentiment analysis would likely be needed.

In closing, our research highlights how era, platform, and textual features interweave to shape user perceptions. We confirm that older, nostalgia-rich releases often command higher ratings, while modern, high-volume platforms exhibit greater variability and somewhat lower averages. Future directions could expand on our findings by incorporating more robust NLP methods to analyze game descriptions, investigating genre-specific dynamics, or factoring in external influences (like marketing budgets and critic reviews). Nonetheless, our work underscores that while some broad patterns in user ratings are clear—particularly the distinctions across time and platform—these patterns emerge from a complex mix of game design choices, marketing strategies, and evolving player expectations.
