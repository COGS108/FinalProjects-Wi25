## Background and Prior Work

Public transportation plays a key role in New York City (NYC), providing a reliable, affordable, and efficient means of commuting. The city's extensive transit network, which includes subways, buses, and commuter rail services, helps reduce reliance on personal vehicles, alleviating road and traffic congestion. However, the use of public transportation is not uniform across all boroughs of NYC. Factors such as income levels and population density influence the usage of public transportation and the number of motor vehicle collisions in different zip codes. Neighborhoods with greater access to public transportation may see a reduced reliance on personal vehicles, while areas with limited transit options may experience higher traffic volumes and congestion. Understanding these relationships is crucial for improving transportation planning, enhancing safety measures, and ensuring equitable access to mobility options for all New Yorkers.

<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)A report from the Office of the New York State Comptroller (OSC) highlights that nearly 48% of NYC households earning less than $51,000 rely on public transportation, compared to only 38% of higher-income households. This disparity reflects the role of public transit in providing essential mobility for lower-income communities. However, despite the availability of transit, many low-income residents still experience financial strain due to fare costs. <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)Programs such as Fair Fares NYC, which provides discounted MetroCards to qualified residents, aim to make public transportation more affordable. <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)As of 2024, over 348,000 New Yorkers are enrolled in this program, highlighting the widespread need for discounted fares. Ensuring that transit remains accessible and cost-effective is essential, especially for communities that depend on public transit as their main mode of transportation.

<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2)Over the past decade, transportation costs in NYC have risen by 56%, driven by inflation and changing commuting habits during the COVID-19 pandemic, as more people preferred to commute using personal vehicles. Car reliance varies significantly by borough; in Manhattan households, only 21% owned at least one vehicle, whereas 82.6% of households in Staten Island and 61.4% of households in Queens owned a vehicle. <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3)This disparity reflects the differences in public transportation availability, as Staten Island has limited transit options compared to Manhattan's extensive subway network. These differences in personal transportation access may contribute to variations in motor vehicle collisions throughout NYC, as areas with higher car dependency are likely to experience increased traffic and accident rates, while areas with more public transportation access may see a reduced reliance on personal vehicles, correlating to fewer motor vehicle collisions.

Additionally, neighborhoods with limited public transit options face greater challenges related to traffic congestion, potentially leading to longer commute times and higher transportation expenses. <a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4)Limited transit access forces residents to rely on personal vehicles, resulting in higher costs for fuel, maintenance, insurance, and tolls. Addressing these disparities by improving transit infrastructure in underserved areas could help reduce car dependency and enhance overall road safety.

Understanding these factors is essential for designing more efficient and accessible transportation networks that help reduce motor vehicle collisions. This study aims to analyze how income levels and population density across NYC zip codes relate to public transportation ridership and motor vehicle collisions from September to November 2023. By examining data on transit ridership, income, population density, and motor vehicle collisions, this research will identify trends that can reduce traffic-related incidents and improve accessibility across NYC.

```python
zip_shapes = gpd.read_file('zip_code_shapefile/geo_export_89c9cb6b-92fe-4a07-b7a0-144ff0ca2a26.shp')
```

```python
# create a list of days in our three month period that will be used to calculate ridership data
days_list = []

for month in range(9, 12):
    for day in range(1, 26, 4):
        days_list.append((month, day, day+3))

    if (month==10):
        days_list.append((month, 29, 31))
    else:
        days_list.append((month, 29, 30))
```

```python
# takes a station id as input, outputs filtered dataframe with only observations for a spectific station
def get_data_for_station(station_id):

    all_data_list = []

    for day in days_list:

        m,d1,d2 = day

        start_time = f'2023-{m}-{d1}'
        end_time = f'2023-{m}-{d2}'
        url_base = f"https://data.ny.gov/resource/wujg-7c2s.json?$query=SELECT%0A%20%20%60transit_timestamp%60%2C%0A%20%20%60station_complex_id%60%2C%0A%20%20%60borough%60%2C%0A%20%20%60payment_method%60%2C%0A%20%20%60fare_class_category%60%2C%0A%20%20%60ridership%60%2C%0A%20%20%60longitude%60%2C%0A%20%20%60latitude%60%0AWHERE%0A%20%20caseless_eq(%60station_complex_id%60%2C%20%22{station_id}%22)%0A%20%20AND%20(%60transit_timestamp%60%0A%20%20%20%20%20%20%20%20%20BETWEEN%20%22{start_time}T00%3A00%3A00%22%20%3A%3A%20floating_timestamp%0A%20%20%20%20%20%20%20%20%20AND%20%22{end_time}T23%3A59%3A59%22%20%3A%3A%20floating_timestamp)"
        try:
            response = requests.get(url_base)
        except Exception as e:
            error_code = type(e).__name__
            error_message = str(e)
            print(f"Error Code: {error_code}")
            print(f"Error Message: {error_message}")
            break

        data = response.json()
        if not data:
            break

        all_data_list.extend(data)  # append observations

    station_days_df = pd.DataFrame(all_data_list)
    print(f"Fetched {len(all_data_list)} total records for station {station_id}.")
    return station_days_df
```

```python
# read in a list of stations in nyc
stations_df = pd.read_csv('subway_folder/stations_list.csv')
```

```python
# removes the time from the timestamp so that we can work the year, month, day
def truncate_timestamp(timestamp):
    return timestamp.split('T')[0]

# turns our ridership column into a float from an object so that it is workable
def ridership_int(ridership):
    return float(ridership)
```

```python
# clean up a resulting dataframe to include only the info we need and sum up all the ridership values per day
def clean_df(input_df):
    df_modified = input_df
    store_input = input_df
    df_modified['transit_timestamp'] = df_modified['transit_timestamp'].apply(truncate_timestamp)
    df_modified['ridership'] = df_modified['ridership'].apply(ridership_int)
    df_modified = df_modified.drop_duplicates(subset=['transit_timestamp']).sort_values(['transit_timestamp']).reset_index()
    store_input = input_df.loc[:, ['transit_timestamp', 'ridership']]
    df_modified['ridership'] = store_input.groupby(['transit_timestamp']).sum().sort_values(['transit_timestamp']).reset_index()['ridership']
    return df_modified.get(['transit_timestamp', 'station_complex_id', 'borough', 'ridership', 'longitude', 'latitude'])
```

## Merge all Datasets

```python
# zip code data set that we needed to use to create a base dataframe of matching zip codes
ny_modzcta = pd.read_csv('https://data.cityofnewyork.us/resource/pri4-ifjk.csv')
```

```python
ny_modzcta = ny_modzcta.get(['modzcta'])
ny_modzcta.head()
```

```python
# read in all our cleaned datasets
zip_collisions = pd.read_csv('collisions_folder/zip_collisions.csv')
zip_income = pd.read_csv('income_folder/zipcode_income.csv')
zip_population = pd.read_csv('population_folder/zipcode_population.csv')
zip_subway = pd.read_csv('subway_folder/subway_ridership_agg.csv')
```

```python
# rid columns not needed for merge
zip_collisions = zip_collisions.drop(['Unnamed: 0', 'date', 'time', 'borough', 'latitude', 'longitude'], axis=1)
zip_collisions.head()
```

```python
# get total number of persons killed, injured, and number of collisions per zip code
total_collisions = zip_collisions.groupby('zip_code').count().drop(columns=['num_persons_injured']).rename(columns={'num_persons_killed': 'total_collisions'})
collisions_per_zip = zip_collisions.groupby('zip_code').sum().drop([99999], axis=0).rename(columns={'num_persons_killed': 'total_persons_killed', 'num_persons_injured': 'total_persons_injured'})
collisions_per_zip = collisions_per_zip.merge(total_collisions, left_index=True, right_index=True)
collisions_per_zip.head()
```

```python
# clean up the income data set a bit more to only inclue what we need in merged
zip_income = zip_income.drop(zip_income.columns[1:12], axis=1)
zip_income = zip_income.drop(zip_income.columns[3:], axis=1)
zip_income.head()
```

```python
# clean up population data a bit more to only include info needed for merge
zip_population = zip_population.drop(['Unnamed: 0', 'borough'], axis=1)
zip_population.head()
```

```python
# get average number of stops per zipcode
num_stops = zip_subway.groupby('zip_code').count().get(['borough']).rename(columns={'borough':'num_stops'})

# get average ridership per zipcode
zip_subway = zip_subway.get(['ridership', 'zip_code'])
zip_subway = zip_subway.groupby('zip_code').mean().rename(columns={'ridership': 'avg_ridership'})
zip_subway['avg_ridership'] = zip_subway['avg_ridership'].apply(int)
zip_subway = zip_subway.merge(num_stops, left_index=True, right_index=True)

zip_subway.head()
```

```python
# merge income data set (which had the least amount of zip codes) with our modzcta list to get zip codes only in nyc
base_merge = zip_income.merge(ny_modzcta, left_on='ZCTA', right_on='modzcta').drop(['modzcta'], axis=1)
base_merge
```

```python
# merge the base df with collisions df
base_collisions_merge = base_merge.merge(collisions_per_zip, left_on='ZCTA', right_index=True, how='left')
base_collisions_merge
```

```python
# merge new base with subway ridership
base_subway_merge = base_collisions_merge.merge(zip_subway, left_on='ZCTA', right_index=True, how='left')
base_subway_merge
```

```python
# merge new base with population to get our final merged dataset of information per zip code
data_per_zip = base_subway_merge.merge(zip_population, left_on='ZCTA', right_on='zip', how='left').drop(columns=['zip'])
data_per_zip
```

Since there is such varation in collisions over time it is best to use a Negative Binomial Regression model. As they are good for count data, where events are discrete and can have large fluctuations over time and this is presented in our dataset.

1. Background on Negative Binomial Regression:
    - They handle overdispersion, or extra variance
    - They work well for count data, which is what our collisions dataset is made of
    - They can include outside factors, in our case, income, ridership, and population

2. We will predict our model off of our 3 month window in 2023 and it will be predicting 2024 collisions.

```python
# get dataframe and rename columns
all_data_df.rename(columns={"median_income(dollars)": "median_income"}, inplace=True)
all_data_df.rename(columns={"mean_income(dollars)": "mean_income"}, inplace=True) 

imputed_all_data_df = all_data_df.fillna(all_data_df.median(numeric_only=True))
```

Preparing the dataset for modeling by selecting relevant features (X) and target variable (Y). It standardizes the features using StandardScaler() to ensure consistent scaling. The dataset is split into 70% training and 30% testing using train_test_split(), with random_state=42 for reproducability.

```python
# input features and target variable
X = imputed_all_data_df[["median_income", "avg_ridership", "population"]]
y = imputed_all_data_df["total_collisions"]

# Standardize inputs
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 70% train 30% test
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)

# convert back into df
train_df = pd.DataFrame(X_train, columns=["median_income", "avg_ridership", "population"])
train_df["total_collisions"] = y_train.values
```

This is an implementation of a Negative Binomial Regression model to predict total collisions. The model summary indicates that median income and population have p-values less than 0.05, meaning that they are statistically significant predictors of total collisions. 

```python
# negative binomial regression model
nb_model = smf.glm(
    formula="total_collisions ~ median_income + avg_ridership + population",
    data=train_df,
    family=sm.families.NegativeBinomial()
).fit()

# run on test data
test_df = pd.DataFrame(X_test, columns=["median_income", "avg_ridership", "population"])
y_pred_nb = nb_model.predict(test_df) 

print(nb_model.summary())

# median income and population have p val less than 0.05
```

This is an implementation of a Neural Network with two hidden layers using ReLu activation function and an output layer with Softplus to ensure positive predictions. It compiles the model with the Adam optimizer and MSLE loss, then trains the model.

```python
# Neural Network Implementation

# nn model
model = Sequential([
    Dense(32, activation="relu", input_shape=(X_train.shape[1],)),
    Dense(16, activation="relu"),
    Dense(1, activation="softplus")  # ensures positive predictions
])

# using MSLE loss function
model.compile(optimizer=keras.optimizers.Adam(learning_rate=0.0005), loss=tf.keras.losses.MeanSquaredLogarithmicError(), metrics=["mae"])

# train
model.fit(X_train, y_train, epochs=100, batch_size=8, verbose=0)

# run on test data
y_pred_nn = model.predict(X_test).flatten() 
```

Here we are calculating the model metrics for both the Neurel Network model and the Negative Binomial Regression Model. Then we create a baseline model to have a reference point to compare the other two models to.

```python
# negative binomial regression model metrics
mae_nb = mean_absolute_error(y_test, y_pred_nb)
mse_nb = mean_squared_error(y_test, y_pred_nb)
rmse_nb = np.sqrt(mse_nb)
r2_nb = r2_score(y_test, y_pred_nb)

# nn model metrics
mae_nn = mean_absolute_error(y_test, y_pred_nn)
mse_nn = mean_squared_error(y_test, y_pred_nn)
rmse_nn = np.sqrt(mse_nn)
r2_nn = r2_score(y_test, y_pred_nn)

# baseline model (original data) metrics
y_baseline = np.full_like(y_test, y_train.mean())
mae_baseline = mean_absolute_error(y_test, y_baseline)
mse_baseline = mean_squared_error(y_test, y_baseline)
rmse_baseline = np.sqrt(mse_baseline)
r2_baseline = r2_score(y_test, y_baseline)

print("Model Performance Comparison:")
print(f"Baseline: MAE={mae_baseline:.2f}, MSE={mse_baseline:.2f}, R^2={r2_baseline:.4f}") 
print(f"Negative Binomial Regression: MAE={mae_nb:.2f}, MSE={mse_nb:.2f}, RMSE={rmse_nb:.2f}, R^2={r2_nb:.4f}")
print(f"Neural Network Model: MAE={mae_nn:.2f}, MSE={mse_nn:.2f}, RMSE={rmse_nn:.2f}, R^2={r2_nn:.4f}")
```

#### Model Findings: 
To analyze the relationship between income levels, population density, public transportation ridership, and motor vehicle collisions across New York City zip codes from September to November 2023, we developed two models: a Negative Binomial Regression (NBR) model and a Neural Network (NN) model. The NBR model outperformed the NN model, likely due to overfitting and the small training dataset.
