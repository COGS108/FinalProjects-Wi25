## Background and Prior Work

Zodiac signs: a topic often debated on whether or not they play any influences in one’s personality and actions. Widely known by many, the belief is based on twelve signs, which are each associated with its own characteristics and traits shared by the people who were born around the same time. Believers of this field say that these traits are based on where the position of celestial bodies are at the time of a person’s birth. While some people consider this as a meaningful tool, many also are skeptical and disbelieving of the idea. As such, given the popularity of the belief, we wanted to ask the question of whether astrological signs have any substantial influence on political behavior, particularly among leaders of the world. 

A study we explored was conducted by the Sunlight Foundation<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1), where they looked at data from the U.S. Congress. The researchers investigated whether the zodiac signs of Congress members correlated with their level of partisanship and collegiality. In order to do this, they collected and organized a bulk dataset with information regarding Congress members and their birthdates; they then categorized them by their Zodiac signs and compared this to the partisan and collegial affiliation. The conclusive data they got from their research showed that there is a direct causal relationship between astrology and legislative productivity. This study gave us a foundation to further explore the plausibility of whether astrological beliefs actually influence political decision making and behavior. While previous research on this topic has been limited to the Congressional members of the U.S., our proposed analysis seeks to widen the scope, examining what trends (if any) appear in leaders around the world. To do so, we will look at different leading positions (i.e. President, Prime Minister, King/Queen, Supreme Leader, etc.) in world countries listed in data by the CIA.

While we do not personally believe in the astrological beliefs of zodiac signs, we were intrigued by the novelty of this research and as such sought out additional information. One source that we found was from Acharya Ganesh, a popular astronomer in India, that informed us of common traits and characteristics associated with each zodiac sign.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2)  The astronomer also listed some notable leaders associated with each sign, and how the personality traits of their sign made them suited for their political leadership. This information gave us some casual insights on why people believe specific zodiac signs are suited for certain political roles. This raised important questions about whether people's zodiac signs could actually influence their leadership styles and political views.

Beyond the zodiac signs themselves, discussing the birth dates of political leaders also opens for a broader possibility about the age of leadership and its implications. In a 2015 study done by Navarro et al., relative age effects (RAE) were found to influence eighth grade students in Chile, affecting their academic performance in reading, mathematics, social studies, and science<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3). Students that were relatively younger than their peers tended to score lower than the relatively older ones, contributing to the idea that these age related effects play a role in performance. By keeping these age-related effects in mind, we can explore how zodiac signs associated with birthdays early in the year may be related to increased approval ratings among world leaders.

```python
# The code of the following function is partially generated by ChatGPT to help add inline comments to our functions and increase legibiility.

# Create Wiki Link from the Names
def create_link(names):

    # Dictionary to store special cases
    name_mapping = {
        'Gabriel BORIC Font': '/wiki/Gabriel_Boric',
        'Samuel Weymouth Tapley SEATON': '/wiki/Tapley_Seaton',
        'YOON Suk Yeol': '/wiki/Yoon_Suk_Yeol',
        'Vjosa OSMANI-Sadriu': '/wiki/Vjosa_Osmani',
        "TUIMALEALI'IFANO Va'aleto'a Eti Sualauvi II": '/wiki/Tuimalealiʻifano_Vaʻaletoʻa_Sualauvi_II',
        'FELIPE VI': '/wiki/Felipe_VI',
        'Kasym-Zhomart TOKAYEV': '/wiki/Kassym-Jomart_Tokayev',
        'MSWATI III': '/wiki/Mswati_III',
        'Muhammad Yunes MANFI': '/wiki/Muhammad_Yunus',
        'Abdelfattah ELSISI': '/wiki/Abdel_Fattah_el-Sisi',
        'Alexander VAN DER BELLEN': '/wiki/Alexander_Van_der_Bellen',
        'Mario ABDO BENITEZ': '/wiki/Mario_Abdo_Benítez',
        'HAMAD bin Isa Al Khalifa': '/wiki/Hamad_bin_Isa_Al_Khalifa',
        'Marlon OCHOA': '/wiki/Marlon_Ochoa',
        'HASSANAL Bolkiah,Sir': '/wiki/Hassanal_Bolkiah',
        'Denis SASSOU-Nguesso': '/wiki/Denis_Sassou_Nguesso',
        'Umaro Sissoco EMBALO': '/wiki/Umaro_Sissoco_Embaló',
        'Bashar al-ASAD': '/wiki/Bashar_al-Assad',
        'CARL XVI GUSTAF': '/wiki/Carl_XVI_Gustaf',
        'Michael Daniel HIGGINS': '/wiki/Michael_D._Higgins',
        'Daniel ORTEGA Saavedra': '/wiki/Daniel_Ortega',
        'David VUNAGI,Rev.': '/wiki/David_Vunagi',
        'Neville CENAC,Sir': '/wiki/Neville_Cenac',
        'TUPOU VI': '/wiki/Tupou_VI',
        'Laurentino CORTIZO Cohen': '/wiki/Laurentino_Cortizo',
        'ABDALLAH II': '/wiki/Abdullah_II_of_Jordan',
        'Yitzhak HERZOG': '/wiki/Isaac_Herzog',
        'Bob DADAE,Grand Chief Sir': '/wiki/Bob_Dadae',
        'Samia Suluhu HASSAN': '/wiki/Samia_Suluhu_Hassan',
        'Frank-Walter STEINMEIER': '/wiki/Frank-Walter_Steinmeier',
        'Tofiga Vaevalu FALANI,Rev.': '/wiki/Tofiga_Vaevalu_Falani',
        'SALMAN bin Abd al-Aziz Al Saud': '/wiki/Salman_of_Saudi_Arabia',
        'SAHLE-WORK Zewde': '/wiki/Sahle-Work_Zewde',
        'MUHAMMAD BIN ZAYID Al Nuhayyan': '/wiki/Mohamed_bin_Zayed_Al_Nahyan',
        'Nicolas MADURO Moros': '/wiki/Nicolás_Maduro',
        'HAYTHAM bin Tariq Al Said': '/wiki/Haitham_bin_Tariq',
        'Philip DAVIS': '/wiki/Philip_Davis_(Bahamian_politician)',
        'TAMIM bin Hamad Al Thani': '/wiki/Tamim_bin_Hamad_Al_Thani',
        'Abdul HAMID': '/wiki/Mohamad_Abdul_Hamid',
        'Judge Paula-Mae WEEKES (Ret.)': '/wiki/Paula-Mae_Weekes',
        'Mohamed Irfaan ALI,Brig. Gen. (Ret.)': '/wiki/Irfaan_Ali',
        'Luis ARCE Catacora': '/wiki/Luis_Arce',
        'Jose RAMOS-HORTA': '/wiki/José_Ramos-Horta',
        'Mark BROWN': '/wiki/Mark_Brown_(Cook_Islands)',
        'ISAIAS Afworki': '/wiki/Isaias_Afwerki',
        'Nana AKUFO-ADDO': '/wiki/Nana_Akufo-Addo',
        'Teodoro OBIANG Nguema Mbasogo,Brig. Gen. (Ret.)': '/wiki/Teodoro_Obiang_Nguema_Mbasogo',
        'ALBERT II,Prince': '/wiki/Albert_II,_Prince_of_Monaco',
        'Pedro CASTILLO Terrones': '/wiki/Pedro_Castillo',
        'Mohamed Ould Cheikh GHAZOUANI': '/wiki/Mohamed_Ould_Ghazouani',
        'Luis Alberto LACALLE POU': '/wiki/Luis_Lacalle_Pou',
        'KIM Jong Un': '/wiki/Kim_Jong_Un',
        "Al-Sultan ABDULLAH Ri'ayatuddin Al-Mustafa Billah Shah": '/wiki/Abdullah_of_Pahang',
        'Drupadi MURMU': '/wiki/Droupadi_Murmu',
        'Surangel WHIPPS, Jr.': '/wiki/Surangel_Whipps_Jr.',
        'MARGRETHE II': '/wiki/Margrethe_II',
        'MIN AUNG HLAING,Sr. Gen.': '/wiki/Min_Aung_Hlaing',
        'Andres Manuel LOPEZ OBRADOR': '/wiki/Andrés_Manuel_López_Obrador',
        'Gustavo PETRO Urrego': '/wiki/Gustavo_Petro',
        'Marcelo REBELO DE SOUSA': '/wiki/Marcelo_Rebelo_de_Sousa',
        'NAWAF al-Ahmad al-Jabir al-Sabah': '/wiki/Nawaf_Al-Ahmad_Al-Jaber_Al-Sabah',
        'Muhammadu BUHARI,Gen. (Ret.)': '/wiki/Muhammadu_Buhari',
        'Rashad Muhammad al-ALIMI,Dr.': '/wiki/Rashad_al-Alimi',
        'MOHAMMED VI': '/wiki/Mohammed_VI_of_Morocco',
        'Abd al-Fatah al-BURHAN Abd al-Rahman,Gen.': '/wiki/Abdel_Fattah_al-Burhan',
        'Salva KIIR Mayardit': '/wiki/Salva_Kiir_Mayardit',
        'Alcyion Cynthia KIRO': '/wiki/Cindy_Kiro',
        'Ali BONGO Ondimba': '/wiki/Ali_Bongo',
        'Arif ALVI': '/wiki/Arif_Alvi',
        'Rodrigo Alberto de Jesus CHAVES Robles': '/wiki/Rodrigo_Chaves_Robles',
        'Paul-Henri Sandaogo DAMIBA': '/wiki/Paul-Henri_Sandaogo_Damiba',
        'Idriss DEBY Itno,Lt. Gen.': '/wiki/Idriss_Déby',
        'Faustin-Archange TOUADERA': '/wiki/Faustin-Archange_Touadéra',
        'David John HURLEY,Gen (Ret.)': '/wiki/David_Hurley',
        'Miguel DIAZ-CANEL Bermudez': '/wiki/Miguel_Díaz-Canel',
        'Bidya Devi BHANDARI': '/wiki/Bidya_Devi_Bhandari',
        'LETSIE III': '/wiki/Letsie_III',
        'Ali Hoseini-KHAMENEI,Ayatollah': '/wiki/Ali_Khamenei',
        'HANS ADAM II,Prince': '/wiki/Hans-Adam_II,_Prince_of_Liechtenstein',
        'HAYBATULLAH Akhundzada': '/wiki/Hibatullah_Akhundzada',
        'Michael HIGGINS': '/wiki/Michael_D._Higgins',
        'Carlos Manuel VILA NOVA': '/wiki/Carlos_Vila_Nova',
        'Dina BOLUARTE Zegarra': '/wiki/Dina_Boluarte',
        'Mahamat Idriss DEBY Itno,Lt. Gen.': '/wiki/Mahamat_Déby',
        'Natasa PIRC MUSAR': '/wiki/Nataša_Pirc_Musar',
        'Raymond NDONG SIMA': '/wiki/Raymond_Ndong_Sima',
        'Daniel NOBOA Azin': '/wiki/Daniel_Noboa',
        'Cesar Bernardo AREVALO de Leon': '/wiki/Bernardo_Arévalo',
        'Santiago PENA Palacios': '/wiki/Santiago_Peña',
        'Mohammed SHAHABUDDIN Chuppu': '/wiki/Mohammed_Shahabuddin',
        'Sultan IBRAHIM Iskandar': '/wiki/Ibrahim_Iskandar_of_Johor',
        'Ram Chandra POUDEL': '/wiki/Ram_Chandra_Poudel',
        'Bassirou Diomaye FAYE': '/wiki/Bassirou_Diomaye_Faye',
        'Muhammad al-MANFI': '/wiki/Mohamed_al-Menfi',
        'Myriam Spiteri DEBONO': '/wiki/Myriam_Spiteri_Debono',
        'Abdourahmane TIANI,Brig. Gen.': '/wiki/Abdourahamane_Tchiani',
        'Transitional Presidential Council': '/wiki/Transitional_Presidential_Council',
        'HAITHAM bin Tarik Al Said': '/wiki/Haitham_bin_Tariq',
        'MISHAL al-Ahmad al-Jabir al-Sabah': '/wiki/Mishal_Al-Ahmad_Al-Jaber_Al-Sabah',
        'Gordana SILJANOVSKA-DAVKOVA': '/wiki/Gordana_Siljanovska-Davkova',
        'William Samoei RUTO,Dr.': '/wiki/William_Ruto',
        'Philip Edward DAVIS': '/wiki/Philip_Davis_(Bahamian_politician)',
        'Mohamed Irfaan ALI,Dr.': '/wiki/Irfaan_Ali',
        'Felix-Antoine TSHISEKEDI Tshilombohe': '/wiki/Félix_Tshisekedi',
        'Anura Kumara DISSANAYAKE (Acting)': '/wiki/Anura_Kumara_Dissanayake',
        'Iris Xiomara CASTRO de Zelaya': '/wiki/Xiomara_Castro',
        'Cyril CHARLES': '/wiki/Errol_Charles',
        'Jose Raul MULINO Quintero': '/wiki/José_Raúl_Mulino',
        'PRABOWO Subianto Djojohadikusumo': '/wiki/Prabowo_Subianto',
        'TAYE Atske Selassie': '/wiki/Taye_Atske_Selassie',
        'Mohamed MUIZZU,Dr.': '/wiki/Mohamed_Muizzu',
        'Claudia SHEINBAUM Pardo': '/wiki/Claudia_Sheinbaum',
        'Rodney WILLIAMS,Dr.': '/wiki/Rodney_Williams',
    }

    # Check if the name is in the special cases dictionary
    if names in name_mapping:
        return name_mapping[names]

    if type(names) == float:
        return None

    # Split the name into parts
    names_split = names.split(' ')

    # Standard formatting for two or more word names
    first_name = names_split[0].capitalize()
    last_name = names_split[-1].capitalize()

    # If the name has multiple words, format as "First_Last"
    if len(names_split) > 1:
        return f'/wiki/{first_name}_{last_name}'

    # If it's a single word, just return it
    return f'/wiki/{first_name}'


# The code of this function is partially generated by ChatGPT to help process errors and make function usage clearer.
# Find Birthday from Wikipedia
def get_ruler_birthday(wiki_link):
    """
    Extracts the birth date of a ruler from their Wikipedia page.
    
    Args:
        wiki_link (str): The Wikipedia URL of the ruler.
        
    Returns:
        str: The birth date if found, else None.
    """
    try:
        url = f"https://en.m.wikipedia.org{wiki_link}"
        req = requests.get(url)
        soup = BeautifulSoup(req.content, "html.parser")

        # Find the infobox table (usually class is 'infobox vcard')
        infobox = soup.find("table", class_="infobox vcard")

        if infobox:
            # Find the row that contains "Born"
            for row in infobox.find_all("tr"):
                header = row.find("th")
                if header and "Born" in header.text:
                    birth_date = row.find("td")
                    return birth_date.text.strip() if birth_date else None
    except:
        print('Error with link')
        print(f'link: https://en.m.wikipedia.org{wiki_link}')

    return None  # Return None if no birth date is found

# Grab all info between parentheses taken from wiki
def between_parentheses(text):
    return re.findall(r"\((.*?)\)", text)

# Birthdates will appear as the first values if there are multiple values that have parentheses
def grab_bday(processed_bday_info):
    if len(processed_bday_info) > 1:
        return processed_bday_info[0]
    return None

# Dictionary to turn month to month number
month_dict = {
    'january': '01',
    'february': '02',
    'march': '03',
    'april': '04',
    'may': '05',
    'june': '06',
    'july': '07',
    'august': '08',
    'september': '09',
    'october': '10',
    'november': '11',
    'december': '12'
}

# Function that takes month and year and produces dataframe from CIA data
def get_cia_country_names(year, month):
    data = []
    for country in cia_countries_set:
    
        country = country.strip()
        country_storage = country

        if country[0:3] == 'The':
            country = country[4:] + '-the'
        if "'" in country:
            country = country.replace("'", '')
        if '*' in country:
            country = country.replace('*', '')
        if '(' in country:
            country = country.replace('(', '')
        if ')' in country:
            country = country.replace(')', '')
        if " " in country:
            country = country.replace(' ', '-')
        country = country.lower()

        # Exceptions
        if country == 'democratic-republic-of-the-congo':
            country = 'congo-democratic-republic-of-the'
        elif country == 'republic-of-the-congo':
            country = 'congo-republic-of-the'
        elif country == 'czech-republic':
            country = 'czechia'
        elif country == 'south-korea':
            country = 'korea-south'
        elif country == 'north-korea':
            country = 'korea-north'
        elif country == 'federated-states-of-micronesia':
            country = 'micronesia-federated-states-of'


        #print(f"https://www.cia.gov/resources/world-leaders/historical-data/{year}-{month}/{country}/")

        try:
            cia_link = f"https://www.cia.gov/resources/world-leaders/historical-data/{year}-{month}/{country}/"
            req = requests.get(cia_link)
            soup = BeautifulSoup(req.content, "html.parser")

            position = soup.find('h4').get_text(strip=True)
            name = soup.find('p').get_text(strip=True)

            data.append([country_storage, name, position])

        except:
            print(country + '- Not Completed')

    temp_df = pd.DataFrame(data, columns=["Country", "Name", "Position"])

    temp_df['Link'] = temp_df['Name'].apply(create_link)

    temp_df['date'] = [f'{year}-{month_dict[month]}-01']*temp_df.shape[0]

    temp_df['Birthday Info'] = temp_df['Link'].apply(get_ruler_birthday)

    temp_df['Processing Birthday Info'] = temp_df['Birthday Info'].apply(str).apply(between_parentheses)
    temp_df['Birthday'] = temp_df['Processing Birthday Info'].apply(grab_bday)
    temp_df = temp_df.drop('Processing Birthday Info', axis=1)

    return temp_df

# Function to determine Zodiac sign
def Zodiac_Sign(birthday) -> str:
    if not isinstance(birthday, str):  # Ensure it's a string
        return "Invalid date"
    
    try:
        # Parse input date
        date = datetime.strptime(birthday.strip(), "%Y-%m-%d")  # Remove any extra spaces
        day, month = date.day, date.month
    except ValueError:
        return np.nan

    # Zodiac sign ranges
    zodiac_signs = [
        ("Capricorn", (1, 1), (1, 19)),
        ("Capricorn", (12, 22), (12, 31)),
        ("Aquarius", (1, 20), (2, 18)),
        ("Pisces", (2, 19), (3, 20)),
        ("Aries", (3, 21), (4, 19)),
        ("Taurus", (4, 20), (5, 20)),
        ("Gemini", (5, 21), (6, 20)),
        ("Cancer", (6, 21), (7, 22)),
        ("Leo", (7, 23), (8, 22)),
        ("Virgo", (8, 23), (9, 22)),
        ("Libra", (9, 23), (10, 22)),
        ("Scorpio", (10, 23), (11, 21)),
        ("Sagittarius", (11, 22), (12, 21))
    ]
    
    # Determine Zodiac sign
    for sign, start, end in zodiac_signs:
        if (month == start[0] and day >= start[1]) or (month == end[0] and day <= end[1]):
            return sign
    
    return None
    
```

```python
# Same get_ruler_birthday as before
def get_ruler_birthday(wiki_link):
    """
    Extracts the birth date of a ruler from their Wikipedia page.
    
    Args:
        wiki_link (str): The Wikipedia URL of the ruler.
        
    Returns:
        str: The birth date if found, else None.
    """
    try:
        url = f"https://en.m.wikipedia.org{wiki_link}"
        req = requests.get(url)
        soup = BeautifulSoup(req.content, "html.parser")

        # Find the infobox table (usually class is 'infobox vcard')
        infobox = soup.find("table", class_="infobox vcard")

        if infobox:
            # Find the row that contains "Born"
            for row in infobox.find_all("tr"):
                header = row.find("th")
                if header and "Born" in header.text:
                    birth_date = row.find("td")
                    return birth_date.text.strip() if birth_date else None
    except:
        print('Error with link')
        print(f'link: https://en.m.wikipedia.org{wiki_link}')

    return None  # Return None if no birth date is found

# Example usage:
hos_link = '/wiki/Charles_III'
birthday = get_ruler_birthday(hos_link)
print(f"Birthday: {birthday}")
```

To assess whether this deviation is statistically significant, we will apply a chi-square goodness-of-fit test, which compares the observed distribution to the expected equal distribution.

*Note: We chose for this figure and the next chi-square GOF test to analyze only the February 2025 data to avoid artificially inflating the sample size. Including data from previous months would likely introduce redundant observations, as leadership changes month-to-month are minimal. Repeatedly counting the same leaders could bias the results and overstate statistical significance. Additionally, we have chosen to compare our results to a uniform distribution.
