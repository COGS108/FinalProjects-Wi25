## Background and Prior Work
While regions with high photovoltaic (PV) potential are well positioned to generate solar energy, actual production often falls short due to economic and infrastructural limitations. Prior research has indicated that high income countries with moderate solar potential can outperform lower income, high potential nations. This is largely due to greater investment capacity and supporitive policy frameworks. These findings suggest that solar potential alone does not determine solar energy output; conversely, a country's economic resources, energy infrastructure, and policy environmental play fundemental roles. Understanding such factors yields essential perspectives for the barriers to solar adoption and the promotion for more equitable and efficient renewable energy strategies. The subsequent central themes emerge from existing studies:

1. Discrepancies Between UV Index & Solar Output: 

    High UV index regions - particularly portions of Africa and South America - typically underperform in solar generation (output). This dictation of underperformance is primarily due to economic constraints, inadequate infrastructure, and limited access to investment capital. Comparatively - countries like Germany and Japan - with moderate UV indices have results indicating significant solar energy outputs. Such success is largely attributed to continued policy support, technological innovation, and investments in renewable energy infrastructure. It can be observed that countries like Japan have invested approximately $1.5 billion into the development of ultra-thin, flexible perovskite solar panels in means of reducing dependency on fossil fuels, directly challenging China’s dominance in renewable energy markets<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1).

2. Economic Factors & Solar Energy:

    A country’s economic status substantially influences its capacity to invest in solar energy. For instance, it can be claimed that high-income countries often have the financial resources to support large-scale solar projects, along with subsidized research and development. These cases however, are notable exceptions. Affluent countries have been seen to have abundant fossil fuel resources but underinvest in solar energy due to existing energy dependence and economic interest. Conversely, emerging economies with much lower GDP per capita - such as Vietnam and India - have made great strides in solar energy adoption. These advancements are often driven by strategic governmental policies, international partnerships, and a pressing need to exceed or meet growing energy demands in a sustainable manner. As previously mentioned, China is a prime example of a nation investing significant amounts into renewable energy positioning them as a leader in environmental transition; despite being the world’s largest CO2 emitter   <a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2). 

3. Policy & Infrastructure Development:

    Policy frameworks and infrastructure development are fundamental determinants of solar generation. Countries that have implemented supportive policies (i.e. feed-in tariffs, tax incentives, and renewable energy mandates) have experienced accelerated growth in solar installations. The International Energy Agency (IEA) reports that power sector investment in solar photovoltaic (PV) technology is projected to exceed USD 500 billion in 2024, surpassing all other generation sources combined <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3). Furthermore, modernizing electrical grids to accommodate the variability of solar energy is pivotal for integrating renewables into national energy systems. The IEA even highlights that recent progress has been a great success, with 2023 being a record year for renewable electricity capacity growth, adding about 560 GW <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3). 

To conclude, while a high UV index provides a natural advantage for solar energy generation, the compounding interplay of factors (such as economic resources, policy decisions, and infrastructure readiness) contribute a decisive role in actual solar production levels. Understanding such factors permits the formation of strategies to enhance solar energy adoption globally.

# Functions

To find countries in each dataset that do not match our list of countries above, we will create a function to find the mismatched and print the names out. Then, we can either change the names of the countries or delete the regions. 

```python
def find_unmatched_countries(df, country_column, country_list):
    dataset_countries = set(df[country_column].unique())
    country_list_set = set(country_list)
    
    unmatched_countries = dataset_countries - country_list_set 
    return unmatched_countries

```

Before further analyzing data, a few major outliers could skew our results. To account for this, we will try normalizing the data using a method called Yeo-Johnson from Power Transformations. 

Normalizing the data often helps when the scales of different columns vary widely — like here, where GDP per capita, energy demand, and solar generation are on very different scales. By bringing them all to a similar range (often between 0 and 1 or with a mean of 0 and standard deviation of 1), we avoid certain columns dominating models and make patterns clearer in visualizations.

```python
normalized_data = merged_df.copy(deep=True)

numeric_cols = merged_df.select_dtypes(include=['number']).columns

for col in numeric_cols:
    pt = PowerTransformer(method='yeo-johnson')
    data = normalized_data[col].to_numpy().reshape(-1, 1)
    pt.fit(data)
    normalized_data[col] = pt.transform(data)

normalized_data.head()
```

Takeaway: The model has a weak fit, meaning other factors (e.g., industrialization, total population, energy policies) likely influence energy demand.
