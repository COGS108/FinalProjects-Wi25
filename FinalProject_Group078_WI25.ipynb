## Background and Prior Work

Technology and screen time have frequently been a hot topic in regards to mental health. As the world grows more digital, our technological reliance ever increases; work, entertainment, and social interaction seem to all be coupled in a magic screen. As such, the potential psychological effects of prolonged exposure necessitate further investigation. Many researchers debate if there is any correlation at all. So many influencing factors exist for anxiety, depression, or of the like, such that it's difficult to pinpoint exactly the cause. Some researchers document that screen time is associated with decreased levels of mental well-being, as they found individuals with higher screen time reported less positive feelings. <a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) Further studies state that due to various confounding factors, screen time might be more of a symptom when compared to typical lifestyle factors like sleep or exercise.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4)

Examining individuals and their screen time allows for a deeper understanding of the developmental implications of mental health outcomes, providing tools for studying this issue as a whole. A longitudinal study done by researchers looking at variance in screen time and mental well-being divided screen time into two qualitative categories: recreational (referring to media) and non-recreational (referring to academia). Factors of physical activity and mental health outcomes by self-survey were used to establish findings of negative associations with recreational use and mental decline, and no associations with non-recreational use and mental decline<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1). Could the issue stem not from screen time, but rather specific apps? Certain applications promote the notion of “doom scrolling,” where algorithms invoke compulsive behavior to increase interaction. This oftentimes results in poorer reported mental health.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) There are productivity and mindfulness apps that intend to support the user, so it's imperative to account for positive effects as well. Some studies found some promise in apps having a positive correlation when used correctly <a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2). Overall, due to the rapid fast-paced nature of technology and consumerism advancements, it's critical to understand the multifaceted psychological impacts.

### Section 2 - Correlations between Screen Time and Depression Frequency

With distributions analyzed, we can now move on to checking for correlation between screen time and depression frequency.

```python
#Create summary table grouping by age
age_bins = [18, 25, 35, 45, 55, 65]
age_labels = ["18-24", "25-34", "35-44", "45-54", "55-64"]

df_phone.loc[:, "Age_Group"] = pd.cut(df_phone["Age"], bins=age_bins, labels=age_labels, right=False)
df_mental.loc[:, "Age_Group"] = pd.cut(df_mental["age"], bins=age_bins, labels=age_labels, right=False)

#Find average screen time and depression frequency per grouping
screen_time_avg = df_phone.groupby("Age_Group", observed=False)["Daily_Screen_Time_Hours"].mean()
depression_avg = df_mental.groupby("Age_Group", observed=False)["depression_frequency"].mean()

df_grouped = pd.DataFrame({"Avg_Screen_Time": screen_time_avg, "Avg_Depression_Frequency": depression_avg}).reset_index()
df_grouped = df_grouped.dropna()
```

```python
print(df_grouped)
```

```python
#Computing Pearson and Spearman correlation between screen time and depression frequency 
p_corr, _ = pearsonr(df_grouped["Avg_Screen_Time"], df_grouped["Avg_Depression_Frequency"])

print(f"Pearsonr correlation: {p_corr}")

s_corr, _ = spearmanr(df_grouped["Avg_Screen_Time"], df_grouped["Avg_Depression_Frequency"])
print(f"Spearmanr correlation: {s_corr}")
```

```python
#Visualizing trend
sns.regplot(x=df_grouped["Avg_Screen_Time"], 
            y=df_grouped["Avg_Depression_Frequency"],
            ci=None
            )
plt.xlabel("Avg_Screen_Time")
plt.ylabel("Avg_Depression_Frequency")
```

The linear regression line shows a positive correlation, where the upward trend aligns with the positive pearson correlation score. As average screen time increases, average depression frequency also increases.

The data points aren't perfectly aligned, which also coincides with the small spearman score. This indicates that the relationship is not very strong.

In summary, higher screen time (3.5+ hours) is linked to higher depression across age groups, but the effect isn't uniform. 


### Section 3 - Comparison across App Categories

To answer our second hypothesis about app categories, we will compare depression frequency against social media, gaming, and productivity usage.

```python
#Aggregate data by age group for comparison 
screen_time_apps = df_phone.groupby("Age_Group", observed=False)[["Social_Media_Usage_Hours",
                                                                  "Gaming_App_Usage_Hours",
                                                                  "Productivity_App_Usage_Hours"
                                                                  ]].mean()

depression_avg = df_mental.groupby("Age_Group", observed=False)["depression_frequency"].mean()
df_apps_grouped = screen_time_apps.merge(depression_avg, on="Age_Group")
```

```python
#Find correlations for each app category at the group level
correlation_results = {}

for app in ["Social_Media_Usage_Hours", "Gaming_App_Usage_Hours", "Productivity_App_Usage_Hours"]:
    pearson_corr, _ = pearsonr(df_apps_grouped[app], df_apps_grouped["depression_frequency"])
    spearman_corr, _ = spearmanr(df_apps_grouped[app], df_apps_grouped["depression_frequency"])
    correlation_results[app] = {"Pearson": pearson_corr, "Spearman": spearman_corr}


df_corr_results = pd.DataFrame.from_dict(correlation_results, orient="index")
print(df_corr_results)
```

```python
#Loop to graph each category. Each point is one age group (18-24, 25-34, etc)
for app in ["Social_Media_Usage_Hours", "Gaming_App_Usage_Hours", "Productivity_App_Usage_Hours"]:
    plt.figure(figsize=(8,5))
    sns.regplot(x=df_apps_grouped[app], 
                y=df_apps_grouped["depression_frequency"], 
                scatter_kws={"s": 100}, 
                line_kws={"color": "red"}, 
                ci=None)
    plt.xlabel(f"{app.replace('_', ' ')} (Hours)")
    plt.ylabel("Depression Frequency (1-5)")
    plt.title(f"{app.replace('_', ' ')} vs. Depression Frequency")
    plt.show()
```

Based on the linear regression lines plotted above, the summaries for each app category are the following:

Social: Weak negative pearson score (~-0.1) indicates weak negative correlation/little direct relationship. Low spearman score (-0.3) suggests weak ranked trend.

Gaming: Moderate negative pearson score (~-0.4) indicates more gaming is associated with lower depression. Moderate negative spearman score (-0.4) suggests clearer ranked trend, where depression frequency lowers with gaming time. 

Productivity: Almost no pearson correlation (~0.02), similar with spearman result (0.2). This suggests that productivity apps likely don't have a strong impact on mental health.

```python
#Geolocator object for getting coords
geolocator = Nominatim(user_agent="geo_mapper")
```

```python
#Helper function to find each location coords
def coords(city):
    try:
        location = geolocator.geocode(city, timeout=10)
        if location:
            return pd.Series([location.latitude, location.longitude])
        else:
            return pd.Series([None, None])
    except:
        return pd.Series([None, None])
```

#### Linear Regression

With a proposed goal to try to predict depression frequency based on app usage, linear regression may be a good starting point (despite dataset limitations).

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.metrics import mean_absolute_error
```

```python
#Selecting our independent and dependent variables
X = df_apps_grouped[["Social_Media_Usage_Hours", "Gaming_App_Usage_Hours", "Productivity_App_Usage_Hours"]]
y = df_apps_grouped["depression_frequency"]

print(X)

#Splitting data into 60% training and 40% testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=42)
```

```python
#Checking the variance of our variables
print(f"Y variance: { y_test.var()}")
print(f"X variance: { X_test.var()}")
```

```python
#Initialize, train, and have the model make predictions
model = LinearRegression()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
#Scatter plot of actual vs predicted depression frequency
plt.scatter(y_test, y_pred, alpha=0.6, label="Predicted vs Actual")
plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], linestyle='dashed', color='red', label="Ideal Fit Line")  # Perfect fit line
plt.xlabel("Actual Depression Frequency")
plt.ylabel("Predicted Depression Frequency")
plt.title("Linear Regression: Actual vs Predicted Values")
plt.legend()
plt.show()
```

Here we see that linear regression struggles to correctly predict due to non-linear nature of correlation. This could be indicative of an influence of multiple factors on depression frequency.

```python
#Model evaluation to understand issue
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
```

```python
print(f"Mean Squared Error: {mse}")
print(f"r^2 score: {r2}")
```

A negative r squared score indicates that linear regression isnt fitting the data well.

```python
print(f"Training size: {len(X_train)}, Test size: {len(X_test)}")
print("Total Dataset Size:", len(X))
```

#### Random Forest

The Random Forest model may prove much more adequate due to its handling of small datasets and non-linear relationships. 

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
```

```python
#Fresh dataset with all features
df_model = pd.read_csv('data/mental_health.csv')
df_model.head()
```

```python
import re

#Helper function to deal with string format of average screen time
def extract_lower_bound(time_str):
    if pd.isna(time_str):
        return None  # Handle missing values
    if "less than" in time_str:
        return 1.0  # Assume 1 hours
    elif "more than" in time_str:
        return 5.0  # Assume at least 5 hours
    else:
        match = re.search(r'\d+', time_str)  # Extract first number found
        return float(match.group()) if match else None
```

```python
#Picking features to be used for depression_frequency prediction

X = df_model[["average_social_media_time", "social_media_without_purpose", 
              "easily_distracted", "bothered_by_worries", "difficulty_concentrating"]]


y = df_model["depression_frequency"]


#Splitting data into 70% training and 30% testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

#### Gradient Boosting


Due to the issues from other models, gradient boosting's ability to sequentially improve itself may prove the best to capture the complex relationship from a small datasize. 

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
#Drop NaN values
X_train = X_train.dropna()
y_train = y_train.loc[X_train.index]
X_test = X_test.dropna()
y_test = y_test.loc[X_test.index]

X_train.shape
y_test.shape
```

```python
#Initializing Gradient Boosting model and attempt to fine tune the parameters for our data
gb = GradientBoostingRegressor(
    n_estimators=500,  # Increase trees
    learning_rate=0.05,  # Smaller step size
    max_depth=3,  # Control model complexity
    subsample=0.8,  # Helps prevent overfitting
    random_state=42
)


#Training model and predict test values
gb.fit(X_train, y_train)
y_pred_gb = gb.predict(X_test)
```

```python
#Evaluating model
mae_gb = mean_absolute_error(y_test, y_pred_gb)
print(f"Tuned MAE: {mae_gb}")
```

# Ethics & Privacy

One potential bias that needs to be addressed would be sampling, as it's important to have a solid representation of the population we want to examine. More specifically would be undercoverage as the data could be skewed if for example younger individuals tend to report, while older people might opt out. More generally would be the inherent nature of self reporting studies. People may overestimate their screen time or have their own reasons for purposely misreporting. 

In order to alleviate some of these potential issues, we could use a preliminary EDA to ensure no imbalances between groups. A longitudinal study could greatly mitigate obstacles through tracking participants over time. For proper representation, a diverse range of participants will help with socioeconomic differences. Though not able to eliminate the bias entirely, informed consent may aid in navigating overestimations and misreporting by promoting clarity and genuinity in what information the longitudinal study is seeking. 

# Discusison and Conclusion

Our analysis helps refine the understanding of screen time’s relationship with mental health by breaking down app usage by category and age group. The data suggests that prolonged exposure to social media is correlated with higher levels of stress and anxiety, particularly among younger users. Regression analyses help indicate a strong positive correlation between social media usage and self-reported depression frequency, supporting previous studies that highlight the role of social comparison and algorithm-driven content in influencing mental health. 

In contrast, gaming applications show a more nuanced relationship. While high gaming hours correlate with slightly higher depression scores, the trend is weaker compared to social media. This suggests that gaming may have mixed effects depending on the individual and context, with some users potentially benefiting from gaming as a stress reliever. 

Though a different trend emerges when examining productivity applications, which display a neutral to slightly positive impact on mental well-being. Individuals who engage more with productivity and mindfulness applications report better mental health scores, suggesting that intentional and goal-oriented app usage may contribute to well-being. 

By including the age-based analysis we can see that younger individuals (ages 18-24) tend to experience stronger negative correlations between screen time and mental health compared to older age groups. Indicating that as individuals age, their digital habits and coping mechanisms adapt, which may influence their resilience to the negative effects of screen time. 

These findings align with previous research that highlights the role of digital habits in shaping mental well-being. The nuance in our results suggest that blanket recommendations to reduce screen time may not be as effective as promoting mindful engagement with technology.  So, instead of simply limiting screen time, interventions should focus on encouraging balanced and intentional usage of different types of applications. 

Building on these insights, it is important to acknowledge we had limitations in our study, including potential self-reporting biases, as participants’ screen time and mental health data were based on personal assessments. Additionally, our study does not establish causation  though correlations are evident, external factors such as lifestyle, personality, and offline social interactions may also contribute to the observed effects. Future research could benefit from longitudinal studies that track screen time and mental health metrics over an extended period, as well as experimental studies that assess the direct impact of specific app usage patterns on emotional well-being. 

In conclusion, this study underscores the importance of distinguishing between different types of screen time when evaluating their effects on mental health. The evidence from our statistical analyses suggests that social media usage is more strongly linked to negative health indicators, whereas productivity apps may mitigate some of these effects. The inclusion of age-group analysis also suggests that younger users may be more vulnerable to screen time’s negative effects. By fostering awareness and promoting healthier digital habits, individuals can mitigate the potential negative impact of excessive screen use while maximizing the benefits of technology for mental well-being.
