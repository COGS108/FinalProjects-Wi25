# Abstract

```python
def remove_after_whitespace(text):
    # Ensure the text is a string before processing
    if isinstance(text, str) and ' ' in text:
        return text.split(' ')[0]
    return text  # Return the original text (or NaN) if no whitespace is found or it's not a string
```

Split the google trendes into quarters. This way, the "Term" column allows us to be in sync with our SETs/CAPES dataset when we plot all the data versus time. This makes our datasets tidy.
```python
df_filtered['Term'] = df_filtered['Term'].replace('S321', 'S121')
gradedf_unique_terms = df_filtered['Term'].unique()
print(gradedf_unique_terms)
```

```python
def assign_quarter(date):
    year = date.year
    month = date.month
    day = date.day
    
    # Winter Quarter: January to February (early March for some)
    if month in [1, 2] or (month == 3 and day <= 24):
        return f"WI{str(year)[-2:]}"
    
    # Spring Quarter: April to early June
    elif month in [4, 5] or (month == 6 and day < 16):
        return f"SP{str(year)[-2:]}"
    
    # Summer Quarter: June to August (split into S1, S2)
    elif month in [6, 7, 8] or (month == 9 and day < 16):
        if month == 6:
            return f"S1{str(year)[-2:]}"  # S1 starts in June
        else:
            return f"S2{str(year)[-2:]}"  # S2 starts in August
    
    # Fall Quarter: September to early December
    elif month in [9, 10, 11] or (month == 12 and day < 15):
        return f"FA{str(year)[-2:]}"
    
    return None

google_df["Quarter"] = google_df["Week"].apply(assign_quarter)

google_df = google_df.dropna(subset=["Quarter"])

google_df = google_df.groupby("Quarter")["chatgpt: (United States)"].mean().reset_index()

google_df.rename(columns={'Quarter': 'Term'}, inplace=True)

display(google_df)
```

```python
# List of all courses to be included in the analysis
courses_of_interest = [
   "MATH 10A - Calculus I", "MATH 10B - Calculus II", "MATH 10C - Calculus III", "MATH 18 - Linear Algebra",
   "MATH 20A - Calculus/Science & Engineering", "MATH 20B - Calculus/Science & Engineering",
   "COGS 1 - Introduction to Cognitive Sci", "COGS 10 - Cognitv Consequence/Technology", "DSGN 1 - Design of Everyday Things",
   "COGS 13 - Fld Methods: Cognition in Wild", "COGS 14A - Intro. to Research Methods", "COGS 14B - Intro. to Statistical Analysis",
   "COGS 17 - Neurobiology of Cognition", "COGS 18 - Introduction to Python", "COGS 100 - Cyborgs Now and in the Future",
   "COGS 101A - Sensation and Perception", "COGS 101B - Learning, Memory and Attention", "COGS 101C - Language",
   "COGS 107A - Neuroanatomy and Physiology", "COGS 107B - Systems Neuroscience", "COGS 107C - Cognitive Neuroscience",
   "COGS 108 - Data Science in Practice", "BILD 62 - Intro to Python for Biologists", "CSE 6R - Intr/CompSci&Obj-Ori:Python",
   "CSE 8A - Intro to Programming 1", "CSE 11 - Accel. Intro to Programming"
]

# Filters the classes from the merged dataframe based on courses of interest
filtered_courses_df = df_filtered[df_filtered['Course'].str.contains('|'.join(courses_of_interest), na=False)].copy()

# Removes sections from the course names
filtered_courses_df.loc[:, 'Course'] = filtered_courses_df['Course'].str.split('(' , n=1).str[0]

# Group by Term and Course, then calculates the average grade received
avg_grade_per_term = filtered_courses_df.groupby(['Term', 'Course']).agg({
   'Avg Grade Received': 'mean'
}).reset_index()

during_covid_terms = ['SP20', 'S120', 'S220', 'FA20', 'WI21', 'SP21', 'S121', 'S221', 'FA21']
pre_chatgpt_post_covid = ['WI22', 'SP22', 'S122', 'S222']

# Categorize terms
avg_grade_per_term['Category'] = avg_grade_per_term['Term'].apply(
    lambda x: 'During COVID' if x in during_covid_terms
    else 'Pre-ChatGPT' if int(x[-2:]) < 20 or (int(x[-2:]) == 20 and x[:2] == 'WI') or x in pre_chatgpt_post_covid
    else 'Post-ChatGPT' if int(x[-2:]) > 22 or (int(x[-2:]) == 22 and x[:2] in ['FA', 'WI']) 
    else 'Other'
)

# Filter out 'During COVID' to focus only on Pre- and Post-ChatGPT
filtered_df = avg_grade_per_term[avg_grade_per_term['Category'].isin(['Pre-ChatGPT', 'Post-ChatGPT'])]
avg_grade_by_course = filtered_df.groupby(['Course', 'Category']).agg({
    'Avg Grade Received': 'mean'
}).reset_index()

# Display results
display(avg_grade_by_course)
```

```python
def gpa_to_letter(grade):
    if grade == 4.0: return 'A'  # A+ and A both map to 4.0
    elif grade >= 3.7: return 'A-'
    elif grade >= 3.3: return 'B+'
    elif grade >= 3.0: return 'B'
    elif grade >= 2.7: return 'B-'
    elif grade >= 2.3: return 'C+'
    elif grade >= 2.0: return 'C'
    elif grade >= 1.7: return 'C-'
    elif grade >= 1.0: return 'D'
    else: return 'F'

# Apply function to convert GPA to letter grades
avg_grade_by_course['Letter Grade'] = avg_grade_by_course['Avg Grade Received'].apply(gpa_to_letter)
```

### Pearson's Correlation

To answer our question of whether or not grades in cognitive science major requirement classes have changed after the release of ChatGPT, we can use Pearson's correlation to determine if grades and ChatGPT popularity are correlated. The Pearson's Coefficient, r, ranges from -1 to 1 and measures the strength and the direction of the correlation. Since we hypothesized that the increased usage of ChatGPT should be correlated to an increase in grades, we expect to see a positive correlation here, r > 0.

```python
# First we want to remove all the quarters before Fall 2022, which was when ChatGPT was released
# We also want to remove all the recent quarters that we don't have grade data for yet
avg_grades_per_term_df = avg_grades_per_term.loc[[26, 29, 30, 31, 34, 35, 36,39]] 
display(avg_grades_per_term_df)
```

Analyzing the graph, we can see that there aren't that many data points to work with. Each point represents a quarter and its corresponding ChatGPT search volume and the average of all the average grades received in that quarter. We can see that there actually seems to be a negative correlation such that as ChatGPT search volume increases, average grade decreases. This is the opposite of what we initially hypothesized. Now we can calculate the pearson coeeficient to see if our analysis of this graph is correct.

```python
from scipy.stats import pearsonr

corr, p_value = pearsonr(avg_grades_per_term_df["chatgpt: (United States)"], avg_grades_per_term_df["Avg Grade Received"])
print(f"Pearson's correlation coefficient: {corr}")
print(f"P-value: {p_value}")
```

From the results of the Pearson's coeffiecient calculation, there does seems to have a moderate negative correlation between ChatGPT search volume and average grade received. This means that as ChatGPT search volume increases, average grades tend to decrease. However, the p-value is quite high.

```python
if p_value < 0.01:
    print("There is a significant difference!")
else:
    print("There is not a significant difference!")
```

The p-value is quite high, which tells us that this correlation might have occurred by chance. There's not enough statistical evidence to conclude that ChatGPT search volume and grades are truly related in a meaningful way. This disproves our initial analysis of the graph that showed a slight downward trend or relationship between the two variables. The results of the Pearson's coeffiecient also disproves our hypothesis as well since we believed there would be a relationship between the two variables, but there doesn't seem to be one.

### T-Test

T-tests are used to determine whether there is a statistically significant difference between the means of two groups. By comparing the average grade received between Pre-ChatGPT and Post-ChatGPT, we can see if there is an increase in grades like we've observed in our graphs and whether or not that increase happened due to chance alone.

```python
import patsy
import statsmodels.api as sm
import scipy.stats as stats
from scipy.stats import ttest_ind, chisquare, normaltest
```

# Discusison and Conclusion

The assumption that ChatGPT's search volume shown on Google Trends accurately reflects its use among UCSD students for academic purposes is one of the main limitations of our analysis. Search interest does not directly represent how students use ChatGPT in their coursework, but it can be a good indicator of engagement. Some students may use ChatGPT regularly for other purposes that are non-school related, while others might use it solely for school-related purposes. We cannot directly track whether or not a person or student is using ChatGPT for academic purposes alone through Google Trends. The measure of ChatGPT interest reflects on popularity and overall search activity rather than focusing solely on academic usage since Google Trends normalizes search volume based on overall search activity. It is overall more difficult to directly compare ChatGPT search trends with grade trends as a result of this normalization.

Our project focuses on UCSD students enrolled in cognitive science major requirement courses, utilizing datasets from UCSD CAPES and SETs. While our dataset spans multiple years and courses, its generalizability is limited. Grading policies, instructor differences, and course difficulty vary across institutions, meaning our conclusions may not be applicable to every university. Even within UCSD, variations in course structure and instruction may influence grading trends, further complicating a direct analysis of ChatGPT’s impact on academic performance.

Another key consideration is the presence of many confounding variables that our data collection and analysis could not fully control. While AI has revolutionized productivity by automating certain tasks, academic course structure may have also evolved with it. Many professors have likely adjusted their course policies in response to AI's growing prevalence, which may explain the minimal fluctuations in average grades received. Additionally, while ChatGPT's usage is widespread, students utilize AI in different ways (and some may not use it at all). This variation in usage is a factor that our sample cannot account for. 

Given these considerations, we fail to reject the null hypothesis (H₀). The t-test results indicated that the mean difference in grades before and after ChatGPT’s release was not statistically significant, suggesting that any observed differences in grades likely occurred by chance rather than as a direct result of ChatGPT usage. While we initially hypothesized that ChatGPT would contribute to grade inflation, our findings suggest that other factors—such as grading policies, course difficulty, or shifts in study habits—were likely more influential. Future research could examine longer-term trends, specific student behaviors, or qualitative insights into how students use AI for academic work.

Overall, the data suggests that the initial spike in grades was more attributable to the COVID-19 pandemic and remote learning rather than the availability of ChatGPT. Although some variations in grade distributions were observed post-ChatGPT, our statistical tests did not find any significant increase or decrease in grades following its release. Contrary to our expectations, the Pearson correlation coefficient even indicated a slight negative relationship between ChatGPT search interest and average grades, though this effect was not statistically significant. These results suggest that while AI tools like ChatGPT may influence how students learn, they have not directly contributed to higher grades in cognitive science courses.
